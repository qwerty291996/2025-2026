## 4. Heatmap — Issiqlik Xaritasi

**Nazariy asos:** Heatmap — ikki o'lchovli matritsa ma'lumotlarini rang intensivligi orqali ko'rsatadi. Inson ko'zi rang farqini raqam farqidan 10x tezroq idrok etadi.

**Qachon ishlatiladi:**
```
Korrelyatsiya matritsasi  → qaysi o'zgaruvchilar bog'liq?
Vaqt × Kategoriya         → qaysi kunda qaysi tovar ko'p sotilgan?
Geografik ma'lumot        → qaysi hududda ko'rsatkich yuqori?
Confusion matrix (ML)     → model qaysi klassni adashtirmoqda?
```

**Korrelyatsiya Heatmap — tushuntirish:**
```
Korrelyatsiya koeffitsienti:
  +1.0  → Mukammal musbat bog'liqlik
  +0.7  → Kuchli musbat bog'liqlik
  +0.3  → Zaif musbat bog'liqlik
   0.0  → Bog'liqlik yo'q
  -0.3  → Zaif manfiy bog'liqlik
  -0.7  → Kuchli manfiy bog'liqlik
  -1.0  → Mukammal manfiy bog'liqlik

Heatmap da:
  To'q qizil  → +1.0 ga yaqin
  Oq          →  0.0 ga yaqin
  To'q ko'k   → -1.0 ga yaqin
```

**cmap tanlash falsafasi:**
```
"coolwarm"  → Diverging (ikki qarama-qarshi tomon) — korrelyatsiya uchun
"viridis"   → Sequential (bir yo'nalish) — davomiy qiymatlar uchun
"Blues"     → Sequential, bir rang — oddiy taqsimot uchun
"Set2"      → Categorical (kategoriyalar) — guruhlar uchun
```

## 5. Pairplot — Ko'p O'zgaruvchili Tahlil

**Nazariy asos:** n ta o'zgaruvchi uchun n×n grafik matritsasi yaratadi. Diagonal → har o'zgaruvchining taqsimoti, diagonal emas → ikki o'zgaruvchi orasidagi munosabat.

```
Ma'lumotlarni dastlabki o'rganish (EDA — Exploratory Data Analysis)
Qaysi o'zgaruvchilar korrelyatsiyaga ega — vizual aniqlash
Guruhlar orasidagi farqni ko'rish (hue parametri bilan)
Machine Learning uchun feature tanlash
```

**Pairplot nima ko'rsatadi:**
```
Diagonal (scatter emas):
  → Har o'zgaruvchining yolg'iz taqsimoti
  → KDE yoki histogram

Diagonal emas (scatter):
  → Chap pastga → o'ng yuqoriga: musbat korrelyatsiya
  → Aniq klasterlar: ma'lumotda guruhlar bor
  → Kuchli naqsh yo'q: o'zgaruvchilar mustaqil
```

## 6. Boxplot — Statistik Taqqoslash

**Nazariy asos:** Boxplot (quti grafik) — beshta statistik ko'rsatkichni bir vaqtda ko'rsatadi. John Tukey (1969) tomonidan ixtiro qilingan.

```
Boxplot anatomiyasi:

          ┌─────────────────┐
     ──── │ Yuqori kvartil  │ Q3 (75-persentil)
          │                 │
  ─────── │    Mediana      │ Q2 (50-persentil)
          │                 │
     ──── │ Quyi kvartil    │ Q1 (25-persentil)
          └─────────────────┘
               ↑
           IQR = Q3 - Q1  (Interquartile Range)

Whisker (mo'ylov):
  Yuqori: Q3 + 1.5 × IQR
  Quyi:   Q1 - 1.5 × IQR

Whiskerdan tashqari nuqtalar → ANOMALIYALAR (outliers)
```

**Boxplot nimani ko'rsatadi:**
```
Quti balandligi → Ma'lumot tarqoqligi (keng = ko'p tarqoq)
Mediana joylashuvi → Nosimmetriklik (chapga/o'ngga og'ish)
Whisker uzunligi → Dumlar og'irligi
Outlier nuqtalar → Anomal qiymatlar
```

**Violin plot — Boxplot ning kuchli varianti:**
```
Boxplot:       Faqat 5 statistik ko'rsatkich
Violin plot:   Boxplot + taqsimot shakli (KDE)

Violin plot qachon afzal:
  → Taqsimot bir cho'qqili yoki ko'p cho'qqilimi — ko'rish kerak bo'lsa
  → Guruhlar ichki tuzilishini solishtirish kerak bo'lsa
```

```python
# Boxplot:
sns.boxplot(
    data=df,
    x="kategoriya",
    y="qiymat",
    hue="guruh",       # Rang bo'yicha ajratish
    palette="Set2",    # Rang palatrasi
    width=0.6,         # Quti kengligi
    flierprops={"marker": "o", "markersize": 4}  # Outlier ko'rinishi
)

# Violin plot:
sns.violinplot(
    data=df,
    x="kategoriya",
    y="qiymat",
    inner="box",       # Ichida: "box", "quart", "point"
    split=True,        # hue bo'lsa: ikki tomonga bo'lish
)
```

## 1. Chiziqli Regressiya — Nazariy Asos

Chiziqli regressiya — bir yoki bir nechta **mustaqil o'zgaruvchilar** (X) asosida **bog'liq o'zgaruvchini** (Y) bashorat qilish usuli. Asosiy g'oya: ma'lumotlar orasidan **eng yaxshi to'g'ri chiziq** topish.

```
Hayotiy misollar:
  X = ta'lim yillari   → Y = oylik maosh
  X = uy maydoni (m²)  → Y = uy narxi
  X = reklama xarajati → Y = sotuvlar hajmi
  X = temperatura      → Y = elektr sarfi
```

**Matematik shakl:**

```
Oddiy chiziqli regressiya (1 ta X):
  Y = β₀ + β₁·X + ε

Ko'p o'zgaruvchili (n ta X):
  Y = β₀ + β₁·X₁ + β₂·X₂ + ... + βₙ·Xₙ + ε

Bu yerda:
  Y   → Bashorat qilinuvchi qiymat (target)
  X   → Kiruvchi o'zgaruvchi (feature)
  β₀  → Intercept (kesma — X=0 da Y qiymati)
  β₁  → Slope (qiyalik — X 1 ga oshsa Y qancha o'zgaradi)
  ε   → Xato (model tushuntirolmagan qism)
```

---

## 2. OLS — Eng Kichik Kvadratlar Usuli

**OLS (Ordinary Least Squares)** — chiziqli regressiyaning matematik yechimi. Maqsad: **haqiqiy qiymatlar va bashorat qiymatlar orasidagi farqlar kvadratlari yig'indisini minimallash.**

```
Har nuqta uchun xato (qoldiq — residual):
  eᵢ = Yᵢ - Ŷᵢ
  Yᵢ  → haqiqiy qiymat
  Ŷᵢ  → model bashorati

OLS minimallashi kerak bo'lgan:
  SSE = Σ(Yᵢ - Ŷᵢ)² → minimal bo'lsin

Nima uchun kvadrat?
  → Musbat va manfiy xatolar bir-birini bekor qilmasin
  → Katta xatolarga ko'proq jarima (kvadrat katta farqni kuchaytiradi)
  → Matematik differentsiallash uchun qulay
```

**Nima uchun absolyut qiymat emas?**
```
|eᵢ| → MAE — differensiallashda qiyin (0 da uzilish)
eᵢ²  → MSE — har yerda differensiallanadi → yechish oson

OLS analitik yechimi (matritsa ko'rinishida):
  β = (XᵀX)⁻¹ · Xᵀ · Y

Bu formula β₀ va β₁ ni bir qadamda topadi!
Iteratsiya shart emas — bu OLS ning kuchli tomoni.
```

## 3. Geometrik Talqin

```
Y
│         *    *
│      *     *
│   *    *          ← Haqiqiy nuqtalar
│      *
│  *
└──────────────── X

Chiziqli regressiya:
  → Bu nuqtalar orasidan "eng yaxshi" to'g'ri chiziq topadi
  → "Eng yaxshi" = vertikal masofalar kvadrati minimalli

Vertikal masofa = residual (qoldiq)

Y
│         *↑ e₃
│      *  |    *↓ e₄
│   * ↑───────────── Ŷ (regressiya chizig'i)
│   ↓e₁ *
│  *↑e₂
└──────────────── X
```

---

## 4. Koeffitsiyentlarni Talqin Etish

Bu qism **eng muhim nazariy qism** — koeffitsiyentlarni noto'g'ri talqin etish keng tarqalgan xato.

### β₀ — Intercept (Kesma):
```
Talqin: "X = 0 bo'lganda Y ning kutilgan qiymati"

Misol: Y = 20,000 + 500·X
  (Y = maosh ming so'mda, X = tajriba yillarda)
  β₀ = 20,000 → tajribasiz xodim (X=0) ning maoshi 20,000

Ehtiyot:
  Ko'p holatlarda X=0 mantiqsiz (uy maydoni 0 m² bo'lmaydi)
  Bunday holda β₀ ni talqin qilish xato — u faqat
  chiziqni "joylashtirish" uchun kerak
```

### β₁ — Slope (Qiyalik):
```
Talqin: "X 1 birlikka oshsa, Y o'rtacha β₁ ga o'zgaradi,
         boshqa barcha omillar o'zgarmas bo'lganda"

Misol: Y = 20,000 + 500·X
  β₁ = 500 → Har yil tajriba maoshni o'rtacha 500 ga oshiradi

Ko'p o'zgaruvchili holatda:
  Y = 20,000 + 500·X₁ + 2,000·X₂
  (X₁ = tajriba, X₂ = ta'lim darajasi)
  β₁ = 500 → X₂ o'zgarmas bo'lganda, tajriba 1 yilga
              oshsa maosh 500 ga oshadi

"Boshqa omillar o'zgarmas" — bu juda muhim shart!
```

### Standartlashtirilgan koeffitsiyentlar:
```
Muammo: X₁ = tajriba (0-40 yil), X₂ = ta'lim (0-5 daraja)
  β₁ = 500,  β₂ = 2,000
  "Ta'lim muhimroq" deb bo'lmaydi — birliklar har xil!

Yechim — Standardizatsiya (Z-score):
  Xᵢ* = (Xᵢ - μ) / σ

Standartlashtirilgan β → "Qaysi o'zgaruvchi muhimroq?" — to'g'ri taqqoslash
```

## 5. Model Baholash — R²

**R² (Determinatsiya koeffitsienti)** — model ma'lumotdagi o'zgarishning qancha ulushini tushuntiradi.

```
Matematik ta'rif:

  SS_tot = Σ(Yᵢ - Ȳ)²    ← Umumiy o'zgarish (Y o'rtachadan farqi)
  SS_res = Σ(Yᵢ - Ŷᵢ)²   ← Tushuntirilmagan o'zgarish (qoldiqlar)

  R² = 1 - SS_res / SS_tot

Talqin:
  R² = 0.85 → Model Y dagi o'zgarishlarning 85% ini tushuntiradi
  R² = 1.0  → Mukammal moslik (amalda shubhali — overfit!)
  R² = 0.0  → Model hech narsani tushuntirolmaydi
  R² < 0    → Model o'rtachadan ham yomoni (juda yomon!)
```

**R² qiymatini talqin etish:**
```
Soha           | Yaxshi R²
───────────────|──────────
Ijtimoiy fanlar| 0.3 — 0.5  (insonlar bashorat qilib bo'lmaydi)
Iqtisodiyot    | 0.5 — 0.7
Muhandislik    | 0.8 — 0.95
Fizika         | 0.99+      (qonunlar aniq)

R² = 0.9 ijtimoiy fanda ajoyib,
R² = 0.9 fizikada — yomon model!
```

**R² ning cheklovi — Adjusted R²:**
```
Muammo: Har yangi o'zgaruvchi qo'shilsa R² oshadi
        (hatto mantiqsiz o'zgaruvchi ham!)

Yechim — Adjusted R²:
  R²_adj = 1 - (1-R²)·(n-1)/(n-k-1)
  n → kuzatuvlar soni
  k → o'zgaruvchilar soni

  Yangi o'zgaruvchi R² ni oshirsa lekin
  Adjusted R² ni tushirsa → o'zgaruvchi keraksiz!
```

---

## 6. Model Baholash — MSE, RMSE, MAE

```
MSE (Mean Squared Error — O'rtacha Kvadratik Xato):
  MSE = (1/n) · Σ(Yᵢ - Ŷᵢ)²

  → Katta xatolarga ko'proq jarima (kvadrat tufayli)
  → Birlik: Y² (agar Y = so'm bo'lsa, MSE = so'm²)
  → Shuning uchun interpretatsiya qiyin

RMSE (Root MSE — Kvadrat ildiz):
  RMSE = √MSE

  → MSE ning kvadrat ildizi → Y bilan bir xil birlik
  → "O'rtacha xato so'mda qancha?" → RMSE javob beradi
  → Katta xatolarga sezgir (outlierlar ta'sir qiladi)

MAE (Mean Absolute Error):
  MAE = (1/n) · Σ|Yᵢ - Ŷᵢ|

  → O'rtacha mutlaq xato
  → Outlierga kam sezgir (kvadrat yo'q)
  → Talqin qilish oson: "O'rtacha X so'm adashaman"

Qachon qaysi:
  Outlier muhim (yo'qolgan → katta zarar) → RMSE
  Outlier muhim emas, mediana xato → MAE
  Nisbiy taqqoslash → R²
```

## 7. Chiziqli Regressiya Taxminlari (Assumptions)

OLS to'g'ri ishlashi uchun bir nechta shart bajarilishi kerak. Bularni bilmaslik — noto'g'ri model va noto'g'ri talqinga olib keladi:

## 2. Sigmoid Funksiyasi — Yechim

**Sigmoid** (S-shakl) funksiyasi istalgan haqiqiy sonni **(0, 1) oralig'iga** siqib qo'yadi:

```
         1
σ(z) = ──────
        1 + e⁻ᶻ

bu yerda:
  z = β₀ + β₁X₁ + β₂X₂ + ...  (chiziqli kombinatsiya)
  e ≈ 2.718 (Eyler soni)
```

**Sigmoid xususiyatlari:**

```
z → -∞   :  σ(z) → 0    (hech qachon 0 ga teng emas)
z =  0   :  σ(z) = 0.5  (aniq o'rta nuqta)
z → +∞   :  σ(z) → 1    (hech qachon 1 ga teng emas)

Simmetriya:  σ(-z) = 1 - σ(z)

Differensiallash (gradient descent uchun muhim):
  σ'(z) = σ(z) · (1 - σ(z))
  → O'zining qiymati orqali hisoblanadi — juda qulay!
```

**Sigmoid shakli:**
```
1.0  |          ___________
     |        /
0.5  |      /         ← Bu nuqta z=0 da
     |    /
0.0  |___/
     |________________________
          -∞    0    +∞
```

## 3. Logistik Regressiya Qanday Ishlaydi?

```
2 bosqich:

Bosqich 1 — Chiziqli kombinatsiya:
  z = β₀ + β₁X₁ + β₂X₂ + ... + βₙXₙ

Bosqich 2 — Sigmoid orqali ehtimollikka aylantirish:
           1
  P(Y=1) = ─────
           1+e⁻ᶻ

Natija:
  P(Y=1) = 0.87  → 87% ehtimollik bilan sinf 1
  P(Y=0) = 0.13  → 13% ehtimollik bilan sinf 0
  (P(Y=0) = 1 - P(Y=1) — ikkilik holat)
```

**Qaror chegara (Decision Boundary):**

```
Standart:  P(Y=1) ≥ 0.5 → Sinf 1
           P(Y=1) < 0.5 → Sinf 0

P = 0.5 → z = 0 → β₀ + β₁X = 0 → bu qaror chegara chizig'i

Chegara o'zgartirish mumkin:
  Tibbiyotda:  P ≥ 0.3 → "Kasal" (xato o'tkazib yubormaslik muhim)
  Spam:        P ≥ 0.7 → "Spam" (muhim xat spam deb belgilanmasin)

Chegara tanlash → precision va recall o'rtasidagi muvozanat
```

## 4. Odds va Log-Odds — Matematik Asos

Logistik regressiyani chuqur tushunish uchun **odds** tushunchasi kerak:

```
Ehtimollik (Probability):
  P = 0.8 → "80% ehtimollik bilan sodir bo'ladi"

Odds (Nisbat):
  Odds = P / (1-P)
  P=0.8 → Odds = 0.8/0.2 = 4
  Ma'nosi: "4:1 nisbatda sodir bo'ladi"

Log-Odds (Logit):
  logit(P) = ln(Odds) = ln(P / (1-P))

  P=0.5  → Odds=1   → log-odds=0
  P=0.8  → Odds=4   → log-odds=1.39
  P=0.95 → Odds=19  → log-odds=2.94
  P=0.2  → Odds=0.25→ log-odds=-1.39
```

**Logistik regressiyaning haqiqiy shakli:**

```
Logistik regressiya aslida log-odds ni chiziqli modellashtiradi:

  ln(P/(1-P)) = β₀ + β₁X₁ + β₂X₂ + ...

Bu aylantirilganda → Sigmoid formulasiga keladi!

Demak:
  Chiziqli regressiya: Y ni to'g'ridan-to'g'ri modellashtiradi
  Logistik regressiya: log-odds ni chiziqli modellashtiradi,
                       keyin sigmoid orqali ehtimollikka aylantiradi
```

---

## 5. Koeffitsiyentlarni Talqin Etish

Chiziqli regressiyadan farqli — logistik regressiya koeffitsiyentlari **odds ratio** orqali talqin etiladi:

```
Chiziqli regressiya:
  β₁ = 500 → "X 1 ga oshsa Y 500 ga oshadi"
  Oddiy, to'g'ridan-to'g'ri talqin

Logistik regressiya:
  β₁ ning to'g'ridan-to'g'ri talqini qiyin
  (log-odds o'zgarishi — intuitiv emas)

Odds Ratio orqali talqin:
  OR = e^β₁

  β₁ = 0.7 → OR = e^0.7 = 2.01
  Talqin: "X 1 ga oshsa, odds 2 marta oshadi"

  β₁ = -0.5 → OR = e^(-0.5) = 0.61
  Talqin: "X 1 ga oshsa, odds 39% kamayadi"

  β₁ = 0 → OR = 1
  Talqin: "X ning hech qanday ta'siri yo'q"
```

## 6. Model O'rgatish — Maximum Likelihood

Chiziqli regressiya OLS (kvadratlar yig'indisini minimallash) ishlatadi. Logistik regressiya esa **Maximum Likelihood Estimation (MLE)** ishlatadi.

```
OLS nima uchun yaramaydi:
  Y ∈ {0, 1} → kvadratik xato funksiyasi notekis (non-convex)
  → Gradient descent lokal minimumga tushib qoladi

MLE g'oyasi:
  "Model parametrlarini shunday tanlayki, kuzatilgan
   ma'lumotlarning ehtimolligi maksimal bo'lsin"

Log-Loss (Binary Cross-Entropy):
  L = -1/n · Σ[Yᵢ·ln(Ŷᵢ) + (1-Yᵢ)·ln(1-Ŷᵢ)]

  Yᵢ=1 holat: -ln(Ŷᵢ)   → Ŷ=0.9 → xato kichik
                          → Ŷ=0.1 → xato KATTA
  Yᵢ=0 holat: -ln(1-Ŷᵢ) → Ŷ=0.1 → xato kichik
                          → Ŷ=0.9 → xato KATTA

  Log funksiyasi: noto'g'ri ishonchli bashoratga
  katta jarima beradi → model ehtiyotkor bo'ladi
```

## 7. Baholash Metrikalari

R² va MSE klassifikatsiya uchun mos emas. Buning o'rniga:

```
Confusion Matrix:
              Bashorat: 0   Bashorat: 1
  Haqiqiy: 0    TN             FP
  Haqiqiy: 1    FN             TP

  TP = True Positive  (to'g'ri 1)
  TN = True Negative  (to'g'ri 0)
  FP = False Positive (noto'g'ri 1 — I tur xato)
  FN = False Negative (noto'g'ri 0 — II tur xato)

Accuracy (To'g'rilik):
  (TP + TN) / (TP + TN + FP + FN)
  Muammo: Nomutanosib klasslarda yolg'on yuqori ko'rinadi
  (99% kasal emas → hammani "sog'" desam → 99% accuracy!)

Precision (Aniqlik):
  TP / (TP + FP)
  "1 deb aytganimning qanchasi haqiqatan 1?"

Recall (Qamrov):
  TP / (TP + FN)
  "Barcha 1 larning qanchasini topdim?"

F1 Score:
  2 · (Precision · Recall) / (Precision + Recall)
  Precision va Recall o'rtasidagi muvozanat

ROC-AUC:
  Turli chegara qiymatlarida model sifati
  AUC = 1.0 → mukammal, AUC = 0.5 → tasodifiy
```

## 2. K-Means — Asosiy G'oya

**K-Means** — ma'lumotlarni **K ta klasterga** bo'lish algoritmi. Har klaster **sentroid** (markaziy nuqta) atrofida joylashgan nuqtalardan iborat.

**Intuitiv tushuncha:**
```
Tasavvur qiling: Xaritada 100 ta shahar bor.
K=3 ta omborxona qurmoqchisiz.
Maqsad: Har shahar o'ziga eng yaqin omborxonaga tegishli bo'lsin
        va umumiy masofa minimall bo'lsin.

K-Means aynan shu masalani yechadi!
Omborxona joylashuvi → Sentroid
Shaharlar → Ma'lumot nuqtalari
```

**Maqsad funksiyasi (minimallash):**
```
WCSS = Σ Σ ||xᵢ - μₖ||²
       k  xᵢ∈Cₖ

WCSS — Within-Cluster Sum of Squares
μₖ   — k-chi klasterning sentroidi
xᵢ   — k-chi klasterdagi nuqta

Ma'nosi: Har nuqtaning o'z sentroididan kvadratik
         masofalar yig'indisi → bu minimal bo'lsin
```

---

## 3. Algoritm Bosqichlari — Iteratsiya

```
Bosqich 1 — Inisializatsiya:
  K ta sentroidni TASODIFIY tanlash
  (yoki K-Means++ usuli bilan — quyida tushuntiriladi)

Bosqich 2 — Tayinlash (Assignment):
  Har nuqtani ENG YAQIN sentroidga tayinlash
  d(x, μₖ) = √Σ(xᵢ - μₖᵢ)²  ← Evklid masofasi

Bosqich 3 — Yangilash (Update):
  Har klasterning yangi sentroidini hisoblash
  μₖ = (1/|Cₖ|) · Σ xᵢ
                   xᵢ∈Cₖ
  Ya'ni: klasterdagi barcha nuqtalarning o'rtachasi

Bosqich 4 — Konvergensiya tekshiruvi:
  Sentroidlar o'zgardimi?
    Ha  → Bosqich 2 ga qayt
    Yo'q → TUGADI (konvergensiya)
```

**Vizual ko'rinish:**

```
Boshlang'ich:          Tayinlash:           Yangilash:
  * * * ×  * *          *₁*₁*₁×  *₂*₂         *₁*₁*₁●  *₂*₂
  * *   ×    *          *₁*₁  ×₁   *₂          *₁*₁  ●    *₂
  * *×    *  *          *₁*₁×₁   *₂ *₂         *₁*₁●    *₂ *₂
    *  × *   *            *₃ ×₃*₃  *₂            *₃ ●*₃  *₂
  × → tasodifiy         × → sentroid           ● → yangi sentroid
      sentroid          1,2,3 → klaster raqami

Iteratsiya davom etadi... → Sentroidlar barqarorlashadi → KONVERGENSIYA
```

---

## 4. Konvergensiya — Nazariy Kafolat

```
K-Means HAR DOIM konvergensiyaga erishadi.
Nima uchun?

  → Har iteratsiyada WCSS kamayadi yoki o'zgarmaydi
  → WCSS pastdan cheklangan (0 dan kichik bo'la olmaydi)
  → Cheklangan sonli nuqtalar → cheklangan sonli tayinlash kombinatsiyasi
  → Demak: Chekli sonli iteratsiyada WCSS minimal bo'ladi

Lekin: Global minimum emas, LOCAL minimum!
  K-Means boshlang'ich sentroidlarga bog'liq →
  Har xil boshlang'ich → Har xil natija mumkin!
```

**Lokal minimum muammosi:**
```
Tasodifiy inisializatsiya:
  Yaxshi boshlang'ich → Yaxshi klasterlar ✅
  Yomon boshlang'ich → Yomon klasterlar ❌

Yechim 1 — Ko'p marta ishlatish:
  model = KMeans(n_init=10)  ← 10 marta ishlatib, eng yaxshini tanlaydi

Yechim 2 — K-Means++ (zamonaviy standart):
  Sentroidlarni "aqlli" tanlash
```

---

## 5. K-Means++ — Aqlli Inisializatsiya

Oddiy K-Means tasodifiy sentroid tanlaydi → yomon natija mumkin. K-Means++ bu muammoni yechadi:

```
K-Means++ algoritmi:

1. Birinchi sentroidni tasodifiy tanlash

2. Keyingi sentroid uchun:
   Har nuqtaning eng yaqin sentroiddan masofasini hisoblash
   Masofa katta bo'lgan nuqtalar ko'proq ehtimollik bilan tanlanadi

   P(x tanlash) = d(x)² / Σ d(xᵢ)²

3. K ta sentroid to'lguncha 2-bosqichni takrorlash

Natija:
  → Sentroidlar bir-biridan uzoq joylashadi
  → Klasterlar butun fazoni yaxshiroq qamrab oladi
  → Lokal minimum ehtimolligi kamayadi
  → Tezroq konvergensiya
```

```python
# sklearn da K-Means++ standart:
from sklearn.cluster import KMeans

model = KMeans(
    n_clusters=3,
    init="k-means++",   # ← standart, K-Means++ ishlatadi
    n_init=10,          # 10 marta ishlatib, eng yaxshini tanlaydi
    max_iter=300,       # maksimal iteratsiya soni
    random_state=42
)
```

## 6. K Soni Qanday Tanlanadi?

Bu K-Means ning **eng muhim va qiyin** savoli. K oldindan ma'lum bo'lmaydi.

### Elbow Method — Tirsak Usuli:

**Nazariy asos:** K oshgan sari WCSS (Within-Cluster Sum of Squares) kamayadi. Lekin ma'lum K dan keyin kamayish sekinlashadi. Bu "tirsak" nuqtasi optimal K.

```
WCSS
│
│*                      K=1: barcha nuqtalar 1 klasterda → WCSS katta
│  *                    K=2: 2 klaster → WCSS kamayadi
│    *                  K=3: 3 klaster → WCSS yana kamayadi
│      *                K=4: KAMAYISH SEKINLASHDI ← TIRSAK!
│        * * * * *      K=5+: deyarli o'zgarmaydi
└──────────────────── K
  1  2  3  4  5  6

Optimal K = 4 (tirsakning "boshi")
```

**Nima uchun eng katta K tanlamaslik kerak:**
```
K = n (har nuqta o'z klasterida) → WCSS = 0 → Befoydа!
Klaster maqsadi: ma'lumotdagi TABIIY guruhlarni topish,
                 sun'iy bo'lish emas.

Tirsak: "Ko'proq klaster qo'shish sezilarli foyda bermayapti"
         degan signaldir.
```

```python
wcss = []
K_oraliq = range(1, 11)

for k in K_oraliq:
    model = KMeans(n_clusters=k, init="k-means++",
                   n_init=10, random_state=42)
    model.fit(X)
    wcss.append(model.inertia_)   # inertia_ = WCSS

# Grafik
plt.plot(K_oraliq, wcss, "bo-", markersize=8)
plt.xlabel("K (klasterlar soni)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.xticks(K_oraliq)
plt.grid(True, alpha=0.3)
plt.show()
```

## 7. Boshqa K Tanlash Usullari

Elbow method har doim aniq javob bermaydi. Qo'shimcha usullar:

### Silhouette Score:
```
Har nuqta uchun:
  a = O'z klasteridagi boshqa nuqtalarga o'rtacha masofa
  b = Eng yaqin boshqa klasterdagi nuqtalarga o'rtacha masofa

  s = (b - a) / max(a, b)

  s → +1  : Nuqta o'z klasterida juda yaxshi joylashgan
  s →  0  : Nuqta chegarada
  s → -1  : Nuqta noto'g'ri klasterda!

O'rtacha Silhouette Score → yuqori bo'lsa yaxshi
Optimal K → Silhouette Score maksimal bo'lgan K
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []
for k in range(2, 11):   # k=1 uchun hisoblab bo'lmaydi
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X)
    skor = silhouette_score(X, labels)
    silhouette_scores.append(skor)
```

### Gap Statistic:
```
G'oya: Haqiqiy ma'lumotdagi WCSS ni
       tasodifiy ma'lumotdagi WCSS bilan solishtirish

Gap(K) = E[log(WCSS_tasodifiy)] - log(WCSS_haqiqiy)

Gap katta → Haqiqiy klaster tuzilishi bor
Gap kichik → Tasodifiy taqsimotdan farq yo'q

Optimal K → Gap eng katta bo'lgan K
```

## 8. K-Means ning Cheklovlari

```
Cheklov 1 — Doira shaklidagi klasterlar:
  K-Means Evklid masofasiga asoslangan →
  Faqat sferik (doira, shar) shaklni yaxshi topadi
  Egri, uzunchoq, halqa shakldagi klasterlar → yomon natija

  Yechim: DBSCAN, Gaussian Mixture Models

Cheklov 2 — Bir xil o'lchamli klasterlar:
  K-Means klasterlarni teng o'lchamda kutadi
  Bitta katta + bitta kichik klaster → xato bo'linishi mumkin

Cheklov 3 — Outlier ta'siri:
  Sentroid = o'rtacha → Outlierlar sentroidni "tortib" ketadi
  Yechim: K-Medoids (sentroid o'rniga haqiqiy nuqta)

Cheklov 4 — K oldindan belgilanishi:
  Haqiqiy dunyo ma'lumotida K noma'lum → Elbow/Silhouette kerak

Cheklov 5 — Miqyos sezgirligi:
  Katta o'lchamli o'zgaruvchilar masofa hisoblashda dominant bo'ladi
  Yechim: Standartlashtirish (StandardScaler) → SHART!
```

## 9. Standartlashtirish — Nima Uchun Shart?

```
Misol: Mijozlarni klasterlash
  Yosh: 20-60 yil        → o'rtacha 40, tarqoqlik 10
  Daromad: 500-50000 so'm → o'rtacha 10000, tarqoqlik 5000

Masofa hisoblashda:
  d = √((yosh farqi)² + (daromad farqi)²)
  d = √((2)² + (3000)²) ≈ 3000

  Daromad masofasi yosh masofasini "bosib ketadi"!
  Model aslida faqat daromad bo'yicha klasterlaydi.

Yechim — StandardScaler:
  Har o'zgaruvchi: (X - μ) / σ
  Natija: Barcha o'zgaruvchilar bir xil miqyosda (o'rtacha 0, σ=1)
  Endi: Yosh va daromad teng hissa qo'shadi
```

## 1. Train-Test Split — Nima Uchun Kerak?

**Asosiy muammo:** Model o'rgangan ma'lumotda yaxshi ishlashi — bu uning haqiqiy qobiliyatini ko'rsatmaydi. Model ma'lumotni shunchaki **yodlab** olgan bo'lishi mumkin.

```
Analogiya — Imtihon:
  O'quvchi faqat o'tgan yilgi savollarni yodlaydi
  O'tgan yil savollari → 100% javob beradi
  Yangi savollar → hech narsa bilmaydi

  Bu "o'rganish" emas — "yodlash"!

Modelda xuddi shu muammo:
  Train ma'lumotida → R²=0.99 (yodlab olgan)
  Yangi ma'lumotda  → R²=0.45 (haqiqiy bilim past)
```

**Yechim — Ma'lumotni ajratish:**

```
Barcha ma'lumot (100%)
        ↓
┌───────────────┬──────────┐
│  Train (80%)  │ Test(20%)│
│  Model shu    │ Model bu │
│  yerda o'rgan.│ yerda    │
│               │ baholanadi│
└───────────────┴──────────┘

Qoida: Test to'plami MODEL UCHUN "Ko'RINMAS" bo'lishi kerak!
       Train jarayonida test ma'lumoti hech qachon ishlatilmaydi.
```

---

## 2. Train-Validation-Test — Uch Qismga Bo'lish

Amalda ikki qismga emas, **uch qismga** bo'lish professional yondashuv:

```
Barcha ma'lumot (100%)
        ↓
┌──────────────┬────────────┬──────────┐
│ Train (60%)  │  Val (20%) │ Test(20%)│
│ O'rgatish    │ Giperpar.  │ Yakuniy  │
│              │ tanlash    │ baho     │
└──────────────┴────────────┴──────────┘

Train       → Model o'rganadi (β, w parametrlar topiladi)
Validation  → Giperparametrlar tanlanadi (K=?, depth=?, C=?)
Test        → Bir marta, yakuniy baholash uchun

Nima uchun Validation kerak?
  Agar giperparametrlarni test da tanlasak →
  Test ham "ko'rilgan" bo'lib qoladi → natija ishonchsiz!
```

---

## 3. Cross-Validation — Kuchli Baholash Usuli

Kichik ma'lumot to'plamida 80/20 bo'lish ishonchsiz — test to'plami "omadli" yoki "omadsiz" bo'lishi mumkin. **K-Fold Cross-Validation** bu muammoni hal qiladi:

```
5-Fold Cross-Validation:

Ma'lumot: [  1  |  2  |  3  |  4  |  5  ]

Iteratsiya 1: Train[2,3,4,5]  Test[1] → Score₁
Iteratsiya 2: Train[1,3,4,5]  Test[2] → Score₂
Iteratsiya 3: Train[1,2,4,5]  Test[3] → Score₃
Iteratsiya 4: Train[1,2,3,5]  Test[4] → Score₄
Iteratsiya 5: Train[1,2,3,4]  Test[5] → Score₅

Final Score = (Score₁+Score₂+Score₃+Score₄+Score₅) / 5

Afzalligi:
  → Har nuqta bir marta test, to'rt marta train da ishlatiladi
  → Barcha ma'lumot baholashga hissa qo'shadi
  → Natija barqarorroq va ishonchli
```

```python
from sklearn.model_selection import cross_val_score, KFold

kf = KFold(n_splits=5, shuffle=True, random_state=42)
skorlar = cross_val_score(model, X, y, cv=kf, scoring="r2")

print(f"Har fold skori: {skorlar}")
print(f"O'rtacha: {skorlar.mean():.4f} ± {skorlar.std():.4f}")
# ± qiymati kichik bo'lsa → model barqaror
```

## 4. Overfitting — Nazariy Asos

**Overfitting** — model train ma'lumotini juda yaxshi o'rganadi, lekin yangi ma'lumotda yomon ishlaydi. Model **shovqinni ham o'rganib oladi.**

```
Analogiya:
  Talaba darslik savollarini yod oladi (shu jumladan xato bosmalarniyam)
  Imtihonda yangi shakldagi savol → javob bera olmaydi

Model nuqtai nazaridan:
  Ma'lumot = Haqiqiy naqsh + Shovqin (tasodifiy xatolar)

  Underfitting: Model haqiqiy naqshni ham o'rgana olmaydi
  Optimal:      Model haqiqiy naqshni o'rganadi, shovqinni emas
  Overfitting:  Model haqiqiy naqsh + shovqinni ham o'rganadi
```

**Grafik ko'rinishda:**

```
Overfitting chizig'i:
  Y
  │    *        Chiziq har bir nuqtadan o'tadi
  │  *  *       (jumladan tasodifiy shovqin nuqtalardan)
  │*      *     → Train da mukammal
  │  *  *       → Yangi nuqtada falokatli xato
  │    *
  └────────── X

Optimal chiziq:
  Y
  │    *        Chiziq umumiy tendensiyani ushlab turadi
  │  *  *       → Ba'zi nuqtalar chiziqdan uzoq, lekin
  │*-─────*     → Yangi nuqtalar uchun yaxshi bashorat
  │  *  *
  │    *
  └────────── X
```

**Overfitting belgilari:**

```
Train R² = 0.98,  Test R² = 0.52  ← Katta farq → Overfitting!
Train MSE = 12,   Test MSE = 340  ← Katta farq → Overfitting!

Qoida: Train va Test skorlari orasidagi katta farq → Overfitting
```

**Overfitting sabablari:**

```
1. Model juda murakkab:
   → Ko'p parametrli neyron tarmoq
   → Juda chuqur qaror daraxti
   → Ko'p darajali polinom

2. Ma'lumot juda kam:
   → 10 ta kuzatuv, 8 ta o'zgaruvchi → parametrlar ko'p!
   → Model "yodlaydi"

3. O'rgatish juda uzoq (neyron tarmoqda):
   → Iteratsiya ko'p → model shovqinni o'rganib oladi

4. Regularizatsiya yo'q:
   → Koeffitsiyentlarga hech qanday cheklov yo'q
```

## 5. Underfitting — Nazariy Asos

**Underfitting** — model juda sodda, ma'lumotdagi haqiqiy naqshni ham ushlab ololmaydi.

```
Analogiya:
  Talaba darslikni umuman o'qimaydi →
  Eng oddiy savolga ham javob bera olmaydi

Model nuqtai nazaridan:
  Chiziqli model → Egri munosabatni tushuntirolmaydi
  Juda kichik neyron tarmoq → Murakkab naqshni o'rgana olmaydi
```

**Grafik ko'rinishda:**

```
Underfitting:
  Y
  │    *
  │  *  *      To'g'ri chiziq — egri munosabatni
  │*──────*    ushlab ololmaydi
  │  *  *      → Train da ham yomon
  │    *       → Test da ham yomon
  └────────── X

Underfitting belgilari:
  Train R² = 0.45,  Test R² = 0.43  ← Ikkalasi ham past
  (Farq kichik, lekin ikkala natija ham yomon)
```

**Underfitting sabablari:**

```
1. Model juda sodda:
   → Chiziqli model, lekin munosabat egri
   → Kam parametrli model

2. O'rgatish juda qisqa (neyron tarmoqda):
   → Iteratsiya kam → model to'liq o'rgana olmagan

3. Juda kuchli regularizatsiya:
   → Koeffitsiyentlar juda kichraytirilgan
   → Model deyarli hech narsa o'rgana olmaydi

4. Muhim o'zgaruvchilar olib tashlangan:
   → Feature engineering yetarli emas
```

## 6. Bias-Variance Tradeoff — Fundamental Nazariya

Overfitting va underfitting ning matematik asosi:

```
Model xatosi = Bias² + Variance + Qaytarilmas Shovqin

Bias (Sistemli xato):
  → Model qanchalik "noto'g'ri farazlaydi"
  → Underfitting → Yuqori Bias
  → Chiziqli model egri munosabatni o'rgana olmaydi

Variance (Beqarorlik):
  → Model turli train to'plamlarida qanchalik farq qiladi
  → Overfitting → Yuqori Variance
  → Har yangi ma'lumotda model keskin o'zgaradi

Qaytarilmas Shovqin:
  → Ma'lumotdagi tabiiy tasodifiylik
  → Hech qanday model buni bartaraf eta olmaydi

Tradeoff:
  Murakkablik ↑ → Bias ↓, Variance ↑
  Murakkablik ↓ → Bias ↑, Variance ↓

  Maqsad: Ikkalasi ham minimal bo'lgan nuqta → Optimal model
```

```
Xato
│
│\        Umumiy xato
│  \    /
│    \/        ← Optimal murakkablik
│   / \
│  /   \──── Variance
│ /
│/──────────── Bias²
└──────────────────── Model murakkabligi
```

## 7. Overfitting Oldini Olish Usullari

### Usul 1 — Regularizatsiya:
```
G'oya: Koeffitsiyentlar juda katta bo'lmasligiga jarima berish
       Katta koeffitsiyent → Model shovqinga haddan oshiqcha ta'sirchan

L2 (Ridge):
  Xato = MSE + λ·Σβᵢ²
  → Koeffitsiyentlarni kichraytiradi, 0 ga tushirmaydi
  → Barcha o'zgaruvchilar qoladi

L1 (Lasso):
  Xato = MSE + λ·Σ|βᵢ|
  → Ba'zi koeffitsiyentlarni aynan 0 ga tushiradi
  → Avtomatik feature tanlash

λ (lambda) — Regularizatsiya kuchi:
  λ katta → Kuchli regularizatsiya → Underfitting xavfi
  λ kichik → Zaif regularizatsiya → Overfitting xavfi
  Optimal λ → Cross-validation bilan tanlanadi
```

### Usul 2 — Ko'proq ma'lumot:
```
Overfitting ning asosiy sababi ko'pincha ma'lumot kamligi.
Ma'lumot ko'p → Model yodlash o'rniga o'rganishga majbur.

Ma'lumot augmentatsiyasi (rasmlarda):
  → Rasmni aylantirish, kichraytirish, yorug'lik o'zgartirish
  → Bir rasmdan ko'p variant yaratish
```

### Usul 3 — Early Stopping (Neyron tarmoqlarda):
```
Neyron tarmoq o'rgatishda:
  Birinchi epochlar: Train xato ↓, Validation xato ↓ (normal)
  Ma'lum nuqtadan: Train xato ↓ davom etadi
                   Validation xato ↑ boshlanadi ← OVERFITTING!

Early Stopping: Validation xato oshishni boshlaganda to'xtatish

Validation xato
│\
│  \         ← To'xtatish nuqtasi
│    \   /
│      \/
│       \──── Overfitting boshlanmoqda
└──────────── Epoch
```

### Usul 4 — Dropout (Neyron tarmoqlarda):
```
O'rgatishda tasodifiy neyronlarni o'chirish (masalan 50%)
→ Tarmoq har safar boshqacha → Yodlash qiyin
→ Natija: Umumlashtirilgan o'rganish

Test vaqtida: Barcha neyronlar ishlatiladi
```

### Usul 5 — Feature Reduction:
```
Keraksiz o'zgaruvchilar → Overfitting xavfi oshadi
PCA (Principal Component Analysis) → O'lchamni kamaytirish
Feature Importance → Muhim o'zgaruvchilarni tanlash
```

## 8. Underfitting Oldini Olish Usullari

```
1. Murakkablikni oshirish:
   → Ko'proq parametrli model tanlash
   → Qaror daraxti uchun: max_depth oshirish
   → Neyron tarmoq uchun: qatlamlar/neyronlar qo'shish

2. Feature Engineering:
   → Yangi ma'lumot belgilari yaratish
   → X² yoki X₁·X₂ kabi o'zaro ta'sir atamalar
   → Logarifmik transformatsiya

3. Regularizatsiyani kamaytirish:
   → λ ni kichraytirish
   → C ni oshirish (Logistik Regressiyada)

4. Ko'proq o'rgatish:
   → Iteratsiya sonini oshirish
   → Learning rate ni kamaytirish
```

```
Overfitting Learning Curve:
Skor
│  ──────────────── Train (yuqori, barqaror)
│
│           ╱────── Validation (past, sekin oshadi)
│         ╱
│       ╱
└──────────────────── Ma'lumot miqdori
→ Katta bo'shliq: Ko'proq ma'lumot yoki regularizatsiya kerak

Underfitting Learning Curve:
Skor
│       ╱────────── Train (past)
│     ╱
│   ╱─────────────── Validation (train ga yaqin, lekin ikkalasi past)
│ ╱
└──────────────────── Ma'lumot miqdori
→ Ikkalasi past: Murakkablikni oshirish kerak

Optimal Learning Curve:
Skor
│  ──────────── Train
│        ╱───── Validation (train ga yaqinlashadi)
│      ╱
│    ╱
└──────────────────── Ma'lumot miqdori
→ Bo'shliq kichik, ikkalasi yuqori ✅
```

## 10. Xulosa

```
Train-Test Split:
  → Model ko'rmagan ma'lumotda baholash
  → Haqiqiy bashorat qobiliyatini o'lchash
  → 80/20 yoki 60/20/20 (train/val/test)

Overfitting:
  → Train yaxshi, Test yomon → Katta farq
  → Sabab: Murakkab model, kam ma'lumot
  → Yechim: Regularizatsiya, ko'proq ma'lumot,
            Early Stopping, Dropout

Underfitting:
  → Train ham, Test ham yomon → Ikkalasi past
  → Sabab: Sodda model, kam feature
  → Yechim: Murakkab model, Feature Engineering

Bias-Variance:
  Underfitting → Yuqori Bias, Past Variance
  Overfitting  → Past Bias, Yuqori Variance
  Maqsad       → Ikkalasi minimal bo'lgan nuqta

Aniqlash:
  → Train va Test skori farqini ko'rish
  → Learning Curve chizish
  → Cross-Validation ishlatish
```
