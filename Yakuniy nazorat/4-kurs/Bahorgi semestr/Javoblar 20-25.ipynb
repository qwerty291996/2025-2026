## 4. Heatmap — Issiqlik Xaritasi

**Nazariy asos:** Heatmap — ikki o'lchovli matritsa ma'lumotlarini rang intensivligi orqali ko'rsatadi. Inson ko'zi rang farqini raqam farqidan 10x tezroq idrok etadi.

**Qachon ishlatiladi:**
```
Korrelyatsiya matritsasi  → qaysi o'zgaruvchilar bog'liq?
Vaqt × Kategoriya         → qaysi kunda qaysi tovar ko'p sotilgan?
Geografik ma'lumot        → qaysi hududda ko'rsatkich yuqori?
Confusion matrix (ML)     → model qaysi klassni adashtirmoqda?
```

**Korrelyatsiya Heatmap — tushuntirish:**
```
Korrelyatsiya koeffitsienti:
  +1.0  → Mukammal musbat bog'liqlik
  +0.7  → Kuchli musbat bog'liqlik
  +0.3  → Zaif musbat bog'liqlik
   0.0  → Bog'liqlik yo'q
  -0.3  → Zaif manfiy bog'liqlik
  -0.7  → Kuchli manfiy bog'liqlik
  -1.0  → Mukammal manfiy bog'liqlik

Heatmap da:
  To'q qizil  → +1.0 ga yaqin
  Oq          →  0.0 ga yaqin
  To'q ko'k   → -1.0 ga yaqin
```

```python
# Korrelyatsiya heatmap:
korrelatsiya = df.corr()

sns.heatmap(
    korrelatsiya,
    annot=True,        # har katakda raqam ko'rsatish
    fmt=".2f",         # 2 xona aniqlik
    cmap="coolwarm",   # rang xaritasi: qizil-oq-ko'k
    vmin=-1, vmax=1,   # rang chegaralari
    center=0,          # oq rang 0 da
    square=True,       # kvadrat kataklar
    linewidths=0.5,    # kataklararo chiziq
)
```

**cmap tanlash falsafasi:**
```
"coolwarm"  → Diverging (ikki qarama-qarshi tomon) — korrelyatsiya uchun
"viridis"   → Sequential (bir yo'nalish) — davomiy qiymatlar uchun
"Blues"     → Sequential, bir rang — oddiy taqsimot uchun
"Set2"      → Categorical (kategoriyalar) — guruhlar uchun
```

## 5. Pairplot — Ko'p O'zgaruvchili Tahlil

**Nazariy asos:** n ta o'zgaruvchi uchun n×n grafik matritsasi yaratadi. Diagonal → har o'zgaruvchining taqsimoti, diagonal emas → ikki o'zgaruvchi orasidagi munosabat.

**Qachon ishlatiladi:**
```
Ma'lumotlarni dastlabki o'rganish (EDA — Exploratory Data Analysis)
Qaysi o'zgaruvchilar korrelyatsiyaga ega — vizual aniqlash
Guruhlar orasidagi farqni ko'rish (hue parametri bilan)
Machine Learning uchun feature tanlash
```

**Pairplot nima ko'rsatadi:**
```
Diagonal (scatter emas):
  → Har o'zgaruvchining yolg'iz taqsimoti
  → KDE yoki histogram

Diagonal emas (scatter):
  → Chap pastga → o'ng yuqoriga: musbat korrelyatsiya
  → Aniq klasterlar: ma'lumotda guruhlar bor
  → Kuchli naqsh yo'q: o'zgaruvchilar mustaqil
```

```python
sns.pairplot(
    df,
    hue="tur",           # Ranglar bilan guruhlar
    diag_kind="kde",     # Diagonal: "hist" yoki "kde"
    plot_kws={"alpha": 0.6},     # Scatter nuqtalar shaffof
    diag_kws={"fill": True},     # KDE to'ldirilgan
    corner=True          # Faqat pastki uchburchak (tezroq)
)
```

**Nazariy tushuncha — KDE (Kernel Density Estimation):**
```
Histogram → Ma'lumotni "chelaklar"ga bo'ladi → diskret, qo'pol
KDE       → Ma'lumotni silliq egri chiziq bilan ifodalaydi → uzluksiz

KDE aslida: har nuqta atrofida kichik "qo'ng'iroq" shaklida taqsimot
            qo'yiladi, barchasi qo'shiladi → silliq taqsimot egri chizig'i

Qachon KDE: taqsimot shaklini aniq ko'rish kerak
Qachon hist: aniq chastotalarni bilish kerak
```

## 6. Boxplot — Statistik Taqqoslash

**Nazariy asos:** Boxplot (quti grafik) — beshta statistik ko'rsatkichni bir vaqtda ko'rsatadi. John Tukey (1969) tomonidan ixtiro qilingan.

```
Boxplot anatomiyasi:

          ┌─────────────────┐
     ──── │ Yuqori kvartil  │ Q3 (75-persentil)
          │                 │
  ─────── │    Mediana      │ Q2 (50-persentil)
          │                 │
     ──── │ Quyi kvartil    │ Q1 (25-persentil)
          └─────────────────┘
               ↑
           IQR = Q3 - Q1  (Interquartile Range)

Whisker (mo'ylov):
  Yuqori: Q3 + 1.5 × IQR
  Quyi:   Q1 - 1.5 × IQR

Whiskerdan tashqari nuqtalar → ANOMALIYALAR (outliers)
```

**Boxplot nimani ko'rsatadi:**
```
Quti balandligi → Ma'lumot tarqoqligi (keng = ko'p tarqoq)
Mediana joylashuvi → Nosimmetriklik (chapga/o'ngga og'ish)
Whisker uzunligi → Dumlar og'irligi
Outlier nuqtalar → Anomal qiymatlar
```

**Violin plot — Boxplot ning kuchli varianti:**
```
Boxplot:       Faqat 5 statistik ko'rsatkich
Violin plot:   Boxplot + taqsimot shakli (KDE)

Violin plot qachon afzal:
  → Taqsimot bir cho'qqili yoki ko'p cho'qqilimi — ko'rish kerak bo'lsa
  → Guruhlar ichki tuzilishini solishtirish kerak bo'lsa
```

```python
# Boxplot:
sns.boxplot(
    data=df,
    x="kategoriya",
    y="qiymat",
    hue="guruh",       # Rang bo'yicha ajratish
    palette="Set2",    # Rang palatrasi
    width=0.6,         # Quti kengligi
    flierprops={"marker": "o", "markersize": 4}  # Outlier ko'rinishi
)

# Violin plot:
sns.violinplot(
    data=df,
    x="kategoriya",
    y="qiymat",
    inner="box",       # Ichida: "box", "quart", "point"
    split=True,        # hue bo'lsa: ikki tomonga bo'lish
)
```

## 1. Chiziqli Regressiya — Nazariy Asos

Chiziqli regressiya — bir yoki bir nechta **mustaqil o'zgaruvchilar** (X) asosida **bog'liq o'zgaruvchini** (Y) bashorat qilish usuli. Asosiy g'oya: ma'lumotlar orasidan **eng yaxshi to'g'ri chiziq** topish.

```
Hayotiy misollar:
  X = ta'lim yillari   → Y = oylik maosh
  X = uy maydoni (m²)  → Y = uy narxi
  X = reklama xarajati → Y = sotuvlar hajmi
  X = temperatura      → Y = elektr sarfi
```

**Matematik shakl:**

```
Oddiy chiziqli regressiya (1 ta X):
  Y = β₀ + β₁·X + ε

Ko'p o'zgaruvchili (n ta X):
  Y = β₀ + β₁·X₁ + β₂·X₂ + ... + βₙ·Xₙ + ε

Bu yerda:
  Y   → Bashorat qilinuvchi qiymat (target)
  X   → Kiruvchi o'zgaruvchi (feature)
  β₀  → Intercept (kesma — X=0 da Y qiymati)
  β₁  → Slope (qiyalik — X 1 ga oshsa Y qancha o'zgaradi)
  ε   → Xato (model tushuntirolmagan qism)
```

---

## 2. OLS — Eng Kichik Kvadratlar Usuli

**OLS (Ordinary Least Squares)** — chiziqli regressiyaning matematik yechimi. Maqsad: **haqiqiy qiymatlar va bashorat qiymatlar orasidagi farqlar kvadratlari yig'indisini minimallash.**

```
Har nuqta uchun xato (qoldiq — residual):
  eᵢ = Yᵢ - Ŷᵢ
  Yᵢ  → haqiqiy qiymat
  Ŷᵢ  → model bashorati

OLS minimallashi kerak bo'lgan:
  SSE = Σ(Yᵢ - Ŷᵢ)² → minimal bo'lsin

Nima uchun kvadrat?
  → Musbat va manfiy xatolar bir-birini bekor qilmasin
  → Katta xatolarga ko'proq jarima (kvadrat katta farqni kuchaytiradi)
  → Matematik differentsiallash uchun qulay
```

**Nima uchun absolyut qiymat emas?**
```
|eᵢ| → MAE — differensiallashda qiyin (0 da uzilish)
eᵢ²  → MSE — har yerda differensiallanadi → yechish oson

OLS analitik yechimi (matritsa ko'rinishida):
  β = (XᵀX)⁻¹ · Xᵀ · Y

Bu formula β₀ va β₁ ni bir qadamda topadi!
Iteratsiya shart emas — bu OLS ning kuchli tomoni.
```

## 3. Geometrik Talqin

```
Y
│         *    *
│      *     *
│   *    *          ← Haqiqiy nuqtalar
│      *
│  *
└──────────────── X

Chiziqli regressiya:
  → Bu nuqtalar orasidan "eng yaxshi" to'g'ri chiziq topadi
  → "Eng yaxshi" = vertikal masofalar kvadrati minimalli

Vertikal masofa = residual (qoldiq)

Y
│         *↑ e₃
│      *  |    *↓ e₄
│   * ↑───────────── Ŷ (regressiya chizig'i)
│   ↓e₁ *
│  *↑e₂
└──────────────── X
```

---

## 4. Koeffitsiyentlarni Talqin Etish

Bu qism **eng muhim nazariy qism** — koeffitsiyentlarni noto'g'ri talqin etish keng tarqalgan xato.

### β₀ — Intercept (Kesma):
```
Talqin: "X = 0 bo'lganda Y ning kutilgan qiymati"

Misol: Y = 20,000 + 500·X
  (Y = maosh ming so'mda, X = tajriba yillarda)
  β₀ = 20,000 → tajribasiz xodim (X=0) ning maoshi 20,000

Ehtiyot:
  Ko'p holatlarda X=0 mantiqsiz (uy maydoni 0 m² bo'lmaydi)
  Bunday holda β₀ ni talqin qilish xato — u faqat
  chiziqni "joylashtirish" uchun kerak
```

### β₁ — Slope (Qiyalik):
```
Talqin: "X 1 birlikka oshsa, Y o'rtacha β₁ ga o'zgaradi,
         boshqa barcha omillar o'zgarmas bo'lganda"

Misol: Y = 20,000 + 500·X
  β₁ = 500 → Har yil tajriba maoshni o'rtacha 500 ga oshiradi

Ko'p o'zgaruvchili holatda:
  Y = 20,000 + 500·X₁ + 2,000·X₂
  (X₁ = tajriba, X₂ = ta'lim darajasi)
  β₁ = 500 → X₂ o'zgarmas bo'lganda, tajriba 1 yilga
              oshsa maosh 500 ga oshadi

"Boshqa omillar o'zgarmas" — bu juda muhim shart!
```

### Standartlashtirilgan koeffitsiyentlar:
```
Muammo: X₁ = tajriba (0-40 yil), X₂ = ta'lim (0-5 daraja)
  β₁ = 500,  β₂ = 2,000
  "Ta'lim muhimroq" deb bo'lmaydi — birliklar har xil!

Yechim — Standardizatsiya (Z-score):
  Xᵢ* = (Xᵢ - μ) / σ

Standartlashtirilgan β → "Qaysi o'zgaruvchi muhimroq?" — to'g'ri taqqoslash
```

## 5. Model Baholash — R²

**R² (Determinatsiya koeffitsienti)** — model ma'lumotdagi o'zgarishning qancha ulushini tushuntiradi.

```
Matematik ta'rif:

  SS_tot = Σ(Yᵢ - Ȳ)²    ← Umumiy o'zgarish (Y o'rtachadan farqi)
  SS_res = Σ(Yᵢ - Ŷᵢ)²   ← Tushuntirilmagan o'zgarish (qoldiqlar)

  R² = 1 - SS_res / SS_tot

Talqin:
  R² = 0.85 → Model Y dagi o'zgarishlarning 85% ini tushuntiradi
  R² = 1.0  → Mukammal moslik (amalda shubhali — overfit!)
  R² = 0.0  → Model hech narsani tushuntirolmaydi
  R² < 0    → Model o'rtachadan ham yomoni (juda yomon!)
```

**R² qiymatini talqin etish:**
```
Soha           | Yaxshi R²
───────────────|──────────
Ijtimoiy fanlar| 0.3 — 0.5  (insonlar bashorat qilib bo'lmaydi)
Iqtisodiyot    | 0.5 — 0.7
Muhandislik    | 0.8 — 0.95
Fizika         | 0.99+      (qonunlar aniq)

R² = 0.9 ijtimoiy fanda ajoyib,
R² = 0.9 fizikada — yomon model!
```

**R² ning cheklovi — Adjusted R²:**
```
Muammo: Har yangi o'zgaruvchi qo'shilsa R² oshadi
        (hatto mantiqsiz o'zgaruvchi ham!)

Yechim — Adjusted R²:
  R²_adj = 1 - (1-R²)·(n-1)/(n-k-1)
  n → kuzatuvlar soni
  k → o'zgaruvchilar soni

  Yangi o'zgaruvchi R² ni oshirsa lekin
  Adjusted R² ni tushirsa → o'zgaruvchi keraksiz!
```

---

## 6. Model Baholash — MSE, RMSE, MAE

```
MSE (Mean Squared Error — O'rtacha Kvadratik Xato):
  MSE = (1/n) · Σ(Yᵢ - Ŷᵢ)²

  → Katta xatolarga ko'proq jarima (kvadrat tufayli)
  → Birlik: Y² (agar Y = so'm bo'lsa, MSE = so'm²)
  → Shuning uchun interpretatsiya qiyin

RMSE (Root MSE — Kvadrat ildiz):
  RMSE = √MSE

  → MSE ning kvadrat ildizi → Y bilan bir xil birlik
  → "O'rtacha xato so'mda qancha?" → RMSE javob beradi
  → Katta xatolarga sezgir (outlierlar ta'sir qiladi)

MAE (Mean Absolute Error):
  MAE = (1/n) · Σ|Yᵢ - Ŷᵢ|

  → O'rtacha mutlaq xato
  → Outlierga kam sezgir (kvadrat yo'q)
  → Talqin qilish oson: "O'rtacha X so'm adashaman"

Qachon qaysi:
  Outlier muhim (yo'qolgan → katta zarar) → RMSE
  Outlier muhim emas, mediana xato → MAE
  Nisbiy taqqoslash → R²
```

## 7. Chiziqli Regressiya Taxminlari (Assumptions)

OLS to'g'ri ishlashi uchun bir nechta shart bajarilishi kerak. Bularni bilmaslik — noto'g'ri model va noto'g'ri talqinga olib keladi:

```
1. Chiziqlilik:
   X va Y orasidagi munosabat chiziqli bo'lishi kerak
   Tekshirish → scatter plot (egri bo'lsa → transformatsiya kerak)

2. Mustaqillik:
   Kuzatuvlar bir-biridan mustaqil bo'lishi kerak
   Buzilishi → vaqt qatori ma'lumotlari (autocorrelation)

3. Gomoskedastiklik (bir xil dispersiya):
   Qoldiqlarning tarqoqligi X bo'yicha bir xil bo'lishi kerak
   Buzilishi → "geteroskodastiklik" — residual plot da konus shakli

4. Normallik:
   Qoldiqlar normal taqsimlangan bo'lishi kerak
   (katta n da bu shart unchalik muhim emas — CLT tufayli)

5. Multikollinearlik yo'qligi (ko'p o'zgaruvchili holatda):
   X o'zgaruvchilar bir-biri bilan kuchli korrelyatsiyada bo'lmasin
   Tekshirish → VIF (Variance Inflation Factor)
   VIF > 10 → muammo bor
```

## 8. Sklearn da Amaliyot

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

# Ma'lumotni train/test ga bo'lish
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% test
    random_state=42     # takrorlanuvchi natija
)

# Model yaratish va o'rgatish
model = LinearRegression()
model.fit(X_train, y_train)

# Koeffitsiyentlar
print(f"Intercept (β₀): {model.intercept_:.2f}")
print(f"Slope (β₁):     {model.coef_[0]:.2f}")

# Bashorat
y_pred = model.predict(X_test)

# Baholash
r2   = r2_score(y_test, y_pred)
mse  = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print(f"R²:   {r2:.4f}")    # 0.85 → 85% tushuntiradi
print(f"MSE:  {mse:.2f}")
print(f"RMSE: {rmse:.2f}")  # Y bilan bir xil birlik
```

**Train/Test ajratish nima uchun kerak:**
```
Model o'rgangan ma'lumotda yaxshi ishlashi tabiiy —
u o'sha ma'lumotni "yodlab" olgan bo'lishi mumkin.

Haqiqiy sinov: model ko'rmagan ma'lumotda qanday ishlaydi?

Train set → model o'rganadi (β₀, β₁ topiladi)
Test set  → model baholanadi (haqiqiy sifat o'lchanadi)

Overfitting:
  Train R² = 0.99, Test R² = 0.45 → Model yodlagan, o'rganmagan!
```

## 9. Xulosa

```
Chiziqli regressiya:
  Y = β₀ + β₁X + ε

OLS maqsadi:
  Σ(Yᵢ - Ŷᵢ)² → minimal

Koeffitsiyentlar:
  β₀ → X=0 da Y qiymati (ko'p holat mantiqsiz)
  β₁ → X 1 birlikka oshsa Y β₁ ga o'zgaradi

Baholash metrikalari:
  R²   → O'zgarishning qancha % tushuntirildi (0-1)
  MSE  → O'rtacha kvadratik xato (katta xatolarga sezgir)
  RMSE → MSE ildizi, Y bilan bir xil birlik
  MAE  → O'rtacha mutlaq xato, outlierga bardoshli

Muhim shart:
  Train/Test ajratish → overfittingni aniqlash
  Adjusted R²         → ko'p o'zgaruvchili holat uchun
```

> 💡 **Eslab qol:** R² yuqori bo'lishi yaxshi model degani emas. **Test to'plamida** R² yuqori bo'lishi — yaxshi model belgisi. Train da R²=0.99, test da R²=0.5 → model yodlagan, bashorat qila olmaydi!
