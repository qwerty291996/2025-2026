PyQt'da **Layout (joylashtirish) menejerlari** oynadagi vidjetlar (tugmalar, matn maydonlari, rasmlar va h.k.) qanday joylashishini va oyna o'lchami o'zgarganda ularning o'lchamlari qanday moslashishini boshqaradi.

Siz sanab o'tgan 5 ta asosiy layout'ning har biri o'ziga xos vazifani bajaradi. Keling, ularning farqlarini aniq va hayotiy misollar yordamida ko'rib chiqamiz.

---

### 1. `QHBoxLayout` (Gorizontal joylashtirish)

Bu layout vidjetlarni chapdan o'ngga qarab, **yonma-yon (gorizontal)** qatorda joylashtiradi. Agar oyna kengaytirilsa, vidjetlar orasidagi bo'shliq ham shunga mos ravishda kengayadi.

* **Konkret misol:** Dastur oynasining eng pastki qismida joylashgan tugmalar paneli. Tasavvur qiling, sizda **"Saqlash" (Save)** va **"Bekor qilish" (Cancel)** tugmalari bor. Ular yonma-yon turishi kerak. Shunda siz `QHBoxLayout` dan foydalanasiz.
* **Ko'rinishi:** `[ Tugma 1 ]  [ Tugma 2 ]  [ Tugma 3 ]`

### 2. `QVBoxLayout` (Vertikal joylashtirish)

Bu layout vidjetlarni yuqoridan pastga qarab, **ustma-ust (vertikal)** ustun shaklida joylashtiradi.

* **Konkret misol:** Oddiy ro'yxat yoki menyu. Masalan, Telegram'ning sozlamalar bo'limini tasavvur qiling: "Profil", "Xabarnomalar", "Xavfsizlik", "Til" kabi tugmalar ketma-ket, bir-birining tagida joylashgan. Buni `QVBoxLayout` yordamida qilish juda oson. Yoki oddiy MP3 pleyer: yuqorida rasm, uning tagida qo'shiq nomi, eng pastda esa boshqaruv tugmalari (Play/Pause).
* **Ko'rinishi:**

`[ Vidjet 1 ]`
  
`[ Vidjet 2 ]`

`[ Vidjet 3 ]`

### 3. `QGridLayout` (Setka / To'rsimon joylashtirish)

Bu layout vidjetlarni **qatorlar va ustunlar** ga (jadvalga o'xshab) bo'lib joylashtiradi. Siz vidjetni aynan qaysi qator va qaysi ustunda joylashishini (koordinatalarini) aniq ko'rsata olasiz. Shuningdek, bitta vidjet bir nechta katakni egallashi ham mumkin (masalan, ikkita ustunni birlashtirish).

* **Konkret misol:** **Kalkulyator interfeysi**. Kalkulyatorda raqamlar va amallar (0-9, +, -, *, /) 4x4 yoki 5x4 o'lchamdagi aniq setka shaklida joylashgan bo'ladi. Yoki "Shaxmat" doskasi ko'rinishidagi o'yin maydoni.
* **Ko'rinishi:**
  
`[ 7 ] [ 8 ] [ 9 ] [ / ]`

`[ 4 ] [ 5 ] [ 6 ] [ * ]`

`[ 1 ] [ 2 ] [ 3 ] [ - ]`

### 4. `QFormLayout` (Forma ko'rinishida joylashtirish)

Bu layout maxsus ikki ustunli tizim bo'lib, odatda **ma'lumot kiritish formalarini** yaratish uchun ishlatiladi. Chap ustunda vidjetning nomi (Label), o'ng ustunda esa ma'lumot kiritish maydoni (LineEdit, ComboBox va h.k.) joylashadi.

* **Konkret misol:** **Ro'yxatdan o'tish (Registration) yoki Login formasi**. Dastur foydalanuvchidan o'z ma'lumotlarini kiritishni so'raganda ishlatiladi. Garchi buni `QGridLayout` yordamida ham qilsa bo'lsa-da, `QFormLayout` bu jarayonni avtomatlashtiradi va kodni qisqartiradi.
* **Ko'rinishi:**

`Ism:       [_________________]`

`Familiya:  [_________________]`

`Parol:     [_________________]`

### 5. `QStackedLayout` (Qatlamli / Ustma-ust joylashtirish)

Bu layout vidjetlarni huddi kartalar kabi **ustma-ust qatlamlar** shaklida joylashtiradi. Eng muhim farqi: bir vaqtning o'zida ekranda **faqat bitta qatlam (bitta vidjet/oyna) ko'rinib turadi**. Boshqa qatlamlarga o'tish uchun maxsus tugmalar (yoki ComboBox) kerak bo'ladi.

* **Konkret misol:** **Dastur o'rnatish ustasi (Installation Wizard)**. Odatda dastur o'rnatayotganda "Next" (Keyingisi) tugmasini bosasiz, shunda oyna ichidagi ma'lumotlar o'zgaradi, lekin asosiy oyna ramkasi joyida qoladi. Yana bir misol: brauzerlardagi "Tab"lar (garchi tablar uchun tayyor `QTabWidget` bo'lsa ham, u ichkarida aynan mana shu Stacked tizimidan foydalanadi).

### Xulosa qilib aytganda:

Agar sizga vidjetlarni:

* Yonga terish kerak bo'lsa $\rightarrow$ **QHBoxLayout**
* Pastga terish kerak bo'lsa $\rightarrow$ **QVBoxLayout**
* Katakchalarga aniq joylash kerak bo'lsa $\rightarrow$ **QGridLayout**
* So'rovnoma/anketa yasash kerak bo'lsa $\rightarrow$ **QFormLayout**
* Bir xil joyda turli xil sahifalarni almashtirib ko'rsatish kerak bo'lsa $\rightarrow$ **QStackedLayout** dan foydalanasiz.

Katta va murakkab dasturlarda bu layoutlar **bir-birining ichiga joylashtirilib (nested)** ishlatiladi. Masalan, butun oyna `QVBoxLayout` da bo'lib, uning eng pastki qismidagi tugmalar `QHBoxLayout` ga solinishi mumkin.

### 🪟 2-qism: PyQt5 dagi Asosiy Oyna Turlari

Endi kod darajasidagi sinflarga (class) o'tamiz. Bu sinflar oyna qanday ishlashini belgilaydi.

#### 1. `QMainWindow` (Asosiy oyna)

Bu dasturning "yuragi". U faqat bitta bo'ladi (keng qamrovli dasturlarda). Uning o'ziga xos arxitekturasi bor: markazda asosiy ish maydoni (`Central Widget`) bo'ladi, atrofida esa menyular.

* **Konkret misol:** Fotoshop dasturi. Tepada menyular (`File`, `Filter`), chapda asboblar paneli (`Toolbar`), pastda holat paneli (masalan, rasm o'lchami ko'rinib turadi) va o'rtada siz chizayotgan rasm (`Central Widget`).

#### 2. `QDialog` (Dialog oyna)

Foydalanuvchi bilan qisqa muddatli muloqot qilish uchun asosiy oynaning ustidan qalqib chiqadigan (pop-up) oyna. U ko'pincha *Modal* bo'ladi (ya'ni, dialog oynasini yopmaguningizcha, orqadagi asosiy oynani bosolmaysiz).

* **Konkret misol:** Telegram'da "Add Contact" (Kontakt qo'shish) tugmasini bossangiz, ism va raqam kiritish uchun kichik oyna chiqadi. Shu oyna – `QDialog`.

#### 3. `QWidget` (Oddiy widget)

Bu PyQt'dagi barcha vizual elementlarning "otasi". Tugma (`QPushButton`) ham, matn maydoni (`QLineEdit`) ham aslida `QWidget` dan tarqalgan. Agar `QWidget` ni boshqa oynaning ichiga solmasangiz, u mustaqil oyna sifatida ochiladi. U juda universal.

* **Konkret misol:** Dastur ichida bir nechta tablar (varaqlar) bo'lsa, har bir tabning ichidagi kontent alohida `QWidget` hisoblanadi.

#### 4. `QMessageBox` (Xabar oynasi)

Bu `QDialog` ning tayyor va maxsus turi. Uni noldan yasash shart emas, chunki tizimda tayyor shabloni bor. U faqat ma'lumot berish, ogohlantirish yoki xato haqida xabar berish uchun ishlatiladi.

* **Konkret misol:** Hujjatni saqlamasdan yopmoqchi bo'lsangiz chiqadigan: *"Hujjat saqlanmadi. Saqlashni xohlaysizmi? [Yes] [No]"* oynasi. Yoki *"Parol noto'g'ri kiritildi"* degan xato (Error) oynasi.

#### 5. `QFileDialog` (Fayl tanlash oynasi)

Bu ham tayyor oyna. Operatsion tizimning o'zida bor bo'lgan kompyuter ichidan fayl qidirish oynasini chaqirib beradi.

* **Konkret misol:** Dasturda "Open" (Ochish) tugmasini bosganingizda kompyuteringizdagi papkalar ro'yxati chiqib, kerakli rasm yoki faylni tanlashingizni so'raydigan oyna.

#### 6. `QInputDialog` (Input dialog)

Foydalanuvchidan bittagina ma'lumot (matn, raqam yoki ro'yxatdan birini tanlash) so'rash uchun mo'ljallangan juda sodda tayyor oyna.

* **Konkret misol:** Wi-Fi tarmog'iga ulanmoqchi bo'lganingizda, faqatgina *"Parolni kiriting: [________]"* deb chiqadigan kichkina oyna.

```
Foydalanuvchi harakat qiladi (klik, klaviatura...)
           ↓
    OS event yaratadi
           ↓
   Qt event queue ga tushadi
           ↓
    app.exec_() ushlab oladi
           ↓
   Tegishli widget ga yuboradi
           ↓
  Signal emit bo'ladi → Slot chaqiriladi
```

## 5. Signal-Slot Mexanizmi — Eng Muhim Qism!

### Tushuncha:
```
SIGNAL  = "Biror narsa sodir bo'ldi!" degan xabar
SLOT    = O'sha xabarga javob beradigan funksiya
CONNECT = Ikkalasini bog'laydigan ko'prik
```

### Sodda misol:
```python
from PyQt5.QtWidgets import *
import sys

class MisalDarsa(QWidget):
    def __init__(self):
        super().__init__()

        # Tugma yaratish
        self.tugma = QPushButton("Bosing!", self)
        self.tugma.setGeometry(50, 50, 200, 50)

        # SIGNAL → SLOT bog'lash
        self.tugma.clicked.connect(self.tugma_bosildi)
        #          ↑ signal    ↑ connect  ↑ slot (funksiya)

    def tugma_bosildi(self):  # ← Bu SLOT
        print("Tugma bosildi!")

app = QApplication(sys.argv)
darsa = MisalDarsa()
darsa.show()
sys.exit(app.exec_())
```

### Turli Signal turlari:
```python
# Parametrsiz signal
self.tugma.clicked.connect(self.funksiya)

# Parametrli signal (bool)
self.checkbox.stateChanged.connect(self.holat_ozgardi)
def holat_ozgardi(self, holat: bool):
    print(f"Holat: {holat}")

# Parametrli signal (str)
self.lineedit.textChanged.connect(self.matn_ozgardi)
def matn_ozgardi(self, matn: str):
    print(f"Matn: {matn}")

# Parametrli signal (int)
self.slider.valueChanged.connect(self.qiymat_ozgardi)
def qiymat_ozgardi(self, qiymat: int):
    print(f"Qiymat: {qiymat}")
```

**PyQt** dagi slider (kodda `QSlider` deb ataladi) — bu foydalanuvchiga ma'lum bir raqamlar oraliqdan (masalan, 0 dan 100 gacha) kerakli qiymatni sichqoncha bilan surish orqali tanlash imkonini beruvchi grafik interfeys (GUI) elementidir.

Buni kundalik hayotda kompyuter yoki telefonda **ovoz balandligini sozlash** yoki videoni **oldinga/orqaga o'tkazish** uchun ishlatiladigan surma tugmalariga o'xshatishingiz mumkin.

### `QSlider` ning Asosiy Xususiyatlari

Slider bilan ishlashda eng ko'p ishlatiladigan xususiyatlar va funksiyalar:

* **Yo'nalish (Orientation):** Slider yotiq (`Qt.Horizontal`) yoki tikka (`Qt.Vertical`) bo'lishi mumkin.
* **Minimum va Maksimum qiymatlar:** Slider'ning eng kichik va eng katta qiymatlarini belgilash (masalan, `setMinimum(0)` va `setMaximum(100)`).
* **Boshlang'ich qiymat:** Dastur ishga tushganda slider qaysi joyda turishini belgilash (`setValue(50)`).
* **Signal (Hodisa):** Foydalanuvchi sliderni surganda, u `valueChanged` degan signalni yuboradi. Buni biror funksiyaga ulab (connect), qiymat o'zgarganda ekrandagi matnni yoki o'zgaruvchini yangilash mumkin.
* **Chiziqchalar (Ticks):** Slider ostida yoki yonida yo'naltiruvchi chiziqchalarni ko'rsatish (`setTickPosition()`).

### PyQt'da QSlider uchun Sodda Misol

Quyida PyQt5 yordamida yozilgan sodda dastur kodi keltirilgan. Unda slider va bitta matn (Label) bor. Sliderni sursangiz, matndagi raqam o'zgaradi.

```python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QSlider, QVBoxLayout, QLabel
from PyQt5.QtCore import Qt

class SliderIlovasi(QWidget):
    def __init__(self):
        super().__init__()
        
        # Oyna sozlamalari
        self.setWindowTitle("QSlider Misoli")
        self.setGeometry(200, 200, 350, 150)

        # Layout (Elementlarni joylashtirish tartibi)
        layout = QVBoxLayout()

        # Matn (Label) yaratish
        self.label = QLabel("Joriy qiymat: 50", self)
        self.label.setAlignment(Qt.AlignCenter)
        layout.addWidget(self.label)

        # Yotiq (Horizontal) Slider yaratish
        self.slider = QSlider(Qt.Horizontal, self)
        self.slider.setMinimum(0)       # Eng kam qiymat
        self.slider.setMaximum(100)     # Eng ko'p qiymat
        self.slider.setValue(50)        # O'rtada turishi uchun 50 berildi
        
        # Slider ostida chiziqchalar ko'rsatish
        self.slider.setTickPosition(QSlider.TicksBelow)
        self.slider.setTickInterval(10) # Har 10 qadamda bitta chiziq

        # Slider qiymati o'zgarganda update_label funksiyasini ishga tushirish
        self.slider.valueChanged.connect(self.update_label)

        layout.addWidget(self.slider)
        self.setLayout(layout)

    # Slider qiymati o'zgarganda ishlaydigan funksiya
    def update_label(self, value):
        self.label.setText(f"Joriy qiymat: {value}")

if __name__ == '__main__':
    app = QApplication(sys.argv)
    oyna = SliderIlovasi()
    oyna.show()
    sys.exit(app.exec_())

```

### Kod qanday ishlaydi?

1. `QSlider(Qt.Horizontal)` orqali biz o'ngga va chapga suriladigan yotiq slider yaratdik.
2. `setMinimum(0)` va `setMaximum(100)` orqali foydalanuvchi faqat 0 dan 100 gacha bo'lgan raqamlarni tanlay olishini belgiladik.
3. Asosiy qism — `self.slider.valueChanged.connect(self.update_label)`. Bu qator orqali biz dasturga: *"Agar slider o'rnidan jilsa, darhol o'sha yangi raqamni olib `update_label` funksiyasiga ber"*, deb buyruq beryapmiz. U funksiya esa yozuvni yangilaydi.

PyQt'da eng ko'p ishlatiladigan vizual elementlar (vidjetlar) va ularning asosiy signallari:

* **`QPushButton` (Tugma)**
* `clicked` — Tugma bosilganda jo'natiladi.


* **`QLineEdit` (Matn kiritish maydoni)**
* `textChanged` — Maydondagi matn o'zgarganda (har bir harf yozilganda yoki o'chirilganda).
* `returnPressed` — Kiritish maydonida "Enter" tugmasi bosilganda.


* **`QComboBox` (Ochiluvchi ro'yxat)**
* `currentIndexChanged` — Ro'yxatdan boshqa variant (element) tanlanganda.


* **`QCheckBox` (Belgilash katakchasi / Ptichka)**
* `stateChanged` — Katakchaga belgi qo'yilganda yoki olib tashlanganda.


* **`QSpinBox` (Raqam kiritish qutisi)**
* `valueChanged` — Raqam qiymati o'zgarganda (xuddi slider kabi ishlaydi).


* **`QRadioButton` (Yakka tanlov tugmasi)**
* `toggled` — Tugma holati yoqilganda yoki o'chirilganda.



Bu signallarning barchasi kodda `element.signal_nomi.connect(funksiya)` shaklida kerakli jarayonga ulanadi.

## 6. O'zingizning Signal yaratish

```python
from PyQt5.QtCore import QObject, pyqtSignal

class Hisoblagich(QObject):

    # O'z signalimizni e'lon qilamiz
    qiymat_ozgardi = pyqtSignal(int)        # int yuboradi
    matn_yuborildi = pyqtSignal(str)        # str yuboradi
    togri_javob    = pyqtSignal(int, str)   # int va str yuboradi
    oddiy_signal   = pyqtSignal()           # hech narsa yubormaydi

    def __init__(self):
        super().__init__()
        self.qiymat = 0

    def oshir(self):
        self.qiymat += 1
        self.qiymat_ozgardi.emit(self.qiymat)  # ← signalni "otish"
        self.matn_yuborildi.emit(f"Qiymat: {self.qiymat}")

# Ishlatish:
hisoblagich = Hisoblagich()
hisoblagich.qiymat_ozgardi.connect(lambda x: print(f"Yangi qiymat: {x}"))
hisoblagich.matn_yuborildi.connect(print)
```

## QThread nima?

**QThread** — bu PyQt'da **parallel (bir vaqtda) ishlaydigan alohida "ip"** (thread) yaratish uchun ishlatiladigan sinf.

---

### Oddiy misol bilan tushuntirish

Tasavvur qiling: siz restoranda ovqat buyurtma qildingiz.

- **Asosiy thread** = ofitsiant (foydalanuvchi bilan gaplashadi, UI ni boshqaradi)
- **QThread** = oshpaz (orqa tomonda ovqat tayyorlaydi)

Agar oshpaz yo'q bo'lsa, ofitsiant ovqat tayyorlanguncha **hamma narsani to'xtatib** kutishi kerak edi. Bu esa dasturda **UI ning "muzlab" qolishiga** olib keladi.

---

### Qachon kerak bo'ladi?

| Holat | Muammo | Yechim |
|-------|--------|--------|
| Katta fayl yuklash | UI muzlaydi | QThread |
| Internet so'rovi | Dastur javob bermaydi | QThread |
| Uzoq hisob-kitob | Tugmalar ishlamaydi | QThread |

### Sodda kod misoli


````python
from PyQt5.QtCore import QThread, pyqtSignal
from PyQt5.QtWidgets import QApplication, QMainWindow, QPushButton, QLabel
import time

# 1. QThread dan meros olamiz
class IshchiThread(QThread):
    natija = pyqtSignal(str)  # Signal — tugaganda xabar yuboradi

    def run(self):
        # Bu qism orqa fonda ishlaydi
        time.sleep(3)  # Og'ir ish simulyatsiyasi
        self.natija.emit("Ish tugadi! ✅")

# 2. Asosiy oyna
class Oyna(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("QThread misoli")
        self.resize(300, 150)

        self.label = QLabel("Tayyor", self)
        self.label.move(100, 40)

        self.tugma = QPushButton("Ishni boshlash", self)
        self.tugma.move(80, 80)
        self.tugma.clicked.connect(self.boshlash)

    def boshlash(self):
        self.label.setText("Ishlayapti... ⏳")
        self.thread = IshchiThread()
        self.thread.natija.connect(self.tugallandi)  # Signal ulash
        self.thread.start()  # Threadni ishga tushurish

    def tugallandi(self, xabar):
        self.label.setText(xabar)

app = QApplication([])
oyna = Oyna()
oyna.show()
app.exec_()
````



---

### Asosiy tushunchalar

```
QThread
  ├── run()        → orqa fonda bajariladigan kod shu yerda
  ├── start()      → threadni ishga tushiradi
  ├── quit()       → threadni to'xtatadi
  └── pyqtSignal   → thread → UI ga ma'lumot yuborish yo'li
```

**Muhim qoida:** Thread ichidan to'g'ridan-to'g'ri UI ni o'zgartirish **mumkin emas** — buning uchun **Signal/Slot** mexanizmidan foydalaniladi.

**QMessageBox** — foydalanuvchiga **qisqa xabar ko'rsatish** yoki **qaror qabul qildirish** uchun ishlatiladigan tayyor dialog oynasi.

```
Dastur ──→ "Faylni o'chirasizmi?" ──→ Foydalanuvchi: Ha / Yo'q
Dastur ──→ "Xato yuz berdi!"      ──→ Foydalanuvchi: OK
Dastur ──→ "Saqlandi ✅"           ──→ Foydalanuvchi: OK
```

---

## 2. Ishlatishning 2 usuli

```
1. Statik funksiyalar  → Tez, 1 qator kod  (ko'p ishlatiladi)
2. Ob'ekt yaratish     → Moslashuvchan, kengaytirish mumkin
```

## 3. Statik Funksiyalar — Tez Ishlatish

### 📘 information — Ma'lumot berish
```python
from PyQt5.QtWidgets import QMessageBox

QMessageBox.information(
    self,                        # ota widget
    "Sarlavha",                  # oyna sarlavhasi
    "Fayl muvaffaqiyatli saqlandi!",  # xabar
    QMessageBox.Ok               # tugmalar (ixtiyoriy)
)
```
> 🔵 Ko'k "i" belgisi — oddiy ma'lumot

---

### ⚠️ warning — Ogohlantirish
```python
QMessageBox.warning(
    self,
    "Ogohlantirish",
    "Fayl o'zgartirilgan! Saqlamasdan chiqmoqchimisiz?",
    QMessageBox.Yes | QMessageBox.No  # ikki tugma
)
```
> 🟡 Sariq uchburchak — ehtiyot bo'lish kerak

---

### ❌ critical — Jiddiy xato
```python
QMessageBox.critical(
    self,
    "Xato",
    "Ma'lumotlar bazasiga ulanib bo'lmadi!",
    QMessageBox.Ok
)
```
> 🔴 Qizil X — jiddiy muammo

---

### ❓ question — Savol berish
```python
javob = QMessageBox.question(
    self,
    "Tasdiqlash",
    "Haqiqatan ham o'chirmoqchimisiz?",
    QMessageBox.Yes | QMessageBox.No,
    QMessageBox.No   # default tugma
)

if javob == QMessageBox.Yes:
    print("O'chirildi!")
else:
    print("Bekor qilindi!")
```
> 🟣 So'roq belgisi — foydalanuvchidan javob so'rash

## 6. Ob'ekt Yaratib Ishlatish — Moslashuvchan Usul

```python
from PyQt5.QtWidgets import QMessageBox
from PyQt5.QtGui import QIcon, QPixmap
from PyQt5.QtCore import Qt

xabar = QMessageBox()

# Sarlavha va matn
xabar.setWindowTitle("Muhim Xabar")
xabar.setText("Asosiy xabar matni")                    # katta matn
xabar.setInformativeText("Qo'shimcha tushuntirish...")  # kichik matn
xabar.setDetailedText("Log: xato 404...\nFile: main.py") # "Batafsil" tugmasi

# Ikon turi
xabar.setIcon(QMessageBox.Information)  # yoki Warning, Critical, Question

# Tugmalar
xabar.setStandardButtons(QMessageBox.Yes | QMessageBox.No | QMessageBox.Cancel)
xabar.setDefaultButton(QMessageBox.Yes)   # Enter bosilganda
xabar.setEscapeButton(QMessageBox.Cancel) # Esc bosilganda

# Ko'rsatish va javob olish
natija = xabar.exec_()

if natija == QMessageBox.Yes:
    print("Ha!")
```

## 9. To'liq Amaliy Misol

```python
import sys
from PyQt5.QtWidgets import *
from PyQt5.QtGui import QIcon
from PyQt5.QtCore import Qt


class AsosiyOyna(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("QMessageBox Namunalar")
        self.setFixedSize(420, 380)
        self.setStyleSheet("background-color: #f0f2f5;")
        self.interfeysni_yaratish()

    def interfeysni_yaratish(self):
        markaziy = QWidget()
        self.setCentralWidget(markaziy)
        tartib = QVBoxLayout(markaziy)
        tartib.setContentsMargins(40, 30, 40, 30)
        tartib.setSpacing(14)

        sarlavha = QLabel("QMessageBox Turlari")
        sarlavha.setAlignment(Qt.AlignCenter)
        sarlavha.setStyleSheet("""
            font-size: 20px; font-weight: bold;
            color: white; background-color: #2c3e50;
            border-radius: 10px; padding: 12px;
        """)
        tartib.addWidget(sarlavha)

        tugmalar = [
            ("🔵  Information",  "#2980b9", self.info_korsot),
            ("🟡  Warning",      "#f39c12", self.warning_korsot),
            ("🔴  Critical",     "#e74c3c", self.critical_korsot),
            ("❓  Question",     "#8e44ad", self.question_korsot),
            ("⚙️  Moslashtirilgan", "#27ae60", self.custom_korsot),
        ]

        for matn, rang, funksiya in tugmalar:
            tugma = QPushButton(matn)
            tugma.setMinimumHeight(42)
            tugma.setStyleSheet(f"""
                QPushButton {{
                    background-color: {rang};
                    color: white;
                    font-size: 14px;
                    font-weight: bold;
                    border-radius: 8px;
                    border: none;
                }}
                QPushButton:hover {{ opacity: 0.85; }}
                QPushButton:pressed {{ padding-top: 2px; }}
            """)
            tugma.clicked.connect(funksiya)
            tartib.addWidget(tugma)

    # ── Statik funksiyalar ──────────────────────────────

    def info_korsot(self):
        QMessageBox.information(
            self,
            "Ma'lumot",
            "✅ Fayl muvaffaqiyatli saqlandi!\nYo'l: /home/user/hujjat.txt",
            QMessageBox.Ok
        )

    def warning_korsot(self):
        javob = QMessageBox.warning(
            self,
            "Ogohlantirish",
            "⚠️ Faylda saqlanmagan o'zgarishlar bor.\n"
            "Saqlamasdan chiqmoqchimisiz?",
            QMessageBox.Save | QMessageBox.Discard | QMessageBox.Cancel,
            QMessageBox.Save   # default
        )
        if javob == QMessageBox.Save:
            self.natija_label_yangilar("💾 Saqlandi va chiqildi")
        elif javob == QMessageBox.Discard:
            self.natija_label_yangilar("🗑️ Saqlanmadi, chiqildi")
        else:
            self.natija_label_yangilar("↩️ Bekor qilindi")

    def critical_korsot(self):
        xabar = QMessageBox(self)
        xabar.setWindowTitle("Kritik Xato")
        xabar.setIcon(QMessageBox.Critical)
        xabar.setText("❌ Ulanishda xatolik yuz berdi!")
        xabar.setInformativeText(
            "Ma'lumotlar bazasiga ulanib bo'lmadi."
        )
        xabar.setDetailedText(
            "Xato: Connection refused\n"
            "Host: localhost:5432\n"
            "Vaqt: 2024-01-15 14:30:22\n"
            "Qayta urinib ko'ring yoki administratorga murojaat qiling."
        )
        xabar.setStandardButtons(QMessageBox.Retry | QMessageBox.Abort)
        xabar.setDefaultButton(QMessageBox.Retry)

        natija = xabar.exec_()
        if natija == QMessageBox.Retry:
            self.natija_label_yangilar("🔄 Qayta urinilmoqda...")
        else:
            self.natija_label_yangilar("🛑 To'xtatildi")

    def question_korsot(self):
        javob = QMessageBox.question(
            self,
            "Tasdiqlash",
            "🗑️ Tanlangan 5 ta faylni o'chirmoqchimisiz?\n"
            "Bu amalni qaytarib bo'lmaydi!",
            QMessageBox.Yes | QMessageBox.No,
            QMessageBox.No   # xavfsiz default
        )
        if javob == QMessageBox.Yes:
            self.natija_label_yangilar("🗑️ Fayllar o'chirildi!")
        else:
            self.natija_label_yangilar("✋ O'chirish bekor qilindi")

    def custom_korsot(self):
        """Ob'ekt yaratib, to'liq moslashtirilgan"""
        xabar = QMessageBox(self)
        xabar.setWindowTitle("Yangilanish mavjud")
        xabar.setText("🚀 Yangi versiya mavjud: v2.0.1")
        xabar.setInformativeText(
            "Yangilanishni hozir o'rnatmoqchimisiz?"
        )
        xabar.setDetailedText(
            "Yangiliklar:\n"
            "- Tezlik 40% oshirildi\n"
            "- 12 ta xato tuzatildi\n"
            "- Yangi dark mode qo'shildi"
        )
        xabar.setIcon(QMessageBox.Information)
        xabar.setStandardButtons(
            QMessageBox.Yes | QMessageBox.No | QMessageBox.Cancel
        )
        xabar.setDefaultButton(QMessageBox.Yes)
        xabar.setEscapeButton(QMessageBox.Cancel)

        xabar.setStyleSheet("""
            QMessageBox {
                background-color: #1e272e;
            }
            QMessageBox QLabel {
                color: #dfe6e9;
                font-size: 13px;
                min-width: 280px;
            }
            QPushButton {
                background-color: #0984e3;
                color: white;
                border-radius: 6px;
                padding: 6px 18px;
                font-weight: bold;
                min-width: 75px;
            }
            QPushButton:hover {
                background-color: #74b9ff;
            }
        """)

        natija = xabar.exec_()
        if natija == QMessageBox.Yes:
            self.natija_label_yangilar("⬇️ Yangilanish boshlandi...")
        elif natija == QMessageBox.No:
            self.natija_label_yangilar("⏭️ Keyinroq eslatiladi")
        else:
            self.natija_label_yangilar("❌ Bekor qilindi")

    def natija_label_yangilar(self, matn):
        QMessageBox.information(self, "Natija", matn)


app = QApplication(sys.argv)
oyna = AsosiyOyna()
oyna.show()
sys.exit(app.exec_())
```

## 10. Xulosa Jadvali

| Tur | Statik funksiya | Ikona | Qachon ishlatish |
|-----|----------------|-------|-----------------|
| Information | `QMessageBox.information()` | 🔵 | Muvaffaqiyat, ma'lumot |
| Warning | `QMessageBox.warning()` | 🟡 | Ehtiyot bo'lish kerak |
| Critical | `QMessageBox.critical()` | 🔴 | Jiddiy xato |
| Question | `QMessageBox.question()` | ❓ | Foydalanuvchidan qaror |

```
Statik funksiya  → 1 qator, tez, sodda
Ob'ekt yaratish  → setDetailedText, StyleSheet, maxsus tugmalar kerak bo'lganda
```

> 💡 **Eslab qol:** `question()` dan qaytgan qiymatni **har doim** tekshir — foydalanuvchi `Yes` yoki `No` bosganini bilish shart!

```
QMainWindow (oyna)
    ↓
QMenuBar (menyu paneli: Fayl, Tahrirlash, Yordamasi...)
    ↓
```

**Analogy:**
```
Restorandagi menyu
├─ Taomlar bo'limi (QMenuBar)
│  ├─ Birinchi taom (QMenu)
│  │  ├─ Shorba (QAction)
│  │  ├─ Salatasi (QAction)
│  │  └─ ...
│  ├─ Ikkinchi taom (QMenu)
│  └─ ...
```

## 4. QMenu — Ichki Menyular

```python
class AsosiyOyna(QMainWindow):
    def __init__(self):
        super().__init__()

        menubar = self.menuBar()

        # === FAYL MENYU ===
        fayl_menyu = menubar.addMenu("📁 Fayl")

        # Harakat yaratish
        yangi_action = QAction("Yangi", self)
        yangi_action.setShortcut("Ctrl+N")
        yangi_action.triggered.connect(self.yangi_fayl)

        ochish_action = QAction("Ochish", self)
        ochish_action.setShortcut("Ctrl+O")
        ochish_action.triggered.connect(self.fayl_ochish)

        saqlash_action = QAction("Saqlash", self)
        saqlash_action.setShortcut("Ctrl+S")
        saqlash_action.triggered.connect(self.fayl_saqlash)

        chiqish_action = QAction("Chiqish", self)
        chiqish_action.setShortcut("Ctrl+Q")
        chiqish_action.triggered.connect(self.close)

        # Menyuga harakat qo'shish
        fayl_menyu.addAction(yangi_action)
        fayl_menyu.addAction(ochish_action)
        fayl_menyu.addSeparator()      # ──── chiziq
        fayl_menyu.addAction(saqlash_action)
        fayl_menyu.addSeparator()
        fayl_menyu.addAction(chiqish_action)

        # === TAHRIRLASH MENYU ===
        tahrirlash_menyu = menubar.addMenu("✏️ Tahrirlash")

        kesish_action = QAction("Kesish", self)
        kesish_action.setShortcut("Ctrl+X")
        tahrirlash_menyu.addAction(kesish_action)

        nusxa_action = QAction("Nusxa", self)
        nusxa_action.setShortcut("Ctrl+C")
        tahrirlash_menyu.addAction(nusxa_action)

        yapistir_action = QAction("Yapistir", self)
        yapistir_action.setShortcut("Ctrl+V")
        tahrirlash_menyu.addAction(yapistir_action)

    def yangi_fayl(self):
        print("Yangi fayl yaratildi!")

    def fayl_ochish(self):
        print("Fayl oynasi ochildi!")

    def fayl_saqlash(self):
        print("Fayl saqlandi!")
```

## 9. To'liq Amaliy Misol

```python
import sys
from PyQt5.QtWidgets import *
from PyQt5.QtGui import QIcon, QKeySequence, QPixmap
from PyQt5.QtCore import Qt


class NotePadApp(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("NotePad - Menyu Namunasi")
        self.setGeometry(200, 100, 700, 500)
        self.setStyleSheet("background-color: #f5f5f5;")

        # Markaziy widget
        self.text_edit = QPlainTextEdit()
        self.text_edit.setPlainText("Matnni shu yerga yozing...")
        self.setCentralWidget(self.text_edit)

        # Status bar
        self.statusBar().showMessage("Tayyor", 5000)

        # Menyu yaratish
        self.menyu_yaratish()

    def menyu_yaratish(self):
        menubar = self.menuBar()

        # ════════════════ FAYL MENYU ════════════════
        fayl_menyu = menubar.addMenu("📁 Fayl")

        # Yangi
        yangi_action = QAction("Yangi", self)
        yangi_action.setShortcut(QKeySequence.New)  # Ctrl+N
        yangi_action.setStatusTip("Yangi fayl yaratish")
        yangi_action.setIcon(self.get_icon(QStyle.SP_FileIcon))
        yangi_action.triggered.connect(self.yangi_fayl)
        fayl_menyu.addAction(yangi_action)

        # Ochish
        ochish_action = QAction("Ochish", self)
        ochish_action.setShortcut(QKeySequence.Open)  # Ctrl+O
        ochish_action.setStatusTip("Faylni ochish")
        ochish_action.setIcon(self.get_icon(QStyle.SP_DirOpenIcon))
        ochish_action.triggered.connect(self.fayl_ochish)
        fayl_menyu.addAction(ochish_action)

        # Saqlash
        saqlash_action = QAction("Saqlash", self)
```

```python
def yangi_fayl(self):
    """Yangi fayl yaratish"""
    if self.text_area.document().isModified():
        javob = QMessageBox.question(
            self, "Saqlash",
            "O'zgarishlarni saqlaysizmi?",
            QMessageBox.Save | QMessageBox.Discard | QMessageBox.Cancel
        )
        if javob == QMessageBox.Save:
            self.fayl_saqlash()
        elif javob == QMessageBox.Cancel:
            return

    self.text_area.clear()
    self.joriy_fayl = None
    self.setWindowTitle("Yangi fayl — MatnMuharriri")
    self.statusBar().showMessage("Yangi fayl", 2000)


def fayl_ochish(self):
    """Faylni diskdan ochish"""
    fayl_nomi, _ = QFileDialog.getOpenFileName(
        self,
        "Faylni ochish",
        "",                                      # boshlang'ich papka
        "Matn fayllar (*.txt);;Barcha fayllar (*)"
    )
    if fayl_nomi:
        try:
            with open(fayl_nomi, 'r', encoding='utf-8') as f:
                self.text_area.setPlainText(f.read())
            self.joriy_fayl = fayl_nomi
            self.setWindowTitle(f"{fayl_nomi} — MatnMuharriri")
            self.statusBar().showMessage(f"Ochildi: {fayl_nomi}", 3000)
        except Exception as e:
            QMessageBox.critical(self, "Xato", f"Fayl ochilmadi:\n{e}")


def fayl_saqlash(self):
    """Joriy faylga saqlash"""
    if self.joriy_fayl:
        self._saqlash(self.joriy_fayl)
    else:
        self.fayl_saqlash_sifatida()  # Yangi fayl → dialog ochiladi


def fayl_saqlash_sifatida(self):
    """Yangi nom bilan saqlash"""
    fayl_nomi, _ = QFileDialog.getSaveFileName(
        self,
        "Saqlash sifatida",
        "",
        "Matn fayllar (*.txt);;Barcha fayllar (*)"
    )
    if fayl_nomi:
        self._saqlash(fayl_nomi)


def _saqlash(self, fayl_nomi):
    """Ichki saqlash funksiyasi"""
    try:
        with open(fayl_nomi, 'w', encoding='utf-8') as f:
            f.write(self.text_area.toPlainText())
        self.joriy_fayl = fayl_nomi
        self.setWindowTitle(f"{fayl_nomi} — MatnMuharriri")
        self.text_area.document().setModified(False)  # ← o'zgarish belgisi o'chadi
        self.statusBar().showMessage(f"Saqlandi: {fayl_nomi}", 3000)
    except Exception as e:
        QMessageBox.critical(self, "Xato", f"Saqlab bo'lmadi:\n{e}")


def chiqish(self):
    """Dasturdan chiqish"""
    if self.text_area.document().isModified():
        javob = QMessageBox.question(
            self, "Chiqish",
            "Saqlanmagan o'zgarishlar bor. Chiqasizmi?",
            QMessageBox.Save | QMessageBox.Discard | QMessageBox.Cancel
        )
        if javob == QMessageBox.Save:
            self.fayl_saqlash()
        elif javob == QMessageBox.Cancel:
            return
    QApplication.quit()
```

```python
                    self.text_area.setPlainText(f.read())
                self.joriy_fayl = nom
                self.setWindowTitle(f"{nom} — MatnMuharriri")
            except Exception as e:
                QMessageBox.critical(self, "Xato", f"Ochilmadi:\n{e}")

    def fayl_saqlash(self):
        if self.joriy_fayl: self._saqlash(self.joriy_fayl)
        else: self.fayl_saqlash_sifatida()

    def fayl_saqlash_sifatida(self):
        nom, _ = QFileDialog.getSaveFileName(
            self, "Saqlash sifatida", "",
            "Matn fayllar (*.txt);;Barcha fayllar (*)")
        if nom: self._saqlash(nom)

    def _saqlash(self, nom):
        try:
            with open(nom, 'w', encoding='utf-8') as f:
                f.write(self.text_area.toPlainText())
            self.joriy_fayl = nom
            self.setWindowTitle(f"{nom} — MatnMuharriri")
            self.text_area.document().setModified(False)
            self.statusBar().showMessage(f"Saqlandi: {nom}", 3000)
        except Exception as e:
            QMessageBox.critical(self, "Xato", f"Saqlanmadi:\n{e}")

    def chiqish(self):
        if self.text_area.document().isModified():
            j = QMessageBox.question(self, "Chiqish",
                    "Saqlanmagan o'zgarishlar bor. Chiqasizmi?",
                    QMessageBox.Save | QMessageBox.Discard | QMessageBox.Cancel)
            if j == QMessageBox.Save:    self.fayl_saqlash()
            elif j == QMessageBox.Cancel: return
        QApplication.quit()

    def closeEvent(self, event):
        """Oyna X tugmasi bosilganda"""
        if self.text_area.document().isModified():
            j = QMessageBox.question(self, "Chiqish",
                    "Saqlanmagan o'zgarishlar bor. Chiqasizmi?",
                    QMessageBox.Save | QMessageBox.Discard | QMessageBox.Cancel)
            if j == QMessageBox.Save:
                self.fayl_saqlash()
                event.accept()
            elif j == QMessageBox.Discard:
                event.accept()
            else:
                event.ignore()   # ← oyna yopilmaydi!
        else:
            event.accept()


if __name__ == "__main__":
    app = QApplication(sys.argv)
    oyna = MatnMuharriri()
    oyna.show()
    sys.exit(app.exec_())
```
