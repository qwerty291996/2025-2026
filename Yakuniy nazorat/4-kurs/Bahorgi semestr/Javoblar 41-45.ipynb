## 4. Dot Product — Skalyar Ko'paytma

```
Nazariy asos:

Vektor dot product:
  a·b = Σ aᵢbᵢ = a₁b₁ + a₂b₂ + ... + aₙbₙ
  a·b = ||a|| · ||b|| · cos(θ)

  Geometrik ma'no:
    a·b = 0 → Vektorlar perpendikulyar (ortogonal)
    a·b > 0 → Bir tomonga yo'nalgan (θ < 90°)
    a·b < 0 → Qarama-qarshi tomonda (θ > 90°)

Matritsa ko'paytmasi:
  (AB)ᵢⱼ = Σₖ AᵢₖBₖⱼ
  A: (m×k), B: (k×n) → AB: (m×n)
  Ichki o'lchamlar teng bo'lishi shart!
```

```python
# Vektor dot product:
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

d1 = np.dot(a, b)      # 1×4 + 2×5 + 3×6 = 32
d2 = a @ b             # Python 3.5+ operator
d3 = (a * b).sum()     # Element-wise × keyin yig'ish
# Barchasi: 32

# Matritsa ko'paytmasi:
A = np.array([[1, 2],
              [3, 4]])   # (2,2)
B = np.array([[5, 6],
              [7, 8]])   # (2,2)

C = A @ B   # yoki np.dot(A, B)
# [[1×5+2×7, 1×6+2×8],   [[19, 22],
#  [3×5+4×7, 3×6+4×8]] =  [43, 50]]
```

## 5. Inverse — Teskari Matritsa

```
Nazariy asos:
  A × A⁻¹ = A⁻¹ × A = I (Birlik matritsa)

  Mavjudlik sharti:
  → Faqat KVADRAT matritsa (n×n)
  → Determinant ≠ 0  (det(A) ≠ 0)
  → "Singular bo'lmagan" matritsa

  det(A) = 0 → Teskari mavjud emas (Singular matritsa)
  → Qatorlar chiziqli bog'liq (multikolinearlik!)

2×2 holat:
  A = [[a, b],    A⁻¹ = 1/det(A) × [[d, -b],
       [c, d]]                        [-c, a]]
  det(A) = ad - bc
```

```python
import numpy as np

A = np.array([[4, 7],
              [2, 6]])

# Determinant:
det = np.linalg.det(A)
print(f"det(A) = {det:.1f}")   # 10.0 ≠ 0 → Teskari mavjud

# Teskari matritsa:
A_inv = np.linalg.inv(A)
print(A_inv)
# [[ 0.6, -0.7],
#  [-0.2,  0.4]]

# Tekshirish: A × A⁻¹ = I
I = A @ A_inv
print(np.round(I, 10))
# [[1., 0.],
#  [0., 1.]]  ← Birlik matritsa ✅

# Singular matritsa (det=0):
B = np.array([[1, 2],
              [2, 4]])   # 2-qator = 2 × 1-qator
det_B = np.linalg.det(B)   # ≈ 0
# np.linalg.inv(B) → LinAlgError: Singular matrix!
```

**Pseudoinverse — Singular holatlar uchun:**

```python
# Moore-Penrose Pseudoinverse:
# Har qanday matritsa uchun ishlaydi (kvadrat bo'lmasa ham!)
A_pinv = np.linalg.pinv(A)

# Nima uchun kerak:
# Chiziqli regressiyada n < p bo'lsa (namuna < xususiyat)
# → XᵀX → Singular → inv() ishlamaydi
# → pinv() → Yechim topadi (eng kichik normali)
```

## 6. OLS — Regressiya Koeffitsiyentlarini Analitik Hisoblash

```
Chiziqli regressiya modeli:
  y = Xβ + ε

  X: (n × p) ma'lumot matritsasi
  y: (n × 1) maqsad vektor
  β: (p × 1) koeffitsiyentlar (topish kerak!)
  ε: (n × 1) xatolar

OLS maqsadi:
  min ||y - Xβ||² = min Σ(yᵢ - Xᵢβ)²

Analitik yechim (Normal tenglamalar):
  ∂/∂β [||y - Xβ||²] = 0
  -2Xᵀ(y - Xβ) = 0
  Xᵀy = XᵀXβ
  (XᵀX)β = Xᵀy
  β = (XᵀX)⁻¹Xᵀy  ← Normal tenglamalar yechimi!
```

**Qadamba-qadam hisoblash:**

```python
import numpy as np

def ols_manual(X, y):
    """
    OLS koeffitsiyentlarini analitik hisoblash
    β = (XᵀX)⁻¹Xᵀy
    """
    # Bias uchun birlar ustuni qo'shish:
    n = X.shape[0]
    X_b = np.column_stack([np.ones(n), X])
    # X_b: (n, p+1) → β₀ uchun +1 ustun

    # Bosqichma-bosqich:
    Xt   = X_b.T                          # Transpose: (p+1, n)
    XtX  = Xt @ X_b                       # (p+1, p+1)
    XtX_inv = np.linalg.inv(XtX)          # (p+1, p+1)
    Xty  = Xt @ y                         # (p+1,)
    beta = XtX_inv @ Xty                  # (p+1,) ← Natija!

    return beta

# Misol — Uy narxi:
np.random.seed(42)
n = 100

maydon    = np.random.uniform(50, 200, n)     # m²
xonalar   = np.random.randint(2, 6, n)        # ta
masofa    = np.random.uniform(1, 20, n)       # km

# Haqiqiy model: narx = 1000 + 2*maydon + 50*xonalar - 30*masofa + shovqin
narx = (1000 + 2*maydon + 50*xonalar
        - 30*masofa + np.random.randn(n)*50)

X = np.column_stack([maydon, xonalar, masofa])
y = narx

# OLS hisoblash:
beta = ols_manual(X, y)

print("Topilgan koeffitsiyentlar:")
print(f"  β₀ (intercept): {beta[0]:.2f}  (haqiqiy: 1000)")
print(f"  β₁ (maydon):    {beta[1]:.2f}  (haqiqiy: 2)")
print(f"  β₂ (xonalar):   {beta[2]:.2f}  (haqiqiy: 50)")
print(f"  β₃ (masofa):    {beta[3]:.2f}  (haqiqiy: -30)")

# Sklearn bilan taqqoslash:
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X, y)
print(f"\nSklearn: {[lr.intercept_] + list(lr.coef_)}")
# Deyarli bir xil! ✅
```

## 7. OLS Muammolari va Yechimlar

**Muammo 1 — XᵀX Singular:**

```python
# Multikolinearlik → det(XᵀX) ≈ 0 → inv() ishlamaydi

# Yechim 1 — Pseudoinverse:
def ols_pinv(X, y):
    X_b = np.column_stack([np.ones(len(X)), X])
    beta = np.linalg.pinv(X_b) @ y
    return beta

# Yechim 2 — SVD (Singular Value Decomposition):
def ols_svd(X, y):
    X_b = np.column_stack([np.ones(len(X)), X])
    # X_b = U Σ Vᵀ → β = V Σ⁻¹ Uᵀ y
    U, S, Vt = np.linalg.svd(X_b, full_matrices=False)
    beta = Vt.T @ np.diag(1/S) @ U.T @ y
    return beta

# Yechim 3 — Ridge (L2 regularizatsiya):
def ridge_analitik(X, y, alpha=1.0):
    X_b = np.column_stack([np.ones(len(X)), X])
    p = X_b.shape[1]
    # (XᵀX + αI)⁻¹Xᵀy
    beta = np.linalg.inv(X_b.T @ X_b + alpha*np.eye(p)) @ X_b.T @ y
    return beta
    # αI → XᵀX ga qo'shiladi → Hech qachon singular bo'lmaydi!
```

**Muammo 2 — Katta n va p:**

```
n=1,000,000 namuna, p=1,000 xususiyat:
  XᵀX → (1000×1000) matritsa
  inv(XᵀX) → O(p³) = O(10⁹) operatsiya → Juda sekin!

Yechim → Gradient Descent:
  Analitik: β = (XᵀX)⁻¹Xᵀy → Katta n da sekin
  GD: Iterativ yangilash → Katta n da tezroq

Qoida:
  n va p kichik (<10,000) → Analitik (OLS)
  n yoki p katta            → Gradient Descent
```

## 5. Rolling Window — Harakatlanuvchi Oyna

```
Nazariy asos:
  Har nuqta uchun → Oxirgi W ta nuqta oynasi → Statistika

  W=3 misol:
  Ma'lumot: [10, 20, 30, 40, 50, 60]
  O'rtacha:
    [1]:  NaN (yetarli ma'lumot yo'q)
    [2]:  NaN
    [3]: (10+20+30)/3 = 20
    [4]: (20+30+40)/3 = 30
    [5]: (30+40+50)/3 = 40
    [6]: (40+50+60)/3 = 50

  "Siljib boruvchi o'rtacha" (Moving Average)
```

```python
# Moving Average (Harakatlanuvchi o'rtacha):
aksiya["MA_7"]  = aksiya["narx"].rolling(window=7).mean()
aksiya["MA_30"] = aksiya["narx"].rolling(window=30).mean()
aksiya["MA_90"] = aksiya["narx"].rolling(window=90).mean()

# Rolling standart og'ish (Volatillik):
aksiya["volatillik"] = aksiya["narx"].rolling(window=30).std()

# Rolling min/max (Qo'llab-quvvatlash va Qarshilik darajalari):
aksiya["min_30"] = aksiya["narx"].rolling(window=30).min()
aksiya["max_30"] = aksiya["narx"].rolling(window=30).max()

# Rolling korrelyatsiya:
aksiya["narx_2"] = aksiya["narx"] * 0.8 + np.random.randn(len(aksiya)) * 10
rolling_corr = aksiya["narx"].rolling(window=30).corr(aksiya["narx_2"])

# Custom rolling funksiya:
def sharpe_ratio(x, rf=0.02/252):
    """30 kunlik Sharpe nisbati"""
    daromad = x.pct_change().dropna()
    if daromad.std() == 0:
        return np.nan
    return (daromad.mean() - rf) / daromad.std() * np.sqrt(252)

aksiya["sharpe"] = aksiya["narx"].rolling(window=30).apply(sharpe_ratio)
```

**Exponential Moving Average (EMA):**

```python
# EMA — Yangi ma'lumotlarga ko'proq og'irlik beradi!
aksiya["EMA_12"] = aksiya["narx"].ewm(span=12, adjust=False).mean()
aksiya["EMA_26"] = aksiya["narx"].ewm(span=26, adjust=False).mean()

# MACD (Moving Average Convergence Divergence):
aksiya["MACD"] = aksiya["EMA_12"] - aksiya["EMA_26"]
aksiya["Signal"] = aksiya["MACD"].ewm(span=9, adjust=False).mean()
aksiya["Histogram"] = aksiya["MACD"] - aksiya["Signal"]

# SMA vs EMA farqi:
# SMA: Barcha W oyna nuqtalari teng og'irlik
# EMA: Yangi nuqtalar ko'proq og'irlik → Tezroq moslashadi
# Trend aniqlashda EMA keng ishlatiladi
```

## 6. Lag Xususiyatlar — ML uchun

```python
# Vaqt qatorini ML uchun tayyorlash:
# Kechagi qiymat bugungi qiymatni bashorat qiladi?

df = aksiya[["narx"]].copy()

# Lag xususiyatlar (orqaga siljish):
for lag in [1, 2, 3, 5, 10]:
    df[f"narx_lag{lag}"] = df["narx"].shift(lag)
    # shift(1) → Bir kun orqaga
    # Bugungi qiymat ← Kechagi qiymat

# Diff xususiyatlar (o'zgarish):
df["narx_diff1"] = df["narx"].diff(1)      # Kunlik o'zgarish
df["narx_diff5"] = df["narx"].diff(5)      # 5 kunlik o'zgarish

# Foizli o'zgarish:
df["daromad_1d"] = df["narx"].pct_change(1)    # 1 kunlik daromad
df["daromad_5d"] = df["narx"].pct_change(5)    # 5 kunlik daromad

# Rolling xususiyatlar:
df["ma_7"]  = df["narx"].rolling(7).mean()
df["ma_30"] = df["narx"].rolling(30).mean()
df["std_7"] = df["narx"].rolling(7).std()

# Maqsad: Ertangi narx
df["target"] = df["narx"].shift(-1)   # Bir kun olg'a

# NaN larni olib tashlash:
df_clean = df.dropna()

print(f"Xususiyatlar: {df_clean.columns.tolist()}")
print(f"O'lcham: {df_clean.shape}")
```

## 8. Texnik Indikatorlar — Savdo Tahlili

```python
def texnik_indikatorlar(df, narx_ustun="narx"):
    d = df.copy()
    n = narx_ustun

    # RSI (Relative Strength Index):
    delta = d[n].diff()
    o_sish  = delta.clip(lower=0)
    tushish = (-delta).clip(lower=0)
    rs = o_sish.rolling(14).mean() / tushish.rolling(14).mean()
    d["RSI"] = 100 - (100 / (1 + rs))
    # RSI > 70 → Haddan oshiq sotib olingan (Sell signal)
    # RSI < 30 → Haddan oshiq sotilgan (Buy signal)

    # Bollinger Bands:
    o_rtacha = d[n].rolling(20).mean()
    std      = d[n].rolling(20).std()
    d["BB_yuqori"]  = o_rtacha + 2*std
    d["BB_o_rta"]   = o_rtacha
    d["BB_quyi"]    = o_rtacha - 2*std
    d["BB_kenglik"] = d["BB_yuqori"] - d["BB_quyi"]
    # Kenglik kichik → Narx tez o'zgarishi mumkin (Squeeze)

    # MACD:
    ema12 = d[n].ewm(span=12).mean()
    ema26 = d[n].ewm(span=26).mean()
    d["MACD"]   = ema12 - ema26
    d["Signal"] = d["MACD"].ewm(span=9).mean()
    d["MACD_hist"] = d["MACD"] - d["Signal"]

    # Momentum:
    d["momentum_10"] = d[n] / d[n].shift(10) - 1

    # Average True Range (ATR) — Volatillik:
    high = d[n].rolling(1).max()
    low  = d[n].rolling(1).min()
    tr   = high - low
    d["ATR"] = tr.rolling(14).mean()

    return d

aksiya_ind = texnik_indikatorlar(aksiya)
```

## 1. Logistik Regressiya — Arxitektura va Ishlash Prinsipi

**Ta'rif:** Logistik regressiya — chiziqli chegara (linear decision boundary) orqali ikki sinfni ajratuvchi klassifikatsiya algoritmi. Chiqishda **sigmoid funksiyasi** orqali ehtimollik beradi.

```
Arxitektura:

  x₁ ──┐
  x₂ ──┤──── [w₁x₁ + w₂x₂ + w₃x₃ + b] ──→ sigmoid(z) ──→ P(y=1)
  x₃ ──┘

  z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b    ← chiziqli kombinatsiya
  P(y=1) = σ(z) = 1 / (1 + e⁻ᶻ)        ← sigmoid funksiyasi

  Chiqish: 0.0 – 1.0 oraliq (ehtimollik)
  Qaror:   P ≥ 0.5 → sinf 1, P < 0.5 → sinf 0
```

### Sigmoid funksiyasining roli:

```
sigmoid(z):

  z → -∞  :  σ(z) → 0.0   (aniq sinf 0)
  z =  -2 :  σ(z) = 0.12
  z =   0 :  σ(z) = 0.50  (noaniq chegara)
  z =  +2 :  σ(z) = 0.88
  z → +∞  :  σ(z) → 1.0   (aniq sinf 1)

  Nima uchun sigmoid? Chiziqli chiqish (-∞, +∞) ni
  ehtimollikka [0,1] o'tkazadi.
```

### Logistik regressiyaning chegarasi:

```
FAQAT chiziqli ajratish (linear decision boundary):

     x₂
      │      ●●●              Logistik regressiya
      │   ●●●  ●●             faqat SHU chiziqni
      │ ●●●                   chizishi mumkin:
      │───────────────  ←──── w₁x₁ + w₂x₂ + b = 0
      │          ○○○
      │       ○○○○○
      │    ○○○○○
      └──────────────── x₁

  ● va ○ chiziq bilan ajralsa → Logistik Regressiya ishlaydi ✅
  Ajralsa → ishlamaydi ❌
```

## 2. MLP — Arxitektura va Ishlash Prinsipi

**Ta'rif:** MLP — bir nechta qatlamdan (layer) iborat sun'iy neyron tarmoq. Har qatlam o'zidan oldingi qatlamning chiqishini qayta ishlaydi. **Nochiziqli (non-linear) chegaralar** chizishi mumkin.

```
Arxitektura (2 ta yashirin qatlam):

  KIRISH       YASHIRIN-1     YASHIRIN-2     CHIQISH
  QATLAM       (4 neytron)    (3 neytron)    QATLAM

  x₁ ────────▶ [N₁] ────────▶ [N₅] ───────▶ [Out] ──▶ P(y=1)
              ╳ [N₂] ──────╳─▶ [N₆] ───────▶
  x₂ ────────▶ [N₃] ────────▶ [N₇] ───────▶
              ╳ [N₄] ──────╳─▶
  x₃ ────────▶

  Har neytron:
    z = w₁x₁ + w₂x₂ + ... + b   ← chiziqli
    a = ReLU(z)                   ← nochiziqli aktivatsiya

  Nochiziqlilik qayerdan keladi?
    Aktivatsiya funksiyasi (ReLU, Tanh, Sigmoid) tufayli!
    Aktivatsiyasiz → faqat chiziqli model (qancha qatlam
    bo'lmasin, natija chiziqli kombinatsiya)
```

### MLP ning nochiziqli chegarasi:

```
MLP qanday chegaralar chizishi mumkin:

  Oddiy                XOR muammo            Murakkab
  ────────────        ──────────────         ─────────────
     ● ●                ● ──── ○              ●●●
   ────────              │      │            ○─────●
     ○ ○                ○ ──── ●            ●   ○○○

  Logistik Reg: ✅      Logistik Reg: ❌     Logistik Reg: ❌
  MLP:          ✅      MLP:           ✅     MLP:           ✅

  XOR muammo — klassik misol:
    (0,0)→0, (1,1)→0, (0,1)→1, (1,0)→1
    Hech qanday chiziq bu 4 nuqtani to'g'ri ajrata olmaydi!
    MLP — ajratadi, chunki nochiziqli chegara chizadi.
```

## 5. Occam's Razor Prinsipi

**Ta'rif (William of Ockham, XIV asr):** "Kerak bo'lmagan murakkablikni qo'shma. Ikki tushuntirish bir xil natija berse — soddarog'ini tanlа."

```
Lotincha asl: "Entia non sunt multiplicanda praeter necessitatem"
             "Zaruratdan ortiq narsalarni ko'paytirma"

ML dagi tarjimasi:

  Ikki model bir xil aniqlikda ishlasa →
  SODDA modelni tanlash kerak.

  Nima uchun?
  ┌─────────────────────────────────────────────────────┐
  │  1. GENERALIZATSIYA                                 │
  │     Sodda model yangi ma'lumotga yaxshiroq          │
  │     moslashadi — overtfit kamroq                    │
  │                                                     │
  │  2. INTERPRETATSIYA                                 │
  │     Logistik regressiyada w₁=0.73 (yosh) degani:   │
  │     "Yosh 1 birlikka oshsa, log-odds 0.73 oshadi"  │
  │     MLP da bu imkonsiz                              │
  │                                                     │
  │  3. TEZLIK                                          │
  │     Sodda model o'qitish ham, bashorat ham tez      │
  │                                                     │
  │  4. DEBUG                                           │
  │     Nima uchun xato berdi — tushunish oson          │
  └─────────────────────────────────────────────────────┘
```

### Occam's Razor va Bias-Variance:

```
Murakkablik o'qida:

  Xato
   │
   │   Jami xato
   │╲         ╱
   │  ╲  ✅  ╱   ← maqbul nuqta
   │   ╲   ╱
   │Bias ╲╱ Variance
   │      ╳
   │     ╱ ╲
   │    ╱   ╲
   └──────────────── Murakkablik

  Bias (qo'shimcha xato):
    Sodda model → ma'lumotni tushuna olmaydi → underfitting
    Logistik Reg XOR muammoda → yuqori bias

  Variance (beqarorlik):
    Murakkab model → shovqinni ham o'rganadi → overfitting
    MLP kam datada → yuqori variance

  Occam's Razor = Minimum kerakli murakkablikni tanlash
                = Bias-Variance muvozanatining optimal nuqtasi
```

## 6. Qachon Qaysi Modelni Tanlash Kerak?

## 7. Amaliy Qiyoslash

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import make_classification, make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# ── Chiziqli ajraladigan dataset ─────────────────────────────
X_lin, y_lin = make_classification(
    n_samples=500, n_features=2,
    n_informative=2, n_redundant=0, random_state=42
)

# ── Nochiziqli dataset (yarim doiralar) ───────────────────────
X_moon, y_moon = make_moons(n_samples=500, noise=0.2, random_state=42)

for nom, X, y in [("Chiziqli", X_lin, y_lin),
                  ("Nochiziqli (moons)", X_moon, y_moon)]:

    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=0.3, random_state=42
    )
    sc = StandardScaler()
    X_tr = sc.fit_transform(X_tr)
    X_te = sc.transform(X_te)

    # Logistik Regressiya
    lr = LogisticRegression()
    lr.fit(X_tr, y_tr)
    lr_acc = accuracy_score(y_te, lr.predict(X_te))

    # MLP
    mlp = MLPClassifier(hidden_layer_sizes=(64, 32),
                        max_iter=500, random_state=42)
    mlp.fit(X_tr, y_tr)
    mlp_acc = accuracy_score(y_te, mlp.predict(X_te))

    print(f"\n{nom} dataset:")
    print(f"  Logistik Regressiya : {lr_acc:.3f}")
    print(f"  MLP                 : {mlp_acc:.3f}")

# Natija:
# Chiziqli dataset:
#   Logistik Regressiya : 0.947   ← deyarli bir xil!
#   MLP                 : 0.953
#
# Nochiziqli (moons) dataset:
#   Logistik Regressiya : 0.873   ← 5% past
#   MLP                 : 0.947   ← aniqroq
#
# → Chiziqli datada MLP "ortiqcha" — Occam's Razor!
# → Nochiziqli datada MLP zarur
```

## 3. Outlier — Aniqlash va Qaror Qabul Qilish

**Ta'rif:** Outlier — boshqa qiymatlardan juda uzoqda joylashgan, g'ayritabiiy kuzatish. Asl ma'lumot bo'lishi ham, xato kirish ham mumkin.

### Outlier qayerdan keladi?

```
Manba                   Misol                    Qaror
──────────────────────  ─────────────────────    ───────────────
O'lchov xatosi          Yosh = 250               O'chirish
Kiritish xatosi         Maosh = 999999 (9 bosgan)O'chirish
Haqiqiy hodisa          Millioner mijoz          SAQLASH
Tizim xatosi            Sensor: -999 (xato kod)  O'chirish/NaN
Noyob hodisa            Pandemiya davrida sotuv   Saqlash + belgi
```

### Outlier aniqlash metodlari:

```
1. BOXPLOT METODI (IQR):
   Q1 = 25-persentil, Q3 = 75-persentil
   IQR = Q3 - Q1

   Chegara:
     Quyi = Q1 - 1.5 × IQR
     Yuqori = Q3 + 1.5 × IQR
   Chegaradan tashqari → outlier

   Boxplot ko'rinishi:
   ────|──[─────────|─────────]──────────●──●── x
       ↑  ↑         ↑         ↑          ↑
     quyi Q1      median     Q3      outlier

2. Z-SCORE METODI:
   z = (x - μ) / σ
   |z| > 3 → outlier (3σ qoidasi)
   Normal taqsimotda 99.7% qiymat [-3σ, +3σ] ichida

3. SCATTER PLOT:
   Ikki feature o'rtasidagi munosabatda
   "uzoq" nuqtalar ko'z bilan aniqlanadi
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

df = pd.read_csv('titanic.csv')

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Boxplot — IQR asosida outlier
sns.boxplot(y=df['Fare'], ax=axes[0], color='lightblue')
axes[0].set_title('Fare Boxplot')
# → 512, 263 kabi yuqori qiymatlar ko'rinadi

# Scatter — ikki feature orasida outlier
axes[1].scatter(df['Age'], df['Fare'],
                alpha=0.5, color='steelblue', s=15)
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Fare')
axes[1].set_title('Age vs Fare — Scatter')
# → Yuqori chap: yoshli, qimmat chipta — outlier?

# Z-score hisoblash
z_scores = np.abs((df['Fare'] - df['Fare'].mean()) / df['Fare'].std())
outlier_soni = (z_scores > 3).sum()
axes[2].hist(z_scores, bins=50, color='coral', edgecolor='white')
axes[2].axvline(x=3, color='red', linestyle='--', label='z=3 chegara')
axes[2].set_title(f'Z-Score (Outlier: {outlier_soni} ta)')
axes[2].legend()

plt.tight_layout()
plt.show()
```

### Outlier ga nima qilish kerak?

```
┌──────────────────┬──────────────────────────────────────────┐
│ Holat            │ Qaror                                    │
├──────────────────┼──────────────────────────────────────────┤
│ Xato kiritish    │ O'chirish yoki NaN ga almashtirish       │
│ Haqiqiy, noyob   │ Saqlash — model uchun muhim ma'lumot     │
│ Noaniq           │ "outlier_flag" ustuni qo'shib saqlash    │
│ Model uchun      │ RobustScaler (Q1-Q3 asosida) ishlatish   │
│ scaling kerak    │ — outlier ta'sirini kamaytiradi          │
│ Linear model     │ Log/sqrt transformatsiya qo'llash        │
│ Tree-based model │ Outlier ta'sir qilmaydi — saqlash        │
└──────────────────┴──────────────────────────────────────────┘
```

## 4. Korrelyatsiya Tahlili

**Ta'rif:** Korrelyatsiya — ikki feature o'rtasidagi **chiziqli bog'liqlik** darajasi. Pearson korrelyatsiyasi `-1` dan `+1` gacha qiymat oladi.

```
Korrelyatsiya koeffitsiyenti (r):

  r = +1.0  : To'liq musbat bog'liqlik
              (x oshsa y albatta oshadi)
  r = +0.7  : Kuchli musbat bog'liqlik
  r = +0.3  : Zaif musbat bog'liqlik
  r =  0.0  : Bog'liqlik yo'q (yoki nochiziqli)
  r = -0.3  : Zaif manfiy bog'liqlik
  r = -0.7  : Kuchli manfiy bog'liqlik
  r = -1.0  : To'liq manfiy bog'liqlik
              (x oshsa y albatta kamayadi)
```

### Korrelyatsiya va ML:

```
FEATURE SELECTION uchun:
  Target bilan yuqori korrelyatsiya → muhim feature
  Target bilan past korrelyatsiya   → keraksiz bo'lishi mumkin

  Misol — Uy narxi:
    maydon        → r = +0.89 ✅  (kuchli, saqlash)
    xonalar soni  → r = +0.70 ✅  (saqlash)
    yil qurilgan  → r = +0.52 ✅  (saqlash)
    ID raqami     → r = +0.01 ❌  (o'chirish — tasodifiy)

MULTICOLLINEARITY uchun:
  Ikki feature o'rtasida yuqori korrelyatsiya (|r| > 0.85):
  → Ikkalasi bir xil ma'lumot beradi
  → Modelga ikkalasini berish ortiqcha
  → Linear modelda koeffitsientlar beqaror bo'ladi
  → Birini o'chirish kerak

  Misol:
    xonalar_soni ↔ maydon  : r = +0.92 ← MULTICOLLINEARITY!
    → Birini o'chirish yoki PCA ishlatish
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('titanic.csv')[['Survived','Pclass',
                                  'Age','SibSp','Parch','Fare']]
df.dropna(inplace=True)

# Korrelyatsiya matritsa
corr = df.corr()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Heatmap
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, ax=axes[0],
            linewidths=0.5, square=True)
axes[0].set_title('Korrelyatsiya Heatmap')

# Target bilan korrelyatsiya (Feature importance ko'rinishida)
target_corr = corr['Survived'].drop('Survived').sort_values()
colors = ['coral' if v < 0 else 'steelblue' for v in target_corr]
axes[1].barh(target_corr.index, target_corr.values, color=colors)
axes[1].axvline(x=0, color='black', linewidth=0.8)
axes[1].set_title("'Survived' bilan Korrelyatsiya")
axes[1].set_xlabel('Korrelyatsiya koeffitsiyenti')

plt.tight_layout()
plt.show()

# Natija tahlili:
# Fare     → +0.26  (yuqori chipta = 1-sinf = omon qolish ko'p)
# Pclass   → -0.34  (3-sinf = past Pclass qiymati = kam omon qolish)
# Age      → -0.08  (kichik ta'sir)
# SibSp    → -0.04  (deyarli ta'sir yo'q)
```

### Korrelyatsiya ≠ Sabab-oqibat:

```
MUHIM OGOHLANTIRISH:

  Misol 1: Muzqaymoq sotilishi ↔ cho'mishlar soni: r = +0.95
           Sabab: Ikkalasi ham yoz issig'iga bog'liq (confounding variable)
           Muzqaymoq cho'mish sababchisi EMAS!

  Misol 2: Oyoq o'lchami ↔ o'qish ko'nikma: r = +0.80
           Sabab: Ikkalasi ham YOSH bilan bog'liq
           Katta oyoq o'qishni yaxshilamaydi!

  ML da ahamiyati:
    Korrelyatsiya yuqori → feature muhim bo'lishi MUMKIN
    Lekin sabab-oqibat munosabati tahlil qilinishi SHART
    → Domain knowledge (soha bilimlari) zarur
```

## 6. Model Baholash Vizualizatsiyalari

### Confusion Matrix:

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

def cm_chiz(y_true, y_pred, sinflar):
    cm = confusion_matrix(y_true, y_pred)

    # Foizga o'tkazish (har qator bo'yicha)
    cm_foiz = cm.astype(float) / cm.sum(axis=1)[:, np.newaxis] * 100

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # Sonlar bilan
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=sinflar, yticklabels=sinflar, ax=axes[0])
    axes[0].set_title('Confusion Matrix (Son)')
    axes[0].set_ylabel('Haqiqiy')
    axes[0].set_xlabel('Bashorat')

    # Foiz bilan
    sns.heatmap(cm_foiz, annot=True, fmt='.1f', cmap='Oranges',
                xticklabels=sinflar, yticklabels=sinflar, ax=axes[1])
    axes[1].set_title('Confusion Matrix (%)')
    axes[1].set_ylabel('Haqiqiy')
    axes[1].set_xlabel('Bashorat')

    plt.tight_layout()
    plt.show()

# Nima ko'rinadi:
# Diagonal:     To'g'ri bashoratlar (yuqori bo'lsin)
# Off-diagonal: Xatolar (past bo'lsin)
# FP (False Positive): spam emas → spam dedi
# FN (False Negative): spam     → spam emas dedi
```

### Learning Curve — Overfitting ko'rish:

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import learning_curve
from sklearn.neural_network import MLPClassifier

def learning_curve_chiz(model, X, y):
    train_hajmlar, train_ball, test_ball = learning_curve(
        model, X, y,
        train_sizes=np.linspace(0.1, 1.0, 10),
        cv=5, scoring='accuracy', n_jobs=-1
    )

    train_ort = train_ball.mean(axis=1)
    test_ort  = test_ball.mean(axis=1)

    plt.figure(figsize=(8, 5))
    plt.plot(train_hajmlar, train_ort, 'o-',
             color='steelblue', label='Train')
    plt.plot(train_hajmlar, test_ort,  'o-',
             color='coral',    label='Validation')
    plt.fill_between(train_hajmlar,
                     train_ort - train_ball.std(axis=1),
                     train_ort + train_ball.std(axis=1),
                     alpha=0.15, color='steelblue')
    plt.fill_between(train_hajmlar,
                     test_ort - test_ball.std(axis=1),
                     test_ort + test_ball.std(axis=1),
                     alpha=0.15, color='coral')
    plt.title('Learning Curve')
    plt.xlabel('Train namunalar soni')
    plt.ylabel('Aniqlik')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
```

### Learning Curve tahlili:

```
HOLAT 1 — UNDERFITTING (yuqori Bias):

  Aniqlik
  1.0│
  0.9│
  0.8│  ─ ─ ─ ─ ─ ─ ─         Train
  0.7│  ─────────────          Val
  0.6│
     └────────────────→ Ma'lumot
  Ikkalasi ham past → Model juda sodda
  Yechim: murakkab model, ko'proq feature

HOLAT 2 — OVERFITTING (yuqori Variance):

  Aniqlik
  1.0│────────────────          Train (yuqori)
  0.9│
  0.8│
  0.7│  ─ ─ ─ ─ ─ ─ ─ ─       Val (past)
  0.6│
     └────────────────→ Ma'lumot
  Katta tafovut → Model juda murakkab
  Yechim: regularizatsiya, dropout, ko'proq data

HOLAT 3 — IDEAL:

  Aniqlik
  1.0│           ─────────      Train
  0.9│      ─────────           Val
  0.8│  ─────
     └────────────────→ Ma'lumot
  Yaqinlashmoqda → Yaxshi muvozanat ✅
```

## 5. 4-Bosqich: Preprocessing

**Maqsad:** Ma'lumotni model qabul qila oladigan shaklga keltirish. EDA da topilgan muammolar shu yerda hal qilinadi.

```
PREPROCESSING PIPELINE:

  Xom ma'lumot
      │
      ├── NaN to'ldirish
      │     Son    → GroupBy median
      │     Kat    → Mode / "Unknown"
      │
      ├── Yangi feature yaratish (Feature Engineering)
      │     tenure_group = pd.cut(tenure, bins=[...])
      │     charges_per_month = TotalCharges / tenure
      │
      ├── Kategorik kodlash
      │     Ordinal  → OrdinalEncoder
      │     Nominal  → OneHotEncoder (drop='first')
      │
      ├── Transformation
      │     Skewed ustunlar → np.log1p()
      │
      ├── Feature selection
      │     Korrelyatsiya < 0.05 → o'chirish
      │     Multicollinearity > 0.85 → birini o'chirish
      │
      └── Scaling
            Linear/SVM/NN → StandardScaler
            Neural Net    → MinMaxScaler
            Tree-based    → Shart emas
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (StandardScaler,
                                    OneHotEncoder,
                                    OrdinalEncoder)
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np

df = pd.read_csv('churn.csv')

# ── Feature Engineering ─────────────────────────────────────
df['charges_per_month'] = np.where(
    df['tenure'] > 0,
    df['TotalCharges'] / df['tenure'],
    df['MonthlyCharges']
)
df['log_tenure'] = np.log1p(df['tenure'])

# ── Target ajratish ─────────────────────────────────────────
X = df.drop(columns=['Churn', 'customerID'])
y = (df['Churn'] == 'Yes').astype(int)

# ── AVVAL bo'lish, KEYIN scaling ─────────────────────────────
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2,
    stratify=y,           # sinf nisbati saqlansin
    random_state=42
)

# ── Ustunlarni ajratish ─────────────────────────────────────
son_ustunlar = X.select_dtypes(
    include=[np.number]).columns.tolist()
nominal_kat  = ['InternetService', 'PaymentMethod']
ordinal_kat  = ['Contract']

# ── Pipeline ─────────────────────────────────────────────────
son_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler',  StandardScaler())
])

nominal_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(drop='first',
                               sparse_output=False))
])

ordinal_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(
        categories=[['Month-to-month',
                     'One year',
                     'Two year']]))
])

preprocessor = ColumnTransformer([
    ('son',     son_pipeline,     son_ustunlar),
    ('nominal', nominal_pipeline, nominal_kat),
    ('ordinal', ordinal_pipeline, ordinal_kat)
])

# Faqat fit — train dan!
X_train_proc = preprocessor.fit_transform(X_train)
X_test_proc  = preprocessor.transform(X_test)
print(f"Train: {X_train_proc.shape}")
print(f"Test:  {X_test_proc.shape}")
```

## 6. 5-Bosqich: Model Tanlash va O'qitish

### Model tanlash strategiyasi:

```
QOIDA: Sodda → Murakkab (Occam's Razor)

  1. BASELINE MODEL
     → DummyClassifier (eng sodda, hech narsa o'rganmaydi)
     → Bu eng past chegara — har qanday model bundan yaxshi bo'lsin

  2. ODDIY MODEL
     → Logistik Regressiya
     → Tez, tushunarli, kuchli baseline

  3. TREE-BASED MODEL
     → Random Forest, XGBoost
     → Ko'p hollarda eng yaxshi natija

  4. MURAKKAB MODEL
     → MLP, faqat zarur bo'lganda

TANLOV KRITERIYLARI:
  ┌─────────────────┬────────────────────────────────────┐
  │ Holat           │ Model                              │
  ├─────────────────┼────────────────────────────────────┤
  │ Interpretatsiya │ Logistik Regressiya                │
  │ kerak (bank)    │ (koeffitsientlar aniq ma'noli)     │
  ├─────────────────┼────────────────────────────────────┤
  │ Tez natija kerak│ Random Forest (default parametr)  │
  ├─────────────────┼────────────────────────────────────┤
  │ Aniqlik muhim   │ XGBoost + Tuning                  │
  ├─────────────────┼────────────────────────────────────┤
  │ Ko'p ma'lumot,  │ MLP / Deep Learning               │
  │ nochiziqli      │                                    │
  └─────────────────┴────────────────────────────────────┘
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (classification_report,
                              roc_auc_score,
                              f1_score)
import xgboost as xgb
import time

modellar = {
    'Baseline (Dummy)': DummyClassifier(
        strategy='most_frequent'),

    'Logistik Regressiya': LogisticRegression(
        class_weight='balanced',   # imbalanced uchun
        max_iter=1000,
        random_state=42),

    'Random Forest': RandomForestClassifier(
        n_estimators=200,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1),

    'XGBoost': xgb.XGBClassifier(
        n_estimators=200,
        scale_pos_weight=5,        # imbalanced: 85/15 → ~5.6
        eval_metric='logloss',
        random_state=42,
        verbosity=0)
}

natijalar = {}
for nom, model in modellar.items():
    boshlanish = time.time()
    model.fit(X_train_proc, y_train)
    vaqt = time.time() - boshlanish

    y_pred      = model.predict(X_test_proc)
    y_pred_proba= model.predict_proba(X_test_proc)[:,1]

    natijalar[nom] = {
        'F1':    f1_score(y_test, y_pred, average='weighted'),
        'AUC':   roc_auc_score(y_test, y_pred_proba),
        'Vaqt':  f"{vaqt:.2f}s"
    }

    print(f"\n{'═'*40}")
    print(f"{nom}")
    print(f"{'═'*40}")
    print(classification_report(y_test, y_pred,
          target_names=['Qoldi','Ketdi']))
    print(f"AUC-ROC: {natijalar[nom]['AUC']:.4f}")
    print(f"Vaqt:    {natijalar[nom]['Vaqt']}")

# Natija qiyosi:
# Baseline:            AUC=0.50  F1=0.72  (kam)
# Logistik Regressiya: AUC=0.84  F1=0.79
# Random Forest:       AUC=0.91  F1=0.85
# XGBoost:             AUC=0.93  F1=0.87  ← eng yaxshi
```

### Hyperparameter Tuning:

```python
from sklearn.model_selection import RandomizedSearchCV
import xgboost as xgb

# XGBoost uchun parametrlar oralig'i
param_tarqatish = {
    'n_estimators':     [100, 200, 300, 500],
    'max_depth':        [3, 4, 5, 6, 7],
    'learning_rate':    [0.01, 0.05, 0.1, 0.2],
    'subsample':        [0.6, 0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.6, 0.7, 0.8, 0.9, 1.0],
    'min_child_weight': [1, 3, 5]
}

xgb_model = xgb.XGBClassifier(
    scale_pos_weight=5,
    eval_metric='logloss',
    random_state=42,
    verbosity=0
)

# RandomizedSearch — GridSearch dan tezroq
qidiruv = RandomizedSearchCV(
    xgb_model,
    param_tarqatish,
    n_iter=50,       # 50 ta tasodifiy kombinatsiya
    cv=5,            # 5-fold cross-validation
    scoring='roc_auc',
    n_jobs=-1,
    random_state=42,
    verbose=1
)

qidiruv.fit(X_train_proc, y_train)

print(f"Eng yaxshi parametrlar: {qidiruv.best_params_}")
print(f"Eng yaxshi AUC: {qidiruv.best_score_:.4f}")
eng_yaxshi = qidiruv.best_estimator_
```

```python
# ── Bashorat funksiyasi ─────────────────────────────────────
import joblib
import pandas as pd
import numpy as np

class ChurnPredictor:
    """Production uchun wrapper klass"""

    def __init__(self, preprocessor_yol, model_yol,
                       metadata_yol):
        self.preprocessor = joblib.load(preprocessor_yol)
        self.model        = joblib.load(model_yol)
        with open(metadata_yol) as f:
            self.meta = json.load(f)
        self.threshold = self.meta['threshold']

    def bashorat(self, mijoz_data: dict) -> dict:
        """
        Bitta mijoz uchun churn ehtimolligi.
        mijoz_data: {'tenure': 12, 'Contract': 'Month-to-month', ...}
        """
        df = pd.DataFrame([mijoz_data])
        X_proc = self.preprocessor.transform(df)
        ehtimol = self.model.predict_proba(X_proc)[0, 1]
        qaror   = int(ehtimol >= self.threshold)

        return {
            'churn_ehtimoli': round(float(ehtimol), 4),
            'qaror':          'KETADI' if qaror else 'QOLADI',
            'xavf_darajasi':  'YUQORI' if ehtimol > 0.7
                              else 'O\'RTA' if ehtimol > 0.4
                              else 'PAST',
            'model_versiya':  self.meta['versiya']
        }

# Ishlatish:
predictor = ChurnPredictor(
    'preprocessor.pkl',
    'xgboost_churn.pkl',
    'model_metadata.json'
)

yangi_mijoz = {
    'tenure': 3,
    'MonthlyCharges': 75.5,
    'TotalCharges': 226.5,
    'Contract': 'Month-to-month',
    'InternetService': 'Fiber optic',
    'PaymentMethod': 'Electronic check'
}

natija = predictor.bashorat(yangi_mijoz)
print(natija)
# {'churn_ehtimoli': 0.7821,
#  'qaror': 'KETADI',
#  'xavf_darajasi': 'YUQORI',
#  'model_versiya': '1.0.0'}
```

## 9. 8-Bosqich: Monitoring va Retraining

```
DEPLOYMENT DAN KEYIN MUAMMOLAR:

  DATA DRIFT:
    Ma'lumot xususiyatlari o'zgaradi
    Misol: Pandемiya → mijoz xulqi butun o'zgardi
    → Model 2019 da o'rgatilgan, 2021 da yomon ishlaydi

  CONCEPT DRIFT:
    Target munosabati o'zgaradi
    Misol: "Spam" tushunchasi 5 yilda o'zgardi
    → Eski labellar yangi haqiqatni aks ettirmaydi

  MODEL DEGRADATION:
    Asta-sekin aniqlik pasayadi
    → Har oy AUC kuzatilsin

MONITORING TIZIMI:
  ┌────────────────────────────────────────────────────────┐
  │ Metrika          │ Ogohlantirish  │ Amal              │
  ├──────────────────┼────────────────┼───────────────────┤
  │ AUC-ROC          │ < 0.85         │ Retraining        │
  │ Precision        │ < 0.75         │ Threshold tuning  │
  │ Recall           │ < 0.80         │ Retraining        │
  │ Feature drift    │ PSI > 0.2      │ Feature check     │
  │ Prediction drift │ Taqsimot o'zgardi│ Model yangilash  │
  └──────────────────┴────────────────┴───────────────────┘

RETRAINING STRATEGIYALARI:
  Davriy (Scheduled):  Har oy yangi data bilan qayta o'qitish
  Triggered:           Aniqlik chegaradan tushsa → trigger
  Online Learning:     Har yangi ma'lumot → model yangilanadi
                       (Kafka, river kutubxonasi)
```

## 10. Xulosa

```
ML LOYIHA TO'LIQ JARAYONI:

  1. MUAMMONI ANIQLASH
       Biznes savol → ML savol
       Metrikani OLDINDAN belgilash (Recall ≥ 0.80)
       ML turi: klassifikatsiya/regressiya/klaster?

  2. MA'LUMOT YIG'ISH
       Miqdor: ≥1000 (oddiy), ≥10000 (murakkab)
       Sifat: NaN foizi, imbalance, yangilik

  3. EDA
       Taqsimot → transformation qaror
       Outlier → tozalash qaror
       Korrelyatsiya → feature tanlov qaror
       Vizualizatsiya → MODEL qaror (chiziqli/yo'q)

  4. PREPROCESSING (Pipeline!)
       NaN → median/mode/guruh
       Kategorik → OrdinalEncoder/OneHotEncoder
       Scaling → StandardScaler/MinMaxScaler
       AVVAL split, KEYIN fit_transform(train)!

  5. MODEL TANLASH
       Sodda → Murakkab (Occam's Razor)
       Baseline → LogReg → RF → XGBoost → MLP

  6. BAHOLASH
       Confusion matrix, ROC, PR curve
       Learning curve → overfit/underfit
       Threshold tuning → biznes metrikasiga moslash

  7. DEPLOYMENT
       joblib.dump() → saqlash
       Wrapper klass → ishlatish
       Metadata → versiya, metrikalar

  8. MONITORING
       Data drift, concept drift kuzatish
       Metrika pasaysa → retraining trigger
```

> 💡 **Eslab qol:** ML loyihada **80% vaqt — ma'lumot bilan** (yig'ish, EDA, preprocessing), faqat **20% — model bilan** o'tadi. Ko'pchilik buни teskari o'ylaydi. Eng kuchli model ham noto'g'ri tayyorlangan ma'lumot bilan yaxshi ishlamaydi — **"Garbage in, garbage out"** prinsipi ML da mutlaq haqiqat.
