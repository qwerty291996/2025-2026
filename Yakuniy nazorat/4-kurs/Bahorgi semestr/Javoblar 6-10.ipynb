## 1. Socket nima?

**Socket** — bu ikki dastur o'rtasida tarmoq orqali ma'lumot almashish uchun ishlatiladigan **eshik** (endpoint). Xuddi telefon kabi: biri qo'ngiroq qiladi, ikkinchisi qabul qiladi, keyin ikkalasi gaplashadi.

```
Kompyuter A                    Kompyuter B
(Klient)                       (Server)
   │                               │
   │  ←———— Tarmoq (IP:Port) ————→ │
   │                               │
[Socket]  ←——— Ma'lumot ———→  [Socket]
```

---

## 2. TCP nima va nima uchun ishlatiladi?

**TCP (Transmission Control Protocol)** — ishonchli, tartibli, xatosiz ma'lumot uzatish protokoli.

```
TCP kafolatlari:
  ✅ Ma'lumot yetib boradi (yo'qolmaydi)
  ✅ Tartib saqlanadi (1,2,3 → 1,2,3)
  ✅ Xatolar tekshiriladi
  ✅ Ulanish o'rnatiladi (握手 — handshake)

Qachon ishlatiladi:
  → Veb saytlar (HTTP)
  → Elektron pochta (SMTP)
  → Fayl uzatish (FTP)
  → Chat dasturlari
```

**TCP vs UDP:**
```
TCP → Sekinroq, lekin ishonchli   (xat jo'natish kabi)
UDP → Tezroq, lekin ishonchsiz    (radio eshittirish kabi)
```

## 3. TCP Ulanish Jarayoni — 3 Bosqichli Handshake

```
Klient              Server
  │                   │
  │──── SYN ─────────→│   "Ulanmoqchiman"
  │                   │
  │←─── SYN+ACK ──────│   "Mayli, tayyorman"
  │                   │
  │──── ACK ─────────→│   "Yaxshi, boshlaylik"
  │                   │
  │←══ Ulanish ═══════│   Ma'lumot almashish boshlandi
```

Bu jarayon avtomatik — Python socket kutubxonasi buni o'zi bajaradi.

## 5. Har bir Metodning Vazifasi

### 🔵 socket() — Eshik yaratish
```python
import socket

# AF_INET  = IPv4 manzil oilasi
# SOCK_STREAM = TCP (oqim asosida)
server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
```
**Nazariy asos:** Socket yaratilganda OS xotirada yangi aloqa nuqtasi ajratadi. `AF_INET` — IPv4 manzillar (192.168.x.x), `SOCK_STREAM` — TCP ning belgisi (oqim uzluksiz keladi).

---

### 🔵 bind() — Manzil va port biriktirish
```python
server.bind(("0.0.0.0", 12345))
# "0.0.0.0" = barcha tarmoq interfeyslarini tinglash
# 12345     = port raqami
```
**Nazariy asos:** `bind()` serverga "Men shu port orqali kelgan ulanishlarni qabul qilaman" deydi. Port — 0 dan 65535 gacha. 0-1023 = tizim portlari (HTTP=80, HTTPS=443). Bizning dasturlar uchun 1024+ ishlatiladi. `"0.0.0.0"` — barcha tarmoq kartalarini tinglaydi, `"127.0.0.1"` — faqat o'z kompyuterini.

---

### 🔵 listen() — Ulanishlarni kutish rejimi
```python
server.listen(5)
# 5 = navbat uzunligi (backlog)
```
**Nazariy asos:** `listen()` socketni **passiv rejimga** o'tkazadi — endi u ulanishlarni kutadi. `5` — bir vaqtda navbatda kutishi mumkin bo'lgan ulanishlar soni. Agar 5 ta ulanish navbatda bo'lsa, 6-chi ulanish rad etiladi.

---

### 🔵 accept() — Ulanishni qabul qilish
```python
klient_socket, klient_manzil = server.accept()
# accept() — yangi ulanish kelguncha TO'XTAB TURADI (blocking)
# klient_socket  = yangi ulanish bilan ishlash uchun socket
# klient_manzil  = (IP, port) — klientning manzili
```
**Nazariy asos:** `accept()` — **blokirovchi** funksiya, ya'ni ulanish kelmaguncha dastur bu qatorda to'xtab turadi. Ulanish kelganda OS yangi socket ob'ektini yaratadi. Asl server socketi esa keyingi ulanishlarni kutishda davom etadi. Bu ikki socket — turli vazifalar uchun!

```
server_socket  → Faqat yangi ulanishlarni qabul qiladi
klient_socket  → Shu klient bilan gaplashish uchun
```

---

### 🔵 connect() — Serverga ulanish (klient tomoni)
```python
klient = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
klient.connect(("127.0.0.1", 12345))
```
**Nazariy asos:** `connect()` TCP handshake ni boshlaydi. Bu yerda OS SYN paketi yuboradi, SYN+ACK kutadi, ACK qaytaradi. Agar server mavjud bo'lmasa — `ConnectionRefusedError` xatosi chiqadi. Agar server band bo'lsa — `TimeoutError`.

---

### 🔵 send() / recv() — Ma'lumot yuborish va qabul qilish
```python
# Yuborish
xabar = "Salom Server!".encode("utf-8")  # str → bytes
klient.send(xabar)

# Qabul qilish
data = klient_socket.recv(1024)          # max 1024 bayt
matn = data.decode("utf-8")             # bytes → str
```
**Nazariy asos:**

`send()` — ma'lumotni TCP bufferiga qo'yadi, OS uni paketlarga bo'lib yuboradi. **Muhim:** `send()` barcha ma'lumotni bir anda yubormagan bo'lishi mumkin! Katta ma'lumotlar uchun `sendall()` ishlatiladi.

`recv()` — TCP bufferidan ma'lumot oladi. `1024` — bir safarda maksimal necha bayt olinishini bildiradi. Agar ma'lumot 1024 baytdan ko'p bo'lsa — bir necha marta `recv()` chaqirilishi kerak. `recv()` ham blokirovchi — ma'lumot kelgunga qadar kutadi.

```python
# To'g'ri katta ma'lumot yuborish:
klient.sendall(katta_matn.encode("utf-8"))  # ← hammasi yetib borishini kafolatlaydi

# To'g'ri katta ma'lumot qabul qilish:
barcha_data = b""
while True:
    qism = klient_socket.recv(4096)
    if not qism:
        break
    barcha_data += qism
```

### 🔵 close() — Ulanishni yopish
```python
klient_socket.close()
server.close()
```
**Nazariy asos:** `close()` TCP **FIN** paketi yuboradi — "Men tugatdim" degan signal. Bu TCP ning to'g'ri yopilishi (graceful shutdown). OS socket uchun ajratilgan xotirani bo'shatadi.
```python
with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
    s.connect(("127.0.0.1", 12345))
    s.sendall(b"Salom!")
    # with bloki tugaganda s.close() avtomatik chaqiriladi
```

## 3. try-except Strategiyalari

### ❌ Noto'g'ri usul — Hamma narsani ushlab tashlash:
```python
try:
    klient.connect(("127.0.0.1", 12345))
    klient.send(data)
except Exception as e:
    print("Xato!")   # ← nima xato ekanini bilmaymiz
```

**Muammo:** Qaysi xato ekani noma'lum, har xil xatolar har xil yechim talab qiladi.

### ✅ To'g'ri usul — Xato turini aniq ajratish:
```python
import socket

def serverga_ulan(host, port, timeout=5.0):
    klient = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    klient.settimeout(timeout)   # ← timeout o'rnatish shart!

    try:
        klient.connect((host, port))
        return klient

    except ConnectionRefusedError:
        # Server o'chiq yoki port yopiq
        print(f"❌ Server {host}:{port} da ishlamayapti.")
        return None

    except socket.timeout:
        # Timeout — server javob bermadi
        print(f"⏱️ {timeout} soniya ichida server javob bermadi.")
        return None

    except OSError as e:
        # Tarmoq yo'q, noto'g'ri manzil va boshqalar
        print(f"🌐 Tarmoq xatosi: {e}")
        return None
```

## 5. BrokenPipe — Himoya Usullari

`BrokenPipe` ni oldindan sezib bo'lmaydi — faqat yuborganda bilinadi. Shuning uchun **har bir send() himoyalanishi shart:**

```python
def xavfsiz_yuborish(sock, data: bytes) -> bool:
    """
    True  → muvaffaqiyatli yuborildi
    False → ulanish uzilgan
    """
    try:
        sock.sendall(data)
        return True

    except BrokenPipeError:
        print("📡 Ulanish uzildi (BrokenPipe)")
        return False

    except ConnectionResetError:
        print("📡 Server ulanishni uzdi (ConnectionReset)")
        return False

    except socket.timeout:
        print("⏱️ Yuborish vaqti o'tdi")
        return False

    except OSError as e:
        print(f"❌ Yuborish xatosi: {e}")
        return False
```

## 6. Qayta Urinish (Retry) Strategiyasi

Tarmoq xatoliklari ko'pincha **vaqtinchalik** bo'ladi. Shuning uchun darhol xato deb hisoblamasdan — **qayta urinish** tavsiya etiladi:

```python
import time

def qayta_urinib_ulan(host, port, urinishlar=3, kutish=2.0):
    """
    Ulanish muvaffaqiyatsiz bo'lsa, bir necha marta qayta urinadi.
    Har urinish orasida kutish vaqti ortib boradi (exponential backoff).
    """
    for i in range(1, urinishlar + 1):
        print(f"🔄 Urinish {i}/{urinishlar}...")
        try:
            sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
            sock.settimeout(5.0)
            sock.connect((host, port))
            print("✅ Ulanish muvaffaqiyatli!")
            return sock

        except ConnectionRefusedError:
            print(f"❌ Server javob bermadi ({i}-urinish)")

        except socket.timeout:
            print(f"⏱️ Timeout ({i}-urinish)")

        except OSError as e:
            print(f"🌐 Tarmoq xatosi: {e}")

        if i < urinishlar:
            vaqt = kutish * i   # 2, 4, 6 soniya (exponential backoff)
            print(f"⏳ {vaqt} soniyadan so'ng qayta urinamiz...")
            time.sleep(vaqt)

    print("🚫 Barcha urinishlar muvaffaqiyatsiz!")
    return None
```

**Exponential backoff nima?**
```
1-urinish → darhol
2-urinish → 2 soniya kutish
3-urinish → 4 soniya kutish
4-urinish → 8 soniya kutish

Sabab: Server yuklanganida har soniyada urinish
       serverga yanada ko'proq yuk tushiradi.
       Kutish vaqtini oshirish serverni "nafas olishiga" imkon beradi.
```

## 8. To'liq Himoyalangan Klient Namunasi

```python
import socket
import logging
import time

logging.basicConfig(
    filename="tarmoq.log",
    level=logging.ERROR,
    format="%(asctime)s — %(levelname)s — %(message)s"
)

class HimoyalangaKlient:
    def __init__(self, host, port):
        self.host    = host
        self.port    = port
        self.sock    = None

    def ulan(self) -> bool:
        try:
            self.sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
            self.sock.settimeout(5.0)
            self.sock.connect((self.host, self.port))
            return True

        except ConnectionRefusedError as e:
            logging.error(f"Ulanish rad etildi: {e}")
            print("❌ Server ishlamayapti.")
            return False

        except socket.timeout as e:
            logging.error(f"Ulanish timeout: {e}")
            print("⏱️ Server javob bermadi.")
            return False

        except OSError as e:
            logging.error(f"Tarmoq xatosi: {e}")
            print(f"🌐 Tarmoq muammosi: {e}")
            return False

    def yuborish(self, xabar: str) -> str | None:
        if not self.sock:
            print("⚠️ Avval ulanish o'rnating!")
            return None
        try:
            self.sock.settimeout(10.0)
            self.sock.sendall(xabar.encode("utf-8"))
            javob = self.sock.recv(4096).decode("utf-8")
            return javob

        except BrokenPipeError:
            logging.error("BrokenPipe: ulanish uzildi")
            print("📡 Ulanish uzildi, qayta ulanish...")
            self.sock = None
            return None

        except ConnectionResetError:
            logging.error("ConnectionReset: server uzdi")
            print("📡 Server ulanishni uzdi.")
            self.sock = None
            return None

        except socket.timeout:
            logging.error("Recv timeout")
            print("⏱️ Javob kutish vaqti o'tdi.")
            return None

    def yopish(self):
        if self.sock:
            try:
                self.sock.close()
            except OSError:
                pass
            finally:
                self.sock = None
                print("🔌 Ulanish yopildi.")
```

## 2. Threading — Ko'p Oqimli Yondashuv

**G'oya:** Har bir klient uchun alohida **thread** (oqim) yaratiladi. Har bir thread mustaqil ishlaydi.

```
Server
├── Thread 1 → Klient A bilan ishlaydi
├── Thread 2 → Klient B bilan ishlaydi
├── Thread 3 → Klient C bilan ishlaydi
└── Thread 4 → Klient D bilan ishlaydi
```

**Qanday ishlaydi:**
```
accept() → yangi klient keldi
    ↓
threading.Thread(target=klient_handler) → yangi thread yaratiladi
    ↓
Thread o'z ichida recv()/send() ni chaqiradi
    ↓
Thread blokirovchiga kirganda (recv) → OS boshqa threadni ishlatadi
```

**Nazariy asos — Blokirovchi I/O:**
Threading da `recv()` chaqirilganda thread **uxlaydi** — OS uni to'xtatib, boshqa threadni ishga tushiradi. Ma'lumot kelganda OS yana shu threadni uyg'otadi. Bu **OS darajasida kontekst almashtirish** (context switching) deyiladi.

```python
import threading
import socket

klientlar = []   # barcha klientlar ro'yxati
lock = threading.Lock()  # ro'yxatga xavfsiz murojaat uchun

def klient_handler(sock, manzil):
    with lock:
        klientlar.append(sock)

    try:
        while True:
            xabar = sock.recv(1024)
            if not xabar:
                break
            # Barcha klientlarga yuborish
            with lock:
                for k in klientlar:
                    if k != sock:
                        try:
                            k.send(xabar)
                        except:
                            pass
    finally:
        with lock:
            klientlar.remove(sock)
        sock.close()

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.bind(("0.0.0.0", 12345))
server.listen(10)

while True:
    sock, manzil = server.accept()
    t = threading.Thread(target=klient_handler, args=(sock, manzil))
    t.daemon = True   # asosiy dastur tugaganda thread ham tugaydi
    t.start()
```

## 3. Threading ning Muammolari

### 🔴 GIL (Global Interpreter Lock)
Python da **GIL** mavjud — bir vaqtda faqat bitta thread Python kodini bajara oladi. Bu threading ning samaradorligini cheklaydi.

```
C++/Java threading:   Thread 1 va Thread 2 haqiqatan parallel
Python threading:     Thread 1 ishlaydi → GIL → Thread 2 ishlaydi
                      (aslida navbat bilan, parallel emas!)
```

**Lekin:** Tarmoq operatsiyalarida (`recv`, `send`) GIL **bo'shatiladi** — chunki bu Python kodi emas, OS kodi. Shuning uchun chat server uchun threading hali ham samarali.

### 🔴 Xotira va tezlik muammosi
```
100 klient  →  100 thread  →  ~100MB xotira
1000 klient →  1000 thread →  ~1GB xotira  ← muammo!

Thread yaratish vaqti: ~1ms
Thread kontekst almashtirish: ~10-100 mikrosekund
```

### 🔴 Race Condition — Mushtarak ma'lumotlarga xavf
```python
# ❌ XAVFLI — lock ishlatilmagan:
klientlar.append(sock)   # Thread 1
klientlar.remove(sock)   # Thread 2 — bir vaqtda!
# → Ro'yxat buzilishi mumkin

# ✅ XAVFSIZ — lock bilan:
with lock:
    klientlar.append(sock)
```

## 4. Asyncio — Asinxron Yondashuv

**G'oya:** Bitta thread ichida ko'plab vazifalarni **navbat bilan** bajarish. Bir vazifa kutayotganda, boshqasi ishlaydi.

```
Asyncio Event Loop (bitta thread):

Klient A recv() kutmoqda →
    Klient B ga xabar yuborilmoqda →
        Klient C recv() kutmoqda →
            Klient A dan xabar keldi! →
                Klient A ni qayta ishlatish
```

**Nazariy asos — Non-blocking I/O:**
Asyncio da `await recv()` chaqirilganda thread **uxlamaydi** — faqat "men hozir band emasman, boshqasi ishlasin" deydi. Bu **kooperativ multitasking** — vazifalar o'zlari qachon "bo'shashlari"ni belgilaydi.

```
Threading:  OS qaror qiladi qaysi thread ishlashini (preemptive)
Asyncio:    Kod o'zi qaror qiladi qachon boshqasiga joy berish (cooperative)
```

```python
import asyncio

klientlar = set()

async def klient_handler(reader, writer):
    manzil = writer.get_extra_info("peername")
    klientlar.add(writer)

    try:
        while True:
            xabar = await reader.read(1024)  # ← await: boshqalar ishlayversin
            if not xabar:
                break

            # Barcha klientlarga yuborish
            for k in klientlar:
                if k != writer:
                    k.write(xabar)
                    await k.drain()  # ← buffer to'lmasligi uchun

    finally:
        klientlar.discard(writer)
        writer.close()

async def main():
    server = await asyncio.start_server(
        klient_handler, "0.0.0.0", 12345
    )
    async with server:
        await server.serve_forever()

asyncio.run(main())
```

## 5. Asyncio ning Afzalligi va Kamchiligi

**Afzalligi — Resurslar:**
```
1000 klient, Threading:  1000 thread  → ~1GB xotira
1000 klient, Asyncio:    1 thread     → ~50MB xotira

Asyncio 20x kam xotira sarflaydi!
```

**Afzalligi — Tezlik:**
```
Thread yaratish:          ~1ms
Asyncio task yaratish:    ~1 mikrosekund (1000x tez!)
```

**Kamchiligi — Murakkablik:**
```python
# Threading: oddiy, tushunarli
def handler():
    data = sock.recv(1024)   # to'g'ridan-to'g'ri

# Asyncio: async/await tushunish kerak
async def handler():
    data = await reader.read(1024)   # nima uchun await?
```

**Kamchiligi — Blokirovchi kod xavfi:**
```python
# ❌ ASYNCIO DA XAVFLI — blokirovchi funksiya:
async def handler():
    import time
    time.sleep(5)   # ← BUTUN serverni muzlatadi!
    # Asyncio bitta thread — bu yerda hamma kutib qoladi

# ✅ TO'G'RI:
async def handler():
    await asyncio.sleep(5)  # ← faqat bu task kutadi
```
