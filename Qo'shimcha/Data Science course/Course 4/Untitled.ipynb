Ushbu darsda internetning asosiy muloqot tili bo'lgan **HTTP protokoli**, **URL** tuzilishi va **Request/Response** (So'rov va Javob) jarayoni haqida so'z boradi. Matnning asosiy qismlarini quyidagicha tushuntirish mumkin:

---

### **1. URL nima? (Uniform Resource Locator)**

URL — bu internetdagi resursning (fayl, rasm yoki sahifa) manzilidir. Uni uchta asosiy qismga bo'lish mumkin:

* **Scheme (Sema):** Protokol turi (masalan, `http://` yoki `https://`).
* **Base URL (Asosiy manzil):** Serverning internetdagi nomi (masalan, `www.ibm.com`).
* **Route (Yo'nalish):** Server ichidagi aniq faylning joylashgan o'rni (masalan, `/images/logo.png`).

---

### **2. HTTP Request va Response jarayoni**

Mijoz (brauzer) va server o'rtasidagi muloqot ikki bosqichdan iborat:

* **HTTP Request (So'rov):** Siz biror saytga kirganingizda, brauzeringiz serverga so'rov yuboradi.
* **GET metodi:** Eng ko'p tarqalgan metod bo'lib, serverdan ma'lumot olish (masalan, `index.html` faylini) uchun ishlatiladi.
* **Header (Sarlavha):** So'rov haqida qo'shimcha ma'lumotlarni o'z ichiga oladi.


* **HTTP Response (Javob):** Server so'rovni qabul qiladi va natijani qaytaradi.
* **Status Line:** Javob holati (masalan, `200 OK`).
* **Body:** So'ralgan ma'lumotning o'zi (HTML kod, rasm yoki matn).



---

### **3. HTTP Status kodlari**

Server qaytargan uch xonali son so'rovning natijasini bildiradi:

* **100-lar (Informational):** Hammasi joyida, davom etavering.
* **200-lar (Success):** So'rov muvaffaqiyatli bajarildi (masalan, **200 OK**).
* **400-lar (Client Error):** Mijoz tomondan xatolik (masalan, **401 Unauthorized** — kirish taqiqlangan yoki mashhur **404 Not Found**).
* **500-lar (Server Error):** Serverda xatolik yuz berdi.

---

### **4. HTTP Metodlari**

Siz serverga nima qilish kerakligini metodlar orqali aytasiz:

* **GET:** Ma'lumot olish.
* **POST:** Serverga yangi ma'lumot yuborish (masalan, ro'yxatdan o'tish formasi).
* **PUT:** Mavjud ma'lumotni yangilash.
* **DELETE:** Ma'lumotni o'chirish.

**Xulosa:** HTTP protokoli — bu mijoz va server qanday tilda gaplashishini belgilaydigan qoidalar to'plamidir. Kelgusi darsda Python yordamida ushbu metodlarni qanday qo'llashni o'rganamiz.



Ushbu darsda Python-ning eng ommabop kutubxonalaridan biri — **Requests** yordamida **GET** va **POST** so'rovlarini qanday yuborish haqida so'z boradi. Quyida asosiy tushunchalarning qisqacha tahlili keltirilgan:

---

### **1. Requests kutubxonasi**

Python-da HTTP protokoli bilan ishlash uchun `requests` kutubxonasi eng qulay vosita hisoblanadi. Uni ishlatish uchun avval import qilish kerak:

```python
import requests

```

---

### **2. GET so'rovi (Ma'lumot olish)**

GET so'rovi serverdan ma'lumot olish uchun ishlatiladi.

* **Status kodi:** `r.status_code` orqali tekshiriladi (200 — muvaffaqiyatli).
* **Headerlar:** `r.headers` javob sarlavhalarini lug'at ko'rinishida qaytaradi.
* **Query String:** URL ning oxiriga qo'shiladigan ma'lumotlar (masalan, `?name=Joseph&ID=123`). Python-da buni `params` argumenti orqali lug'at ko'rinishida yuborish oson.

---

### **3. POST so'rovi (Ma'lumot yuborish)**

POST so'rovi serverga ma'lumot yuborish uchun ishlatiladi. GET dan farqli o'laroq, POST ma'lumotlarni URL ichida emas, balki so'rovning **tanasida (body)** yuboradi.

* **Xavfsizlik:** Ma'lumotlar URL da ko'rinmaydi.
* **Amalga oshirish:** `requests.post(url, data=payload)` ko'rinishida bajariladi.

---

### **4. GET va POST o'rtasidagi asosiy farqlar**

| Xususiyat | GET | POST |
| --- | --- | --- |
| **Ma'lumot joylashuvi** | URL ichida (Query string) | So'rov tanasida (Request body) |
| **Request Body** | Odatda bo'sh (None) | Ma'lumot bilan to'ldirilgan |
| **Xavfsizlik** | Kamroq (ma'lumot URL da ko'rinadi) | Yuqoriroq (maxfiy ma'lumotlar uchun) |
| **Keshlanish** | Ha, keshlanishi mumkin | Yo'q, odatda keshlanmaydi |

---

### **5. Javobni tahlil qilish (JSON)**

Serverdan kelgan javob ko'pincha JSON formatida bo'ladi. `r.json()` metodi yordamida ushbu ma'lumotni darhol Python lug'atiga aylantirish va kerakli qiymatlarni (masalan, `args` yoki `form` kalitlari orqali) olish mumkin.




Ushbu darsda Web Scraping (veb-sahifalardan ma'lumot yig'ish) asoslari, HTML tuzilishi va Python-ning BeautifulSoup hamda Pandas kutubxonalari yordamida ma'lumotlarni qanday ajratib olish o'rgatiladi.

Quyida darsning qisqacha va tizimlashtirilgan tahlili keltirilgan:

1. Web Scraping nima va u qanday ishlaydi?
Web scraping — bu veb-saytlardan avtomatlashtirilgan tarzda ma'lumotlarni yig'ish jarayonidir. U quyidagi bosqichlardan iborat:

HTTP Request: Scraper serverga GET so'rovini yuboradi.

HTML Retrieval: Server javob sifatida sahifaning HTML kodini qaytaradi.

Parsing: HTML kodi tarkibiy qismlarga (teglar, atributlar) bo'linadi.

Extraction: Kerakli ma'lumotlar (matn, link, rasm) ajratib olinadi.

Storage: Ma'lumotlar CSV, JSON yoki ma'lumotlar bazasida saqlanadi.

2. HTML Tuzilishi va Hujjat Daraxti
HTML kodini daraxt (tree) ko'rinishida tasavvur qilish kerak. Bu navigatsiya qilishni osonlashtiradi:

Root (Ildiz): <html> tegi.

Children (Farzandlar): <head> va <body> teglari <html> ning farzandlari.

Siblings (Aka-ukalar): Bir xil ota-onaga ega teglar (masalan, <h1> va <p> teglari <body> ichida aka-uka hisoblanadi).

HTML Jadvallari: Ma'lumotlar ko'pincha jadvallarda saqlanadi:

<table>: Jadval boshlanishi.

<tr>: Qator (row).

<th>: Sarlavha katagi (header).

<td>: Ma'lumot katagi (data).

3. Python yordamida Web Scraping
BeautifulSoup bilan ishlash
Bu kutubxona HTML kodini tahlil qilish (parsing) uchun ishlatiladi.

````python
import requests
from bs4 import BeautifulSoup
# Specify the URL of the webpage you want to scrape
url = 'https://en.wikipedia.org/wiki/IBM'
# Send an HTTP GET request to the webpage
response = requests.get(url)
# Store the HTML content in a variable
html_content = response.text
# Create a BeautifulSoup object to parse the HTML
soup = BeautifulSoup(html_content, 'html.parser')
# Display a snippet of the HTML content
print(html_content[:500])

# Find all <a> tags (anchor tags) in the HTML
links = soup.find_all('a')
# Iterate through the list of links and print their text
for link in links:
    print(link.text)
````

Xulosa
BeautifulSoup — murakkab HTML tuzilmalarini qidirish va navigatsiya qilish uchun ideal.
