**2D massivlar: Matritsalar**

2D massiv matritsa deb ham ataladi. Ular odatda ma'lumotlar turli satrlarda saqlanadigan to‘rtburchak shaklidagi massivlardir. Yuqorida aytib o‘tilgan barcha amallar 2D massivlar uchun ham amal qiladi.

Quyidagi tasvirlarda ko‘rsatilganidek, skalyar ko‘paytma (dot product) birinchi matritsa satrlari elementlarini ikkinchi matritsa ustunlari elementlariga mos ravishda ko‘paytirish va qo‘shish orqali amalga oshiriladi. Natijada, ko‘paytirishdan hosil bo‘lgan chiqish matritsasi o‘zgargan shaklga ega bo‘ladi.

Umumiy qoida shundan iboratki, $m \times n$ o‘lchamli matritsaning skalyar ko‘paytmasini faqat $n \times p$ o‘lchamli matritsa bilan bajarish mumkin va natijaviy matritsa $m \times p$ shakliga ega bo‘ladi. Quyida ko‘rsatilgan misolda, $4 \times 4$ o‘lchamli matritsani hosil qilish uchun $4 \times 2$ o‘lchamli matritsa $2 \times 4$ o‘lchamli matritsaga ko‘paytiriladi.
