## Numpy nima?

NumPy — bu massivlar, chiziqli algebra, Fure almashtirishlari va matritsalar bilan ishlash uchun mo‘ljallangan Python kutubxonasi. NumPy «Numerical Python» (Raqamli Python) degan ma'noni anglatadi va u ochiq manbali loyihadir. NumPy ichidagi massiv obyekti **ndarray** deb ataladi va u ndarray bilan ishlashni osonlashtiruvchi ko‘plab yordamchi funksiyalarni taqdim etadi.

Massivlar ma'lumotlar fani sohasida juda tez-tez ishlatiladi, chunki bu sohada tezlik va resurslar juda muhim ahamiyatga ega.

NumPy odatda `np` taxallusi (alias) ostida chaqiriladi.

U odatda o‘zgarmas o‘lchamga ega bo‘ladi va uning har bir elementi bir xil turdagi ma'lumotdan iborat bo‘ladi. Biz avval `numpy` ni chaqirib, so‘ngra ro‘yxatni (list) numpy massiviga aylantirishimiz mumkin:

### Kesish

Ro‘yxatlardagi kabi, biz numpy massivini ham kesishimiz (slice) mumkin. Pythonda kesish bu ma'lum bir indeksdan boshqa bir berilgan indeksgacha bo'lgan elementlarni olishni anglatadi.

Biz kesishni quyidagicha yozamiz: `[start:end]`. Oxirgi indeksdagi (end) element natijaga kiritilmaydi.

Biz 1-dan 3-gacha bo'lgan elementlarni tanlab, uni yangi <code>d</code> numpy massiviga quyidagicha biriktirishimiz mumkin:

`````python
# Numpy massivini kesish

d = c[1:4]
d
`````

`````python
# To'rtinchi va beshinchi elementlarga 300 va 400 qiymatlarini beramiz

c[3:5] = 300, 400
c
`````

Biz kesishda qadamlarni ham quyidagicha belgilashimiz mumkin: `[start:end:step]`.

`````python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5:2])
`````

Agar biz boshlang'ich nuqtani (start) bermasak, u 0 deb hisoblanadi.

`````python
print(arr[:4])
`````

Agar biz oxirgi nuqtani (end) bermasak, u massivning oxirigacha (massiv uzunligigacha) deb hisoblanadi.

`````python
print(arr[4:])
`````

Agar biz qadamni (step) bermasak, u 1 deb hisoblanadi.

`````python
print(arr[1:5:])
`````

## Linspace
Matematik funksiyalarni chizish uchun foydali funksiya `linspace` hisoblanadi. Linspace belgilangan interval bo'yicha bir xil masofadagi sonlarni qaytaradi.

**numpy.linspace(start, stop, num = int value)**

start : interval oralig'ining boshi

stop : interval oralig'ining oxiri

num : hosil qilinadigan namunalar soni

`````python
# [-2, 2] oralig'ida 5 ta elementdan iborat numpy massivini yaratish

np.linspace(-2, 2, num=5)
`````

Agar biz `num` parametrini 9 ga o'zgartirsak, -2 dan 2 gacha bo'lgan oraliqda bir xil masofada joylashgan 9 ta sonni olamiz:

`````python
# Import the libraries

import time 
import sys
import numpy as np 

import matplotlib.pyplot as plt


def Plotvec2(a,b):
    ax = plt.axes()# to generate the full window axes
    ax.arrow(0, 0, *a, head_width=0.05, color ='r', head_length=0.1)#Add an arrow to the  a Axes with arrow head width 0.05, color red and arrow head length 0.1
    plt.text(*(a + 0.1), 'a')
    ax.arrow(0, 0, *b, head_width=0.05, color ='b', head_length=0.1)#Add an arrow to the  b Axes with arrow head width 0.05, color blue and arrow head length 0.1
    plt.text(*(b + 0.1), 'b')
    plt.ylim(-2, 2)#set the ylim to bottom(-2), top(2)
    plt.xlim(-2, 2)#set the xlim to left(-2), right(2)
`````

`[-1, 1]` va `[1, 1]` ro'yxatlarini `a` va `b` numpy massivlariga o'tkazing. So'ngra, `Plotvec2` funksiyasi yordamida massivlarni vektorlar sifatida chizing va ularning skalyar ko'paytmasini (dot product) toping:

`````python
# Kodni pastga yozing va bajarish uchun Shift+Enter tugmalarini bosing
`````
