# Pandas kutubxonasi bilan ma'lumotlar tahlili asoslari

Kutubxonalar (libraries) yoki bog'liqliklar (dependencies) — bu muammolarni hal qilishga yordam beradigan oldindan yozilgan kodlar to'plamidir. Ushbu darsda biz ma'lumotlar tahlili uchun eng mashhur kutubxona bo'lgan **Pandas** bilan tanishamiz.

## 1. Kutubxonani yuklash (Import)
Pandas kutubxonasini `import` buyrug'i orqali yuklab olish mumkin. Odatda, kod yozishni osonlashtirish uchun unga `pd` taxallusi beriladi.

```python
import pandas as pd
```

Endi sizda kutubxonaning barcha tayyor funksiyalari va klasslariga kirish imkoniyati mavjud.

## 2. Ma'lumotlarni yuklash va DataFrame
Pandas'da ma'lumotlar bilan ishlashning asosiy usuli — bu **DataFrame** (ma'lumotlar jadvali) hisoblanadi.

### CSV faylni o'qish:
CSV fayl — ma'lumotlarni saqlash uchun eng ko'p qo'llaniladigan fayl turi. Uni `read_csv` funksiyasi orqali yuklaymiz:

```python
csv_path = 'file1.csv'
df = pd.read_csv(csv_path)
```

### Dastlabki qatorlarni ko'rish:
Yuklangan ma'lumotlarning dastlabki 5 ta qatorini ko'rish uchun `head()` metodidan foydalaniladi:

```python
df.head()
```

Excel fayllarini yuklash ham shunga o'xshash bo'lib, `read_excel` funksiyasi orqali amalga oshiriladi.

## 3. Lug'at (Dictionary) orqali DataFrame yaratish
DataFrame'ni oddiy Python lug'atidan ham yaratish mumkin. Bunda lug'at **kalitlari** — ustun nomlari, **qiymatlari** esa — qatorlardagi ma'lumotlar hisoblanadi:

```python
songs = {'Album': ['Thriller', 'Back in Black'], 'Released': [1982, 1980]}
songs_df = pd.DataFrame(songs)
```

## 4. Ma'lumotlarni tanlash va filtrlash (Selection)
Ma'lumotlar to'plamidan ma'lum bir ustunlarni ajratib olish uchun qo'shaloq qavslardan foydalanamiz:

- **Bitta ustunni ajratish:** `x = df[['Artist']]`
- **Bir nechta ustunni ajratish:** `y = df[['Artist', 'Released', 'Genre']]`

## 5. Elementlarga kirish: `iloc` va `loc` metodlari
Ma'lumotlar jadvalidagi aniq bir elementni topish uchun ikki xil usul mavjud:

* **`iloc` (Integer-location):** Faqat sonli indekslar orqali murojaat qiladi.
    * `df.iloc[0, 0]` — birinchi qator, birinchi ustun.
    * `df.iloc[1, 2]` — ikkinchi qator, uchinchi ustun.
* **`loc` (Label-location):** Qator indeksi yoki ustun nomi orqali murojaat qiladi.
    * `df.loc[0, 'Artist']` — 0-indeksdagi qatorning "Artist" ustunidagi qiymati.

## 6. Slicing (Kesib olish)
Jadvalning ma'lum bir qismini ajratib olib, yangi o'zgaruvchiga saqlash mumkin:

```python
# Birinchi ikki qator va birinchi uchta ustunni ajratib olish
Z = df.iloc[0:2, 0:3]

# 'Artist' dan 'Released' ustunigacha bo'lgan qismini ajratish
Z = df.loc[0:2, 'Artist':'Released']
```
