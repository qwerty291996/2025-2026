# Ma’lumotlar tahlili uchun Pandas kutubxonasiga kirish

### **Maqsad:**
* Pandas **Series** nima ekanligini va ularni yaratishni o‘rganish;
* Series ichidagi ma’lumotlarga kirish va ularni manipulyatsiya qilishni tushunish;
* Pandas **DataFrame** yaratish va ular bilan ishlash asoslarini kashf qilish;
* DataFrame'lardagi ma’lumotlarga kirish, o‘zgartirish va tahlil qilishni o‘rganish;
* Umumiy DataFrame atributlari va metodlari haqida tushunchaga ega bo‘lish.

### **Pandas nima?**
Pandas — Python dasturlash tili uchun ma’lumotlarni manipulyatsiya qilish va tahlil qilishga mo‘ljallangan mashhur ochiq kodli kutubxona. U strukturalangan (tuzilmaviy) ma’lumotlar bilan ishlash uchun kuchli va moslashuvchan vositalar to‘plamini taqdim etadi, bu esa uni ma’lumotlar bo‘yicha mutaxassislar, tahlilchilar va muhandislar uchun asosiy vositaga aylantiradi.

Pandas jadval ko‘rinishidagi ma’lumotlar, vaqtli qatorlar (time series) va boshqa turli formatlar bilan ishlash uchun mo‘ljallangan bo‘lib, ko‘plab sohalarda ma’lumotlarni qayta ishlash jarayonining ajralmas qismi hisoblanadi.



**Pandas’ning asosiy xususiyatlari va funksiyalari:**

* **Ma’lumotlar tuzilmalari:** Pandas ikkita asosiy ma’lumot tuzilmasini taklif etadi — **DataFrame** va **Series**.
    * **DataFrame** — bu yorliqlangan o‘qlarga (satrlar va ustunlar) ega bo‘lgan ikki o‘lchamli, o‘lchami o‘zgaruvchan va har xil turdagi ma’lumotlarni o‘z ichiga olishi mumkin bo‘lgan jadval ko‘rinishidagi ma’lumotlar tuzilmasidir.
    * **Series** — bu bir o‘lchamli yorliqlangan massiv bo‘lib, mohiyatan ma’lumotlarning bitta ustuni yoki satridir.
* **Ma’lumotlarni import va eksport qilish:** Pandas CSV fayllari, Excel jadvallari, SQL ma’lumotlar bazalari va boshqa manbalardan ma’lumotlarni o‘qishni osonlashtiradi. Shuningdek, u ma’lumotlarni ushbu formatlarga eksport qilishi mumkin, bu esa ma’lumotlarning uzluksiz almashinuvini ta’minlaydi.
* **Ma’lumotlarni birlashtirish (Merging and Joining):** Siz SQL operatsiyalariga o‘xshash `merge` va `join` metodlari yordamida bir nechta DataFrame'larni birlashtirib, turli manbalardan yanada murakkab ma’lumotlar to‘plamlarini yaratishingiz mumkin.
* **Samarali indekslash:** Pandas ma’lumotlarning aniq satrlari va ustunlariga tezda kirish imkonini beruvchi samarali indekslash va tanlash usullarini taqdim etadi.
* **Maxsus ma’lumotlar tuzilmalari:** Siz o‘zingizning maxsus ehtiyojlaringizga mos keladigan ma’lumotlar tuzilmalarini yaratishingiz va ma’lumotlarni manipulyatsiya qilishingiz mumkin, bu esa Pandas imkoniyatlarini yanada kengaytiradi.

### **Pandas kutubxonasini import qilish**

Pandas kutubxonasini `import` buyrug‘i va undan keyin kutubxona nomini yozish orqali loyihaga qo‘shishingiz mumkin. Kod yozishda qisqalikni ta’minlash maqsadida, odatda Pandas `pd` qisqartmasi orqali import qilinadi.

```python
import pandas as pd
```

---

### **Ma’lumotlarni yuklash**

* Pandas yordamida turli manbalardan, masalan, CSV va Excel fayllaridan ma’lumotlarni yuklab olish mumkin.
* `read_csv` funksiyasi CSV faylidagi ma’lumotlarni Pandas **DataFrame**’iga yuklash uchun ishlatiladi.

**CSV faylini o‘qish uchun sintaksis:**

```python
import pandas as pd

# CSV faylini DataFrame'ga o'qib olish
df = pd.read_csv('sizning_faylingiz.csv')
```


`'sizning_faylingiz.csv'` yozuvi o‘rniga CSV faylingizning haqiqiy manzilini yozing. Fayl Python skriptingiz joylashgan papkada ekanligiga yoki to‘g‘ri fayl yo‘li ko‘rsatilganiga ishonch hosil qiling.

### **Series nima?**

Series — bu Pandas kutubxonasidagi bir o‘lchamli yorliqlangan massivdir. Uni har bir element uchun o‘z yorlig‘i yoki indeksiga ega bo‘lgan bitta ma’lumotlar ustuni deb tasavvur qilish mumkin. Siz Series obyektini turli xil ma’lumot manbalaridan, masalan, ro‘yxatlar (list), NumPy massivlari yoki lug‘atlardan (dictionary) yaratishingiz mumkin.

**Pandas’da Series yaratishga doir oddiy misol:**

```python
import pandas as pd

# Ro'yxatdan Series yaratish
data = [10, 20, 30, 40, 50]
s = pd.Series(data)

print(s)
```

Ushbu misorda biz raqamli ma’lumotlardan iborat `s` nomli Series yaratdik. E’tibor bering, Pandas avtomatik ravishda har bir elementga raqamli indekslarni (0, 1, 2, 3, 4) biriktirdi, lekin zarurat bo‘lsa, siz maxsus yorliqlarni ham belgilashingiz mumkin.

### **Series elementlariga kirish**

Series elementlariga indeks yorliqlari yoki butun sonli pozitsiyalar orqali kirish mumkin. Quyida Series ma’lumotlariga kirishning bir necha keng tarqalgan usullari keltirilgan:

**Yorliq (label) orqali kirish:**
```python
print(s[2])     # 2-yorliqdagi elementga kirish (qiymati: 30)
```

**Pozitsiya (position) orqali kirish:**
```python
print(s.iloc[3]) # 3-pozitsiyadagi elementga kirish (qiymati: 40)
```

**Bir nechta elementga kirish (Slicing):**
```python
print(s[1:4])   # Yorliq orqali elementlar diapazoniga kirish
```

### **Series atributlari va metodlari**

Pandas Series obyekti ma’lumotlarni samarali manipulyatsiya qilish va tahlil qilishga yordam beradigan turli xil atribut va metodlar bilan ta’minlangan. Quyida ularning eng muhimlari keltirilgan:

* **`values`**: Series ma’lumotlarini NumPy massivi ko‘rinishida qaytaradi.
* **`index`**: Series-ning indekslarini (yorliqlarini) qaytaradi.
* **`shape`**: Series o‘lchamlarini ifodalovchi tupleni (tuple) qaytaradi.
* **`size`**: Series-dagi elementlar sonini qaytaradi.
* **`mean()`, `sum()`, `min()`, `max()`**: Ma’lumotlarning statistik xulosalarini hisoblaydi.
* **`unique()`, `nunique()`**: Noyob qiymatlarni yoki noyob qiymatlar sonini aniqlaydi.
* **`sort_values()`, `sort_index()`**: Series-ni qiymatlari yoki indeks yorliqlari bo‘yicha saralaydi.
* **`isnull()`, `notnull()`**: Yo‘qolgan (NaN) yoki mavjud qiymatlarni tekshiradi.
* **`apply()`**: Series-ning har bir elementiga maxsus funksiyani qo‘llaydi.

### **DataFrame nima?**

DataFrame — bu har xil turdagi ma’lumotlarni o‘z ichiga olgan ustunlardan iborat ikki o‘lchamli yorliqlangan ma’lumotlar tuzilmasidir. Uni jadval deb tasavvur qiling: har bir ustun o‘zgaruvchini (variable), har bir satr esa kuzatuvni (observation) yoki ma’lumot nuqtasinini ifodalaydi.



DataFrame'lar turli xil ma’lumotlar, jumladan, CSV fayllari, Excel jadvallari, SQL ma’lumotlar bazalari va boshqalardan olingan strukturalangan ma’lumotlar bilan ishlash uchun juda mos keladi.

### **Lug‘atlardan DataFrame yaratish**

DataFrame'larni lug‘atlar (dictionaries) yordamida yaratish mumkin. Bunda lug‘at kalitlari (keys) ustun nomlari, qiymatlar (values) esa qatorlarni ifodalovchi ro‘yxatlardan iborat bo‘ladi.

**Lug‘atdan DataFrame yaratishga doir misol:**

```python
import pandas as pd

# Lug'at yordamida DataFrame yaratish
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 28],
        'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']}

df = pd.DataFrame(data)

print(df)
```

### **Ustunlarni tanlash**

DataFrame'dan bitta ustunni tanlash uchun ustun nomini kvadrat qavslar ichida ko‘rsatish kifoya. Shunga o‘xshash tarzda bir nechta ustunlarni tanlash orqali yangi DataFrame yaratish ham mumkin.

**Ustunni tanlashga doir misol:**

```python
# 'Name' ustuniga murojaat qilish
print(df['Name']) 
```

### **Satrlarga kirish**

Siz DataFrame satrlariga ularning pozitsiyasi (indeksi) bo‘yicha `.iloc[]` metodidan foydalanib yoki yorlig‘i (label) bo‘yicha `.loc[]` metodidan foydalanib kirishingiz mumkin.

**Satrlarga kirishga doir misol:**

```python
# Uchinchi satrga pozitsiyasi bo'yicha kirish
print(df.iloc[2])   

# Ikkinchi satrga yorlig'i bo'yicha kirish
print(df.loc[1])    
```

### **Kesish (Slicing)**

Muayyan satrlar va ustunlarni tanlab olish uchun DataFrame’larni "kesish" (slice) mumkin.

**Slicing-ga doir misol:**

```python
# Faqat tanlangan ustunlarni ko'rish
print(df[['Name', 'Age']])  

# Faqat tanlangan satrlar diapazonini ko'rish
print(df[1:3])             
```

### **Noyob elementlarni topish**

DataFrame ustunidagi noyob (takrorlanmas) elementlarni aniqlash uchun `unique()` metodidan foydalaniladi.

```python
# 'Age' ustunidagi noyob qiymatlarni aniqlash
unique_ages = df['Age'].unique()
```

### **Shartli filtrlash**

Tengsizlik operatorlari yordamida DataFrame ma’lumotlarini ma’lum bir shartlar asosida filtrlashingiz mumkin. Masalan, yoshi 25 dan katta bo‘lgan barcha foydalanuvchilarni ajratib olish mumkin.

```python
# Yoshi 25 dan katta bo'lganlarni filtrlash
high_above_25 = df[df['Age'] > 25]
```

### **DataFrame'larni saqlash**

DataFrame ma’lumotlarini CSV fayli ko‘rinishida saqlash uchun `.to_csv()` metodidan foydalaniladi va fayl nomi `.csv` kengaytmasi bilan ko‘rsatiladi.

```python
# DataFrame'ni CSV fayliga saqlash
df.to_csv('trading_data.csv', index=False)
```

Pandas kutubxonasi ma’lumotlarni boshqa turli formatlarda saqlash uchun ham maxsus funksiyalarni taqdim etadi.

### **DataFrame atributlari va metodlari**

DataFrame obyekti ma’lumotlarni manipulyatsiya qilish va tahlil qilish uchun ko‘plab atribut hamda metodlarni taqdim etadi:

* **`shape`**: DataFrame o‘lchamlarini (satrlar va ustunlar soni) qaytaradi.
* **`info()`**: DataFrame haqida umumiy ma’lumot, jumladan, ma’lumot turlari va bo‘sh bo‘lmagan qiymatlar sonini beradi.
* **`describe()`**: Raqamli ustunlar uchun statistik xulosalarni (o‘rtacha qiymat, standart chetlanish va h.k.) shakllantiradi.
* **`head()`, `tail()`**: DataFrame’ning dastlabki yoki oxirgi `n` ta qatorini ko‘rsatadi.
* **`mean()`, `sum()`, `min()`, `max()`**: Ustunlar bo‘yicha umumiy statistik ko‘rsatkichlarni hisoblaydi.
* **`sort_values()`**: DataFrame’ni bir yoki bir nechta ustun bo‘yicha saralaydi.
* **`groupby()`**: Ma’lumotlarni agregatsiyalash (guruhlab hisoblash) uchun muayyan ustunlar bo‘yicha guruhlaydi.
* **`fillna()`, `drop()`, `rename()`**: Yo‘qolgan qiymatlarni to‘ldirish, ustunlarni o‘chirish yoki ularni qayta nomlash uchun ishlatiladi.
* **`apply()`**: DataFrame’ning har bir elementi, satri yoki ustuniga maxsus funksiyani qo‘llaydi.

Pandas ushbu misollardan tashqari juda keng ko‘lamli metodlarni taklif etadi. Batafsil ma’lumot uchun Pandas rasmiy veb-saytidagi hujjatlarga murojaat qilishingiz mumkin.

### **Xulosa**

Xulosa qilib aytganda, Pandas **Series** va **DataFrame** tuzilmalarini mukammal o‘rganish Python’da ma’lumotlarni samarali manipulyatsiya qilish va tahlil qilish uchun juda muhimdir. Series bir o‘lchamli yorliqlangan ma’lumotlar bilan ishlash uchun asos bo‘lsa, DataFrames ikki o‘lchamli ma’lumotlar uchun ko‘p qirrali, jadval ko‘rinishidagi strukturani taqdim etadi.

Siz ma’lumotlarni tozalash, o‘rganish, transformatsiya qilish yoki tahlil qilish bilan shug‘ullanasizni, ushbu tuzilmalar va ularning metodlari sizga qimmatli xulosalar chiqarish uchun ma’lumotlarni moslashuvchan boshqarish imkonini beradi. Series va DataFrame’larni o‘z "data science" asboblar to‘plamingizga qo‘shish orqali siz ma’lumotlar bilan bog‘liq keng ko‘lamli vazifalarni bajarishga tayyor bo‘lasiz.

**Pandas orqali ma’lumotlar tahlili bo‘yicha ko‘nikmalaringizni yanada oshirish uchun:**

* **Amaliyot:** O‘rganganlaringizni qo‘llash va tajriba orttirish uchun real ma’lumotlar to‘plamlari (datasets) bilan ishlang.
* **Hujjatlarni o‘rganing:** Qo‘shimcha funksiya va metodlarni kashf qilish uchun Pandas rasmiy saytidagi keng qamrovli hujjatlar bilan tanishib chiqing.
