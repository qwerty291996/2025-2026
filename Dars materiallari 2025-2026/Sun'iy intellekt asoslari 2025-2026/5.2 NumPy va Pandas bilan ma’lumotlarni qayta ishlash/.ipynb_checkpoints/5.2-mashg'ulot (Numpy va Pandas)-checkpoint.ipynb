<h3>1. NumPy nima?</h3>

**NumPy (Numerical Python)** — bu Python dasturlash tilida ilmiy va matematik hisob-kitoblar uchun eng asosiy kutubxonalardan biridir. NumPy massivlar (array) bilan ishlashni osonlashtiradi va an’anaviy Python ro‘yxatlariga qaraganda tezroq, ixcham va samaraliroq hisoblanadi.

NumPy:
* katta hajmdagi sonli ma’lumotlar bilan ishlaydi,
* matritsa va vektorlar ustida amallarni bajaradi,
* statistik, matematik va chiziqli algebra funksiyalarini taqdim etadi,
* boshqa ML kutubxonalari (Pandas, Scikit-Learn, TensorFlow, PyTorch) uchun asos bo‘lib xizmat qiladi.

NumPy amallari oddiy Python looplari orqali emas, balki **C darajasida yozilgan vektorlashgan funksiyalar** orqali bajariladi.

Bu shuni anglatadi:

* Python har bir elementni for bilan aylanmaydi

* Birinchi darajada bitta buyruq barcha elementlarga qo‘llanadi

* CPU’ning SIMD (Single Instruction Multiple Data) registrlari ishlatiladi

**Vektorlashgan hisoblash (vectorized computing)** – bu hisoblash jarayonini **elementma-element emas**, balki **bir nechta elementni bir vaqtning o‘zida** (blok holida) bajaradigan optimallashtirilgan usuldir.

> **Oddiy usul:** bir xil operatsiyani 1 ta elementga
> **Vektorlashgan usul:** shu operatsiyani 1000 ta elementga *bir yo‘la* bajaradi.

---

# Nega bunga “vektorlashgan” deyishadi?

Chunki ma’lumotlar **vektorlar** (ya’ni elementlar ketma-ketligi) ko‘rinishida qayta ishlanadi:

* numpy massivlari
* GPU/CPU SIMD registrlari
* parallel instruktsiyalar

---

# Oddiy misol (Python – oddiy sikl)




````python
result = []
for x in [1, 2, 3, 4]:
    result.append(x * 2)
````

Bu yerda Python **har bir elementni alohida hisoblaydi** → sekinroq.

---

# Vektorlashgan misol (NumPy)

````python
import numpy as np
````

````python
x = np.array([1, 2, 3, 4])
result = x * 2
````

Bu yerda NumPy **SIMD** (Single Instruction, Multiple Data) orqali
→ **barcha elementlarga bir yo‘la** ko‘paytirish bajaradi
→ **juda tez ishlaydi**.

---

# Vektorlashgan hisoblashning afzalliklari

| Afzallik                  | Izoh                                                |
| ------------------------- | --------------------------------------------------- |
| **Tez ishlaydi**          | Bir nechta elementni bir vaqtning o‘zida hisoblaydi |
| **Kam xotira harakati**   | Loop overhead yo‘q, CPU keshlari yaxshi ishlaydi    |
| **GPUga mos**             | GPUlar minglab operatsiyani paralel bajaradi        |
| **Kod qisqa va chiroyli** | “x * 2” kabi oddiy yoziladi                         |

# Xulosa

**Vektorlashgan hisoblash – bir xil operatsiyani ko‘p elementlarga bir vaqtning o‘zida bajarib, hisoblashni bir necha baravar tezlashtirish texnikasidir.**

Quyida, NumPy dagi **vectorization vs loop** bo‘yicha benchmark (tezlik solishtiruv) kodi keltirilgan.

---

## 1-qism: Oddiy Python ro‘yхati bilan hisoblash

````python
import time

# 10 millionta elementli ro'yxat
n = 10_000_000
py_list = list(range(n))

start = time.time()
result = [x * 2 for x in py_list]
end = time.time()

print("Oddiy Python ro'yxati uchun vaqt:", end - start, "sekund")
````

Bu yerda:

* `py_list` – oddiy Python ro‘yхat
* har bitta elementni `x * 2` qilib **list comprehension** orqali hisoblayapmiz
* `time.time()` orqali qancha vaqt ketganini o‘lchayapmiz

---

## 2-qism: NumPy vektorlashgan hisoblash


````python
import time

n = 10_000_000
np_array = np.arange(n)

start = time.time()
result_np = np_array * 2
end = time.time()

print("NumPy vektorlashgan hisoblash vaqti:", end - start, "sekund")
````

Bu yerda:

* `np_array` – NumPy massivi
* `np_array * 2` – **bitta yozuv bilan** barcha elementlarga operatsiya
* ichida C/Fortran darajasida optimallashtirilgan, SIMD va hokazo

---

## ⏱ 3-qism: %timeit bilan solishtirish (Jupyter uchun maxsus)

Jupyter Notebook’da `timeit` “sehrli” komandasi bor, uni ishlatsak yanada aniqroq bo‘ladi.

### Python ro‘yхati:

````python
py_list = list(range(1_000_000))

%timeit [x * 2 for x in py_list]
````

### NumPy vektorlashgan:



````python
np_array = np.arange(1_000_000)

%timeit np_array * 2
````

Natijada ko‘rasizki:

* NumPy vektorlashgan usuli **bir necha baravar tezroq** chiqadi.

# **1. List → NumPy array**

Python ro‘yxatini (list) to‘g‘ridan-to‘g‘ri massivga aylantirish eng oson variant.

### ✔ Oddiy list


````python
lst = [1, 2, 3, 4]
arr = np.array(lst)
arr
````

### ✔ Ichma-ich list (2D list)

````python
lst2 = [[1, 2, 3],
        [4, 5, 6]]

arr2 = np.array(lst2)
arr2
````

**Natija:** 2×3 matritsa.

**Eslatma:**
Har bir ichki ro‘yxat uzunligi teng bo‘lishi kerak, aks holda NumPy `dtype=object` bo‘lgan “gibrid” massiv yaratadi.

---

# **2. Tuple → NumPy array**

Tuple ham listga o‘xshaydi, faqat o‘zgarmas (immutable). NumPy uni ham 1×1 qabul qiladi:

### ✔ Oddiy tuple

````python
tpl = (10, 20, 30)
arr = np.array(tpl)
arr
````

### ✔ Ichki tuple

````python
tpl2 = ((1, 2), (3, 4), (5, 6))
arr2 = np.array(tpl2)
arr2
````

NumPy buni 3×2 massiv deb oladi.

# **3. Dictionary → NumPy array**

**3 xil turi** mavjud.

---

## **3.1. Faqat qiymatlar (values)ni arrayga aylantirish**

Eng keng tarqalgan amaliy yechim:

````python
d = {"a": 1, "b": 2, "c": 3}
arr = np.array(list(d.values()))
arr
````

---
## **3.2. Kalitlar (keys)ni arrayga aylantirish**


````python
np.array(list(d.keys()))
````

---

## **3.3. Dictionary elementlarini juftlik (key-value pair) sifatida array qilish**


````python
arr = np.array(list(d.items()))
arr
````

Bu massiv **2 ustunli** bo‘ladi: key va value.

---
# **Eslatma: Dictionary → np.array() bevosita berilsa nima bo‘ladi?**

````python
np.array({"a":1, "b":2})
````

Bu **xato emas**, lekin foydasiz, chunki **object-type** massiv boʻladi.
Shuning uchun dictionaryni oldin **list yoki tuple** ko‘rinishga o‘tkazib olish kerak.

---

# **4. Aralash ma’lumotlar (mixed types)**

Agar list, tuple yoki dict qiymatlari turli tipda bo‘lsa:



````python
np.array([1, "salom", 3.5])
````

NumPy **hammasini string**ga aylantiradi yoki `dtype=object` qiladi.

## 1. Massiv yaratish metodlari

### 1.1. `np.array`

**Vazifasi:** Python list/tuple/np array’dan yangi NumPy massivi yaratadi.
```python
np.array(object, dtype=None, copy=True)
```
* `object` – list, tuple, boshqa array va h.k.
* `dtype` – element turi (`np.int32`, `np.float64`, `np.bool_`, …)
* `copy` – `True` bo‘lsa, nusxa yaratadi; `False` bo‘lsa, imkon bo‘lsa, xotirani bo‘lishadi.

**Misol:**

````python
lst = [1, 2, 3]
a = np.array(lst)          # dtype avtomatik
b = np.array(lst, dtype=float)  # float massiv

print(a, a.dtype)   # [1 2 3] int64 (masalan)
print(b, b.dtype)   # [1. 2. 3.] float64
````

---

### 1.2. `np.zeros`, `np.ones`, `np.full`

**Vazifasi:** Berilgan shaklda 0, 1 yoki ma’lum qiymatga to‘ldirilgan massiv yaratadi.

```python
np.zeros(shape, dtype=float)
np.ones(shape, dtype=float)
np.full(shape, fill_value, dtype=None)
```
* `shape` – o‘lcham: `3`, `(2,3)`, `(2,2,2)` va boshqalar
* `fill_value` – to‘ldiriladigan qiymat
* `dtype` – element turi

**Misol:**

````python
z = np.zeros((2, 3))               # 2x3 nol
o = np.ones((2, 2), dtype=int)     # 2x2 birlik, int
f = np.full((1, 4), 7.5)           # 1x4, hamma element 7.5

print(z)
print(o)
print(f)
````

---

### 1.3. `np.arange`, `np.linspace`

#### `np.arange`

**Vazifasi:** `range`ga o‘xshash, lekin array qaytaradi.
```python
np.arange(start, stop, step, dtype=None)
```

* `start` – boshlanish (standart: 0)
* `stop` – to‘xtash (kiritilmaydi)
* `step` – qadam (standart: 1)
* `dtype` – turi

**Misol:**

````python
a = np.arange(5)         # 0,1,2,3,4
b = np.arange(1, 10, 2)  # 1,3,5,7,9
print(a)
print(b)
````

#### `np.linspace`

**Vazifasi:** [start, stop] oraliqni berilgan `num` bo‘lakka bo‘lib, massiv qaytaradi.

```python
np.linspace(start, stop, num=50, endpoint=True)
```

* `start` – boshlanish
* `stop` – oxiri
* `num` – elementlar soni
* `endpoint` – `True` bo‘lsa, `stop` ham kiradi

**Misol:**

````python
x = np.linspace(0, 1, 5)
print(x)   # [0.   0.25 0.5  0.75 1.  ]
````

---

### 1.4. `np.eye`

**Vazifasi:** Diagonal bo‘yicha 1 bo‘lgan kvadrat / to‘rtburchak matritsa.

```python
np.eye(N, M=None, k=0, dtype=float)
```

* `N` – satrlar soni
* `M` – ustunlar soni (standart: `N`)
* `k` – qaysi diagonal: 0 – asosiy, 1 – yuqori, -1 – pastki
* `dtype` – element turi

**Misol:**

```python
I = np.eye(3)
print(I)
# [[1. 0. 0.]
#  [0. 1. 0.]
#  [0. 0. 1.]]

J = np.eye(3, 4, k=1)
print(J)
```

---

### 1.5. Tasodifiy massivlar — `np.random.rand`, `np.random.randint`

#### `np.random.rand`

**Vazifasi:** [0, 1) oraliqda bir xil taqsimlangan tasodifiy float’lar.

```python
np.random.rand(d0, d1, ...)
```

**Misol:**

```python
r = np.random.rand(2, 3)   # 2x3 tasodifiy son
print(r)
```

#### `np.random.randint`

**Vazifasi:** [low, high) oraliqda tasodifiy integer’lar.

```python
np.random.randint(low, high=None, size=None)
```

* `low` – pastki chegarasi (kiritiladi)
* `high` – yuqori chegarasi (kiritilmaydi)
* `size` – shakl (masalan `(2,3)`)

**Misol:**

```python
ri = np.random.randint(0, 10, size=(2, 4))
print(ri)
```

## 2. Massiv haqida ma’lumot metodlari

a = np.arange(12).reshape(3, 4)

### 2.1. `.ndim`, `.shape`, `.size`, `.dtype`

**Vazifasi:**

* `ndim` – o‘lchamlar soni (1D, 2D, 3D…)
* `shape` – har bir o‘lcham bo‘yicha uzunlik

* `dtype` – elementlar turi

**Misol:**

```python
print("ndim:", a.ndim)    # 2
print("shape:", a.shape)  # (3, 4)
print("size:", a.size)    # 12
print("dtype:", a.dtype)  # int64 va hokazo
```

## 3. Shakl va o‘lcham bilan ishlash

### 3.1. `reshape`

**Vazifasi:** Massiv shaklini (o‘lchamini) o‘zgartirish.

```python
a.reshape(newshape)
```

* `newshape` – yangi shakl. Jami elementlar soni o‘zgarmasligi shart.
* `-1` ishlatilsa, o‘sha joy avtomatik hisoblanadi.

**Misol:**

```python
b = np.arange(12)
c = b.reshape(3, 4)
d = b.reshape(2, -1)   # 2 qator, ustunlarni o‘zi topadi

print(c)
print(d)
```

---

### 3.2. `ravel` va `flatten`

**Ikkalasi ham tekis (1D) ko‘rinishga keltiradi, farqi:**

* `ravel()` – imkon bo‘lsa view qaytaradi (xotira bo‘lishadi)
* `flatten()` – har doim nusxa (copy)

```python
b = np.arange(6).reshape(2, 3)
r = b.ravel()
f = b.flatten()

print(r)
print(f)
```

---

### 3.3. `transpose` va `.T`

**Vazifasi:** O‘qlarni (axis) almashtirish.

2D matritsa uchun `.T` — transponirlangan matritsa.

```python
a = np.arange(6).reshape(2, 3)
print(a)
print(a.T)
```

Ko‘p o‘lchamli massiv uchun:

```python
np.transpose(a, axes=(1, 0))
```

---

### 3.4. `swapaxes`

**Vazifasi:** Ikki o‘qning o‘rnini almashtirish.

```python
np.swapaxes(a, axis1, axis2)
```

---

### 3.5. `expand_dims` va `squeeze`

#### `np.expand_dims`

**Vazifasi:** Yangi o‘lcham qo‘shish (1D → 2D kabi).

```python
x = np.array([1, 2, 3])
y = np.expand_dims(x, axis=0)
print(x.shape)  # (3,)
print(y.shape)  # (1, 3)
```

#### `np.squeeze`

**Vazifasi:** O‘lchami 1 bo‘lgan o‘qlarni olib tashlaydi.

```python
z = np.array([[[1, 2, 3]]])   # shape (1,1,3)
print(z.shape)
zs = np.squeeze(z)
print(zs.shape)                # (3,)
```

## 4. Tip o‘zgartirish — `astype`

**Vazifasi:** Elementlarning ma’lumot turini o‘zgartirish.

```python
a.astype(dtype)
```

* `dtype` – `np.float32`, `np.int64`, `'float'`, `'int'`, `'bool'` va h.k.

**Misol:**

```python
a = np.array([1.2, 3.7, 4.0])
b = a.astype(int)
c = a.astype(bool)

print(b, b.dtype)  # [1 3 4] int64
print(c, c.dtype)  # [ True  True  True] bool
```

## 5. Indekslash va tanlash

### 5.1. Oddiy slicing

```python
a = np.arange(10)
print(a[2:7])      # indeks 2..6
print(a[::2])      # har 2-ta element
```

2D massivda:

```python
b = np.arange(12).reshape(3,4)
print(b[1, 2])     # 2-qator, 3-ustun
print(b[0:2, 1:3]) # 1-2-qatorlar, 2-3-ustunlar
print(b[:, 0])     # barcha qatorlar, 1-ustun
```

---

### 5.2. Boolean indexing

**Vazifasi:** Shart bo‘yicha element tanlash.

```python
a = np.array([1, 2, 3, 4, 5])
mask = a > 3
print(mask)        # [False False False  True  True]
print(a[mask])     # [4 5]
```

To‘g‘ridan-to‘g‘ri:

```python
print(a[a % 2 == 0])   # juftlar
```

---

### 5.3. Fancy indexing

**Vazifasi:** Indekslar list/array’i bo‘yicha element tanlash.

```python
a = np.array([10, 20, 30, 40, 50])
idx = [0, 2, 4]
print(a[idx])   # [10 30 50]
```

2D misol:

```python
b = np.arange(12).reshape(3,4)
print(b[[0, 2], [1, 3]])  # (0,1) va (2,3) elementlar
```

---

### 5.4. `np.where`

**Vazifasi:**

* 1-variant: Shart bo‘yicha indekslar
* 2-variant: Shart bo‘yicha qiymat tanlash (if-else)

```python
a = np.array([1, 2, 3, 4, 5])

idx = np.where(a > 3)
print(idx)          # (array([3, 4]),)

b = np.where(a > 3, a, 0)
print(b)            # [0 0 0 4 5]
```

---

## 6. Arifmetik amallar va ufunc’lar

### 6.1. Elementma-element arifmetika

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b)    # [11 22 33]
print(a - b)    # [-9 -18 -27]
print(a * b)    # [10 40 90]
print(b / a)    # [10. 10. 10.]
print(a ** 2)   # [1 4 9]
```

Broadcasting:

## 8. Saralash va qidirish

### 8.1. `np.sort` va `.sort`

**`np.sort(a)`** – yangi saralangan massiv qaytaradi (copy).
**`a.sort()`** – massivni o‘zida saralaydi (in-place).

```python
a = np.array([3, 1, 4, 1, 5])
print(np.sort(a))   # [1 1 3 4 5]
a.sort()
print(a)           # [1 1 3 4 5]
```

---

### 8.2. `np.argsort`

**Vazifasi:** Saralash uchun indekslar tartibini qaytaradi.

```python
a = np.array([30, 10, 20])
idx = np.argsort(a)
print(idx)       # [1 2 0]
print(a[idx])    # [10 20 30]
```

---

### 8.3. `np.unique`

**Vazifasi:** Takrorlanmaydigan elementlar.

```python
a = np.array([1, 2, 2, 3, 3, 3])
u = np.unique(a)
print(u)   # [1 2 3]
```

Parametrlar:

```python
np.unique(a, return_counts=True)
```

* `return_counts=True` bo‘lsa, har bir qiymat nechta ekanini ham qaytaradi.

```python
values, counts = np.unique(a, return_counts=True)
print(values)   # [1 2 3]
print(counts)   # [1 1 3]
```

## 10. Nusxa va view — `.copy`, `view`

### 10.1. `copy`

```python
a = np.array([1, 2, 3])
b = a          # reference (xotira bo‘lishadi)
c = a.copy()   # yangi nusxa

a[0] = 99
print(b)   # [99  2  3]
print(c)   # [1 2 3]
```

`copy()` — alohida xotira.

`view()` — xuddi o‘sha xotiraga boshqa qarash.

Ko‘p hollarda `.copy()` bilan ishlash xavfsizroq.
