# **1. Ustunni indeks (raqam) bo‘yicha chaqirish – `df.iloc` orqali**

Agar siz 0-ustun yoki 2-ustunni chaqirmoqchi bo‘lsangiz:

```python
df.iloc[:, [0, 2]]
```

Buni Seaborn’da ishlatish:

```python
sns.lineplot(data=df.iloc[:, [0, 2]])
```

`:` → barcha qatorlar
`[0, 2]` → 0-ustun va 2-ustun

---

# **2. Faqat bitta ustun kerak bo‘lsa**

```python
sns.lineplot(data=df.iloc[:, 0])    # 0-ustun
sns.lineplot(data=df.iloc[:, 2])    # 2-ustun
```

---
# 🟦 Qo‘shimcha: Agar ustunlarni nom bilan chaqirmoqchi bo‘lsangiz

```python
sns.lineplot(data=df[['col1', 'col3']])
```

---


````python
plt.figure(figsize = (20, 6)) # plt.figure((16, 6)) - xatolik qaytaradi

plt.title("Muzeylarga tashriflar soni")
plt.xlabel("Sanalar")
plt.ylabel("Odamlar soni")
plt.grid(alpha = 0.3, color = "k")

sns.lineplot(data=df.iloc[:, [0, 2]])
plt.xticks(rotation = 90)

plt.show()
````

# 🟩 **`loc` nima?**

`loc` — Pandas DataFrame’da **qator va ustunlarni NOMI bo‘yicha** tanlash uchun ishlatiladi.

Sintaksis:

```python
df.loc[ qatorlar , ustunlar ]
```

* Qatorlar → index nomi (yoki shart)
* Ustunlar → column nomi

---

# 1️⃣ **QATORLARNI tanlash (`loc` bilan)**

## ✔ 1. Bitta qatorni nomi bo‘yicha olish

```python
df.loc['row1']
```

## ✔ 2. Bir nechta qator

```python
df.loc[['row1', 'row3', 'row5']]
```

## ✔ 3. Qatorlar bo‘yicha interval

```python
df.loc['row2' : 'row6']
```

Diapazon **oxirigacha kiradi** (INTEGER iloc’dan farqli ravishda).

---

# 2️⃣ **USTUNLARNI tanlash (`loc` bilan)**

## ✔ 1. Bitta ustun

```python
df.loc[:, 'Age']
```

## ✔ 2. Bir nechta ustun

```python
df.loc[:, ['Age', 'Salary']]
```

## ✔ 3. Ustunlar diapazoni

```python
df.loc[:, 'Age' : 'Salary']
```

---

# 3️⃣ **Qator + ustun tanlash**

## ✔ 1. Bitta katakni olish

```python
df.loc['row1', 'Age']
```

## ✔ 2. Bir nechta qator + bir nechta ustun

```python
df.loc[['row1', 'row3'], ['Age', 'Salary']]
```

---

# 4️⃣ **Shartlar bilan `loc` ishlatish**

## ✔ 1. Filtr: `Age > 30`

```python
df.loc[df['Age'] > 30]
```

## ✔ 2. Filtr + ustun tanlash

```python
df.loc[df['Age'] > 30, ['Name', 'Age']]
```

## ✔ 3. Ko‘p shartlar

```python
df.loc[(df['Age'] > 30) & (df['Salary'] < 5000)]
```

---

# 5️⃣ **Qiymatlarni o‘zgartirish (`loc` orqali)**

## ✔ 1. Bitta katakni o‘zgartirish

```python
df.loc['row1', 'Age'] = 99
```

## ✔ 2. Shart bo‘yicha ustun qiymatini o‘zgartirish

```python
df.loc[df['Age'] < 18, 'Status'] = 'Minor'
```

---

# 6️⃣ **`loc` va `iloc` farqi**

| Xususiyat       | `loc`              | `iloc`           |
| --------------- | ------------------ | ---------------- |
| Indekslash turi | nom bo‘yicha       | raqam bo‘yicha   |
| Qator diapazoni | yakunini ham oladi | yakunini olmaydi |
| Sintaksis       | `df.loc['A':'C']`  | `df.iloc[0:2]`   |

---

# 7️⃣ **Mini DataFrame bilan misollar**

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 35, 40],
    "Salary": [3000, 5000, 7000]
}, index=["Ali", "Vali", "Sodiq"])
```

### ✔ Ali’ning yoshini olish:

```python
df.loc['Ali', 'Age']
```

### ✔ Ali va Vali’ning maoshlarini olish:

```python
df.loc[['Ali', 'Vali'], ['Salary']]
```

### ✔ Salary > 4000 bo‘lganlarni olish:

```python
df.loc[df['Salary'] > 4000]
```

# Seabornda BARPLOT

````python
df = pd.read_csv("Files/uzbekistan.csv", index_col = 0)
df
````

`index_col` — bu `pd.read_csv()` funksiyasida **CSV fayldan qaysi ustunni DataFrame indeksiga aylantirish kerakligini** belgilaydi.

| Parametr             | Ma’nosi                                                   |
| -------------------- | --------------------------------------------------------- |
| `index_col=0`        | CSV dagi birinchi ustunni DataFrame indeksiga aylantiradi |
| `index_col=1`        | ikkinchi ustun indeks bo‘ladi                             |
| `index_col="Region"` | ustunni nomi orqali indeksga aylantirish                  |

---

````python
plt.figure(figsize = (16, 6))
sns.barplot(x = df.index, y = df['Aholisi'])

plt.title("O'zbekiston Respublikasi tarkibidagi hududlar aholi soni")
plt.xlabel("Hudud")
plt.ylabel("Aholi soni (million)")
plt.grid(alpha = 0.2, color = "grey")
plt.xticks(rotation = 90)
plt.show()
````

````python
plt.figure(figsize=(16, 6))

sns.barplot(
    x=df.index,
    y=df['Aholisi'],
    hue=df.index,                     
    palette=sns.color_palette("deep", len(df)),
    legend=False                     
)

plt.title("O'zbekiston Respublikasi tarkibidagi hududlar aholi soni")
plt.xlabel("Hudud")
plt.ylabel("Aholi soni (million)")
plt.grid(alpha=0.2, color="grey")
plt.xticks(rotation=90)
plt.show()
````

````python
plt.figure(figsize=(16, 6))

colors = ["red", "green", "blue", "orange", "purple", "brown",
          "pink", "gray", "yellow", "cyan", "magenta", "black", "lime"]

sns.barplot(
    x=df.index,
    y=df["Aholisi"],
    hue=df.index,
    palette=colors,
    legend=False
)


plt.title("O'zbekiston Respublikasi tarkibidagi hududlar aholi soni")
plt.xlabel("Hudud")
plt.ylabel("Aholi soni (million)")
plt.grid(alpha=0.2, color="grey")
plt.xticks(rotation=90)
plt.show()
````
