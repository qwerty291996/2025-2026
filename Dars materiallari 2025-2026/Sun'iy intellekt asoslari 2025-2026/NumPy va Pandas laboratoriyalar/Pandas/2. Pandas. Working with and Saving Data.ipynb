# Pandas: Ma'lumotlarni saralash va saqlash

Ma'lumotlar jadvali (DataFrame) bilan ishlaganda, biz ko'pincha ma'lum bir shartga mos keladigan qismlarni ajratib olamiz va ularni yangi formatlarda saqlaymiz.

## 1. Takrorlanmas qiymatlarni topish (`unique`)
Tasavvur qiling, sizda 13 ta rangli blok bor va ulardan nechtasi noyob (takrorlanmas) ekanini bilmoqchisiz. Agar ma'lumotlar soni millionlab bo'lsa, buni qo'lda qilish imkonsiz. Pandas'da `unique()` metodi ushbu vazifani bajaradi.

Masalan, albomlar jadvalidagi noyob yillarni topish uchun:

```python
# 'Released' ustunidagi noyob yillarni ko'rish
df['Released'].unique()
```

## 2. Ma'lumotlarni filtrlash (Boolean Indexing)
Aytaylik, bizga faqat 1980-yildan keyin chiqqan albomlar kerak. Buni amalga oshirish uchun **mantiqiy operatorlardan** foydalanamiz.

### Qadamli tushuntirish:
1. **Shartni aniqlash:** Avval ustunni tanlaymiz va shartni yozamiz. Natijada `True` (to'g'ri) va `False` (noto'g'ri) qiymatlardan iborat qator (Series) hosil bo'ladi.
2. **Filtrlash:** Ushbu mantiqiy shartni asosiy jadvalga qavs ichida uzatamiz.

```python
# 1979-yildan keyin chiqqan albomlar uchun mantiqiy shart
shart = df['Released'] > 1979

# Shartga mos keladigan qatorlarni ajratib olib, yangi jadvalga saqlash
df1 = df[shart]
```

Endi `df1` jadvalida faqat 1980 va undan keyingi yillarda chiqqan albomlar saqlanadi.

## 3. Natijalarni saqlash (`to_csv`)
Yaratilgan yangi ma'lumotlar jadvalini kompyuterga fayl ko'rinishida saqlash juda oson. Buning uchun `to_csv` metodidan foydalaniladi:

```python
# Yangi jadvalni CSV formatida saqlash
df1.to_csv('new_albums.csv')
```

> **Muhim:** Fayl nomini yozayotganda `.csv` kengaytmasini yozishni unutmang. Shuningdek, Pandas'da ma'lumotlarni Excel, JSON va boshqa ko'plab formatlarda saqlash imkoniyati ham mavjud.
