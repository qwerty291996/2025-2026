### **1-mashq: Pandas: DataFrame va Series**

**Pandas** — bu Python dasturlash tili asosida yaratilgan, ma’lumotlar tahlili uchun mo‘ljallangan mashhur kutubxonadir. Pandas ma’lumotlar bilan ishlash uchun asosan ikkita ma’lumot tuzilmasini taqdim etadi:

* DataFrame
* Series

**DataFrame** — bu ikki o‘lchamli ma’lumotlar tuzilmasi, ya’ni ma’lumotlar satrlar va ustunlar ko‘rinishida jadval shaklida joylashadi.

* Pandas DataFrame mavjud xotiradagi ma’lumotlar to‘plamini yuklash orqali yaratiladi.
* Ma’lumotlar manbasi SQL ma’lumotlar bazasi, CSV fayli, Excel fayli va h.k. bo‘lishi mumkin.
* Shuningdek, u ro‘yxatlar, lug‘atlar va lug‘atlar ro‘yxatidan ham yaratilishi mumkin.

**Series** — bu indekslangan ma’lumotlarning bir o‘lchamli massividir.
U ikkita asosiy komponentdan iborat:
1. Haqiqiy ma’lumotlar massivi.
2. Indekslar yoki ma’lumot yorliqlarining bog‘langan massivi.

Indeks alohida ma’lumot qiymatlariga kirish uchun ishlatiladi. Shuningdek, siz DataFrame-ning bitta ustunini **Series** sifatida ajratib olishingiz mumkin. Siz Pandas Series-ni bir o‘lchamli (1-D) DataFrame deb tasavvur qilishingiz mumkin.

````python
# Keling, Pandas kutubxonasini import qilamiz
import pandas as pd
````

Pandas-ni import qilganingizdan so'ng, ma'lumotlarni yaratish va tahlil qilish uchun undagi o'rnatilgan funksiyalardan foydalanishingiz mumkin.

**Ushbu amaliy laboratoriya mashg'ulotida biz lug'at (dictionary) asosida DataFrame yaratishni o'rganamiz.**

Quyidagi kalitlar va qiymatlardan iborat 'x' lug'atini ko'rib chiqaylik.

### 2-mashq: `loc()` va `iloc()` funksiyalari

`loc()` — bu yorliqlarga (label) asoslangan ma'lumotlarni tanlash usuli bo'lib, u orqali biz tanlamoqchi bo'lgan satr yoki ustunning **nomini** ko'rsatishimiz kerak. Bu usul ko'rsatilgan diapazonning oxirgi elementini ham o'z ichiga oladi.

Tushunish uchun sodda sintaksis:

 - loc[satr_nomi, ustun_nomi]

`iloc()` — bu indekslarga (index) asoslangan tanlash usuli bo'lib, u orqali ma'lum bir satr yoki ustunni tanlash uchun **butun sonli indeksni** ko'rsatishimiz kerak. Bu usul ko'rsatilgan diapazonning oxirgi elementini o'z ichiga olmaydi.

Tushunish uchun sodda sintaksis:
   
 - iloc[satr_indeksi, ustun_indeksi]



<h4 id="data">Keling, bularga oid ba'zi misollarni ko'rib chiqamiz.</h4>

````python
# Birinchi satr va birinchi ustundagi qiymatga kirish

df.iloc[0, 0]
````

````python
# Birinchi satr va uchinchi ustundagi qiymatga kirish

df.iloc[0, 2]
````

````python
# Ustun nomidan foydalanib qiymatga kirish

df.loc[0, 'Salary']
````

### 3-mashq: Kesish (Slicing)

Kesish (Slicing) — DataFrame-dan ma'lum bir satrlar va/yoki ustunlar to'plamini tanlash uchun `[]` operatoridan foydalanadi.

Satrlar to'plamini kesib olish uchun quyidagi sintaksisdan foydalanasiz: `data[start:stop]`.

Bu yerda **start** (boshlanish) hisobga olinadigan indeksni, **stop** (to'xtash) esa siz tanlamoqchi bo'lgan satrdan KEYINGI bir qadamdagi indeksni anglatadi. Kesishni indeks orqali ham, ustun nomi orqali ham amalga oshirishingiz mumkin.

> **ESLATMA:** Pandas-da kesishda boshlang'ich chegara (`start`) natijaga kiritiladi.

Shunday qilib, agar siz 0, 1 va 2-satrlarni tanlamoqchi bo'lsangiz, kodingiz quyidagicha ko'rinadi: `df.iloc[0:3]`.

Bu Python-ga 0-indeksdan boshlashni va 0, 1, 2-satrlarni tanlashni, lekin 3-indeksni kiritmaslikni buyurayotganingizni anglatadi.

> **ESLATMA:** Yorliqlar (labels) DataFrame-da mavjud bo'lishi kerak, aks holda `KeyError` xatosini olasiz.

Yorliqlar bo'yicha indekslash (ya'ni `loc()` dan foydalanish) butun sonlar bo'yicha indekslashdan (ya'ni `iloc()` dan foydalanish) farq qiladi. `loc()` yordamida ham boshlang'ich, ham oxirgi chegara natijaga **kiritiladi**. `loc()` dan foydalanganda butun sonlarni ishlatish mumkin, lekin bu sonlar pozitsiyani emas, balki indeks yorlig'ini (nomini) anglatadi.

Masalan, `loc()` dan foydalanib `1:4` ni tanlash, `iloc()` yordamida `1:4` satrlarni tanlashdan farqli natija beradi.



<h4 id="data">Biz, shuningdek, DataFrame ichidagi satr va ustun joylashuvi hamda iloc indekslash yordamida aniq bir ma'lumot qiymatini tanlashimiz mumkin.</h4>

````python
# keling, eski df dataframe-dan foydalanib kesishni (slicing) bajaramiz
df.iloc[0:2, 0:3]
````

````python
# indeks ustuni 0, 1, 2 yorliqlariga ega bo'lgan eski df dataframe-da 
# loc() funksiyasi yordamida kesishni (slicing) bajaramiz
df.loc[0:2,'ID':'Department']
````

````python
# indeks ustuni Name bo'lgan va Rose, John hamda Jane yorliqlariga ega 
# yangi df2 dataframe-da loc() funksiyasi yordamida kesishni bajaramiz
df2.loc['Rose':'Jane', 'ID':'Department']
````
