```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Ma'lumotlarni shakllantirish
# Bosqichlar: An'anaviy -> Raqamli -> Interaktiv -> SI Ekotizimi
data = {
    'Bosqichlar': [
        'An’anaviy\n(Kitob/Ma’ruza)', 
        'Raqamli\n(LMS/Elektron)', 
        'Interaktiv\n(Simulyator/Kiber)', 
        'SI Ekotizimi\n(Tanib olish/Kiber-himoya)'
    ],
    'Texnologik murakkablik': [10, 30, 60, 95],
    'Ta’lim samaradorligi': [15, 45, 75, 92]
}

df = pd.DataFrame(data)

# Grafik uslubi
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# Ikki xil ko'rsatkichni chizish
sns.lineplot(x='Bosqichlar', y='Texnologik murakkablik', data=df, marker='o', 
             label='Texnologik murakkablik (Kiber/IT)', linewidth=3, color='#1b4332')
sns.lineplot(x='Bosqichlar', y='Ta’lim samaradorligi', data=df, marker='s', 
             label='Ta’lim samaradorligi (Qaror qabul qilish)', linewidth=3, color='#ce2029')

# Grafik sarlavhasi va yozuvlar
plt.title('Harbiy ta’lim tizimining raqamli evolyutsiyasi', fontsize=16, fontweight='bold', pad=20)
plt.ylabel('Rivojlanish darajasi (%)', fontsize=12)
plt.xlabel('Evolyutsion bosqichlar', fontsize=12)
plt.ylim(0, 110)

# Har bir nuqtaga foiz qiymatlarini qo'shish
for i in range(len(df)):
    plt.text(i, df['Texnologik murakkablik'][i] + 3, f"{df['Texnologik murakkablik'][i]}%", 
             ha='center', fontweight='bold', color='#1b4332')
    plt.text(i, df['Ta’lim samaradorligi'][i] - 7, f"{df['Ta’lim samaradorligi'][i]}%", 
             ha='center', fontweight='bold', color='#ce2029')

plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```
