```python
import matplotlib.pyplot as plt
import seaborn as sns

# Ma'lumotlar
aspects = ['Individual dars rejasi', 'Avtomatik baholash', 'Interaktivlik', 'Ma’lumot tahlili']
impact_scores = [90, 75, 85, 95] # Foizlarda (shartli ravishda)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))
plot = sns.barplot(x=aspects, y=impact_scores, palette="viridis")

plt.title("Sun’iy intellektning o‘quv jarayoni samaradorligiga ta’siri (%)", fontsize=14)
plt.ylabel("Samaradorlik darajasi")
plt.ylim(0, 100)

for p in plot.patches:
    plot.annotate(format(p.get_height(), '.1f'), 
                   (p.get_x() + p.get_width() / 2., p.get_height()), 
                   ha = 'center', va = 'center', 
                   xytext = (0, 9), 
                   textcoords = 'offset points')

plt.show()
```
