```python
import matplotlib.pyplot as plt

# make_graph funksiyasi
def make_graph(tesla_data, tesla_revenue, stock_name):
    # Tesla aksiyalarining yopilish narxlarini olish
    plt.figure(figsize=(10,6))
    plt.plot(tesla_data.index, tesla_data['Close'], label=f'{stock_name} Stock Price')
    
    # Graphning nomi va boshqa o'zgarishlar
    plt.title(f'{stock_name} Stock Data')
    plt.xlabel('Date')
    plt.ylabel('Price in USD')
    plt.legend()
    plt.grid(True)
    plt.show()

# Tesla stock graphini yaratish
make_graph(tesla_data, tesla_revenue, 'Tesla')
```
