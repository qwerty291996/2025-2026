**Ma’lumotlar tavsifini sharhlash**

Natijalar asl ma’lumotlar to‘plamingizdagi har bir ustun uchun 8 ta raqamni ko‘rsatadi. Birinchi raqam — `count` (soni), qancha qatorlarda bo‘sh bo‘lmagan qiymatlar mavjudligini ko‘rsatadi.

Bo‘sh qiymatlar turli sabablarga ko‘ra yuzaga keladi. Masalan, 1 xonali uy so‘rovnoma o‘tkazilganda, 2-yotoqxona hajmi yig‘ilmagan bo‘lishi mumkin. Biz bo‘sh ma’lumotlar mavzusiga keyinroq qaytamiz.

Ikkinchi qiymat — `mean` (o‘rta arifmetik qiymat), ya’ni o‘rtacha ko‘rsatkich. Uning ostida `std` (standart chetlanish) turadi, u qiymatlar son jihatdan qanchalik tarqoq ekanligini o‘lchaydi.

`min`, `25%`, `50%`, `75%` va `max` qiymatlarini sharhlash uchun har bir ustunni eng kichik qiymatdan eng katta qiymatgacha tartiblagan holda tasavvur qiling. Birinchi (eng kichik) qiymat `min` hisoblanadi. Agar siz ro‘yxatning chorak qismigacha borsangiz, siz qiymatlarning 25% dan kattaroq va 75% dan kichikroq bo‘lgan raqamni topasiz. Bu `25%` qiymati (o‘qilishi: "25-persentil"). 50- va 75-persentillar xuddi shunday tarzda aniqlanadi, `max` esa eng katta raqamdir.
