### 9.2. Pufakchali saralashni vizuallashtirish (Visualizing Bubble Sort)

Ushbu paragrafda **PyGame** kutubxonasi yordamida **Bubble Sort** (pufakchali saralash) algoritmini vizuallashtirish usuli koʻrib chiqiladi. Dastur ishga tushirilganda turli balandlikdagi saralanmagan ustunlar hosil boʻladi va "Space" (boʻshliq) tugmasi bosilganda saralash jarayoni boshlanadi. Har bir iteratsiyadan soʻng eng katta qiymatga ega element oxirgi oʻringa oʻtadi. Bubble Sort algoritmi elementlarni birma-bir taqqoslaydi va ularni qiymatlariga qarab tartiblaydi.

**Amalga oshirish bosqichlari:**
1. Asosiy oynani yaratish.
2. Asosiy oynani qora rang bilan toʻldirish.
3. Ma'lum bir oraliq masofada joylashgan ustunlar roʻyxatini koʻrsatuvchi metod yaratish.
4. Foydalanuvchidan tugmalar kiritilishini qabul qilish.
5. Agar "Space" tugmasi bosilsa, saralash jarayonini boshlash.
6. Roʻyxatda Bubble Sort algoritmini qoʻllash.
7. Har bir ichki iteratsiyadan soʻng ekranni qora rang bilan toʻldirish va yangilangan roʻyxatni ustunlar shaklida koʻrsatish uchun "show" metodini chaqirish.

**Dastur kodi:**

```python
!pip install pygame
```

```python
# pygame kutubxonasini yuklash
import pygame
pygame.init()

# oyna o'lchamini belgilash
win = pygame.display.set_mode((500, 400))

# oyna sarlavhasini o'rnatish
pygame.display.set_caption("Bubble sort vizuallashtirish")

# boshlang'ich pozitsiya
x = 40
y = 40

# har bir ustun kengligi
width = 20

# har bir ustun balandligi (saralanadigan ma'lumotlar)
height = [200, 50, 130, 90, 250, 61, 110, 88, 33, 80, 70, 159, 180, 20]

run = True

# ro'yxatni ko'rsatish metodi
def show(height):
    # ro'yxatning har bir elementini takrorlash
    for i in range(len(height)):
        # har bir ustunni tegishli masofa bilan chizish
        pygame.draw.rect(win, (255, 0, 0), (x + 30 * i, y, width, height[i]))

# cheksiz sikl
while run:
    # saralashni boshlash bayrog'i
    execute = False

    # vaqt kechikishi
    pygame.time.delay(10)

    # hodisalarni tekshirish
    for event in pygame.event.get():
        # agar dasturdan chiqish tugmasi bosilsa
        if event.type == pygame.QUIT:
            run = False

    # bosilgan tugmalarni aniqlash
    keys = pygame.key.get_pressed()

    # agar space tugmasi bosilsa
    if keys[pygame.K_SPACE]:
        execute = True

    # agar execute bayrog'i yolg'on bo'lsa
    if execute == False:
        # oynani qora rang bilan to'ldirish
        win.fill((0, 0, 0))
        # ro'yxatni ko'rsatish metodini chaqirish
        show(height)
        # oynani yangilash
        pygame.display.update()
    else:
        # pufakchali saralash usulini boshlash
        for i in range(len(height) - 1):
            for j in range(len(height) - i - 1):
                # agar joriy element keyingisidan katta bo'lsa
                if height[j] > height[j + 1]:
                    # o'rinlarini almashtirish (swap)
                    t = height[j]
                    height[j] = height[j + 1]
                    height[j + 1] = t

                # har bir almashinuvdan so'ng ekranni yangilash
                win.fill((0, 0, 0))
                show(height)
                # ko'rishga ulgurish uchun vaqt kechikishi
                pygame.time.delay(50)
                pygame.display.update()

# pygame oynasini yopish
pygame.quit()
```

### 9.3. Birlashtirib saralashni vizuallashtirish (Merge Sort Visualization)

Ushbu maqolada **Matplotlib** kutubxonasi yordamida **Merge Sort** (birlashtirish orqali saralash) algoritmini vizuallashtirish usuli koʻrib chiqiladi. Algoritmlarni vizuallashtirish ularning ishlash jarayonidagi amallar soni, solishtirish va elementlar oʻrnini almashtirish bosqichlarini tahlil qilish orqali tushunishni osonlashtiradi.

### Umumiy yondashuv:
*   Tasodifiy elementlardan iborat massiv hosil qilinadi.
*   Algoritm massivda chaqiriladi va vizuallashtirish maqsadida `return` oʻrniga `yield` operatoridan foydalaniladi.
*   Solishtirish va oʻrin almashtirishdan keyingi massivning joriy holatlari `yield` orqali uzatiladi. Natijada algoritm **generator** obyektini qaytaradi.
*   Matplotlib kutubxonasining animatsiya funksiyalaridan foydalaniladi.
*   Massiv Matplotlib ustunli diagrammasi (`bar_rects`) koʻrinishida saqlanadi, bunda har bir ustun balandligi mos element qiymatiga teng boʻladi.
*   Matplotlib animatsiyasining oʻrnatilgan `FuncAnimation` metodi konteyner va generator obyektlarini animatsiya yaratuvchi funksiyaga uzatadi. Animatsiyaning har bir kadri (frame) generatorning bitta iteratsiyasiga mos keladi.



### Dasturiy amalga oshirish:

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import matplotlib as mp
import numpy as np
import random

# Animatsiya obyekti o'chib ketmasligi uchun uni global darajada e'lon qilamiz
anim = None 

def mergesort(A, start, end):
    if end <= start:
        return
    mid = start + ((end - start + 1) // 2) - 1
    yield from mergesort(A, start, mid)
    yield from mergesort(A, mid + 1, end)
    yield from merge(A, start, mid, end)

def merge(A, start, mid, end):
    merged = []
    leftIdx = start
    rightIdx = mid + 1
    while leftIdx <= mid and rightIdx <= end:
        if A[leftIdx] < A[rightIdx]:
            merged.append(A[leftIdx])
            leftIdx += 1
        else:
            merged.append(A[rightIdx])
            rightIdx += 1
    while leftIdx <= mid:
        merged.append(A[leftIdx])
        leftIdx += 1
    while rightIdx <= end:
        merged.append(A[rightIdx])
        rightIdx += 1
    for i in range(len(merged)):
        A[start + i] = merged[i]
        yield A

def showGraph():
    global anim # Funksiya ichida global o'zgaruvchidan foydalanamiz
    
    n = 20
    a = [i for i in range(1, n+1)]
    random.shuffle(a)
    generator = mergesort(a, 0, len(a)-1)
    
    fig, ax = plt.subplots()
    bar_rects = ax.bar(range(len(a)), a, align="edge", color="red")
    
    ax.set_xlim(0, len(a))
    ax.set_ylim(0, int(1.1*len(a)))
    
    iteration = [0]
    def animate(A, rects, iteration):
        for rect, val in zip(rects, A):
            rect.set_height(val)
        iteration[0] += 1

    # Animatsiyani o'zgaruvchiga biriktiramiz
    anim = FuncAnimation(fig, func=animate,
                         fargs=(bar_rects, iteration), 
                         frames=generator, 
                         interval=50,
                         repeat=False,
                         cache_frame_data=False)
    plt.show()

showGraph()
```

### 9.4. Tezkor saralashni vizuallashtirish (Quick Sort Visualization)

Algoritmlarni vizuallashtirish, elementlarni solishtirish va o'rinlarini almashtirish uchun bajarilgan amallar sonini tahlil qilish va taqqoslash orqali ularni tushunishni osonlashtiradi. Buning uchun biz massiv elementlarini ifodalovchi ustunli diagrammalarni chizishda matplotlib kutubxonasidan foydalanamiz.

**Yondashuv:**
*   Biz tasodifiy elementlardan iborat massiv yaratamiz.
*   Algoritm ushbu massiv uchun chaqiriladi va vizuallashtirish maqsadida `return` (qaytarish) operatori o'rniga `yield` operatoridan foydalaniladi.
*   Biz solishtirish va o'rin almashtirishlardan so'ng massivning joriy holatlarini `yield` orqali uzatamiz. Shunday qilib, algoritm generator obyektini qaytaradi.
*   Massivdagi solishtirish va o'rin almashtirishlarni vizuallashtirish uchun Matplotlib animatsiyasidan foydalaniladi.
*   Massiv Matplotlib'ning ustunli konteyner obyektida (`bar_rects`) saqlanadi, bunda har bir ustunning o'lchami massivdagi mos elementning qiymatiga teng bo'ladi.
*   Matplotlib animatsiyasining o'rnatilgan `FuncAnimation` metodi konteyner va generator obyektlarini animatsiya yaratish uchun ishlatiladigan funksiyaga uzatadi. Animatsiyaning har bir kadri generatorning bitta iteratsiyasiga mos keladi.
*   Takroran chaqiriladigan animatsiya funksiyasi to'g'rtburchakning (ustunning) balandligini elementlarning qiymatiga teng qilib o'rnatadi.

Quyida ushbu yondashuvning kodda amalga oshirilishi keltirilgan (xatoliklarni oldini olish uchun yopishib qolgan qatorlar to'g'rilab yozildi):

```python
%matplotlib qt
# Barcha modullarni yuklash
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import matplotlib as mp
import numpy as np
import random

# Animatsiya o'chib ketmasligi uchun global o'zgaruvchi e'lon qilamiz
anim = None

# Quick Sort funksiyasi
def quicksort(a, l, r):
    if l >= r:
        return 
    x = a[l]
    j = l
    for i in range(l + 1, r + 1):
        if a[i] <= x:
            j += 1
            a[j], a[i] = a[i], a[j]
        yield a
    a[l], a[j] = a[j], a[l]
    yield a

    yield from quicksort(a, l, j-1)
    yield from quicksort(a, j + 1, r)

def showGraph():
    global anim # Funksiya ichida global o'zgaruvchidan foydalanamiz

    # Massiv o'lchamini kiritish
    n = int(input("Massiv o'lchamini kiriting: \n"))
    a = [i for i in range(1, n + 1)]
    random.shuffle(a)

    generator = quicksort(a, 0, n-1)

    plt.style.use('fivethirtyeight')

    data_normalizer = mp.colors.Normalize()
    color_map = mp.colors.LinearSegmentedColormap(
        "my_map",
        {
            "red": [(0, 1.0, 1.0), (1.0, .5, .5)],
            "green": [(0, 0.5, 0.5), (1.0, 0, 0)],
            "blue": [(0, 0.50, 0.5), (1.0, 0, 0)]
        }
    )

    fig, ax = plt.subplots()

    bar_rects = ax.bar(range(len(a)), a, align="edge", 
                       color=color_map(data_normalizer(range(n))))

    ax.set_xlim(0, len(a))
    ax.set_ylim(0, int(1.1 * len(a)))
    ax.set_title("ALGORITM : Quick Sort\nMA'LUMOTLAR : Tasodifiy", 
                 fontdict={'fontsize': 13, 'fontweight': 'medium', 'color': '#E4365D'})

    text = ax.text(0.01, 0.95, "", transform=ax.transAxes, color="#E4365D")
    iteration = [0]

    def animate(A, rects, iteration):
        for rect, val in zip(rects, A):
            rect.set_height(val)
        iteration[0] += 1
        text.set_text("iteratsiyalar : {}".format(iteration[0]))

    # Animatsiyani yaratish va keshni o'chirish
    anim = FuncAnimation(fig, func=animate,
                         fargs=(bar_rects, iteration), 
                         frames=generator, 
                         interval=50,
                         repeat=False,
                         cache_frame_data=False) # Kesh xatosi yechimi
    plt.show()

# Dasturni ishga tushirish
showGraph()
```

### 9.5. 3D Tezkor saralashni vizuallashtirish (3D Quick Sort Visualization)

Algoritmlarni vizuallashtirish, elementlarni solishtirish va o'rinlarini almashtirish uchun bajarilgan amallar sonini tahlil qilish hamda taqqoslash orqali ularning qanday ishlashini tushunishni osonlashtiradi. Algoritmlarning 3D vizualizatsiyasi kamroq uchraydi, shu sababli biz massiv elementlarini ifodalash va ularni animatsiya qilishda **Matplotlib** yordamida 3 o'lchamli ustunli diagrammalar (bar graphs) chizamiz.

Keling, **Quick Sort** (tezkor saralash) algoritmining 3D vizualizatsiyasini ko'rib chiqamiz:

**Yondashuv:**
1.  Biz tasodifiy elementlardan iborat massiv yaratamiz.
2.  Algoritm ushbu massiv uchun chaqiriladi va vizuallashtirish maqsadida `return` (qaytarish) operatori o'rniga `yield` operatoridan foydalaniladi.
3.  Biz solishtirish va o'rin almashtirishlardan so'ng massivning joriy holatlarini `yield` orqali uzatamiz. Shunday qilib, algoritm generator obyektini qaytaradi.
4.  Massivdagi solishtirish va o'rin almashtirishlarni vizuallashtirish uchun Matplotlib animatsiyasidan foydalaniladi.
5.  Shundan so'ng biz grafikni chizamiz, u `Poly3dCollection` obyektini qaytaradi va keyingi animatsiyalar aynan shu obyekt yordamida amalga oshiriladi.

```python
%matplotlib qt
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from mpl_toolkits.mplot3d import axes3d
import matplotlib as mp
import numpy as np
import random

# Animatsiya obyekti o'chib ketmasligi uchun global o'zgaruvchi
anim = None

def quicksort(a, l, r):
    """Tezkor saralash algoritmi generator ko'rinishida."""
    if l >= r:
        return 
    x = a[l]
    j = l
    for i in range(l + 1, r + 1):
        if a[i] <= x:
            j += 1
            a[j], a[i] = a[i], a[j]
        yield a
    a[l], a[j] = a[j], a[l]
    yield a
    
    yield from quicksort(a, l, j - 1)
    yield from quicksort(a, j + 1, r)

def showGraph():
    global anim
  
    # Kiruvchi ma'lumotlar
    try:
        n = int(input("Massiv o'lchamini kiriting (masalan, 20):\n"))
    except ValueError:
        n = 20
        print("Xato kiritish! Standart 20 o'lcham tanlandi.")

    a = [i for i in range(1, n + 1)]
    random.shuffle(a)
    
    generator = quicksort(a, 0, n - 1)
    algoName = 'Quick Sort 3D'
    
    plt.style.use('fivethirtyeight')
    
    # Ranglar palitrasi
    data_normalizer = mp.colors.Normalize(vmin=1, vmax=n)
    color_map = plt.get_cmap('magma') # Yaxshiroq ko'rinish uchun standart palitra

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(projection='3d')
    
    # Animatsiya uchun kerakli parametrlar
    z = np.zeros(n)
    dx = np.ones(n) * 0.8 # Ustunlar bir-biriga yopishib qolmasligi uchun
    dy = np.ones(n) * 0.8
    iteration = [0]

    def animate(A, iteration):
        # 1. Sahnani to'liq tozalash
        ax.clear()
        
        # 2. Ustunlarni chizish
        # dz - bu ustun balandligi, u A (massiv) qiymatlariga teng
        dz = np.array(A)
        colors = color_map(data_normalizer(dz))
        
        ax.bar3d(range(len(A)), np.zeros(n), z, dx, dy, dz, color=colors)
        
        # 3. O'qlarni va sarlavhani qayta sozlash (tozalashdan keyin shart)
        ax.set_xlim(-1, n)
        ax.set_ylim(-1, 1) # Y o'qi bizda tekis
        ax.set_zlim(0, int(1.1 * n))
        
        ax.set_title(f"ALGORITM : {algoName}\nIteratsiyalar : {iteration[0]}", 
                     fontdict={'fontsize': 13, 'color' : '#E4365D'})
        
        iteration[0] += 1

    # Animatsiyani ishga tushirish
    # 3D renderlash og'ir bo'lgani uchun intervalni 100ms qildik
    anim = FuncAnimation(fig, 
                         func=animate,
                         fargs=(iteration,), 
                         frames=generator, 
                         interval=100,
                         repeat=False,
                         cache_frame_data=False)
    
    plt.show()

if __name__ == '__main__':
    showGraph()
```

### 9.6. 3D Birlashtirib saralashni vizuallashtirish (3D Merge Sort Visualization)

Algoritmlarni vizuallashtirish elementlarni solishtirish va o'rinlarini almashtirishda bajariladigan amallar sonini tahlil qilish orqali ularning ishlashini tushunishga yordam beradi. 3D vizualizatsiya kamroq qo'llanilsa-da, Matplotlib yordamida massiv elementlarini ifodalash va ularni animatsiya qilish mumkin.

*   Tasodifiy elementlardan iborat massiv yaratiladi.
*   Vizuallashtirish uchun algoritmda `return` o'rniga `yield` operatori ishlatiladi.
*   Elementlarni solishtirish va almashtirishdan keyingi joriy holatlar `yield` qilinadi, natijada algoritm generator obyektini qaytaradi.
*   Matplotlib animatsiyasi massivning o'zgarish jarayonini ko'rsatish uchun xizmat qiladi.
*   Diagramma chizilganda u `Poly3dCollection` obyektini qaytaradi va aynan shu obyekt yordamida keyingi animatsiyalar bajariladi.

Quyida **Merge Sort** (birlashtirish orqali saralash) algoritmining 3D vizualizatsiyasi uchun to'g'rilangan kod varianti keltirilgan:

```python
%matplotlib qt
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import matplotlib as mp
import numpy as np
import random

# Animatsiya obyekti o'chib ketmasligi uchun global o'zgaruvchi
anim = None

# Massivni bo'lish uchun merge sort funksiyasi
def mergesort(A, start, end):
    if end <= start:
        return

    mid = start + ((end - start + 1) // 2) - 1

    # yield from operatori merge funksiyasidan massivni olish uchun ishlatiladi
    yield from mergesort(A, start, mid)
    yield from mergesort(A, mid + 1, end)
    yield from merge(A, start, mid, end)

# Massivni birlashtirish funksiyasi
def merge(A, start, mid, end):
    merged = []
    leftIdx = start
    rightIdx = mid + 1

    while leftIdx <= mid and rightIdx <= end:
        if A[leftIdx] < A[rightIdx]:
            merged.append(A[leftIdx])
            leftIdx += 1
        else:
            merged.append(A[rightIdx])
            rightIdx += 1

    while leftIdx <= mid:
        merged.append(A[leftIdx])
        leftIdx += 1

    while rightIdx <= end:
        merged.append(A[rightIdx])
        rightIdx += 1

    for i in range(len(merged)):
        A[start + i] = merged[i]
        yield A

# Grafikni chizish funksiyasi
def showGraph():
    global anim

    # Tasodifiy qiymatlar uchun o'lchamni kiritish
    try:
        n = int(input("Massiv o'lchamini kiriting:\n"))
    except ValueError:
        n = 20
        print("Xato! Standart 20 o'lcham tanlandi.")

    a = [i for i in range(1, n + 1)]
    random.shuffle(a)
    datasetName = 'Tasodifiy'

    generator = mergesort(a, 0, len(a)-1)
    algoName = 'Merge Sort 3D'

    plt.style.use('fivethirtyeight')
    data_normalizer = mp.colors.Normalize(vmin=1, vmax=n)
    color_map = plt.get_cmap('magma')

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(projection = '3d')

    z = np.zeros(n)
    dx = np.ones(n) * 0.8
    dy = np.ones(n) * 0.8
    iteration = [0]

    def animate(A, iteration):
        # Sahnani tozalash
        ax.clear()
        
        # Ustunlarni chizish
        dz = np.array(A)
        colors = color_map(data_normalizer(dz))
        
        ax.bar3d(range(len(A)), np.zeros(n), z, dx, dy, dz, color=colors)

        # O'qlarni va sarlavhani qayta sozlash
        ax.set_xlim(-1, n)
        ax.set_ylim(-1, 1)
        ax.set_zlim(0, int(1.1 * n))
        
        ax.set_title(f"ALGORITM : {algoName}\nMA'LUMOTLAR : {datasetName}", 
                     fontdict={'fontsize': 13, 'color' : '#E4365D'})
        
        iteration[0] += 1
        ax.text2D(0.1, 0.95, f"iteratsiyalar : {iteration[0]}", 
                  transform=ax.transAxes, color="#E4365D")
            
    # Animatsiyani ishga tushirish
    anim = FuncAnimation(fig, func = animate,
                         fargs =(iteration,), 
                         frames = generator, 
                         interval = 100,
                         repeat = False,
                         cache_frame_data = False)
    plt.show()

if __name__ == '__main__':
    showGraph()
```

### 9.7. 3D Kiritish orqali saralashni vizuallashtirish (3D Insertion Sort Visualization)

Algoritmlarni vizuallashtirish elementlarni solishtirish va o'rinlarini almashtirishda bajariladigan amallar sonini tahlil qilish orqali ularning ishlashini tushunishni osonlashtiradi. Algoritmlarning 3D vizualizatsiyasi kamroq tarqalgan bo'lsa-da, ushbu usulda **Matplotlib** yordamida massiv elementlarini ifodalash va ularni animatsiya qilish uchun 3 o'lchamli ustunli grafiklardan foydalaniladi.

### Amalga oshirish yondashuvi:
*   Tasodifiy elementlardan iborat massiv yaratiladi.
*   Vizuallashtirish maqsadida algoritmda `return` o'rniga `yield` operatori ishlatiladi, natijada algoritm generator obyektini qaytaradi.
*   Solishtirish va almashtirishlardan so'ng massivning joriy holatlari `yield` orqali uzatiladi.
*   Massivdagi o'zgarishlarni ko'rsatish uchun Matplotlib animatsiyasidan foydalaniladi.
*   Grafik chizilganda u `Poly3dCollection` obyektini qaytaradi va aynan shu obyekt yordamida animatsiya davom ettiriladi.

Quyida **Insertion Sort** (joylashtirish orqali saralash) algoritmining 3D vizualizatsiyasi uchun to'g'rilangan (Python 3.13 va Jupyter Notebook muhitiga moslashtirilgan) kod varianti keltirilgan:

```python
%matplotlib qt
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import matplotlib as mp
import numpy as np
import random

# Animatsiya obyekti o'chib ketmasligi uchun global o'zgaruvchi
anim = None

# Insertion sort algoritmi generator ko'rinishida
def insertionsort(a):
    for j in range(1, len(a)):
        key = a[j]
        i = j - 1
        
        while(i >= 0 and a[i] > key):
            a[i + 1] = a[i]
            i -= 1
            yield a # Joriy holatni uzatish
        a[i + 1] = key
        yield a # Har bir asosiy qadamdan keyin holatni uzatish

# Grafikni chizish funksiyasi
def showGraph(n):
    global anim # Funksiya ichida global o'zgaruvchidan foydalanamiz
  
    # Tasodifiy qiymatlarni yaratish
    a = [i for i in range(1, n + 1)]
    random.shuffle(a)
    datasetName = 'Tasodifiy'
    algoName = 'Insertion Sort'
    
    # Generator obyektini olish
    generator = insertionsort(a)

    # Grafik uslubi
    plt.style.use('fivethirtyeight')

    # Ustunlar ranglarini sozlash
    data_normalizer = mp.colors.Normalize(vmin=1, vmax=n)
    color_map = plt.get_cmap('magma')

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(projection='3d')
    
    # Z o'qi qiymatlari va ustunlar pozitsiyasi
    z = np.zeros(n)
    dx = np.ones(n) * 0.8
    dy = np.ones(n) * 0.8
    iteration = [0]

    # Animatsiya funksiyasi
    def animate(A, iteration):
        # Sahnani tozalash
        ax.clear()
        
        # Ustunlarni chizish (A massiv elementlari dz - balandlik sifatida)
        dz = np.array(A)
        colors = color_map(data_normalizer(dz))
        
        ax.bar3d(range(len(A)), np.zeros(n), z, dx, dy, dz, color=colors)
        
        # O'qlarni va sarlavhani qayta sozlash
        ax.set_xlim(-1, n)
        ax.set_ylim(-1, 1)
        ax.set_zlim(0, int(1.1 * n))
        
        ax.set_title(f"ALGORITM : {algoName}\nMA'LUMOTLAR : {datasetName}", 
                     fontdict={'fontsize': 13, 'color': '#E4365D'})
        
        iteration[0] += 1
        ax.text2D(0.1, 0.95, f"iteratsiyalar : {iteration[0]}", 
                  transform=ax.transAxes, color="#E4365D")

    # Animatsiyani ishga tushirish
    anim = FuncAnimation(fig, func=animate,
                         fargs=(iteration,),
                         frames=generator, 
                         interval=100,
                         repeat=False,
                         cache_frame_data=False)
    plt.show()

# Dasturni ishga tushirish
if __name__ == "__main__":
    n = 11 # Massiv o'lchami
    showGraph(n)
```

### 9.8. Harakatlanuvchi o‘rtacha qiymatga ega vaqtli qatorlar grafigini yaratish (Create a Time Series Plot with Rolling Average)

Vaqtli qatorlar grafigi (Time Series Plot) ma'lum bir vaqt oralig'ida ma'lumotlar to'plamidagi turli tendensiyalarni kuzatish uchun ishlatiladi. Bunday muammolarda ma'lumotlar vaqt bo'yicha tartiblangan bo'ladi va ma'lumotlar to'plamidagi vaqt birligiga (kun, oy, soniya, soat va h.k.) qarab o'zgarishi mumkin.

Vaqtli qator ma'lumotlarini chizishda bu tebranishlar grafikdagi eng yuqori (peaks) va eng past (troughs) nuqtalar haqida aniq tushunchaga ega bo'lishimizga to'sqinlik qilishi mumkin. Shuning uchun, ma'lumotlardan aniq qiymat olish uchun biz vaqtli qator grafigini yaratishda **sirpanuvchi o'rtacha** (rolling average) konsepsiyasidan foydalanamiz.

### Sirpanuvchi o'rtacha haqida
Sirpanuvchi o'rtacha yoki harakatlanuvchi o'rtacha (moving average) — bu oxirgi 'n' ta qiymatning oddiy o'rtacha arifmetigidir. U boshqa yo'l bilan aniqlash qiyin bo'lgan tendensiyalarni topishda va uzoq muddatli tendensiyalarni aniqlashda yordam beradi.

**Hisoblash usuli va xususiyatlari:**
*   Sirpanuvchi o'rtachani oldingi 'n' ta qiymatni qo'shib, ularni 'n' soniga bo'lish orqali hisoblash mumkin.
*   Bunda sirpanuvchi o'rtachaning birinchi (n-1) ta qiymati **NaN** (mavjud bo'lmagan son) bo'ladi.
*   Pandas kutubxonasida hisoblash sintaksisi: `pandas.DataFrame.rolling(n).mean()`.

### 9.9. Turli tovushlarni grafikda tasvirlash (Plotting Different Sounds)

Ushbu paragrafda biz Python va Matplotlib yordamida tovush to'lqinlarini vizuallashtirish usulini ko'rib chiqamiz.

**Kerakli modullar**

**Yondashuv**
*   `matplotlib`, `numpy`, `wave` va `sys` modullarini yuklab oling.
*   Audio faylni `wave.open()` metodi yordamida oching.
*   Ochilgan tovush to'lqinining barcha kadrlarini (frames) `readframes()` funksiyasi yordamida o'qing.
*   Kadrlar tezligini (frame rate) `getframerate()` funksiyasi yordamida o'zgaruvchiga saqlang.
*   Va nihoyat, kadrlar tezligidan foydalanib, X o'qini soniyalarda chizing.
*   Olingan grafikni chizish uchun `matplotlib.figure()` funksiyasidan foydalaning.
*   Talabga muvofiq o'q nomlaridan (labels) foydalaning.

Quyida uning kodda amalga oshirilishi keltirilgan:

```python
# modullarni yuklash
import matplotlib.pyplot as plt
import numpy as np
import wave, sys

# tovush to'lqinlarini ko'rsatuvchi funksiya
def visualize(path: str):
  
    # audio faylni o'qish
    raw = wave.open(path)
    
    # barcha kadrlarni o'qiydi 
    # -1 barcha yoki maksimal kadrlarni bildiradi
    signal = raw.readframes(-1)
    signal = np.frombuffer(signal, dtype="int16")
    
    # kadrlar tezligini oladi
    f_rate = raw.getframerate()

    # X o'qini soniyalarda chizish uchun 
    # kadrlar tezligini olishingiz va 
    # audio fayl o'lchami bilan chiziqli ravishda 
    # ajratilgan Vaqt Vektorini yaratish uchun 
    # uni signal o'lchamiga bo'lishingiz kerak
    time = np.linspace(
        0, # boshlanishi
        len(signal) / f_rate,
        num=len(signal)
    )

    # matplotlib yordamida grafik chizish
    # yangi oyna (figure) yaratadi
    plt.figure(1)
    
    # grafik sarlavhasi
    plt.title("Sound Wave (Tovush to'lqini)")
    
    # x o'qi nomi
    plt.xlabel("Time (Vaqt)")
   
    # chizish jarayoni
    plt.plot(time, signal)
    
    # grafikni yangi oynada ko'rsatadi
    plt.show()

    # shuningdek, siz grafikni quyidagi 
    # kod yordamida saqlashingiz ham mumkin:
    # plt.savefig('fayl_nomi.png')

if __name__ == "__main__":
  
    # buyruqlar qatori orqali uzatilgan qiymatni (fayl manzilini) oladi
    path = sys.argv[1]

    visualize(path)
```
