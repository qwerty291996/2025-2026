### 4.1. 3D grafiklar yaratish (Creating 3D Plots)

Uchta o‘zgaruvchini o‘z ichiga olgan ma’lumotlarni vizuallashtirish ko‘pincha ikki o‘lchamli (2D) grafiklar ko‘rsata olmaydigan murakkab bog‘liqliklar va qonuniyatlarni yaxshiroq tushunish uchun **uch o‘lchamli (3D) chizmalarni** talab qiladi. Python-ning Matplotlib kutubxonasi o‘zining `mpl_toolkits.mplot3d` vositalar to‘plami (toolkit) orqali 3D vizuallashtirishni juda yaxshi qo‘llab-quvvatlaydi.

3D grafiklarni yaratishni boshlash uchun birinchi asosiy qadam — chizma o‘qlarida (axes) 3D proyeksiyasini yoqish orqali muhitni sozlab olishdir. 

**Misol: Matplotlib yordamida 3D o‘qlarni chizish**

```python
import matplotlib.pyplot as plt

# Yangi shakl (figure) yaratish
fig = plt.figure()

# Shaklga 3D proyeksiyali o‘qlarni (axes) qo‘shish
ax = plt.axes(projection='3d')

# Grafikni ekranga chiqarish
plt.show()
```

**Izoh:**

*   `plt.figure()` – Barcha grafik elementlari uchun asosiy idish (konteyner) bo‘lib xizmat qiladigan yangi shakl (figure) obyektini yaratadi.
*   `fig.add_subplot(111, projection='3d')` (yoki yuqoridagi kodda ko‘rsatilgandek `plt.axes(projection='3d')`) – Shaklga 3D proyeksiyasi yoqilgan o‘qlar to‘plamini qo‘shadi. Bu yerdagi `111` raqami *"1 qator, 1 ustun, birinchi ichki grafik (subplot)"* degan ma’noni anglatadi. Har ikki usul ham bitta natijani — bo‘sh 3D maydonini beradi.
*   `plt.show()` – Chizma oynasini yuklaydi va 3D o‘qlarni vizual tarzda ekranga chiqaradi. Natijada x, y va z o‘qlariga ega bo‘sh 3D maydon hosil bo‘ladi.

### Matplotlib yordamida uch o‘lchamli (3D) grafik chizishga misol

Uch o‘lchamli grafiklarning eng asosiy va ko‘p ishlatiladigan turlaridan biri bu — 3D chiziqli grafikdir.

### 1. 3D chiziqli grafik (3D Line plot)

**3D chiziqli grafik** uzluksiz yo‘lni (trayektoriyani) vizuallashtirish uchun uch o‘lchamli fazodagi nuqtalarni bir-biriga tuthtashtiradi. Bu usul biror o‘zgaruvchining vaqt yoki 3D makon bo‘ylab qanday o‘zgarishini ko‘rsatish uchun juda qulaydir. 

Quyidagi misolda spiral (spiralsimon) trayektoriyani chizish uchun sinus va kosinus funksiyalaridan foydalanilgan.


```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

# Shakl va 3D o‘qlarni yaratish
fig = plt.figure()
ax = plt.axes(projection='3d')

# Z o‘qi uchun 0 dan 1 gacha bo‘lgan 100 ta nuqta
z = np.linspace(0, 1, 100)

# Spiral hosil qilish uchun X va Y koordinatalarini hisoblash
x = z * np.sin(25 * z)
y = z * np.cos(25 * z)

# 3D chiziqni yashil rangda chizish
ax.plot3D(x, y, z, 'green')
ax.set_title('3D Chiziqli Grafik (3D Line Plot)')

plt.show()
```

**Izoh:**
Biz `np.linspace()` funksiyasi yordamida `z` o‘qi uchun 0 va 1 oralig‘ida 100 ta nuqta yaratib olamiz. So‘ngra, fazoda spiral shaklini hosil qilish uchun ushbu `z` qiymatlari asosida `x = z * np.sin(25 * z)` va `y = z * np.cos(25 * z)` formulalarini hisoblab chiqamiz. Yakunda ushbu 3D spiral `ax.plot3D(x, y, z, 'green')` buyrug‘i yordamida chiziladi.

### 2. 3D Tarqoq grafik (3D Scatter plot)

**3D tarqoq grafik** alohida ma'lumotlar nuqtalarini uch o‘lchamda ko‘rsatadi. Bu tendentsiyalar (trendlar) yoki klasterlarni (to‘dalarni) aniqlash uchun foydalidir. Har bir nuqta (x, y, z) qiymatlariga ega bo‘lgan pozitsiyani ifodalaydi. To‘rtinchi o‘lchamni qo‘shish uchun rangdan foydalanish mumkin.


```python
import matplotlib.pyplot as plt
import numpy as np

# Shakl va 3D o‘qlarni yaratish
fig = plt.figure()
ax = plt.axes(projection='3d')

# Ma'lumotlarni tayyorlash (oldingi misoldagidek)
z = np.linspace(0, 1, 100)
x = z * np.sin(25 * z)
y = z * np.cos(25 * z)

# x va y asosidagi ranglar massivi (to‘rtinchi o‘lcham)
c = x + y

# 3D tarqoq grafikni chizish
ax.scatter(x, y, z, c=c)
ax.set_title('3D Tarqoq Grafik (Scatter Plot)')

plt.show()
```

**Izoh:**
o‘sha bir xil `x`, `y` va `z` qiymatlaridan foydalangan holda, `ax.scatter()` funksiyasi alohida 3D nuqtalarni chizadi. Nuqtalarning ranglari `c = x + y` orqali belgilanadi, bu esa nuqtalar orasidagi o‘zgarishlarni vizuallashtirish uchun to‘rtinchi o‘lchamni qo‘shadi.

### 3. Sirt grafigi (Surface Plot)

**Sirt grafiklari** $X$ va $Y$ qiymatlari to‘ri (grid) bo‘ylab yoyilgan va $Z$ qiymatlari orqali shakllantirilgan uzluksiz, silliq sirtni ko‘rsatadi. Ular ma'lumotlarning aniq topografiyasini (relyefini) taqdim etib, ikkita o‘zgaruvchili funksiyalarni vizuallashtirish uchun juda qulay vosita hisoblanadi.

```python
import numpy as np
import matplotlib.pyplot as plt

# X va Y koordinatalari uchun to‘r (grid) yaratish
x = np.outer(np.linspace(-2, 2, 10), np.ones(10))
y = x.copy().T  # Qator va ustunlarni almashtirish (transponirlash)

# Z qiymatini hisoblash: Z = cos(X^2 + Y^3)
z = np.cos(x**2 + y**3)

# Shakl va 3D o‘qlarni yaratish
fig = plt.figure()
ax = plt.axes(projection='3d')

# Sirt grafigini chizish (rang palitrasi 'viridis', to‘r chiziqlari yashil)
ax.plot_surface(x, y, z, cmap='viridis', edgecolor='green')
ax.set_title('Sirt Grafigi (Surface Plot)')

plt.show()
```

**Izoh:** 
Biz `np.outer()` va `.T` (transponirlash) yordamida $X$ va $Y$ uchun to‘r (grid) yaratamiz, so‘ngra funksiya asosida 

$z = \cos(x^2 + y^3)$ qiymatlarini hisoblab chiqamiz. Sirt o‘ziga xos relyefga ega bo‘lishi uchun `ax.plot_surface()` yordamida chiziladi. Bunda balandlikni farqlash uchun `cmap='viridis'` ranglar palitrasi va yuzadagi to‘r chiziqlarini ajratib ko‘rsatish uchun `edgecolor='green'` ishlatilgan.

### 4. Karkas grafigi (Wireframe Plot)

**Karkas (simli to‘r) grafigi** sirt grafigiga juda o‘xshaydi, lekin u sirtni to‘liq bo‘yamasdan, uning faqat qirralarini yoki "skeletini" ko‘rsatadi. U sirtning 3D tuzilishini ranglar bilan chalg‘imasdan tushunish va relyef qanday o‘zgarishini kuzatish uchun juda foydalidir.

```python
import matplotlib.pyplot as plt
import numpy as np

# Z qiymatini hisoblash uchun funksiya
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# X va Y uchun -1 dan 5 gacha bo‘lgan oraliqda nuqtalar yaratish
x = np.linspace(-1, 5, 10)
y = np.linspace(-1, 5, 10)

# Koordinatalar to‘rini (meshgrid) hosil qilish
X, Y = np.meshgrid(x, y)

# Z qiymatlarini hisoblash
Z = f(X, Y)

# Shakl va 3D o‘qlarni yaratish
fig = plt.figure()
ax = plt.axes(projection='3d')

# Karkas grafigini yashil rangda chizish
ax.plot_wireframe(X, Y, Z, color='green')
ax.set_title("Karkas grafigi (Wireframe Plot)")

plt.show()
```

**Izoh:** 
Biz bu yerda $f(x, y) = \sin(\sqrt{x^2 + y^2})$ funksiyasini e'lon qilamiz. So‘ngra `np.meshgrid()` orqali $X$ va $Y$ uchun to‘r yaratib olamiz va bu qiymatlarni funksiyaga berib, $Z$ ni hisoblab chiqamiz. `ax.plot_wireframe()` metodi xuddi ustidan to‘r tashlangandek, 3D sirtni yashil chiziqlardan iborat karkas ko‘rinishida ekranga chiqarib beradi.

### 5. 3D Kontur grafigi (Contour plot in 3D)

Ushbu grafik balandlik yoki chuqurlik darajalarini ta'kidlash uchun 3D sirtini kontur chiziqlari (yoki balandlikka qarab o‘zgaruvchi rang sxemalari) bilan birlashtiradi. Bu ma'lumotlarning shakli va gradient o‘zgarishlarini (qayerda qiyalik tik yoki yotiq ekanligini) 3D fazoda aniqroq tushunishga yordam beradi.


```python
import matplotlib.pyplot as plt
import numpy as np

# Z qiymatini hisoblash uchun funksiya
def fun(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# X va Y uchun -10 dan 10 gacha bo‘lgan oraliqda 40 ta nuqta yaratish
x = np.linspace(-10, 10, 40)
y = np.linspace(-10, 10, 40)

# Koordinatalar to‘rini (meshgrid) hosil qilish
X, Y = np.meshgrid(x, y)

# Z qiymatlarini funksiya orqali hisoblash
Z = fun(X, Y)

# Shakl o‘lchamini kattalashtirib (10x8) yaratish
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection='3d')

# 3D sirt grafigini chizish
# alpha=0.8 - sirtning biroz shaffof bo‘lishini ta'minlaydi
# cmap='cool' - sovuq ranglar palitrasi (balandlikka qarab o‘zgaradi)
ax.plot_surface(X, Y, Z, cmap='cool', alpha=0.8)

# Sarlavha va o‘qlarga nom berish
ax.set_title("3D Kontur grafigi (3D Contour Plot)")
ax.set_xlabel("X o‘qi")
ax.set_ylabel("Y o‘qi")
ax.set_zlabel("Z o‘qi")

plt.show()
```

**Izoh:**
Biz bu yerda $f(x, y) = \sin(\sqrt{x^2 + y^2})$ funksiyasini e'lon qildik va $X$ hamda $Y$ uchun nisbatan kengroq (`-10` dan `10` gacha) zich to‘r yaratib oldik. Sirt `ax.plot_surface()` yordamida chiziladi. Bunda `alpha=0.8` sirtga qisman shaffoflik berib, uning orqa qismlarini ham ko‘rish imkonini yaratadi. Rivojlangan Matplotlib dasturlarida bunga qo‘shimcha ravishda `ax.contour3D()` funksiyasi ham tez-tez ishlatiladi.

### 6. Uchburchakli sirt grafigi (Surface Triangulation Plot)

Ushbu grafik tarqoq yoki to‘rli ma’lumotlardan 3D sirtni qurish uchun uchburchakli to‘rlardan (meshes) foydalanadi. Sirt noto‘g‘ri shaklda bo‘lganda yoki to‘rtburchak bo‘lmagan to‘rlardan foydalanilganda ushbu usul juda qulaydir.


```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.tri import Triangulation

# Z qiymatini hisoblash uchun funksiya
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# X va Y uchun -6 dan 6 gacha bo‘lgan oraliqda 30 ta nuqta yaratish
x = np.linspace(-6, 6, 30)
y = np.linspace(-6, 6, 30)

# Koordinatalar to‘rini hosil qilish
X, Y = np.meshgrid(x, y)

# Z qiymatlarini hisoblash
Z = f(X, Y)

# Matrisalarni bir o‘lchamli massivga o‘tkazib (ravel), 
# triangulyatsiya obyektini yaratish
tri = Triangulation(X.ravel(), Y.ravel())

# Shakl o‘lchamini kattalashtirib yaratish
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Uchburchakli sirt grafigini chizish
ax.plot_trisurf(tri, Z.ravel(), cmap='cool', edgecolor='none', alpha=0.8)

# Sarlavha va o‘qlarga nom berish
ax.set_title("Uchburchakli sirt grafigi (Surface Triangulation Plot)")
ax.set_xlabel("X o‘qi")
ax.set_ylabel("Y o‘qi")
ax.set_zlabel("Z o‘qi")

plt.show()
```

*(Siz tashlagan matnda Output ostida "3D contour plot..." deb yozilgan ekan, bu ehtimol oldingi misoldan qolib ketgan xatolik bo‘lishi mumkin, asrida bu Triangulyatsiya grafigi hisoblanadi)*

**Izoh:**
Funksiyani e’lon qilib, `np.meshgrid()` yordamida $X$ va $Y$ to‘rini yaratganimizdan so‘ng, biz ularni `.ravel()` metodi yordamida bir o‘lchamli ro‘yxatga tekislaymiz (yoyamiz) va shundan so‘ng `Triangulation` obyektini yaratamiz. Sirt ranglar palitrasi (`cmap`) va shaffoflik (`alpha`) qo‘llanilgan holda `ax.plot_trisurf()` funksiyasi orqali chiziladi.

### 7. Myobius tasmasi grafigi (Möbius Strip Plot)

Myobius tasmasi (Möbius strip) — bu buralgan, bir tomonlama sirt bo‘lib, topologiyadagi mashhur tushuncha hisoblanadi. Ushbu grafik uning 3D geometriyasini vizuallashtirib, matematika va san’atning qanday go‘zal tarzda uyg‘unlashishi mumkinligini ko‘rsatadi.

```python
import numpy as np
import matplotlib.pyplot as plt

R = 2
# Aylana va tasma kengligi bo‘ylab oraliqlarni yaratish
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(-1, 1, 100)

# Koordinatalar to‘rini hosil qilish
u, v = np.meshgrid(u, v)

# Parametrik tenglamalar yordamida X, Y va Z ni hisoblash
x = (R + v * np.cos(u / 2)) * np.cos(u)
y = (R + v * np.cos(u / 2)) * np.sin(u)
z = v * np.sin(u / 2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# 3D sirtni chizish (alpha=0.5 shaffoflikni ta'minlaydi)
ax.plot_surface(x, y, z, alpha=0.5)

# O‘qlarga nom berish
ax.set_xlabel("X o‘qi")
ax.set_ylabel("Y o‘qi")
ax.set_zlabel("Z o‘qi")
ax.set_title("Myobius tasmasi (Möbius Strip)")

# O‘qlar chegarasini moslashtirish
ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_zlim([-3, 3])

plt.show()
```

**Izoh:** Biz aylana va tasma kengligini qamrab olish uchun `u` va `v` parametrlarini yaratamiz, ularni to‘rga (mesh) keltiramiz va parametrik tenglamalar yordamida `x`, `y` va `z` qiymatlarini hisoblab chiqamiz. Buralgan tasma shaffoflik va maxsus o‘q chegaralari qo‘llanilgan holda `ax.plot_surface()` yordamida chiziladi.

### 4.2. Sirtli va to‘rsimon grafiklar (Surface & Wireframe Plots)

### 3D Sirt grafigi (Surface Plot)

Sirt grafigi uch o‘lchamli ma’lumotlar to‘plamini ifodalaydi. U alohida ma’lumotlar nuqtalarini ko‘rsatish o‘rniga, ikkita mustaqil 

$X$ va $Y$ o‘zgaruvchilari hamda belgilangan qaram $Z$ o‘zgaruvchisi o‘rtasidagi funksional bog‘liqlikni tavsiflaydi. U kontur grafigining hamrohi hisoblanadi. U karkas (wireframe) grafigiga o‘xshaydi, lekin karkasning har bir yuzi bo‘yalgan ko‘pburchakdan iborat bo‘ladi. Bu vizuallashtirilayotgan sirtning topologiyasini yaratishga yordam beradi.

### 3D Sirt grafigini yaratish

Matplotlib'ning `mpl_toolkits.mplot3d` to‘plamidagi `axes3d` 3D sirt grafiklarini yaratish uchun zarur funksiyalarni taqdim etadi. Sirt grafiklari `ax.plot_surface()` funksiyasi yordamida yaratiladi.

**Sintaksis:**
`ax.plot_surface(X, Y, Z)`

Bu yerda $X$ va $Y$ nuqtalarning 2D massivi, $Z$ esa balandliklarning 2D massividir. `ax.plot_surface()` funksiyasining ba’zi qo‘shimcha atributlari quyida keltirilgan:

| Atribut | Tavsif |
| :--- | :--- |
| `X, Y, Z` | Ma’lumotlar qiymatlarining 2D massivlari |
| `cstride` | Ustun qadami (step size) massivi |
| `rstride` | Qator qadami (step size) massivi |
| `ccount` | Foydalaniladigan ustunlar soni, standart qiymati 50 |
| `rcount` | Foydalaniladigan qatorlar soni, standart qiymati 50 |
| `color` | Sirtning rangi |
| `cmap` | Sirt uchun ranglar palitrasi (colormap) |
| `norm` | Ranglar palitrasi qiymatlarini normallashtirish instansi |
| `vmin` | Xaritaning minimal qiymati |
| `vmax` | Xaritaning maksimal qiymati |
| `facecolors` | Alohida sirtlarning yuz rangi |
| `shade` | Yuz rangiga soya berish (mantiqiy qiymat) |

**Misol:** Keling, yuqoridagi funksiyadan foydalanib 3D sirtni yaratamiz.


```python
# Kutubxonalarni yuklash
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

# Ma’lumotlar to‘plamini yaratish
x = np.outer(np.linspace(-3, 3, 32), np.ones(32))
y = x.copy().T # transponirlash
z = (np.sin(x**2) + np.cos(y**2))

# Shaklni yaratish
fig = plt.figure(figsize=(14, 9))
ax = plt.axes(projection='3d')

# Grafikni chizish
ax.plot_surface(x, y, z)

# Grafikni ko‘rsatish
plt.show()
```

### Gradient sirt grafigi

Gradient sirt grafigi 3D sirt grafigi va 2D kontur grafigining uyg‘unlashuvidan iborat. Ushbu grafikda 3D sirt xuddi 2D kontur grafigi kabi ranglanadi. Sirtdagi baland qismlar past qismlardan farqli ranglarga ega bo‘ladi.

**Sintaksis:**
`surf = ax.plot_surface(X, Y, Z, cmap=, linewidth=0, antialiased=False)`

`cmap=` atributi sirtning rangini belgilaydi. Shuningdek, `fig.colorbar` funksiyasini chaqirish orqali ranglar shkalasini ham qo‘shish mumkin. Quyidagi kod gradient sirt grafigini yaratadi:

### 4.3. Issiqlik xaritalari va kontur grafiklar (Heatmaps & Contour Plots)

### Kontur grafiklari (Contour plots)

Kontur grafiklari (shuningdek, daraja grafiklari deb ham ataladi) ko‘p o‘zgaruvchili tahlil o‘tkazish va 3D grafiklarni 2D maydonda vizuallashtirish uchun vositadir. Agar biz $X$ va $Y$ ni o‘zimiz chizmoqchi bo‘lgan o‘zgaruvchilar deb hisoblasak, natijaviy $Z$ qiymati $X-Y$ tekisligida qatlamlar (kesmalar) sifatida chiziladi, shu sababli konturlar ba’zan Z-kesmalar yoki izo-reaksiya deb ham ataladi. Kontur grafiklari zichlikni, balandliklarni yoki tog‘ relyefini vizuallashtirishda, shuningdek meteorologiya sohasida keng qo‘llaniladi. Bunday keng qo‘llanilishi sababli `matplotlib.pyplot` kontur grafiklarini chizishni osonlashtirish uchun `contour` metodini taqdim etadi.

### `matplotlib.pyplot.contour`

`matplotlib.pyplot.contour()` funksiyasi odatda $Z = f(X, Y)$ bo‘lganda, ya’ni $Z$ kiritilgan $X$ va $Y$ ning funksiyasi sifatida o‘zgarganda foydalidir. Shuningdek, bo‘yalgan konturlarni chizish imkonini beruvchi `contourf()` funksiyasi ham mavjud.

**Sintaksis:**
`matplotlib.pyplot.contour([X, Y, ] Z, [levels], **kwargs)`

**Parametrlar:**

*   **X, Y**: $Z$ bilan bir xil shakldagi 2D numpy massivlari yoki `len(X) == M` va `len(Y) == N` bo‘lgan 1D massivlar (bu yerda $M$ va $N$ mos ravishda $Z$ massivining qatorlari va ustunlaridir).
*   **Z**: Kontur chiziladigan balandlik qiymatlari. Shakli $(M, N)$ bo‘lishi kerak.
*   **levels**: Kontur chiziqlari yoki hududlarining soni va joylashuvini belgilaydi.

**Qaytaradi:** `QuadContourSet`

Quyidagi misol `matplotlib.pyplot` dagi `contour()` funksiyasining ishlashini ko‘rsatadi:

**Misol 1:** Faqat kontur chiziqlarini ko‘rsatadigan `contour()` funksiyasidan foydalanib kontur grafigini chizish.

```python
# matplotlib funksiyalarini yuklash
import matplotlib.pyplot as plt
import numpy as np

feature_x = np.arange(0, 50, 2)
feature_y = np.arange(0, 50, 3)

# O‘zgaruvchilarning 2D to‘rini (grid) yaratish
[X, Y] = np.meshgrid(feature_x, feature_y)

fig, ax = plt.subplots(1, 1)

Z = np.cos(X / 2) + np.sin(Y / 4)

# Kontur chiziqlarini chizish
ax.contour(X, Y, Z)

ax.set_title('Kontur grafigi')
ax.set_xlabel('feature_x')
ax.set_ylabel('feature_y')

plt.show()
```

**Misol 2:** Bo‘yalgan konturlarni chizuvchi `contourf()` funksiyasidan foydalanib kontur grafigini chizish.


```python
# Matplotlib funksiyalarini yuklash
import matplotlib.pyplot as plt
import numpy as np

feature_x = np.linspace(-5.0, 3.0, 70)
feature_y = np.linspace(-5.0, 3.0, 70)

# O‘zgaruvchilarning 2D to‘rini yaratish
[X, Y] = np.meshgrid(feature_x, feature_y)

fig, ax = plt.subplots(1, 1)

Z = X ** 2 + Y ** 2

# Bo‘yalgan kontur grafigini chizish
ax.contourf(X, Y, Z)

ax.set_title("Bo‘yalgan kontur grafigi")
ax.set_xlabel('feature_x')
ax.set_ylabel('feature_y')

plt.show()
```

Bu yondashuv turli balandlik yoki zichlik darajalarini vizual tarzda tez farqlash imkonini beradi.

### 4.4. Grafiklarni animatsiya qilish (Animating Plots)

Python'ning Matplotlib kutubxonasi funksiyalar yoki shakllarning grafiklarini chizish vositasi bo‘lib, undan animatsiya vositasi sifatida ham foydalanish mumkin. Chizilgan grafiklarga animatsiyalar qo‘shilganda vizuallashtirish yanada kuchayadi. Matplotlib'da animatsiyalar `Animation` sinfi yordamida ikki xil usulda yaratilishi mumkin:

1.  **Funksiyani qayta-qayta chaqirish orqali:** Takroran ishga tushirilganda animatsiya hosil qiluvchi maxsus funksiyadan foydalanadi (eng keng tarqalgan usul).
2.  **O‘zgarmas obyektlardan foydalanish orqali:** Ba’zi animatsiyalangan grafik obyektlar boshqalari bilan birlashib animatsiya sahnasini beradi.

**Muhim eslatma:** Biz har doim animatsiya obyektiga bo‘lgan havolani (reference) saqlab qolishimiz kerak, aks holda animatsiya to‘xtab qoladi. Buning sababi shundaki, `Animation` sinfi obyektga faqat bitta havola saqlaydi va agar o‘zgaruvchi biror joyga biriktirilmasa, u xotiradan avtomat tarzda o‘chiriladi (garbage collection).

**Talab qilinadigan o‘rnatmalar:** `numpy`, `matplotlib` va `ffmpeg` (animatsiyani video sifatida saqlash uchun).

**Misol:** Uzluksiz harakatlanuvchi sinus to‘lqinini yaratish.


```python
!pip install ffmpeg
```

```python
from matplotlib import pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation

# Grafik chiziladigan shaklni (figure) ishga tushirish
fig = plt.figure()

# X va Y o‘qlarini chegaralarini belgilash
axis = plt.axes(xlim=(0, 4), ylim=(-2, 2))

# Chiziq o‘zgaruvchisini ishga tushirish (bo‘sh ma'lumot bilan)
line, = axis.plot([], [], lw=3)

# Dastlabki fonni (bo‘sh chiziqni) tayyorlovchi funksiya
def init():
    line.set_data([], [])
    return line,

# Animatsiya funksiyasi: har bir freym (i) uchun chaqiriladi
def animate(i):
    x = np.linspace(0, 4, 1000)
    
    # Sinus grafigini i qiymatiga qarab siljitish
    y = np.sin(2 * np.pi * (x - 0.01 * i))
    
    # Yangi x va y ni chiziqqa o‘rnatish
    line.set_data(x, y)
    return line,

# Animatsiya obyektini yaratish
anim = FuncAnimation(fig, animate, init_func=init,
                     frames=200, interval=20, blit=True)

# Animatsiyani kompyuterga saqlash (ffmpeg talab qilinadi)
anim.save('continuousSineWave.gif', writer='pillow', fps=30)

# Agar shunchaki ekranda ko‘rmoqchi bo‘lsangiz:
# plt.show()
```

**Izoh:** Bu yerda `init()` funksiyasi grafikning boshlang‘ich (bo‘sh) holatini tayyorlaydi. `animate(i)` funksiyasi esa har bir kadr uchun takroran chaqiriladi. Undagi `x - 0.01 * i` qismi to‘lqinning gorizontal bo‘ylab doimiy ravishda siljishini, ya’ni harakatlanish effektini ta’minlaydi. `FuncAnimation` esa bu jarayonni ko‘rsatilgan oraliq (interval) bo‘yicha avtomatik tarzda yangilab turadi.

Dastlab, kerakli kutubxonalarni yuklagandan so‘ng, butun animatsiya chiziladigan bo‘sh shakl yoki oynani o‘rnatamiz. So‘ng grafikning $X$ va $Y$ koordinatalarini o‘z ichiga oluvchi `line` o‘zgaruvchisini e’lon qilamiz. Ular dastlab bo‘sh qoldiriladi, chunki undagi ma’lumotlar animatsiya sababli doimiy ravishda o‘zgarib turadi. Nihoyat, $i$ argumentini qabul qiluvchi `animate(i)` animatsiya funksiyasini yozamiz. Bu yerda $i$ kadr (frame) raqami deb ataladi va uning yordamida $i$ ning qiymatiga qarab doimiy ravishda o‘zgarib turuvchi shaklni yaratamiz. Oxirgi qatorda 500 ta kadr va 20 millisekund oraliqqa ega animatsiyani yaratish uchun `FuncAnimation` funksiyasidan foydalanamiz.


```python
import matplotlib.animation as animation
import matplotlib.pyplot as plt
import numpy as np

# Animatsiya uchun bo‘sh oyna yaratish
fig = plt.figure()
axis = plt.axes(xlim=(-50, 50), ylim=(-50, 50))
line, = axis.plot([], [], lw=2)

def init():
    line.set_data([], [])
    return line,

# X va Y koordinatalari uchun bo‘sh ro‘yxatlarni e’lon qilish
xdata, ydata = [], []

# Animatsiya funksiyasi
def animate(i):
    # t - kadr raqamiga qarab o‘zgaruvchi parametr
    t = 0.1 * i
    
    # Chiziladigan X va Y qiymatlari
    x = t * np.sin(t)
    y = t * np.cos(t)
    
    # Oldin bo‘sh bo‘lgan X va Y ro‘yxatlariga qiymatlarni qo‘shish
    xdata.append(x)
    ydata.append(y)
    line.set_data(xdata, ydata)
    
    return line,

# Animatsiya funksiyasini chaqirish
anim = animation.FuncAnimation(fig, animate, init_func=init,
                               frames=500, interval=20, blit=True)

# Animatsiyani kompyuterga saqlash
anim.save('growingCoil.gif', writer='pillow', fps=30)
```

**Izoh:** Ushbu misolda $x = t \sin(t)$ va $y = t \cos(t)$ parametrik tenglamalaridan foydalanib o‘suvchi spiral (Arximed spirali) chiziladi. Oldingi misoldan farqli o‘laroq, bu yerda hisoblangan nuqtalar avvalgilarining o‘rnini egallamaydi, balki `xdata` va `ydata` ro‘yxatlariga qo‘shilib boriladi (`append()`). Bu esa chiziqning doimiy ravishda o‘sib borish effektini yaratadi.

Ushbu grafik uzluksiz o‘sib boruvchi spiral animatsiyasi hisoblanadi. Kutubxonalarni yuklashda `import matplotlib.animation as animation` yoki `from matplotlib.animation import FuncAnimation` usullaridan ixtiyoriy birini tanlash mumkin, ikkisi ham bir xil natija beradi.

`animate` funksiyasi kadr raqami $i$ ga qarab o‘zgaradi. Spiral – sinus va kosinusning kompozit funksiyasi bo‘lib, $X$ o‘qida sinus va $Y$ o‘qida kosinus funksiyalarini birlashtirish natijasida hosil qilinadi.
