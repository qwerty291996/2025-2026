### Matplotlib'da ma’lumotlarni Vizuallashtirish Turlari

### 1. Ustunli Diagramma (Bar Plot)

**Ustunli diagramma** – bu turli xil toifalar (kategoriyalar) o‘rtasidagi qiymatlarni solishtirish uchun eng zo‘r vosita. U ma’lumotlarning miqdorini to‘g‘ri to‘rtburchak shaklidagi ustunlar balandligi orqali ifodalab beradi.

**Misol:**

````python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Namunaviy ma’lumotlar lug‘atini (dictionary) yaratamiz
data = {
    'Category': ['A', 'B', 'C', 'D'], 
    'Values': [20, 34, 30, 35]
}

# 2. Uni Pandas DataFrame jadvaliga o‘giramiz
df = pd.DataFrame(data)

# 3. Ustunli diagrammani (Bar Chart) chizamiz
# x o‘qiga kategoriyalarni, y o‘qiga qiymatlarni beramiz va osmonrang ('skyblue') ranga bo‘yaymiz
plt.bar(df['Category'], df['Values'], color='skyblue')

# 4. Grafik o‘qlariga tushunarli nomlar beramiz
plt.xlabel("Kategoriya (Category)")
plt.ylabel("Qiymatlar (Values)")

# 5. Grafikning asosiy sarlavhasi
plt.title("Ustunli Diagramma (Bar Chart) Misoli")

# 6. Natijani ekranga chiqarish
plt.show()
````

**Izoh:**
*   **`plt.bar(x, y)`** funksiyasi Matplotlib'ga aynan ustunli grafik chizish kerakligini aytadi. Biz unga `df` (DataFrame) ichidagi tegishli ustunlarni uzatamiz.
*   **`plt.xlabel()` va `plt.ylabel()`** funksiyalari grafikingizni o‘qiyotgan odamga qaysi tomonda nima joylashganini tushuntirib beruvchi yorliqlar (matnlar) yopishtiradi.
*   **`plt.show()`** har qanday Matplotlib chizmasining eng oxirida kelishi shart. Bu "Barcha tayyorgarliklar tugadi, oynani ochib natijani ko‘rsat!" degan buyruqdir.

### 2. Gistogramma (Histogram)

**Gistogramma** — raqamli ma’lumotlarni ularning uchrash chastotasini (qanchalik tez-tez takrorlanishini) ko‘rsatish uchun ma’lum bir oraliqlarga (kataklarga yoki inglizcha *bins* ga) guruhlaydi. U ma’lumotlarning qanday taqsimlanganligini (tarqalishini) tushunish va tendensiyalarni aniqlashga yordam beradi hamda asosan uzluksiz raqamli ma’lumotlar uchun keng qo‘llaniladi.

**Misol:**

````python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Namunaviy raqamli ma’lumotlardan iborat DataFrame yaratamiz
df = pd.DataFrame({'Values': [12, 23, 45, 36, 56, 78, 89, 95, 110, 130]})

# 2. Gistogrammani chizamiz
# bins=5: ma’lumotlarni 5 ta teng oraliqqa bo‘ladi
# color: och qizil rang, edgecolor: ustunlar ajralib turishi uchun qora hoshiya
plt.hist(df['Values'], bins=5, color='lightcoral', edgecolor='black')

# 3. Grafik o‘qlariga tushunarli nomlar beramiz
plt.xlabel("Qiymatlar oralig‘i (Value Range)")
plt.ylabel("Uchrash chastotasi (Frequency)")

# 4. Grafikning asosiy sarlavhasi
plt.title("Gistogramma (Histogram) Misoli")

# 5. Natijani ekranga chiqarish
plt.show()
````

**Kodning ishlash mantig‘i (Dasturchi uchun izoh):**
*   **`plt.hist(x, bins)`** funksiyasi Matplotlib'ga gistogramma chizish kerakligini bildiradi. 
*   **`bins=5` parametridagi sir:** Bu gistogrammaning eng muhim qismi! U bizdagi raqamlarning eng kichigi (12) va eng kattasi (130) o‘rtasidagi masofani teng 5 ta ustunga bo‘lib chiqadi. Keyin esa har bir oraliqqa bizning ro‘yxatdan nechta son tushishini sanab, o‘sha ustunning balandligini belgilaydi.
*   **`edgecolor='black'`**: Ustunli diagramma (`bar chart`) dan farqli o‘laroq, gistogrammada ustunlar bir-biriga yopishgan holda chiqadi. Shuning uchun ularning chetiga qora chiziq tortish chegaralarni aniq ko‘rishga yordam beradi.

### 3. Doiraviy diagramma (Pie Chart)

**Doiraviy diagramma** toifali ma’lumotlarni butun bir qismning ulushlari sifatida ko‘rsatish uchun ishlatiladi. Undagi har bir bo‘lak ma’lumotlar to‘plamining foiz ko‘rsatkichini ifodalaydi.

**Misol:**

````python
import pandas as pd
import matplotlib.pyplot as plt

# 1. ma’lumotlarni yaratamiz
df = pd.DataFrame({'Category': ['A', 'B', 'C'], 'Values': [30, 50, 20]})

# 2. Doiraviy diagrammani chizamiz
plt.pie(df['Values'], 
        labels=df['Category'], 
        autopct='%1.1f%%', 
        colors=['gold', 'lightblue', 'pink'])

# 3. Sarlavha qo‘shish
plt.title("Doiraviy diagramma (Pie Chart) misoli")

# 4. Natijani ko‘rsatish
plt.show()
````

**Izoh:**
*   **`plt.pie(df['Values'])`**: Jadvalning qiymatlar ustuni asosida diagramma bo‘laklarini shakllantiradi.
*   **`labels=df['Category']`**: Har bir bo‘lakka mos keluvchi toifa nomlarini (A, B, C) yopishtiradi.
*   **`autopct='%1.1f%%'`**: Bo‘laklarning butunga nisbatan foiz qiymatini avtomatik hisoblab, diagramma ustida ko‘rsatadi.
*   **`colors`**: Har bir bo‘lak uchun maxsus ranglar (oltin rang, och ko‘k va pushti) berish imkonini beradi.

### 4. Nuqtali grafik (Scatter Plot)

**Scatter plot** ikkita raqamli o‘zgaruvchi o‘rtasidagi bog‘liqlikni vizuallashtiradi, bu bizga korrelyatsiyalarni va ma’lumotlar modellarini aniqlashga yordam beradi.

**Misol:**

````python
import pandas as pd
import matplotlib.pyplot as plt

# ma’lumotlarni DataFrame ko‘rinishida yaratamiz
df = pd.DataFrame({'X': [1, 2, 3, 4, 5], 'Y': [2, 4, 1, 8, 7]})

# Nuqtali grafikni chizamiz
plt.scatter(df['X'], df['Y'], color='green', marker='o‘)

# Grafik elementlarini sozlaymiz
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot Example")

# Natijani ko‘rsatish
plt.show()
````

**Izoh:**
*   **`plt.scatter()`**: Berilgan $X$ va $Y$ qiymatlari asosida nuqtali grafikni hosil qiladi.
*   **`color='green'`**: Nuqtalar rangini yashil qilib belgilaydi.
*   **`marker='o‘`**: Nuqtalar shaklini doira ko‘rinishida tasvirlaydi.

### 5. Chiziqli grafik (Line Chart)

**Chiziqli grafik** ma’lumotlar nuqtalarini uzluksiz chiziq bilan bog‘lash orqali vaqtli ketma-ketlikdagi yoki ketma-ket ma’lumotlardagi tendensiyalar va qonuniyatlarni tahlil qilish uchun juda mos keladi.

**Misol:**

````python
import pandas as pd
import matplotlib.pyplot as plt

# ma’lumotlarni DataFrame ko‘rinishida yaratamiz
df = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'], 
    'Sales': [200, 250, 300, 280, 350]
})

# Chiziqli grafikni chizamiz
plt.plot(df['Month'], df['Sales'], marker='o‘, linestyle='-', color='blue')

# Grafik elementlarini sozlaymiz
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Line Chart Example")

# To‘r (grid) chiziqlarini qo‘shish
plt.grid()

# Natijani ko‘rsatish
plt.show()
````

**Izoh:**
*   **`plt.plot()`**: ma’lumotlar nuqtalarini chiziq bilan tutashtirib chizadi.

### 7.2. Vaqtli qatorlar ma’lumotlarini vizuallashtirish (Visualizing Time Series Data)

Vaqtli ketma-ketlik ma’lumotlari (Time series data) — bu vaqt o‘tishi bilan ketma-ket yig‘ilgan ma’lumotlar to‘plamidir. Ular har kungi aksiyalar narxi yoki har soatlik harorat kabi o‘zgarishlarni ko‘rsatib beradi. Bunday ma’lumotlarni tahlil qilish va vizuallashtirishdan asosiy maqsad — ma'noli qonuniyatlarni ajratib olish va kelajak uchun bashoratlar qilishdir.

### Vaqtli ketma-ketlik tahlilidagi asosiy tushunchalar

Ushbu sohadagi tahlillar bir nechta muhim tushunchalarga tayanadi:
*   **Trend (Tendensiya)**: ma’lumotlarning uzoq muddatli yo‘nalishi (o‘suvchi, kamayuvchi yoki barqaror).
*   **Mavsumiylik (Seasonality)**: Muntazam vaqt oralig‘ida takrorlanib turuvchi qonuniyatlar.
*   **Sirpanuvchi o‘rtacha (Moving average)**: Trendni yaqqolroq ko‘rsatish uchun qisqa muddatli tebranishlarni silliqlashtirish usuli.
*   **Shovqin (Noise)**: Aniq qonuniyatga ega bo‘lmagan tasodifiy o‘zgarishlar.
*   **Statsionarlik (Stationarity)**: Statistik xususiyatlari (o‘rtacha qiymat, dispersiya) vaqt o‘tishi bilan o‘zgarmas qoladigan ketma-ketlik.
*   **Avtokorrelyatsiya**: Vaqtli ketma-ketlikning o‘zi bilan ma’lum bir vaqt kechikishidagi (lagged version) o‘xshashlik darajasi.
*   **Resampling**: ma’lumotlar kuzatuv chastotasini (masalan, kunlikdan oylikka) o‘zgartirish usuli.

### ma’lumot turlari va vizuallashtirish

Vaqtli ketma-ketlik ma’lumotlari ikki qismga bo‘linadi:
1.  **Uzluksiz (Continuous)**: Doimiy diapazondagi qiymatlarni (harorat, sensor ma’lumotlari) o‘z ichiga oladi. Ularni vizuallashtirishda trendlarni ta'kidlash uchun **chiziqli (line plot)** yoki **sohali (area chart)** grafiklardan foydalaniladi.
2.  **Diskret (Discrete)**: Aniq vaqt nuqtalarida qayd etilgan alohida qiymatlar yoki toifalar (hodisalar soni). Ularning chastotasi yoki taqsimotini ko‘rsatish uchun **ustunli (bar chart)** yoki **gistogrammalardan** foydalanish tavsiya etiladi.

Ushbu tahlillar moliya, farmatsevtika va tadqiqot kabi sohalarda qaror qabul qilish jarayonlarini qo‘llab-quvvatlash uchun xizmat qiladi.

Vaqtli ketma-ketliklarni vizuallashtirish bo‘yicha amaliy mashg‘ulotni boshlaymiz. Ushbu jarayon moliya, farmatsevtika va ijtimoiy tarmoqlar kabi sohalarda trendlar hamda mavsumiy qonuniyatlarni topish uchun juda muhimdir.

Vaqtli ketma-ketlik tahlilining yakuniy va eng muhim bosqichlaridan biri bu ma’lumotlardagi ichki qonuniyatlarni aniqlashdir.

### 6-bosqich: Avtokorrelyatsiya orqali mavsumiylikni aniqlash

Mavsumiylikni (Seasonality) aniqlash uchun **Avtokorrelyatsiya funksiyasi (ACF)** grafigidan foydalaniladi. ACF grafigidagi muntazam vaqt oraliqlarida paydo bo‘ladigan cho‘qqilar (peaks) ma’lumotlarda takrorlanuvchi mavsumiy qonuniyatlar borligidan dalolat beradi.

**Misol:**

````python
# Agar 'Date' ustuni indeks bo‘lmasa, uni indeksga o‘tkazishni tekshiramiz
if 'Date' not in df.columns:
    print("'Date' allaqachon indeks qilingan yoki jadvalda mavjud emas.")
else:
    df.set_index('Date', inplace=True)

# ACF grafigini chizish
plt.figure(figsize=(12, 6))
plot_acf(df['High'], lags=40)
plt.xlabel('Lag (Kechikish)')
plt.ylabel('Autocorrelation (Avtokorrelyatsiya)')
plt.title('Avtokorrelyatsiya funksiyasi (ACF) grafigi')
plt.show()
````

**Izoh:**
*   **Avtokorrelyatsiya**: Bu statistik usul bo‘lib, vaqtli ketma-ketlikning o‘zi bilan ma’lum bir vaqt kechikishidagi (lag) holati o‘rtasidagi o‘xshashlik darajasini o‘lchaydi.
*   **`lags=40`**: Grafikda ketma-ketlikning o‘zidan oldingi 40 ta vaqt birligi (masalan, 40 kun) bilan solishtirilishi ko‘rsatiladi.
*   **Mavsumiylikni o‘qish**: Agar grafikdagi chiziqlar ma’lum bir tartibda (masalan, har 7 yoki 30 kunda) yuqoriga ko‘tarilsa, demak ma’lumotlarda haftalik yoki oylik takrorlanuvchi holatlar mavjud.

Vaqtli ketma-ketlik tahlilining yakuniy bosqichi bu ma’lumotlarning statsionarligini rasmiy ravishda tekshirishdir. Statsionarlik – bu ketma-ketlikning statistik xususiyatlari (o‘rtacha qiymat va dispersiya kabi) vaqt o‘tishi bilan o‘zgarmas qolishini anglatadi.

### 7-bosqich: ADF testi yordamida statsionarlikni tekshirish

Statsionarlikni aniqlash uchun biz **Augmented Dickey-Fuller (ADF)** testidan foydalanamiz.

**Misol:**

````python
from statsmodels.tsa.stattools import adfuller

# 'High' ustuni uchun ADF testini o‘tkazish
result = adfuller(df['High'])

print('ADF Statistic:', result[0])
print('p-value:', result[1])
print('Critical Values:', result[4])
````

**Izoh:**
*   **ADF Statistic**: Agar bu qiymat kritik qiymatlardan (Critical Values) kichik bo‘lsa, ma’lumotlar statsionar hisoblanadi.
*   **p-value**: Odatda, agar p-qiymat **0.05** dan kichik bo‘lsa, statsionarlik haqidagi gipoteza qabul qilinadi.
*   **Natija tahlili**: Yuqoridagi ADF statistikasi natijasiga ko‘ra, nol gipoteza qabul qilinadi, ya'ni ma’lumotlar **statsionar emas**.
*   **Keyingi qadam**: ma’lumotlar statsionar bo‘lmagani uchun, bashorat qilish modellarini qo‘llashdan avval **differensiyalash (differencing)** yoki boshqa transformatsiya usullarini qo‘llash talab etiladi.

Vaqtli ketma-ketlik statsionar bo‘lmagan holatda, undagi trend yoki mavsumiylikni yo‘qotish uchun differensiyalash usuli qo‘llaniladi.

### 8-bosqich: Statsionarlikka erishish uchun differensiyalash (Differencing)

Differensiyalash — bu joriy kuzatuv qiymatidan undan oldingi kuzatuv qiymatini ayirish jarayonidir. Bu usul ma’lumotlar to‘plamidagi o‘zgaruvchanlikni barqarorlashtirishga xizmat qiladi.

**Misol:**

````python
# Differensiyalangan qiymatlarni hisoblash va yangi ustunga saqlash
df['high_diff'] = df['High'].diff()

# Grafik o‘lchamini belgilash
plt.figure(figsize=(12, 6))

# Asl ma’lumotlarni chizish
plt.plot(df['High'], label='Original High', color='blue')

# Differensiyalangan ma’lumotlarni chizish (punktir chiziq bilan)
plt.plot(df['high_diff'], label='Differenced High', linestyle='--', color='green')

# Grafik elementlarini sozlash
plt.legend()
plt.title('Original vs Differenced High')
plt.show()
````

**Izoh:**
*   **`df['High'].diff()`**: Har bir qatordagi qiymatdan o‘zidan oldingisini ayirib chiqadi. Birinchi qator uchun oldingi qiymat bo‘lmagani sababli, u `NaN` (bo‘sh qiymat) bo‘lib qoladi.
*   **Vizuallashtirish**: Grafikda asl ko‘k chiziq trendga ega bo‘lsa, yashil punktir chiziq (differensiyalangan ma’lumot) endi nol atrofida tebranuvchi statsionar holatga yaqinlashganini ko‘rish mumkin.
*   **Maqsad**: Ushbu jarayondan so‘ng ma’lumotlar statistik bashorat qilish modellarini (masalan, ARIMA) qo‘llash uchun tayyor holatga keladi.

Vaqtli ketma-ketlik ma’lumotlaridagi qisqa muddatli tebranishlarni kamaytirish va asosiy trendni yaqqolroq ko‘rsatish uchun **sirpanuvchi o‘rtacha (moving average)** usuli qo‘llaniladi.

### 9-bosqich: Sirpanuvchi o‘rtacha yordamida ma’lumotlarni silliqlash

Ushbu usul ma’lum bir vaqt oralig‘idagi (oyna o‘lchami) qiymatlar o‘rtachasini hisoblab, "shovqinlarni" (noise) silliqlashtiradi.

**Misol:**

````python
# Oyna o‘lchamini belgilash (masalan, 120 kunlik)
window_size = 120
# Rolling funksiyasi yordamida sirpanuvchi o‘rtacha hisoblash
df['high_smoothed'] = df['High'].rolling(window=window_size).mean()

plt.figure(figsize=(12, 6))

# Asl ma’lumotlar grafigi
plt.plot(df['High'], label='Original High', color='blue')

# Silliqlashtirilgan (sirpanuvchi o‘rtacha) grafik
plt.plot(df['high_smoothed'], label=f'Moving Average (Window={window_size})', 
         linestyle='--', color='orange')

plt.xlabel('Date')
plt.ylabel('High')
plt.title('Original vs Moving Average')
plt.legend()
plt.show()
````

**Izoh:**
*   **`rolling(window=120)`**: ma’lumotlarni 120 ta qatorli (kunli) bloklarga ajratadi.
*   **`.mean()`**: Har bir 120 kunlik blok uchun o‘rtacha qiymatni hisoblab, ma’lumotlardagi keskin sakrashlarni tekislaydi.
*   **Vizuallashtirish**: To‘q sariq rangli punktir chiziq asl ko‘k chiziqqa nisbatan ancha silliq bo‘lib, uzoq muddatli yo‘nalishni (trendni) ko‘rsatib beradi.
*   **Differensiyalash bilan farqi**: `df['High'].diff()` ketma-ket kuzatuvlar orasidagi farqni hisoblab, o‘zgarishlarni ko‘rsatsa, sirpanuvchi o‘rtacha umumiy harakatni silliqlashga xizmat qiladi.

Vaqtli ketma-ketlik tahlili jarayonida sirpanuvchi o‘rtacha hisoblash ma’lumotlardagi "shovqinlarni" silliqlab, trendni yaqqol ko‘rsatishga xizmat qiladi. Ushbu jarayonning yakunida asl ma’lumotlar va ular ustida bajarilgan matematik amallar (differensiyalash) natijasini solishtirish muhim bosqich hisoblanadi.

### 10-bosqich: Asl va differensiyalangan ma’lumotlarni solishtirish

Ushbu bosqichda biz asl `High` ustuni va differensiyalash natijasida hosil bo‘lgan `high_diff` ustunini yonma-yon jadval ko‘rinishida chiqaramiz. Bu bizga har bir qadamda ma’lumotlar qanday o‘zgarganini raqamlarda ko‘rish imkonini beradi.

**Misol:**

````python
# Asl va differensiyalangan ustunlarni bitta jadvalga birlashtirish
df_combined = pd.concat([df['High'], df['high_diff']], axis=1)

# Jadvalning dastlabki 5 qatorini chop etish
print(df_combined.head())
````

**Izoh:**
*   **`pd.concat([...], axis=1)`**: Ikkita alohida ustunni (Series) yonma-yon qilib, yangi DataFrame hosil qiladi.
*   **Birinchi qator (NaN)**: Differensiyalangan ustunning birinchi qatorida `NaN` qiymati bo‘ladi, chunki birinchi kungi narxdan ayirish uchun undan oldingi kun ma’lumoti mavjud emas.
*   **Silliqlash effekti**: 120 kunlik (bir choraklik) oyna o‘lchami bilan hisoblangan `high_smoothed` seriyasi ma’lumotlarning choraklik o‘rtacha trendini ifodalaydi.
*   **Tahlil**: Ushbu jadval yordamida asl narxlar va kunlik o‘zgarishlar (differensiyalar) o‘rtasidagi bog‘liqlikni raqamli tahlil qilish mumkin.

Differensiyalash natijasida hosil bo‘lgan `high_diff` ustunidagi birinchi qiymat `NaN` bo‘ladi, chunki undan oldingi qiymat mavjud emas. Statistik testlarni (masalan, ADF testi) to‘g‘ri o‘tkazish uchun bu bo‘sh qiymatni o‘chirib tashlash zarur.

### Differensiyalangan ma’lumotlar bilan ADF testini qayta o‘tkazish

Bo‘sh qiymatlarni o‘chirganimizdan so‘ng, ma’lumotlar statsionar holatga kelganligini tekshirish uchun ADF testini qaytadan bajaramiz.

**Misol:**

````python
from statsmodels.tsa.stattools import adfuller

# NaN qiymatlarni o‘chirib tashlash
df.dropna(subset=['high_diff'], inplace=True)

# high_diff ustuni uchun ADF testini o‘tkazish
result = adfuller(df['high_diff'])

print('ADF Statistic:', result[0])
print('p-value:', result[1])
print('Critical Values:', result[4])
````

**Izoh:**
*   **`df.dropna()`**: `high_diff` ustunidagi `NaN` qiymatini olib tashlaydi, bu esa testning xatosiz ishlashini ta'minlaydi.
*   **Statsionarlikning o‘zgarishi**: Asl narxlar (`High`) statsionar bo‘lmasligi mumkin, lekin ularning birinchi tartibli farqi (`high_diff`) ko‘pincha statsionar natija beradi.
*   **Natijani baholash**: Agar yangi `p-value` **0.05** dan kichik bo‘lsa, demak differensiyalash maqsadi amalga oshgan va ma’lumotlar statsionar holatga keltirilgan bo‘ladi.

ADF testi natijalariga ko‘ra, biz nol gipotezani rad etamiz va buni tasdiqlash uchun yetarli dalillarga egamiz.

**Izoh:**
*   **Nol gipotezani rad etish**: Bu differensiyalangan ma’lumotlar (`high_diff`) endi statsionar holatga kelganini anglatadi.
*   **Tahlil**: Birinchi marta `High` ustuni tekshirilganda, ma’lumotlar statsionar emas edi. Differensiyalash amali (joriy qiymatdan oldingisini ayirish) orqali trendlar yo‘qotildi va ma’lumotlar statistik modellarni qo‘llash uchun tayyor holatga keltirildi.

### 7.4. Juftlik grafiklarini yaratish (Creating Pair Plots)

**Pair Plot (Juftliklar grafigi)** – bu ma'lumotlar to'plamidagi turli xil raqamli qiymatlar bir-biriga qanday bog'liqligini ko'rsatuvchi grafik turidir. Seaborn ushbu grafikni tezda yaratish uchun tayyor `pairplot()` funksiyasiga ega bo'lsa-da, **Matplotlib** grafikning ko'rinishi va ishlashini moslashtirish (customization) uchun ko'proq boshqaruv imkonini beradi. 

**Pair Plot** (shuningdek, tarqoqlik matritsasi yoki *scatterplot matrix* deb ham ataladi) quyidagilardan iborat:
*   Har bir raqamli o'zgaruvchilar juftligi uchun **Nuqtali grafiklar** (Scatter plots).
*   Diagonalda joylashgan va alohida o'zgaruvchilarning taqsimlanishini ifodalovchi **Gistogrammalar** (yoki yadro zichligi grafiklari - KDE).

Ushbu vizuallashtirish quyidagilarni aniqlashga yordam beradi:
*   Xususiyatlar (features) o'rtasidagi chiziqli va nochiziqli munosabatlarni.
*   Ma'lumotlar ichidagi klasterlar yoki guruhlarni.
*   Ehtimoliy begona qiymatlarni (outliers).

### Matplotlib yordamida Pair Plot yaratish
Ishni boshlash uchun avvalo kerakli kutubxonalarni chaqirib (import qilib) olishimiz kerak.

````python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
````

*   **`matplotlib.pyplot`**: Vizuallashtirishlarni yaratish uchun ishlatiladi.
*   **`pandas`**: Tuzilgan ma'lumotlar (dataframelar) bilan ishlashga yordam beradi.
*   **`numpy`**: Raqamli ma'lumotlarni yaratish (generatsiya qilish) uchun foydali.

**Misol:**

````python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(42)
data = pd.DataFrame({
    '1-xususiyat': np.random.rand(50),
    '2-xususiyat': np.random.rand(50),
    '3-xususiyat': np.random.rand(50),
    '4-xususiyat': np.random.rand(50)
})

# Xususiyatlar soni
num_features = len(data.columns)

# Kichik grafiklar (Subplots) to'rini yaratish
fig, axes = plt.subplots(num_features, num_features, figsize=(10, 10))

# Har bir xususiyatlar juftligi bo'ylab tsikl
for i in range(num_features):
    for j in range(num_features):
        ax = axes[i, j]
        
        if i == j:
            # Diagonal: Xususiyatning gistogrammasi
            ax.hist(data.iloc[:, i], bins=15, color='skyblue', edgecolor='black')
        else:
            # Xususiyatlar juftligi uchun nuqtali grafik
            ax.scatter(data.iloc[:, j], data.iloc[:, i], alpha=0.7, s=10, color="blue")

        # Chap va pastki o'qlarga yorliqlar (labels) o'rnatish
        if j == 0:
            ax.set_ylabel(data.columns[i], fontsize=10)
        if i == num_features - 1:
            ax.set_xlabel(data.columns[j], fontsize=10)

        # Tozaroq ko'rinish uchun o'q raqamlarini (ticks) olib tashlash
        ax.set_xticks([])
        ax.set_yticks([])

# Joylashuvni to'g'rilash
plt.tight_layout()
plt.show()
````

**Izoh:**
*   **Ma’lumotlarni generatsiya qilish:** 4 ta xususiyat × 50 ta qiymat (0 dan 1 gacha) Pandas `DataFrame`ida saqlanadi (`np.random.seed(42)`).
*   **Kichik grafiklar to‘ri (Subplots Grid):** 4×4 o‘lchamli joylashuv (`plt.subplots()`), diagonalda gistogrammalar (`i == j`) va qolgan joylarda nuqtali grafiklar (scatter plots) (`i ≠ j`).
*   **Gistogrammalar:** Aniqlik uchun 15 ta ustunli (bins), ochiq ko‘k (skyblue) rangga bo‘yalgan va qora chekkalarga ega `ax.hist()`.
*   **Nuqtali grafiklar (Scatter Plots):** O‘zaro bog‘liqliklarni ko‘rsatish uchun `alpha=0.7`, `s=10` parametrli va ko‘k rangli `ax.scatter()`.
*   **Formatlash:** Yorliqlar (labels) faqat eng chap ustunda (`j == 0`) va eng pastki qatorda (`i == num_features - 1`). Toza ko‘rinish uchun o‘q raqamlari (ticks) olib tashlandi. `plt.tight_layout()` grafiklar ustma-ust tushib qolishining oldini oladi.
*   **`plt.show()`** yakuniy vizualizatsiyani ekranga chiqaradi.

### Matplotlib’da Pair Plot (juftliklar grafigi)ning afzalliklari

*   **Moslashtirish imkoniyati (Customizability):** Seaborn’ning `pairplot()` funksiyasidan farqli o‘laroq, Matplotlib grafik uslubi ustidan to‘liq nazoratni ta’minlaydi.
*   **Yaxshiroq integratsiya:** Matplotlib’ga asoslangan yirikroq vizualizatsiyalar ichida muammosiz ishlaydi.
*   **Moslashuvchanlik (Flexibility):** Ranglar, markerlar, chiziq uslublari va izohlar (annotations) kabi elementlarni osongina o‘zgartirish mumkin.

### Pair Plot’ni yaxshilash

Vizuallashtirishni yaxshilash uchun quyidagilarni amalga oshirishni ko‘rib chiqing:
*   Nuqtali grafiklarga regressiya chiziqlarini qo‘shish.
*   Ma’lumotlar to‘plamidagi toifalarni (kategoriyalarni) ajratib ko‘rsatish uchun turli ranglardan foydalanish.
*   Gistogrammalarni yadro zichligini baholash (KDE - Kernel Density Estimation) grafiklari bilan almashtirish.

**Misol: Regressiya chiziqlari qo‘shilgan Pair Plot (Juftliklar grafigi)**

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)
data = pd.DataFrame(np.random.rand(50, 4), columns=['1-xususiyat', '2-xususiyat', '3-xususiyat', '4-xususiyat'])

# Xususiyatlar soni
num_features = len(data.columns)

# Grafik oynasini yaratish
fig, axes = plt.subplots(num_features, num_features, figsize=(10, 10))

# Har bir xususiyatlar juftligi bo‘ylab tsikl
for i in range(num_features):
    for j in range(num_features):
        ax = axes[i, j]

        if i == j:
            # Diagonalda gistogramma chizish
            ax.hist(data.iloc[:, i], bins=10, color="skyblue", edgecolor="black")
        else:
            # Nuqtali grafik (Scatter plot)
            x = data.iloc[:, j]
            y = data.iloc[:, i]
            ax.scatter(x, y, alpha=0.7, s=10, color="blue")

            # Regressiya chizig'ini qo'shish
            m, b = np.polyfit(x, y, 1)  # Chiziqli regressiya
            ax.plot(x, m*x + b, color="red", linewidth=1)

        # Yorliqlar (Labels)
        if j == 0:
            ax.set_ylabel(data.columns[i], fontsize=10)
        if i == num_features - 1:
            ax.set_xlabel(data.columns[j], fontsize=10)

        # Tozaroq ko'rinish uchun o'q belgilashlarini (ticks) yashirish
        ax.set_xticks([])
        ax.set_yticks([])

# Joylashuvni to'g'rilash
plt.tight_layout()
plt.show()
````

**Izoh:**

*   **Ma’lumotlarni tayyorlash:** NumPy yordamida to‘rtta xususiyat (feature) uchun tasodifiy qiymatlar generatsiya qilinadi va ma’lumotlar to‘plami Pandas `DataFrame`ida saqlanadi.
*   **Kichik grafiklar (Subplots) yaratish:** Juftlik munosabatlarini ko‘rsatish uchun 4×4 o‘lchamli kichik grafiklar to‘ri yaratiladi. `plt.subplots(num_features, num_features, figsize=(10, 10))` to‘rning joylashuvini sozlaydi.
*   **Pair Plot’ni chizish:** Agar `i == j` bo‘lsa, `ax.hist()` yordamida diagonalda gistogramma chiziladi. Agar `i ≠ j` bo‘lsa, `ax.scatter()` yordamida nuqtali grafik (scatter plot) yaratiladi.
*   **Regressiya chiziqlarini qo‘shish:** `np.polyfit(x, y, 1)` funksiyasi regressiya chizig‘ining qiyaligi (`m`) va kesishish nuqtasini (`b`) hisoblaydi. `ax.plot(x, m*x + b, color="red", linewidth=1)` funksiyasi nuqtali grafik ustiga qizil rangli regressiya chizig‘ini tushiradi.
*   **Yorliqlar (Labels)** faqat eng chap va eng pastki grafiklarga qo‘shiladi. Toza dizayn hosil qilish uchun o‘q belgilashlari (ticks) yashiriladi.
*   **`plt.tight_layout()`** o‘qishga qulay bo‘lishi uchun grafiklar o‘rtasidagi mos bo‘shliqlarni ta’minlaydi.

### Korrelyatsion issiqlik xaritasi (Correlation heatmap) nima?

U korrelyatsiya qiymatlarini ko‘rsatish uchun rangli katakchalardan foydalanadi va ma’lumotlar ichidagi naqshlar hamda munosabatlarni vizual jihatdan tushunarli qiladi. ### Korrelyatsion issiqlik xaritasini yaratish qadamlari

Korrelyatsion issiqlik xaritasini yaratish uchun quyidagi amallar bajariladi:
1.  Barcha kerakli modullarni chaqirish (import qilish).
2.  Ma’lumotlar to‘plamini (dataset) yuklash.

4.  Seaborn yordamida issiqlik xaritasini chizish.
5.  Matplotlib yordamida issiqlik xaritasini ekranga chiqarish.

**1-misol: Eng ko‘p sotilgan romanlar (Bestseller Novels) ma’lumotlar to‘plami uchun korrelyatsion issiqlik xaritasi**

Ushbu misolda Kaggle platformasidan yuklab olingan va Amazon'dagi eng ko‘p sotilgan romanlar haqidagi ma’lumotlarni o‘z ichiga olgan ma’lumotlar to‘plamidan (dataset) foydalaniladi.

````python
# Kerakli modullarni chaqirish
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Ma’lumotlar to‘plamini yuklash
data = pd.read_csv("Files/bestsellers with categories.csv")

# Korrelyatsiya matritsasini hisoblash
co_mtx = data.corr(numeric_only=True)

# Korrelyatsiya matritsasini ekranga chiqarish
print(co_mtx)

# Korrelyatsion issiqlik xaritasini chizish
sns.heatmap(co_mtx, cmap="YlGnBu", annot=True)

# Issiqlik xaritasini ko‘rsatish
plt.show()
````

**Izoh:**

*   **Ma’lumotlar to‘plamini yuklash:** Ma’lumotlar to‘plamini (dataset) yuklash uchun `pd.read_csv()` funksiyasidan foydalanamiz.
*   **Korrelyatsiya matritsasini hisoblash:** `.corr()` usuli (method) raqamli ustunlar o‘rtasidagi korrelyatsiyani hisoblaydi.
*   **Issiqlik xaritasini chizish:** `sns.heatmap()` ranglar yordamida kodlangan vizualizatsiyani yaratadi.
*   **Issiqlik xaritasini ko‘rsatish:** `plt.show()` issiqlik xaritasini ekranga chiqaradi.
