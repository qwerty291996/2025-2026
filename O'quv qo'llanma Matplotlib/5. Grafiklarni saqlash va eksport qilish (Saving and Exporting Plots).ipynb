## V BOB. Grafiklarni saqlash va eksport qilish (Saving and Exporting Plots)

### 5.1. Grafiklarni PNG, JPG va SVG formatlarida saqlash (Saving Plots as PNG, JPG & SVG)

### Matplotlib grafiklarini tasvir (rasm) fayli sifatida saqlash

Matplotlib turli xil chizmalar va grafiklarni yaratish uchun mo‘ljallangan mashhur Python kutubxonasidir. Grafikni ekranga chiqarish uchun `show()` funksiyasi ishlatilsa-da, u grafikni faylga saqlamaydi. Grafikni rasm sifatida saqlash uchun bir necha usullar mavjud:

*   **`savefig()`** metodidan foydalanish.
*   **`matplotlib.pyplot.imsave()`** metodidan foydalanish.
*   **Pillow** kutubxonasidan foydalanish.

#### 1. `savefig()` metodi yordamida saqlash

`savefig()` — Matplotlib grafiklarini saqlashning eng ommabop usulidir. Ushbu funksiya grafikni kompyuteringizga **PNG, JPEG, SVG** va boshqa formatlarda saqlash imkonini beradi.

Quyidagi misolda ustunli grafik (bar chart) yaratiladi va u `savefig()` yordamida saqlanadi:

````python
import matplotlib.pyplot as plt

# Ma’lumotlar to‘plami
year = ['2010', '2002', '2004', '2006', '2008']
production = [25, 15, 35, 30, 10]

# Ustunli grafikni chizish
plt.bar(year, production)

# Grafikni JPEG fayli sifatida saqlash
plt.savefig("output.jpg")

# Grafikni qo‘shimcha parametrlar bilan saqlash (ixtiyoriy)
plt.savefig("output1.jpg", 
            facecolor='yellow',    # Orqa fon rangi
            bbox_inches="tight",   # Grafik atrofidagi bo'shliqlarni qisqartirish
            pad_inches=0.3,        # Grafik chetidagi masofa
            transparent=True)      # Shaffoflikni yoqish
````

**Izoh:** `savefig()` funksiyasi orqali grafikni nafaqat oddiy holatda, balki qo‘shimcha sozlamalar bilan ham saqlash mumkin. Masalan, `bbox_inches="tight"` parametri grafik atrofidagi ortiqcha oq joylarni olib tashlaydi, `transparent=True` esa orqa fonni shaffof qiladi.

#### 2. `matplotlib.pyplot.imsave()` metodidan foydalanish

`imsave()` funksiyasi grafikni tasvir fayli sifatida saqlashning yana bir usulidir. U odatda 2D massivlarni tasvir fayllari sifatida saqlashda qo‘llaniladi, bu esa uni tasvir ma’lumotlari bilan ishlashda ayniqsa foydali qiladi.


````python
import matplotlib.pyplot as plt
import numpy as np

# Tasodifiy rasm ma’lumotlarini hosil qilish
img = np.random.rand(100, 100)

# Rasmni saqlash
plt.imsave('sample_image.png', img, cmap='gray')
````

**Izoh:**
*   **`np.random.rand(100, 100)`**: 100x100 o‘lchamdagi tasodifiy sonlardan iborat massiv yaratadi.
*   **`plt.imsave()`**: Ushbu massivni belgilangan nom bilan saqlaydi. `cmap='gray'` parametri tasvirni oq-qora (kulrang) formatda saqlashni ta’minlaydi.

#### 3. Pillow kutubxonasidan (PIL) foydalanish

Muayyan holatlarda Matplotlib grafigini Pillow (PIL) tasvir obyektiga aylantirish va saqlash foydali bo‘lishi mumkin. Bu, ayniqsa, Pillow kutubxonasini talab qiladigan tasvirga ishlov berish amaliyotlari bilan ishlashda qo‘l keladi.

*   **BytesIO**: Grafik tasvirini xotirada (memory) saqlash uchun `BytesIO` buferidan foydalaniladi, so‘ngra u PIL orqali qayta ishlanishi mumkin.
*   **fig.savefig()**: Kannvani to‘g‘ridan-to‘g‘ri satrga (string) aylantirish o‘rniga, grafikni PNG formatida buferga saqlaymiz.
*   **Image.open()**: Tasvirni buferdan PIL tasvir obyektiga yuklaymiz.



````python
import matplotlib.pyplot as plt
from PIL import Image
import io

fig, ax = plt.subplots()
ax.plot([1, 2, 3], [1, 2, 3])

# Grafikni BytesIO obyektiga (xotiraga) saqlash
buf = io.BytesIO()
fig.savefig(buf, format='png')

# BytesIO obyektini PIL tasviriga aylantirish
buf.seek(0)
img = Image.open(buf)

# Tasvirni saqlash
img.save('pil_image_save.png')
````

**Izoh:**
Ushbu usul grafikni qattiq diskka fayl sifatida yozmasdan turib, xotiraning o‘zida rasmga aylantirish imkonini beradi. `buf.seek(0)` buyrug‘i bufer ko‘rsatkichini boshiga qaytaradi, shunda `Image.open()` tasvir ma’lumotlarini boshidan o‘qishi mumkin bo‘ladi.

### 5.2. Grafiklarni PDF formatida saqlash (Exporting Plots to PDF)

### Matplotlib grafiklarini PDF formatida saqlash

Matplotlib'da vizuallashtirish bilan ishlaganda, hisobotlar, taqdimotlar yoki chop etish uchun ko‘pincha grafiklarni PDF formatida saqlash zarur bo‘ladi. Matplotlib `savefig()` funksiyasi yordamida aniq va professional ko‘rinishdagi yuqori sifatli grafiklarni eksport qilishning oddiy usulini taqdim etadi.

#### Bitta grafikni eksport qilish uchun `savefig()` dan foydalanish

Matplotlib grafigini PDF sifatida eksport qilishning eng oson yo‘li `savefig()` funksiyasidan foydalanishdir. Ushbu usul bir qator kod yordamida shaklni (figure) yuqori sifatda saqlash imkonini beradi.

````python
import matplotlib.pyplot as plt

# Oddiy grafik yaratish
plt.plot([1, 2, 3, 4], [10, 20, 25, 30], marker='o', linestyle='--', color='b')
plt.title("Namuna grafigi")

# Grafikni PDF sifatida saqlash
plt.savefig("plot.pdf", format="pdf")

# Grafikni ko'rsatish
plt.show()
````

**Izoh:**
*   `plt.plot(...)` belgilar (markers) bilan oddiy chiziqli grafik yaratadi.
*   `plt.title(...)` grafikaga sarlavha qo‘shadi.
*   `plt.savefig("plot.pdf", format="pdf")` shaklni PDF fayli sifatida saqlaydi.
*   `plt.show()` grafikni ekranda ko‘rsatadi.

Odatiy holda, fayl skript ishlayotgan direktoriya ichida saqlanadi. Agar siz uni ma’lum bir joyda saqlamoqchi bo‘lsangiz, to‘liq yo‘lni ko‘rsatishingiz kerak:

### Bir nechta grafikni bitta PDF fayliga eksport qilish

Agar bir nechta grafikni bitta PDF fayliga saqlash kerak bo‘lsa, Matplotlib **PdfPages** vositasini taqdim etadi. Bu bitta hujjat ichida bir nechta shakllarni (figure) ketma-ket saqlash imkonini beradi.


````python
import matplotlib.pyplot as plt  
from matplotlib.backends.backend_pdf import PdfPages  

# Bir nechta grafikni saqlash uchun PDF fayli yaratish
with PdfPages("multiple_plots.pdf") as pdf:  
    
    # Birinchi grafik
    plt.figure()  
    plt.plot([1, 2, 3], [4, 5, 6], marker='o', linestyle='--', color='r')  
    plt.title("1-grafik")  
    pdf.savefig()  # Birinchi grafikni PDFga saqlash
    plt.close()  # Xotirani bo'shatish uchun shaklni yopish
    
    # Ikkinchi grafik
    plt.figure()  
    plt.bar(["A", "B", "C"], [10, 20, 30], color="g")  
    plt.title("2-grafik")  
    pdf.savefig()  # Ikkinchi grafikni PDFga saqlash
    plt.close()  # Shaklni yopish
````

**Izoh:**
*   **`PdfPages("multiple_plots.pdf")`** bir nechta grafikni saqlash uchun PDF fayli yaratadi.
*   **`plt.figure()`** har bir yangi grafik uchun alohida shaklni ishga tushiradi.
*   **`pdf.savefig()`** ayni damdagi joriy shaklni PDF fayliga bitta sahifa sifatida qo‘shadi.
*   **`plt.close()`** operativ xotirani bo‘shatish uchun grafik shaklini yopadi.

Odatiy holda, fayl skript ishlayotgan direktoriyada saqlanadi. Agar uni boshqa maxsus manzilda saqlashni istasangiz, to‘liq yo‘lni ko‘rsating:
`with PdfPages("C:/Foydalanuvchi/Hujjatlar/multiple_plots.pdf") as pdf:`

### 5.3. Bir nechta grafiklarni saqlash (Saving Multiple Figures)

Ushbu maqolada bir nechta Matplotlib grafiklarini bitta PDF fayliga saqlash usullari ko‘rib chiqiladi. Buning uchun `PdfPages` klassining `savefig()` metodidan foydalaniladi. Bu usul barcha yaratilgan shakllarni (figures) joriy ishchi direktoriyadagi bitta PDF hujjatiga birlashtiradi.

#### Bosqichma-bosqich amalga oshirish

*   **1-bosqich:** Kerakli modullarni yuklab olish.
*   **2-bosqich:** Grafik o‘lchamlarini va sub-plotlar orasidagi masofani (`autolayout`) sozlash.
*   **3-bosqich:** `plt.figure()` yordamida bir nechta shakllarni (masalan: `fig1`, `fig2`, `fig3`) e’lon qilish.
*   **4-bosqich:** `plt.plot()` metodidan foydalanib ma’lumotlarni chizish.
*   **5-bosqich:** Barcha ochiq grafik raqamlarini `plt.get_fignums()` orqali olib, ularni PDFga saqlaydigan maxsus funksiya yaratish.

#### To‘liq kod namunasi

````python
from matplotlib import pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

# 2-bosqich: Grafik parametrlarini sozlash
plt.rcParams["figure.figsize"] = [7.00, 3.50]
plt.rcParams["figure.autolayout"] = True

# 3-bosqich: Shakllarni yaratish
fig1 = plt.figure()
plt.plot([17, 45, 7, 8, 7], color='orange', label='Line 1')
plt.title("Birinchi grafik")

fig2 = plt.figure()
plt.plot([13, 25, 1, 6, 3], color='blue', label='Line 2')
plt.title("Ikkinchi grafik")

fig3 = plt.figure()
plt.plot([22, 11, 2, 1, 23], color='green', label='Line 3')
plt.title("Uchinchi grafik")

# 5-bosqich: Saqlash funksiyasi
def save_image(filename):
    # PdfPages - bu xatolarsiz fayl yaratish uchun PDF ustidagi qobiqdir
    p = PdfPages(filename)
    
    # Mavjud barcha grafik raqamlari ro'yxatini olish
    fig_nums = plt.get_fignums()  
    figs = [plt.figure(n) for n in fig_nums]
    
    # Har bir shakl bo'ylab sikl aylantirish va saqlash
    for fig in figs: 
        fig.savefig(p, format='pdf') 
        
    # PDF obyektini yopish
    p.close()  

# Funksiyani chaqirish
save_image("Save_multiple_plots.pdf")
````

**Asosiy tushunchalar:**
*   **`plt.get_fignums()`**: Dastur davomida yaratilgan va hali yopilmagan barcha grafik raqamlarini ro‘yxat shaklida qaytaradi.
*   **`PdfPages`**: PDF fayli bilan ishlashda to‘qnashuvlarning oldini oluvchi qulay vositadir.
*   **`p.close()`**: Faylni to‘liq yakunlash va xotiradan bo‘shatish uchun ushbu metodni chaqirish shart.
