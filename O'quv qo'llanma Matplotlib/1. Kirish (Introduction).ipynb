* *Interaktiv vidjetlar:* ma’lumotlarni dinamik ravishda manipulyatsiya qilish uchun to‘g‘ridan-to‘g‘ri chizmalarga slayderlar, tugmalar va nazorat katakchalarini (checkboxes) qo‘shish uchun maxsus modulni o‘z ichiga oladi.

### 1.1. Matplotlib-ga kirish (Introduction to Matplotlib)
Matplotlib — bu ma’lumotlardan statik, interaktiv va animatsion vizualizatsiyalar yaratish uchun mo‘ljallangan Python kutubxonasi. U ma’lumotlar qonuniyatlari, tendensiyalari va bog‘liqliklarini samarali tushunishga yordam beradigan moslashuvchan va sozlanishi mumkin bo‘lgan chizma funksiyalarini taqdim etadi.

**Misol**: Matplotlib yordamida oddiy chiziqli grafik yaratamiz, bu ma’lumotlarni vizualizatsiya qilish qanchalik osonligini ko‘rsatadi.

```python
import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]

plt.plot(x, y)
plt.show()
```

**Matplotlib shaklining (Figure) komponentlari yoki qismlari**

**Matplotlib chizmasining anatomiyasi:** U ma’lumotlar vizualizatsiyasi uchun zarur bo‘lgan Matplotlib chizmasining asosiy qismlarini, jumladan, shakllar, o‘qlar, sarlavhalar va legendalarni o‘rganadi.



Matplotlib shaklining qismlari quyidagilardan iborat:

* **`Figure (Shakl):`** Barcha chizma elementlarini o‘z ichiga olgan umumiy konteyner, vizualizatsiya uchun mato (canvas) vazifasini bajaradi.
* **`Axes (Chizma maydoni):`** Shakl ichidagi ma’lumotlar chiziladigan sohalar; har bir shakl bir nechta chizma maydonini o‘z ichiga olishi mumkin.
* **`Axis (o‘q):`** X va Y o‘qlarini ifodalaydi, ma’lumotlarni talqin qilish uchun chegaralar, belgi (tick) joylashuvi va yorliqlarni belgilaydi.
* **`Lines and Markers (Chiziqlar va belgilar):`** Chiziqlar tendensiyalarni ko‘rsatish uchun ma’lumot nuqtalarini birlashtiradi, belgilar esa sochma grafiklar kabi chizmalarda alohida ma’lumot nuqtalarini bildiradi.
* **`Title and Labels (Sarlavha va yorliqlar):`** Sarlavha chizma uchun kontekstni taqdim etadi, o‘q yorliqlari esa har bir o‘qda qanday ma’lumotlar aks ettirilayotganini tavsiflaydi.

**`Matplotlib Pyplot`**

`Pyplot` — Matplotlib ichidagi modul bo‘lib, chizmalar yaratish uchun MATLAB kabi interfeysni taqdim etadi. U joriy shaklning (figure) o‘qlariga chiziqlar, tasvirlar va matn kabi chizma elementlarini qo‘shish jarayonini soddalashtiradi.

**`Pyplot`dan foydalanish bosqichlari**

* *Matplotlib’ni import qilish:* Ishni `matplotlib.pyplot as plt` ko‘rinishida import qilishdan boshlang.
* *Ma’lumotlarni yaratish:* Ma’lumotlaringizni ro‘yxat (list) yoki massiv (array) ko‘rinishida tayyorlang.
* *Ma’lumotlarni chizish:* Chizma yaratish uchun `plt.plot()` funksiyasidan foydalaning.
* *Chizmani moslashtirish:* `plt.title()`, `plt.xlabel()` va `plt.ylabel()` kabi usullar yordamida sarlavhalar, yorliqlar va boshqa elementlarni qo‘shing.
* *Chizmani ko‘rsatish:* Chizmani ekranda chiqarish uchun `plt.show()` funksiyasidan foydalaning.

Keling, oddiy chizmani vizualizatsiya qilamiz va Matplotlib shaklining asosiy komponentlarini tushunib olamiz:

```python
import matplotlib.pyplot as plt

x = [0, 2, 4, 6, 8]
y = [0, 4, 16, 36, 64]

fig, ax = plt.subplots()  
ax.plot(x, y, marker='o', label="Ma’lumot nuqtalari")

ax.set_title("Matplotlib shaklining asosiy komponentlari")
ax.set_xlabel("X o‘qi") 
ax.set_ylabel("Y o‘qi")  

plt.show()
```

### 1. Chiziqli grafik (Line Chart)

Chiziqli grafik asosiy grafik turlaridan biri bo‘lib, uni `plot()` funksiyasi yordamida yaratish mumkin. U turli o‘qlardagi ikki ma’lumot (X va Y) o‘rtasidagi bog‘liqlikni ifodalash uchun ishlatiladi.



**Sintaksis:**
`matplotlib.pyplot.plot(x, y)`


**Parametrlar:**
* **`x, y:`** ma’lumot nuqtalari uchun koordinatalar.

**Misol:**
Ushbu kod Matplotlib yordamida o‘qlari nomlangan va sarlavhaga ega oddiy chiziqli grafikni chizadi.


```python
# ma’lumotlarni shakllantirish
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

# Grafikni yaratish
plt.plot(x, y)

# Grafik bezaklari
plt.title("Line Chart (Chiziqli grafik)")
plt.ylabel('Y-O‘qi')
plt.xlabel('X-O‘qi')

# Natijani ko‘rsatish
plt.show()
```

### **2. Ustunli diagramma (Bar Chart)**

Ustunli diagramma kategoriyali ma’lumotlarni to‘rtburchak ustunlar ko‘rinishida tasvirlaydi, bunda ustunlarning uzunligi ular ifodalayotgan qiymatlarga proporsional bo‘ladi. Turli kategoriyalarni o‘zaro solishtirish uchun u vertikal yoki gorizontal holatda chizilishi mumkin.



**Sintaksis (Syntax):**
`matplotlib.pyplot.bar(x, height)`

**Parametrlar:**
* **`x:`** X o‘qidagi kategoriyalar yoki pozitsiyalar.
* **`height:`** Ustunlarning balandligi (Y o‘qidagi qiymatlar).



**Misol:**
Ushbu kod haftaning turli kunlari uchun jami hisob-kitoblarni ko‘rsatuvchi oddiy ustunli diagramma yaratadi. Bunda X o‘qi kunlarni, Y o‘qi esa jami hisob miqdorini ifodalaydi.

```python
import matplotlib.pyplot as plt

# ma’lumotlar
x = ['Paysh', 'Juma', 'Shanba', 'Yak']
y = [170, 120, 250, 190]

# Diagrammani yaratish
plt.bar(x, y)

# Bezash
plt.title("Ustunli diagramma (Bar Chart)")
plt.xlabel("Kunlar")
plt.ylabel("Jami hisob")

# Ko‘rsatish
plt.show()
```

### 3. Gistogramma (Histogram)

Gistogramma ma’lumotlarni guruhlarga (binlarga) ajratish orqali ularning **taqsimlanishini** ko‘rsatadi. Uni yaratish uchun `hist()` funksiyasidan foydalaniladi; bunda X o‘qi guruhlarni (intervallarni), Y o‘qi esa takrorlanish chastotasini (miqdorini) bildiradi.

**Sintaksis:**
`matplotlib.pyplot.hist(x, bins=None)`

**Parametrlar:**
* **`x:`** Kiruvchi ma’lumotlar.
* **`bins:`** ma’lumotlarni guruhlash uchun intervallar (bloklar) soni.

**Misol:**
Ushbu kod `x` ro‘yxatidagi umumiy hisob-kitob (total bill) qiymatlarining taqsimlanish chastotasini ko‘rsatuvchi gistogrammani chizadi. Unda 10 ta guruhdan foydalanilgan hamda tushunarli bo‘lishi uchun o‘qlar nomlanib, sarlavha qo‘shilgan.


```python
import matplotlib.pyplot as plt

# ma’lumotlar ro‘yxati
x = [7, 8, 9, 10, 10, 12, 12, 12, 13, 14, 14, 15, 16, 16, 17, 18, 18, 19, 20, 20,
     21, 22, 23, 24, 25, 25, 26, 28, 30, 32, 35, 36, 38, 40, 42, 44, 48, 50]

# Gistogramma yaratish
plt.hist(x, bins=10, color='steelblue')

# Grafik elementlari
plt.title("Gistogramma")
plt.xlabel("Umumiy hisob")
plt.ylabel("Chastota")

# Grafikni ko‘rsatish
plt.show()
```

### 4. Tarqoq grafik (Scatter Plot)

Tarqoq grafiklar o‘zgaruvchilar o‘rtasidagi bog‘liqlikni kuzatish uchun ishlatiladi. Matplotlib kutubxonasidagi `scatter()` metodi tarqoq grafik chizish uchun qo‘llaniladi.



**Sintaksis:**
`matplotlib.pyplot.scatter(x, y)`

**Parametrlar:**
* **`x, y:`** Nuqtalarning koordinatalari.

Kod bajarilganda, har bir ma’lumot nuqtasi alohida nuqtalar ko‘rinishida grafikda aks etadi, bu esa kunlar bo‘yicha xarajatlarning tarqalishini ko‘rish imkonini beradi.

### 5. Doiraviy diagramma (Pie Chart)

Doiraviy diagramma — bu ma’lumotlarni ulushlar yoki foizlar ko‘rinishida ko‘rsatish uchun ishlatiladigan doira shaklidagi grafikdir. U `pie()` funksiyasi yordamida yaratiladi, bunda har bir bo‘lak butunning bir qismini ifodalaydi.



**Sintaksis:**
`matplotlib.pyplot.pie(x, labels=None, autopct=None)`

**Parametrlar:**
* **`x:`** Doira bo‘laklari uchun ma’lumot qiymatlari.
* **`labels:`** Har bir bo‘lak uchun nomlar.
* **`autopct:`** Foiz ko‘rsatkichlarini aks ettirish formati (masalan, `'%1.1f%%'`).

**Misol:**
Ushbu kod turli xil avtomobil brendlarining taqsimlanishini ko‘rsatish uchun oddiy doiraviy diagramma yaratadi. Diagrammaning har bir bo‘lagi ma’lumotlar to‘plamidagi har bir brendga tegishli avtomobillar ulushini ifodalaydi. Kod bajarilganda, har bir avtomobil brendi o‘zining ulushiga mos ravishda rangli bo‘laklar ko‘rinishida doira ichida aks etadi.

```python
import pandas as pd

# ma’lumotlar
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR']
data = [23, 10, 35, 15, 12]

# Doiraviy diagrammani yaratish
plt.pie(data, labels=cars)
plt.title("Doiraviy diagramma")

# Grafikni ko‘rsatish
plt.show()
```

### 6. Qutichali grafik (Box Plot)

Qutichali grafik — bu ma’lumotlarning qanchalik yoyilganligini (dispersiyasini) ko‘rsatadigan oddiy grafikdir. U ma’lumotlarning minimal, maksimal, median va kvartil qiymatlarini aks ettiradi, shuningdek, chetlanishlarni (outliers) osongina aniqlashga yordam beradi.



**Sintaksis:**
`matplotlib.pyplot.boxplot(x, notch=False, vert=True)`

**Parametrlar:**
* **`x:`** Grafik chizilishi kerak bo‘lgan ma’lumotlar (odatda ro‘yxat yoki massiv).
* **`notch:`** Agar `True` bo‘lsa, median atrofidagi ishonch oralig‘ini ko‘rsatish uchun "o‘yiq" (notch) chizadi.
* **`vert:`** Agar `True` bo‘lsa, qutichalar vertikal bo‘ladi. Agar `False` bo‘lsa, gorizontal joylashadi.

**Misol:**
Ushbu kod ma’lumotlar taqsimotini ko‘rsatish va uchta guruhni o‘zaro solishtirish uchun Matplotlib yordamida qutichali grafik yaratadi. Kod bajarilganda, har bir guruh uchun ma’lumotlar markazi va tarqalish ko‘lamini ko‘rsatuvchi “mo‘ylovli qutichalar” hosil bo‘ladi.

```python
# ma’lumotlar guruhlari
data = [ [10, 12, 14, 15, 18, 20, 22],
         [8, 9, 11, 13, 17, 19, 21],
         [14, 16, 18, 20, 23, 25, 27] ]

# Qutichali grafikni yaratish
plt.boxplot(data)

# Grafik elementlari
plt.xlabel("Guruhlar")
plt.ylabel("Qiymatlar")
plt.title("Qutichali grafik (Box Plot)")

# Grafikni ko‘rsatish
plt.show()
```

### 7. Issiqlik xaritasi (Heatmap)

Issiqlik xaritasi — bu ma’lumotlarning grafik ko‘rinishi bo‘lib, unda qiymatlar ranglar orqali ifodalanadi. U matritsa ko‘rinishidagi ma’lumotlarda qonuniyatlar, korrelyatsiyalar yoki intensivlikni vizuallashtirishga yordam beradi. Matplotlib kutubxonasida u `imshow()` metodi yordamida yaratiladi.



**Sintaksis:**
`matplotlib.pyplot.imshow(X, cmap='viridis')`

**Parametrlar:**
* **`X:`** 2D massiv (tasvir yoki issiqlik xaritasi sifatida ko‘rsatiladigan ma’lumotlar).
* **`cmap:`** Ranglar xaritasini (koloritni) belgilaydi.

**Misol:**
Ushbu kod `imshow()` funksiyasidan foydalanib, 10×10 o‘lchamdagi tasodifiy ma’lumotlarning issiqlik xaritasini yaratadi. Unda `'viridis'` ranglar xaritasidan foydalanilgan va `colorbar()` metodi ranglar shkalasini qo‘shadi. Kod bajarilganda, har bir katakchaning qiymatiga mos ravishda turli ranglarga bo‘yalgan kvadratik panjara hosil bo‘ladi. To‘qroq yoki ochroq ranglar ma’lumotlarning intensivligini bildiradi.

```python
import numpy as np

# Tasodifiy ma’lumotlar yaratish
np.random.seed(0)
data = np.random.rand(10, 10)

# Issiqlik xaritasini chizish
plt.imshow(data, cmap='viridis', interpolation='nearest')

# Grafik elementlari
plt.colorbar() # Ranglar shkalasi
plt.xlabel('X o\'qi sarlavhasi')
plt.ylabel('Y o\'qi sarlavhasi')
plt.title('Issiqlik xaritasi (Heatmap)')

# Grafikni ko‘rsatish
plt.show()
```

**Izoh (Explanation):**
* **`np.random.seed(0)`**: Har safar bir xil tasodifiy qiymatlar hosil bo‘lishini ta'minlaydi (natijaning takrorlanuvchanligi uchun).
* **`np.random.rand(10, 10)`**: 0 va 1 oralig‘idagi tasodifiy sonlardan iborat 10×10 o‘lchamli massiv (matritsa) yaratadi.

### Ma’lumotlarni vizuallashtirish (Data Visualization)

Ma’lumotlarni vizuallashtirish — bu axborot va ma’lumotlarning chiziqli grafik, ustunli diagramma, doiraviy diagramma kabi tasviriy yoki grafik shakldagi ko‘rinishidir. Ma’lumotlarni vizuallashtirish ma’lumotlardagi asosiy tendensiyalarni tushunish va tahliliy xulosalar chiqarishga yordam beradi, bu esa tashkilotlarga ma’lumotlarga asoslangan qarorlar qabul qilish imkonini beradi.



**Ma’lumotlarni vizuallashtirish muhimligining sabablari:**
* *Qonuniyatlarni aniqlash:* Ma’lumotlarning vizual ko‘rinishi ma’lumotlardagi qonuniyat va tendensiyalarni aniqlashga yordam beradi.
* *Ma’lumotlar tahlili:* Ma’lumotlarning vizual tasviri foydalanuvchilarga ma’lumotlarni saralash va chuqurroq tushunchalarga ega bo‘lish imkonini beradi.
* *Qaror qabul qilish:* Imkoniyatlar va potensial xavflarni tasviriy ko‘rinishda aniqlash orqali tashkilotning qaror qabul qilish jarayonini yaxshilaydi.
* *Xotirada saqlab qolish:* Insonlar raqamli yoki matnli ma’lumotlarga qaraganda grafik ma’lumotlarni yaxshiroq eslab qolishga moyil bo‘ladilar.
* *Samaradorlik:* Foydalanuvchiga asosiy tushunchalar, tendensiyalar va chetlanishlarni (outliers) tezda aniqlash imkonini beradi va ko‘p vaqtni tejaydi.



### Ma’lumotlarni vizuallashtirish turlari

* **Ustunli diagramma (Bar Chart):** Ma’lumotlarni tasvirlash uchun to‘rtburchak ustunlar ishlatiladi. Ular alohida toifalar orasidagi farqni ko‘rsatish yoki vaqt o‘tishi bilan sodir bo‘lgan o‘zgarishlarni ta’kidlash uchun qo‘llaniladi. Gorizontal va vertikal turlari mavjud.
* **Chiziqli grafik (Line Chart):** Ma’lumotlar chiziqlar bilan bog‘langan nuqtalar to‘plami ko‘rinishida ko‘rsatiladi. Ular vaqt kabi uzluksiz o‘q bo‘ylab tendensiya va qonuniyatlarni ko‘rsatish uchun juda mos keladi.
* **Doiraviy diagramma (Pie Chart):** Ma’lumotlarni doira bo‘laklari ko‘rinishida ko‘rsatadi, bunda har bir bo‘lak umumiy toifaning foiz ulushini ifodalaydi. Ular butunning qismlarini ko‘rsatish yoki nisbiy o‘lchamlarni solishtirish uchun foydalidir.
* **Tarqoq grafik (Scatter Plot):** Alohida ma’lumot nuqtalari ikki o‘lchovli setkada nuqtalar sifatida ko‘rsatiladi. Ular ikki o‘zgaruvchi o‘rtasidagi bog‘liqlikni va korrelyatsiyani aniqlash uchun ishlatiladi.
* **Gistogramma (Histogram):** Ma’lumotlarni intervallarga (binlarga) guruhlash orqali ularning chastotasi yoki taqsimlanishini ko‘rsatadi. Uzluksiz ma’lumotlar taqsimotini ko‘rsatish uchun samarali.
* **Issiqlik xaritasi (Heatmap):** Matritsa yoki jadval formatida ma’lumotlar qiymatlari rang intensivligi bilan ifodalanadi. Ular ikki toifali o‘zgaruvchilar o‘rtasidagi aloqalarni ko‘rsatishga yordam beradi.
* **Pufakchali grafik (Bubble Chart):** Tarqoq grafikga o‘xshaydi, lekin uchinchi o‘zgaruvchini ko‘rsatish uchun pufakcha o‘lchami o‘lchovidan ham foydalaniladi.
* **Qutichali grafik (Box Plot):** Ma’lumotlar taqsimotini ko‘rsatish uchun kvartillar, chetlanishlar va qutichadan foydalanadi. Ma’lumotlarning asimmetriyasini (skewness) ko‘rsatish uchun foydali.
* **Ierarxik to‘rtburchaklar (Tree Map):** Ma’lumotlarni ichma-ich joylashgan to‘rtburchaklarga ajratadi, ularning har biri alohida toifani anglatadi. Ierarxik ma’lumotlar uchun mos keladi.
* **Xarita-diagramma (Choropleth Map):** Turli geografik hududlar uchun ma’lumotlar qiymatlarini ko‘rsatishda ranglar bilan kodlangan hududlardan foydalaniladi.
* **So‘zlar buluti (Word Cloud):** Har bir so‘zning o‘lchami uning takrorlanish chastotasi yoki ahamiyatini bildiradi. Matnli ma’lumotlar tahlilida keng qo‘llaniladi.
* **Radar diagrammasi (Radar Chart):** Ma’lumot nuqtalari markaziy nuqtadan chiquvchi o‘qlar (shpols) ustida ko‘rsatiladi. Turli guruhlarning ko‘plab parametrlarini solishtirishda yordam beradi.
* **Tarmoq diagrammasi (Network Diagram):** Tugunlar o‘rtasidagi bog‘lanish va aloqalarni ko‘rsatadi. Murakkab ijtimoiy tarmoqlar yoki tizimlar aloqalarini tasvirlashda qo‘llaniladi.

**Ma’lumotlarni vizuallashtirishning boshqa turlari:** Donut diagrammasi, oqim diagrammasi (flowchart), jadval, geografik xarita, Venn diagrammasi, voronka (funnel) diagrammasi va boshqalar.

### Jupyter Notebook-da ustunli grafik (Bar Graph)
Ustunli grafik ma’lumotlarni turli uzunlikdagi to‘rtburchak ustunlar orqali ifodalaydi va har bir ustun uzunligi u ifodalayotgan qiymatga mos keladi. Bu grafik turlari toifalarni yoki alohida ma’lumot nuqtalarini o‘zaro solishtirish uchun samaralidir.

Jupyter Notebook-da ustunli grafikdan foydalanish uchun quyidagi bosqichlarni bajaring:
1. `matplotlib` modulini chaqiring (import).
2. X o‘qi uchun ma’lumotlarni massivga yuklang.
3. Y o‘qi uchun ma’lumotlarni massivga yuklang.
4. `bar()` funksiyasi yordamida ustunli grafikni chizing. Grafikni moslashtirish uchun `color` (rang), `width` (kenglik) kabi qo‘shimcha parametrlarni sozlashimiz mumkin (pastdagi misolga qarang).
5. `title()` funksiyasi yordamida grafik sarlavhasini belgilang.
6. `show()` metodi yordamida grafikni ko‘rsating.

**Misol:**

```python
# ma’lumotlar
x = [10, 20, 30, 40, 50, 60]
y = [13, 45, 23, 34, 96, 76]

# Grafik sarlavhasi
plt.title('Ustunli grafik (Bar Graph)')

# Grafikni chizish (rangi 'dodgerblue', ustun kengligi 5)
plt.bar(x, y, color='dodgerblue', width=5)

# Grafikni ko‘rsatish
plt.show()
```

### Jupyter Notebook-da doiraviy diagramma (Pie Chart)

Doiraviy diagramma ma’lumotlarni bo‘laklarga bo‘lingan aylana shaklidagi grafik ko‘rinishida aks ettiradi va har bir bo‘lak butunning ma’lum bir qismi yoki foizini ifodalaydi.



Jupyter Notebook-da doiraviy diagrammadan foydalanish uchun quyidagi bosqichlarni bajaring:
1. `matplotlib` modulini chaqiring (import).
2. Ma’lumotlaringiz yorliqlarini (nomlarini) massivga yuklang, masalan: `label = ['olma', 'banan', 'apelsin']`.
3. Qiymatlarni massivga yuklang, masalan: `values = [13, 45, 23, 34, 96, 76]`.
4. `pie()` metodi yordamida doiraviy diagrammani chizing. Diagrammani moslashtirish uchun `explode` (bo‘lakni ajratish), `autopct` (foiz ko‘rsatish), `pctdistance` (foiz masofasi), `startangle` (boshlang‘ich burchak) kabi parametrlarni sozlashimiz mumkin.
5. `title()` funksiyasi yordamida grafik sarlavhasini belgilang.
6. `show()` metodi yordamida grafikni ko‘rsating.

**Misol:**


```python
# ma’lumotlar
x = [35, 20, 30, 40, 50, 30]
y = ['Olma', 'Banan', 'Uzum', 'Apelsin', 'Ananas', 'Ajdaho mevasi']

# Grafik sarlavhasi
plt.title('Doiraviy diagramma (Pie Chart)')

# Diagrammani chizish
plt.pie(x, labels=y)

# Grafikni ko‘rsatish
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# ma’lumotlar to‘plamini yuklash
data = sns.load_dataset("iris")

# Chiziqli grafikni chizish
sns.lineplot(x="sepal_length", y="sepal_width", data=data)

# Grafikni ko‘rsatish
plt.show()
```
