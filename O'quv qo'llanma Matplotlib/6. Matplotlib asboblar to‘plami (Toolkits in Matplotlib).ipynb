### 1. Lineplot (Chiziqli grafik)
**Lineplot** ikkita uzluksiz o‘zgaruvchi orasidagi bog‘liqlikni ko‘rsatish uchun ishlatiladi. U odatda vaqt seriyalari (time-series) yoki tartiblangan ma’lumotlar uchun juda qo‘l keladi.

*   `hue="region"`: Chiziqlarni hududiga qarab turli ranglarga ajratadi.

Vizuallashtirish olamida chiziqli grafiklar vaqt o‘tishi bilan nima sodir bo‘layotganini tushunish uchun eng sodda va samarali usuldir. Xuddi ob-havo ma’lumotlaridagi harorat o‘zgarishini kuzatishdek gap!

### 2. Scatterplot (Tarqoq grafik)

**Scatterplot** ikkita raqamli o‘zgaruvchi o‘rtasidagi munosabatni ko‘rsatish uchun ishlatiladi. U ma’lumotlar ichidagi qonuniyatlarni yoki korrelyatsiyalarni aniqlashga yordam beradi.

### 6. Heatmap (Issiqlik xaritasi)

### 6.2. `mplot3d` orqali 3D grafiklar (3D Plotting with `mplot3d`)

Ushbu paragrafda Matplotlib yordamida 3D grafiklar chizishni o‘rganamiz. 3D grafik yaratishning bir necha yo‘li mavjud bo‘lib, ular orasida bo‘sh kanva yaratish va unga 3D proyeksiyali o‘qlarni qo‘shish yoki `Matplotlib.pyplot.gca()` kabi metodlardan foydalanish kabilar bor.

### Bo‘sh 3D grafik yaratish

Dastlab bo‘sh kanva yaratiladi, so‘ngra grafikning o‘qlari (axes) "3D" proyeksiyasi ko‘rinishida belgilanadi. Bu kanvada 3D o‘qlar shaklini hosil qilishga yordam beradi. Agar grafikni `plt.show()` yordamida ko‘rsatsak, u uchta o‘qqa (x, y, z) ega bo‘lgan bo‘sh fazoni aks ettiradi.

```python
# Kerakli paketlarni yuklab olish
import numpy as np
import matplotlib.pyplot as plt

# Bo‘sh shakl (figure) yaratish
fig = plt.figure()

# O‘qlarni 3D proyeksiyada aniqlash, 
# shunda unga 3D ma’lumotlarni joylashtirish mumkin bo‘ladi
ax = plt.axes(projection="3d")

# Grafikni ko‘rsatish
plt.show()
```

**Kodni tushunish:**
*   **`fig = plt.figure()`**: Grafik chizish uchun asosiy maydonni yaratadi.
*   **`ax = plt.axes(projection="3d")`**: O‘qlarni 3D formatda belgilaydi. Bu Matplotlib'ga biz oddiy 2D grafik emas, balki fazoviy grafik chizmoqchi ekanligimizni bildiradi.
*   **`plt.show()`**: Yaratilgan bo‘sh 3D maydonni ekranda ko‘rsatadi.

Endi bo‘sh kanva yaratish haqida asosiy tushunchaga egamiz. Keyingi bosqichda ushbu fazoga 3D nuqtalar yoki chiziqlarni qanday joylashtirishni ko‘rib chiqishni istaysizmi?

Keling, Matplotlib yordamida 3D grafik chizish bo‘yicha amaliy misolni ko‘rib chiqamiz. Bu safar biz nafaqat chiziq, balki rangli nuqtalar to‘plamini ham fazoda aks ettiramiz.

### 3D grafik chizishga doir misollar

**1-misol: 3D chiziqli va tarqoq grafik**

Ushbu misolda biz 3D fazoda oddiy egri chiziqni chizamiz. Shu bilan birga, X, Y va Z koordinatalariga ega bo‘lgan bir qator nuqtalar to‘plamini ham joylashtiramiz.

```python
# numpy paketini yuklab olish
import numpy as np
# matplotlib paketini yuklab olish
import matplotlib.pyplot as plt
# mplot3d vositasini yuklab olish
from mpl_toolkits import mplot3d

# Bo‘sh kanva yaratish
fig = plt.figure()

# 3D grafiklar chizish uchun proyeksiyani "3d" deb belgilash
ax = plt.axes(projection="3d")

# X, Y, Z nuqtalar ketma-ketligini yaratish
x = [0, 1, 2, 3, 4, 5, 6]
y = [0, 1, 4, 9, 16, 25, 36]
z = [0, 1, 4, 9, 16, 25, 36]

# Koordinatalar asosida 3D chiziqli grafik chizish (qizil rangda)
ax.plot3D(x, y, z, 'red')

# 3D tarqoq grafik (scatter plot) chizish
# Nuqtalar rangi z qiymatiga qarab 'cividis' palitrasida o‘zgaradi
ax.scatter3D(x, y, z, c=z, cmap='cividis')

# Grafikni ko‘rsatish
plt.show()
```

**Kodning tushuntirishi:**

*   **Kutubxonalarni yuklash**: Dastlab 3D grafik chizish uchun Python kutubxonasidan `numpy`, `matplotlib.pyplot` va `mpl_toolkits` paketlarini yuklab olamiz.
*   **Shakl yaratish**: `plt.figure()` yordamida bo‘sh shakl (kanva) yaratiladi.
*   **3D proyeksiya**: O‘qlarni aniqlashda proyeksiyaning 3D formatda bo‘lishini ko‘rsatamiz.
*   **Koordinatalar**: Grafik chizish uchun X, Y va Z o‘qlari vazifasini bajaruvchi ixtiyoriy nuqtalardan iborat 3 ta massiv tayyorlab olamiz.
*   **`ax.plot3D()`**: Ushbu metod yordamida berilgan koordinatalar bo‘yicha qizil rangli 3D chiziq chiziladi. Barcha sozlamalar qavs ichida parametr sifatida uzatiladi.
*   **`ax.scatter3D()`**: Xuddi shu qiymatlar asosida tarqoq grafik ham chiziladi. Bunda har bir nuqtaning rangi uning koordinata qiymatlariga qarab o‘zgarib boradi.
*   **Natija**: Nihoyat, `plt.show()` buyrug‘i bilan hosil bo‘lgan 3D tasvir ekranda ko‘rsatiladi.

**2-misol: `figure.gca()` yordamida 3D grafik chizish**

```python
import matplotlib.pyplot as plt
import numpy as np

# Ma'lumotlarni tayyorlash
x = np.arange(-5, 5, 0.6)
y = np.arange(-5, 5, 0.6)
x, y = np.meshgrid(x, y)
z = np.sin(x * np.pi / 2)

fig = plt.figure()

# Mana shu qatorni o'zgartirdik:
ax = fig.add_subplot(projection='3d')

# Grafikni chizish
ax.plot_wireframe(x, y, z, color="red")

plt.show()
```

**3-misol: Bitta shaklda ikkita 3D grafik yaratish**

Ushbu oxirgi misolda biz bitta figura yoki kanva ichida ikkita 3D grafikni (3D nuqtalarimizni) yaratamiz. Quyida kodning amalga oshirilishi keltirilgan:

```python
import matplotlib.pyplot as plt
import numpy as np

# Chizish uchun bo‘sh shakl yaratish
fig = plt.figure()

# 1x2 o‘qdan iborat sub-plot yaratish va 1-grafikni 3D proyeksiyada belgilash
ax = fig.add_subplot(1, 2, 1, projection='3d')

# X va Y o‘qlari uchun 12 ta elementdan iborat oraliq yaratish
X = np.arange(12)
Y = np.linspace(12, 1)

# X va Y ning meshgrid (to‘rli matritsa) shaklini yaratish
X, Y = np.meshgrid(X, Y)

# X va Y asosida ifoda tuzib, uni Z da saqlash
Z = X * 2 + Y * 3

# X, Y va Z qiymatlari to‘plami bilan simli karkas (wireframe) grafigini chizish
ax.plot_wireframe(X, Y, Z)

# 1x2 o‘qdan iborat 2-sub-plotni yaratish va uni 3D proyeksiya deb belgilash
ax = fig.add_subplot(1, 2, 2, projection='3d')

# X, Y va Z uchun nuqtalar to‘plamini aniqlash
X1 = [1, 2, 1, 4, 3, 2, 7, 5, 9]
Y1 = [8, 2, 7, 4, 3, 6, 1, 8, 9]
Z1 = [1, 2, 4, 7, 9, 6, 7, 6, 9]

# X1, Y1, Z1 nuqtalarini yashil rangda chizish
ax.plot(X1, Y1, Z1, color='green')

# Grafikni ko‘rsatish
plt.show()
```

**Izoh:**

*   Keyin ikkita 3D grafik joylashadigan bo‘sh shakl (figure) yaratiladi va birinchi sub-plot hosil qilinadi.
*   Bir qator nuqtalarni o‘z ichiga olgan **X** va **Y** olingan hamda ular yordamida **meshgrid** (koordinata matritsalari) yaratilgan.
*   So‘ngra **X** va **Y** qiymatlari orqali $$Z = X \cdot 2 + Y \cdot 3$$ ifodasi hosil qilinadi. `plot_wireframe` yordamida ushbu nuqtalar 3D o‘qda chiziladi.
*   Shundan so‘ng, ikkinchi grafikning sub-plot parametrlari (1x2 o‘qdagi ikkinchi o‘rin) belgilanadi.
*   **X1**, **Y1** va **Z1** massivlari ko‘rinishidagi turli elementlar yaratiladi.
*   Ushbu nuqtalar yordamida grafik chiziladi va yakunda ikkita 3D grafikni o‘z ichiga olgan umumiy shakl ko‘rsatiladi.

### Matplotlib yordamida asosiy 3D grafiklar: Sirtli (Surface) va Uchburchakli sirtli (Tri-surface) grafiklar

Ushbu bo‘limda sirtli grafik (Surface Plot) va uchburchakli sirtli (Tri-surface Plot) grafik yaratishni ko‘rib chiqamiz. Biz sub-plotlar tushunchasini hamda keng ko‘lamli nuqtalar uchun grafiklar chizishni tushunish maqsadida har ikkala grafikni bitta shaklga (figure) birlashtiramiz. Simli karkasli (wireframe) grafiklardan farqli o‘laroq, ushbu grafiklarning sirtlari rang bilan to‘ldiriladi.

**Misol:**

```python
# Python kutubxonasidan matplotlib.pyplot'ni yuklab olish
import matplotlib.pyplot as plt
# Python kutubxonasidan numpy paketini yuklab olish
import numpy as np

# Chizish uchun bo‘sh shakl yaratish
fig = plt.figure()

# 1x2 o‘qdan iborat sub-plotni aniqlash va uni 3D proyeksiyali birinchi grafik sifatida belgilash
ax = fig.add_subplot(1, 2, 1, projection='3d')

# Grafikda bitta egri chiziq yaratish uchun -1 dan 1 gacha 0.1 qadam bilan qiymatlar yaratish
x1 = np.arange(-1, 1, 0.1)
y1 = np.arange(-1, 1, 0.1)

# n-o‘lchamli massiv yaratish uchun x1 va y1 bilan mesh grid (to‘rli matritsa) hosil qilish
x1, y1 = np.meshgrid(x1, y1)

# Meshgrid qiymatlari asosida kosinus funksiyasini yaratish
z1 = np.cos(x1 * np.pi / 2)

# x1, y1, z1 nuqtalar asosida qizil rangli sirtli grafik yaratish
ax.plot_surface(x1, y1, z1, color="red")

# 1x2 o‘qdan iborat ikkinchi sub-plotni 3D proyeksiyada aniqlash
ax = fig.add_subplot(1, 2, 2, projection='3d')

# X1, Y1 va Z1 uchun nuqtalar to‘plamini aniqlash
X1 = [1, 2, 1, 4, 3, 2, 7, 5, 9]
Y1 = [8, 2, 7, 4, 3, 6, 1, 8, 9]
Z1 = [1, 2, 4, 7, 9, 6, 7, 6, 9]

# Binafsha rangda X1, Y1, Z1 nuqtalarini uchburchakli sirtli grafik ko'rinishida chizish
ax.plot_trisurf(X1, Y1, Z1, color='purple')

# Grafikni ko‘rsatish
plt.show()
```

**Izoh:**

*   Zarur paketlar (**numpy**, **matplotlib.pyplot**) yuklab olingach, `plt.figure()` yordamida bo‘sh shakl yaratiladi.
*   Ushbu shaklga 1, 2, 1 parametrlariga ega 3D proyeksiyali sub-plot qo‘shiladi, bu esa 1x2 o‘lchamdagi maydonda birinchi grafikni yaratish imkonini beradi.
*   Keyin -1 dan 1 gacha bo‘lgan oraliqda 0.1 farq bilan tekis taqsimlangan qiymatlar hosil qilinib, **x1** va **y1** o‘zgaruvchilarida saqlanadi.
*   `meshgrid` yordamida **x1** va **y1** uchun n-o‘lchamli massiv yaratiladi.
*   **x1** qiymatlari asosida kosinus egri chizig‘i yaratilib, natijalar **z1** da saqlanadi.
*   `ax.plot_surface()` yordamida **x1**, **y1** va **z1** nuqtalariga asoslangan qizil rangli sirtli grafik (surface plot) chiziladi.
*   Shundan so‘ng, xuddi shu 1x2 o‘lchamdagi ikkinchi grafik sub-ploti aniqlanadi.
*   **X1**, **Y1** va **Z1** o‘zgaruvchilari orqali nuqtalar to‘plami yaratilib, ushbu ro‘yxatlarda saqlanadi.
*   `ax.plot_trisurf()` yordamida **X1**, **Y1** va **Z1** nuqtalari asosida binafsha rangli uchburchakli sirt grafigi hosil qilinadi.
*   Yakunda `plt.show()` buyrug‘i orqali tayyor bo‘lgan grafik ekranda ko‘rsatiladi.

### Kontur va to‘ldirilgan kontur grafiklari

**Kontur grafigi** — bu doimiy $z$ qatlamlarini (z-slices) chizish orqali 3D grafikni ko‘rsatish usulidir. **To‘ldirilgan kontur** esa kontur grafiklaridagi chiziqlar orasidagi sohalarni rang bilan to‘ldiradi.

*   **Ikkinchi sub-plot**: Xuddi shu o‘lchamdagi ikkinchi grafik maydoni aniqlanadi. 
*   **Ma’lumotlar va ifodalar**: To‘ldirilgan kontur grafigi uchun **X1**, **Y1**, **Z1** o‘zgaruvchilaridagi oddiy qiymatlar diapazoni kifoya qiladi (matnda meshgrid talab qilinmasligi aytilgan bo‘lsa-da, koddagi $Z1$ ifodasi uchun undan foydalanilgan).

### Ko‘pburchakli grafiklar (Polygon plots)

Ushbu bo‘limda biz ko‘pburchakli grafikni (**polygon plot**) o‘rganamiz. Ushbu grafik turi oldingi misollarda chizilgan grafiklardan biroz farq qiladi. Bunda biz z-o‘qining turli nuqtalarida uzluksiz nuqtalar to‘plamini aks ettiramiz. Keling, amalga oshirish bosqichiga o‘tamiz.

**Misol:**

```python
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.collections import PolyCollection
import matplotlib.pyplot as plt
import numpy as np

# Bo‘sh shakl yaratish
fig = plt.figure()

# Xatolikni tuzatish uchun add_subplot'dan foydalanamiz
ax = fig.add_subplot(projection='3d')

# Ma'lumotlarni yaratish
xs = np.arange(0, 1, 0.1)
verts = []
zs = [0.0, 0.2, 0.4, 0.6, 0.8]

for z in zs:
    ys = np.random.rand(len(xs))
    verts.append(list(zip(xs, ys)))

poly = PolyCollection(verts)
ax.add_collection3d(poly, zs=zs, zdir='y')

plt.show()
```

**Izoh:**

*   Ushbu misolda grafikni yaratish uchun biz 4 ta turli paketlarni (Numpy, matplotlib.pyplot, polycollection, Axes3d) import qilishimiz kerak.
*   Shundan so‘ng, biz yuqoridagi grafik chiziladigan bo‘sh shakl (figure) yaratamiz.
*   `arange` va bir qator ixtiyoriy qiymatlardan foydalanib, biz $x$ va $z$ o‘qlari uchun qiymatlar ro‘yxatini hosil qilamiz.
*   Endi biz $z$ bo‘ylab sikl (loop) aylantiramiz, bunda $z$ ning har bir qiymati uchun $y$ o‘qi bo‘yicha $x$ bilan bir xil uzunlikdagi tasodifiy sonlar ro‘yxatini olamiz.
*   Har bir $z$ uchun chiziq bo‘ylab $x$ ga nisbatan $y$ ni chizamiz.
*   `PolyCollection` yordamida biz barcha chizmalarni har bir grafikda bir xil rang bilan birlashtiramiz.
*   `add_collection3d` yordamida grafikning $x-z$ o‘qida chizilishini belgilaymiz va shuningdek, `zs=0` deb aniqlaymiz.
*   Yakunda biz yuqoridagi grafikni ekranda ko‘rsatamiz.

### Quiver plots (Vektorli maydon grafigi)

**Quiver plot** 3D fazoda belgilangan nuqtalarda yo‘naltirilgan strelkalar (vektorlar) maydonini chizish uchun ishlatiladi. Bu grafik turi vektor maydonlarini vizuallashtirishda juda qo‘l keladi va strelkalarni turli usullar bilan sozlash mumkin.


```python
from mpl_toolkits.mplot3d import axes3d
import matplotlib.pyplot as plt
import numpy as np

# 3D grafik chizish uchun bo'sh shakl yaratish
fig = plt.figure()

# Xatolikni tuzatish uchun add_subplot'dan foydalanamiz
ax = fig.add_subplot(projection='3d')

# X, Y, Z qiymatlari uchun meshgrid yaratish
x, y, z = np.meshgrid([1,2,5,2,4,8,3,3,1],
                      [6,4,3,1,6,2,7,8,2],
                      [1,2,5,2,4,8,3,3,1])

# Yo'nalish vektorlari
u = x*2 + y*3 + z*3
v = (x+3) * (y+5) * (z+7)
w = x + y + z

# Vektorlar maydonini chizish
ax.quiver(x, y, z, u, v, w, length=0.2, normalize=True)

plt.show()
```

**Izoh:**
*   Yuqoridagi kodda biz barcha zarur paketlarni (matplotlib.pyplot, numpy, mpl_toolkit.mplot3d) eng yuqori qismda yuklab olyapmiz.
*   So‘ngra `plt.figure()` yordamida bo‘sh shakl (figure) yaratyapmiz.
*   `fig.gca(projection='3d')` yordamida esa proyeksiyani 3D formatida, ya’ni 3D chizish deb belgilayapmiz.
*   `meshgrid` yordamida X, Y va Z o‘qlari bilan N-o‘lchamli massiv hosil qilyapmiz.
*   Quiver grafigida shunchaki nuqtalarni chizishning o‘zi kifoya emas.
*   Biz vektor yo‘nalishlarining komponentlarini aniq ko‘rsatishimiz kerak.
*   Shuning uchun biz X, Y va Z yordamida turli ifodalar orqali shakllangan $u$, $v$, $z$ (kodda $w$) qiymatlarini olyapmiz.
*   `ax.quiver()` yordamida quiver grafigini chizyapmiz; bunda har bir vektor (quiver) uzunligini 0.1 deb belgilaymiz va strelkalar bir xil uzunlikda bo‘lishi uchun `normalize` parametrini `True` deb aniqlaymiz.

### 3D grafiklarda chiziladigan 2D ma'lumotlar

Ushbu bo‘limda biz 3D grafikning ma'lum bir o‘qi bo‘ylab chiziladigan 2D nuqtalar to‘plamini olamiz. Chunki 3D tekislikda barcha koordinatalarga ega bo‘lmagan 2D nuqtani chizish imkonsiz, shuning uchun 2D nuqtalarni joylashtirish uchun aniq bir o‘qni belgilash lozim.

```python
import numpy as np
import matplotlib.pyplot as plt

# Bo'sh kanva yaratish
fig = plt.figure()

# Tuzatilgan qator: add_subplot yordamida 3D o'qlarni aniqlash
ax = fig.add_subplot(projection='3d')

# O'qlarni nomlash
ax.set_xlabel('X-Axis')
ax.set_ylabel('Y-Axis')
ax.set_zlabel('Z-Axis')

# X va Y uchun qiymatlar
x = [1, 1.2, 1.4, 1.6, 1.8, 2.0, 2.2, 2.4, 2.6, 2.8]
y = [1, 1.2, 1.4, 1.6, 1.8, 2.0, 2.2, 2.4, 2.6, 2.8]

# zdir='z' barcha nuqtalarni zs=0 koordinatasiga mahkamlaydi 
# va (x,y) nuqtalar x-y tekisligida chiziladi
ax.plot(x, y, zs=0, zdir='z')

# zdir='y' barcha nuqtalarni zs=0 koordinatasiga mahkamlaydi 
# va (x,y) nuqtalar x-z tekisligida chiziladi
ax.plot(x, y, zs=0, zdir='y')

plt.show()
```

**Izoh:**

*   Ushbu misolda biz Matplotlib yordamida 3D grafikda 2D ma'lumotlarni aks ettiryapmiz.
*   Buning uchun qiymatlar to‘plamini yaratish va ularni 3D proyeksiyada chizish maqsadida bizga **NumPy** va **matplotlib.pyplot** paketlari kerak bo‘ladi.
*   Barcha zarur paketlarni import qilgandan so‘ng, biz X, Y va Z o‘qlari uchun turli xil qiymatlar diapazonini yaratyapmiz.
*   Endi `ax.plot()` funksiyasi yordamida 2D nuqtalarimizni chizyapmiz; bunda `ax.plot()` ichida parametr sifatida chizilishi kerak bo‘lgan ikkita ma'lumotlar to‘plamini va ular qaysi o‘q (tekislik) bo‘ylab joylashishini ko‘rsatib o‘tamiz.
*   Kodingizda `zs=0` va `zdir` parametrlaridan foydalanilgan: `zdir='z'` nuqtalarni **x-y** tekisligiga, `zdir='y'` esa nuqtalarni **x-z** tekisligiga joylashtiradi.
