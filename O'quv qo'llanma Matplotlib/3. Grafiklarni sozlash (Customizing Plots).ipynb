## III BOB. Grafiklarni sozlash (Customizing Plots)

### 3.1. Markerlar (Markers)

Matplotlib’dagi **markerlar (markers)** moduli grafikdagi alohida ma’lumot nuqtalarini ajratib ko‘rsatishga yordam beradi, bu esa o‘qishni osonlashtiradi va vizual jozibadorlikni oshiradi. Turli xil marker uslublari yordamida foydalanuvchilar ma’lumotlar seriyalarini bir-biridan farqlash va muhim nuqtalarga urg‘u berish uchun grafikni moslashtirishlari mumkin.

Ushbu tushunchani tushunish uchun doira shaklidagi markerlarga ega chiziqli grafik chizilgan quyidagi oddiy misolni ko‘rib chiqing:

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# marker='o‘ orqali har bir nuqtaga doira qo‘yiladi
plt.plot(x, y, marker='o‘, linestyle='-', label='Ma’lumot nuqtalari')

plt.title('Markerlarga ega oddiy grafik')
plt.xlabel('X o‘qi')
plt.ylabel('Y o‘qi')
plt.legend()
plt.show()
```

**Izoh:**
Ushbu misolda chiziqdagi har bir nuqta doira (`'o‘`) bilan belgilangan bo‘lib, bu alohida ma’lumot nuqtalarini aniqlashni osonlashtiradi.

### Matplotlib Markerlar Moduli (Matplotlib.markers)

`matplotlib.markers` moduli Matplotlib'da markerlar bilan ishlash funksiyalarini taqdim etadi. Har bir marker o‘ziga xos belgi yoki belgi bilan bog‘langan bo‘lib, u grafikda ma’lumot nuqtasi qanday ko‘rinishini ifodalaydi. Bu moslashuvchanlik foydalanuvchilarga o‘z afzalliklari yoki talablariga muvofiq grafiklarni shaxsiylashtirish imkonini beradi.



Quyidagi jadvalda Matplotlib'da eng ko‘p qo‘llaniladigan markerlar va ularning tavsifi keltirilgan:

| Marker (Simvol) | Shakli | Tavsifi (Description) |
| :--- | :---: | :--- |
| **"."** | • | Nuqta (Point) |
| **"o"** | ⬤ | Doira (Circle) |
| **"v"** | ▼ | Pastga qaragan uchburchak (Triangle down) |
| **"^"** | ▲ | Tepaga qaragan uchburchak (Triangle up) |
| **"s"** | ■ | Kvadrat (Square) |
| **"p"** | ⬟ | Beshburchak (Pentagon) |
| **"*"** | ★ | Yulduzcha (Star) |
| **"8"** | ⬢ | Sakkizburchak (Octagon) |


Keling, har bir nuqtani yulduzcha ramzi bilan belgilaymiz:

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# marker='*' yulduzcha markerini qo‘shadi
# linestyle='--' chiziqni punktir (uzuq-uzuq) ko‘rinishiga keltiradi
# color='r' rangni qizil qiladi
plt.plot(x, y, marker='*', linestyle='--', color='r', label='Ma’lumot nuqtalari')

plt.title('Markerlar bilan o‘zgartirilgan grafik')
plt.xlabel('X o‘qi')
plt.ylabel('Y o‘qi')
plt.legend()
plt.grid(True) # To‘r chiziqlarini yoqish
plt.show()
```

**Izoh:**
Ushbu kodda chiziqdagi har bir nuqta yulduzcha (`'*'`) shaklida ifodalangan. Shuningdek, `linestyle='--'` yordamida chiziq uzuq-uzuq qilingan va `plt.grid(True)` orqali grafik orqasiga yordamchi to‘r chiziqlari qo‘shilgan. Bu kabi o‘zgarishlar ma’lumotlarni yanada aniqroq tahlil qilish imkonini beradi.

### Bir nechta marker shakllarini vizuallashtirish

Ushbu kod Matplotlib’dagi turli xil markerlarni bitta grafikda namoyish etadi. U marker uslublari ro‘yxati bo‘ylab takrorlanadi (sikl yordamida) va ularni bir xil X o‘qi bo‘ylab, lekin har xil Y qiymatlarida joylashtiradi.


```python
# X o‘qi uchun 1 dan 10 gacha bo‘lgan sonlar
x = range(1, 11)

# Marker uslublari ro‘yxati
markers = ['o‘, 's', '^', 'v', 'D', '*', '+', 'x']

# Har bir markerni sikl orqali chizish
for i, marker in enumerate(markers):
    # [i*2]*10 har bir marker turi uchun alohida gorizontal chiziq yaratadi
    # linestyle='' chiziqni yashirib, faqat markerlarni qoldiradi
    plt.plot(x, [i*2]*10, marker=marker, linestyle='', label=f"Marker: {marker}")

plt.title('Turli xil Matplotlib markerlari')
plt.xlabel('X o‘qi')
plt.ylabel('Y o‘qi (Marker tartibi)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') # Izohni grafik tashqarisiga chiqarish
plt.tight_layout()
plt.show()
```

**Izoh:**
* `enumerate(markers)` funksiyasi ham marker belgisini, ham uning tartib raqamini (`i`) qaytaradi.
* `[i * 2] * 10` ifodasi har bir marker turi uchun o‘ziga xos balandlikni (Y koordinatasini) belgilaydi.
* `linestyle=''` parametri juda muhim — u nuqtalarni birlashtiruvchi chiziqlarni olib tashlaydi va faqat markerlarning o‘zini ko‘rsatadi.
* Bu usul yordamida siz loyihangiz uchun eng mos keladigan marker shaklini osongina tanlab olishingiz mumkin.

### Matplotlib-da `fmt` parametridan foydalanish

Matplotlib-dagi `fmt` (format string) parametri marker, chiziq uslubi va rangni bitta qisqa satr yordamida belgilashning juda qulay usulidir. Bu usul kodni qisqartiradi va grafik ko‘rinishini tezkor sozlash imkonini beradi.

**Sintaksis:** `fmt = '[marker][line][color]'`

* **Marker:** Nuqtalarning uslubini belgilaydi (masalan, `'o‘` — doira, `'s'` — kvadrat).
* **Line (Chiziq):** Chiziq uslubini belgilaydi (masalan, `'-'` — uzluksiz chiziq, `':'` — punktir/nuqtali chiziq).
* **Color (Rang):** Grafik elementlarining rangini belgilaydi (masalan, `'r'` — qizil, `'b'` — ko‘k).



Keling, misolni ko‘rib chiqamiz:

```python
x = [4, 1, 7, 5, 8]

# 'o-r' formati: 'o‘ (doira marker), '-' (to‘g‘ri chiziq), 'r' (qizil rang)
plt.plot(x, 'o-r') 

plt.title('fmt parametri yordamida chizilgan grafik')
plt.xlabel('X o‘qi')
plt.ylabel('Y o‘qi')

plt.show()
```

**Izoh:**
`plt.plot(x, 'o-r')` kodi uchta alohida parametrni (`marker='o‘`, `linestyle='-'`, `color='r'`) bitta kichik format satriga birlashtiradi. Bu usul ayniqsa tezkor prototiplash va oddiy vizualizatsiyalar uchun juda samaralidir.

**Foydali format birikmalari:**
* `'g*--'` : Yashil yulduzchali markerlar va uzuq-uzuq chiziq.
* `'bs:'` : Ko‘k kvadrat markerlar va nuqtali chiziq.
* `'mD-'` : Magenta (to‘q qizil) rangli romb markerlar va uzluksiz chiziq.

Matplotlib-dagi chiziq va ranglar qo‘llanmasi grafik chiziqlari va markerlarning ko‘rinishini moslashtirish imkonini beradi. Vizual aniqlik va estetikani yaxshilash uchun chiziq uslubi va rangini tezda sozlashingiz mumkin.

#### Chiziq uslubini vizuallashtirish

### 3.2. Yorliqlarni qo‘shish (Adding Labels)

### Ustunli diagrammaga (Bar Chart) qiymat belgilarini qo‘shish

**Ustunli diagramma** — bu turli balandlikdagi ustunlar yordamida ma’lumotlarning grafik tasviridir. U turli ma’lumotlar to‘plamlarini vizual ravishda solishtirish uchun keng qo‘llaniladi. Biroq, Matplotlib sukut bo‘yicha (default) har bir ustunda qiymat belgilarini ko‘rsatmaydi, bu esa individual ustunlar ifodalagan aniq qiymatlarni tahlil qilishni qiyinlashtirishi mumkin.

Qiymat belgilarini qo‘shish diagrammaning o‘qilishini (readability) yaxshilaydi va ma’lumotlarni talqin qilishni osonlashtiradi.

### Qiymat belgilarini qo‘shish usullari

Matplotlib-da ustunli diagrammaga qiymat belgilarini qo‘shishda asosan ikkita funksiyadan foydalaniladi:
1.  **`plt.bar()`** — ustunli diagrammani yaratish uchun.
2.  **`plt.text()`** — har bir ustunning ustiga yoki ichiga matnli yorliqlarni qo‘shish uchun.

### 1. Ustunli diagramma yaratish: `plt.bar()`



Ustunli diagrammani chizish uchun `plt.bar()` funksiyasidan foydalaniladi.

**Sintaksis:**
`plt.bar(x, height, color)`

**Parametrlar:**
*   **`x`**: X o‘qida joylashadigan ma’lumotlar qiymatlari.
*   **`height`**: Y o‘qida rejalashtirilgan ma’lumotlar qiymatlari (ustunlar balandligi).
*   **`color`**: Ustunlarning rangini belgilaydi.

### 2. Qiymat belgilarini qo‘shish: `plt.text()`



Ustunlar ustiga matnli yorliqlarni qo‘shish uchun `plt.text()` funksiyasi ishlatiladi.

**Sintaksis:**
`plt.text(x, y, s, ha, bbox)`

**Parametrlar:**
*   **`x, y`**: Matn ko‘rsatiladigan joyning koordinatalari.
*   **`s`**: Ko‘rsatiladigan satr (ya’ni qiymat belgisi).
*   **`ha`**: Gorizontal tekislash (Horizontal alignment) — `'center'`, `'left'` yoki `'right'` bo‘lishi mumkin.
*   **`bbox`**: Matn yaxshi ko‘rinishi uchun uning atrofidagi to‘rtburchak quti.

### Amaliy misol

Quyida ushbu usullarni birlashtirgan holda kod qanday ko‘rinishi keltirilgan:

```python
# Ma'lumotlar
x = ['A', 'B', 'C', 'D']
y = [10, 25, 15, 30]

# Ustunli diagrammani yaratish
plt.bar(x, y, color='skyblue')

# Har bir ustun ustiga qiymatni qo‘shish
for i in range(len(x)):
    plt.text(i, y[i], str(y[i]), ha='center', va='bottom')

plt.title("Qiymat belgilari bilan ustunli diagramma")
plt.show()
```

Ushbu uslub yordamida siz diagrammangizni nafaqat chiroyli, balki tahlil uchun ancha foydali holatga keltira olasiz. Diagrammadagi har bir ustun qanday qiymatga ega ekanligi endi bir qarashda aniq bo‘ladi.


Ustunli diagrammaga qiymat belgilarini qo'shish ma'lumotlarni aniq tahlil qilish uchun juda muhimdir. Quyida ushbu jarayonning bosqichma-bosqich tushuntirilishi keltirilgan:

### Qiymat belgilarini qo'shish bosqichlari

**1-bosqich: Kerakli kutubxonalarni import qilish**
Ishni boshlash uchun Matplotlib kutubxonasini import qilib olamiz:
`import matplotlib.pyplot as plt`

**2-bosqich: Qiymat belgilarini qo'shish uchun funksiya yaratish**
Har bir ustunning tepasiga raqamli qiymatlarni joylashtirish uchun yordamchi funksiya yaratiladi. Bu funksiya:
*   X o'qi qiymatlari (toifalar) va ularga mos keladigan Y o'qi qiymatlari (ustun balandliklari) bo'yicha takrorlanadi (iterate).
*   Yorliqlarni tegishli balandliklarda joylashtirish uchun Matplotlib-ning `text()` funksiyasidan foydalanadi.
*   Yorliqlarni ustunlarning yuqori qismiga yoki o'rtasiga joylashtirish kabi sozlamalarni qo'llab-quvvatlaydi.
*   Yaxshiroq ko'rinishi uchun yorliqlarni ajratilgan quti ichiga olish imkoniyatiga ega.

**3-bosqich: Asosiy funksiyani yaratish**
Ushbu bosqichda:
*   Ma'lumotlar to'plami (toifalar va ularga mos raqamli qiymatlar) aniqlanadi.
*   To'g'ri vizuallashtirishni ta'minlash uchun oyna (figure) o'lchami belgilanadi.
*   Tanlangan rang sxemasi bilan `bar()` funksiyasi yordamida diagramma chiziladi.
*   Ustunlar ustiga qiymatlarni qo'shish uchun avval yaratilgan funksiya chaqiriladi.
*   Sarlavha, X va Y o'qi nomlari kabi qo'shimcha elementlar qo'shiladi.

**1-misol: Qiymat belgilarini ustunlar ustiga qo'shish (standart sozlama)**

```python
# Ustunlar ustiga qiymat belgilarini qo'shish uchun funksiya
def add_labels(x, y):
    for i in range(len(x)):
        # Matnni ustunning aynan balandligi (y[i]) bo'yicha joylashtirish
        plt.text(i, y[i], y[i]) 

# Diagramma uchun ma'lumotlar
x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]

# Ustunli diagrammani yaratish
plt.bar(x, y)

# Qiymat belgilarini qo'shish funksiyasini chaqirish
add_labels(x, y)

# Sarlavha va o'q nomlarini qo'shish
plt.title("Kollejga qabul")
plt.xlabel("Kurslar")
plt.ylabel("Qabul qilinganlar soni")

# Diagrammani ko'rsatish
plt.show()
```

**Eslatma:** Ushbu misolda raqamlar ustunlarning eng yuqori nuqtasida paydo bo'ladi. Matn biroz yuqoriroqda yoki markazda bo'lishi uchun `ha='center'` (gorizontal tekislash) va `va='bottom'` (vertikal tekislash) kabi parametrlarni qo'shish tavsiya etiladi.

**Izoh:** `add_labels(x, y)` funksiyasi ma’lumotlar bo‘ylab takrorlanish (iteratsiya) va qiymatlarni o‘z balandliklarida joylashtirish uchun `plt.text(i, y[i], y[i])` buyrug‘idan foydalanish orqali har bir ustun ustiga raqamli yorliqlarni qo‘shadi. Tushunarli bo‘lishi uchun yorliqlar sukut bo‘yicha markazga tekislanadi.

**2-misol: Qiymat belgilarini ustunlar ustiga markazlashtirish**

Ushbu misolda biz `ha='center'` parametridan foydalanib, raqamlarni ustun markaziga nisbatan tekislaymiz. Bu usul grafikni yanada o‘qishli qiladi.


```python
# Qiymat belgilarini markazlashtirib qo‘shish funksiyasi
def add_labels(x, y):
    for i in range(len(x)):
        # ha='center' matnni markaz bo‘ylab tekislaydi
        plt.text(i, y[i], y[i], ha='center')

# Ma’lumotlar to‘plami
x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]

# Grafik o‘lchamini belgilash
plt.figure(figsize=(10, 5))

# Ustunli diagrammani yaratish
plt.bar(x, y)

# Qiymat belgilarini qo‘shish
add_labels(x, y)

# Sarlavha va o‘qlarga nom berish
plt.title("College Admission")
plt.xlabel("Courses")
plt.ylabel("Number of Admissions")

plt.show()
```

**Izoh:** 
*   `add_labels(x, y)` funksiyasi o‘qishni osonlashtirish uchun `ha='center'` yordamida raqamli belgilarni ustunlar markaziga joylashtiradi. 
*   `plt.figure(figsize=(10, 5))` grafik o‘lchamini o‘rnatadi va `plt.bar(x, y)` ustunlarni chizadi. 
*   Yorliqlar `plt.text(i, y[i], y[i])` orqali o‘z balandliklariga joylashtirilib, ma’lumotlarning aniqligi ta’minlanadi.

**3-misol: Qiymat belgilarini to‘rtburchak quti ichida ko‘rsatish**

Ba’zida ustunli diagramma rangi va matn rangi bir-biriga yaqin bo‘lib qolsa, raqamlarni o‘qish qiyinlashadi. Bunday holatlarda `bbox` parametridan foydalanib, qiymatlarni maxsus quti ichiga olish ma’lumotlarning ko‘rinuvchanligini oshiradi.


```python
# Qiymat belgilarini quti ichida qo‘shish funksiyasi
def add_labels(x, y):
    for i in range(len(x)):
        # facecolor='red' - quti rangi qizil
        # alpha=0.8 - qutining shaffoflik darajasi
        plt.text(i, y[i], y[i], ha='center', 
                 bbox=dict(facecolor='red', alpha=0.8))

# Ma’lumotlar to‘plami
x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]

# Grafik o‘lchamini belgilash
plt.figure(figsize=(10, 5))

# Ustunli diagrammani yaratish
plt.bar(x, y)

# Qiymat belgilarini qo‘shish
add_labels(x, y)

# Sarlavha va o‘qlarga nom berish
plt.title("College Admission")
plt.xlabel("Courses")
plt.ylabel("Number of Admissions")

plt.show()
```

**Izoh:** 
* `add_labels(x, y)` funksiyasi ko‘rinuvchanlikni yaxshilash uchun raqamli belgilarni **qizil rangli qutilar** ichiga joylashtiradi. 
* `plt.bar(x, y)` diagrammani yaratadi va `plt.figure(figsize=(10, 5))` uning hajmini moslashtiradi. 
* Yorliqlar `ha='center'` yordamida markazlashtirilgan holda ko‘rsatiladi. 
* `bbox` ichidagi `alpha` parametri qutining rangini biroz shaffof qilib, grafik elementlari to‘liq yopilib qolmasligini ta’minlaydi.

**4-misol: Qiymat belgilarini har bir ustunning markaziga joylashtirish**

Ayrim hollarda, ayniqsa ustunlar juda baland bo‘lganda, qiymat belgilarini ustunning tepasida emas, balki uning ichida (markazida) ko‘rsatish qulayroq bo‘ladi. Bunda matn koordinatasini belgilashda ustun balandligining yarmi olinadi.



```python
# Qiymat belgilarini har bir ustunning markaziga joylashtirish funksiyasi
def add_labels(x, y):
    for i in range(len(x)):
        # y[i] // 2 orqali matn ustun balandligining yarmiga joylashtiriladi
        plt.text(i, y[i] // 2, y[i], ha='center')  

# Ma’lumotlar to‘plami
x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]

# Grafik o‘lchamini belgilash
plt.figure(figsize=(10, 5))

# Ustunli diagrammani yaratish
plt.bar(x, y)

# Qiymat belgilarini qo‘shish
add_labels(x, y)

# Sarlavha va o‘qlarga nom berish
plt.title("College Admission")
plt.xlabel("Courses")
plt.ylabel("Number of Admissions")

plt.show()
```

**Izoh:**
*   `add_labels(x, y)` funksiyasi yaxshiroq tekislash uchun `y[i] // 2` formulasidan foydalanib, raqamli yorliqlarni ustun balandligining yarmiga joylashtiradi.
*   `plt.bar(x, y)` diagrammani hosil qiladi va `plt.figure(figsize=(10, 5))` uning hajmini o‘rnatadi.
*   Yorliqlar `ha='center'` yordamida gorizontal markazlashtiriladi.
*   Bu usul diagramma tepasida bo‘sh joy kam bo‘lgan holatlarda ma’lumotlarni tartibli ko‘rsatishga yordam beradi.

**5-misol: To‘rtburchak quti ichida markazlashtirilgan yorliqlar**

Ushbu misol avvalgi usullarni birlashtirgan holda, qiymat belgilarini ustunlar markaziga joylashtiradi va ularni yarim shaffof oq qutichaga oladi. Bu usul ayniqsa to‘q rangli ustunlar (masalan, ko‘k) fonida raqamlarning aniq ko‘rinishini ta’minlash uchun juda samarali hisoblanadi.

```python
# Qiymat belgilarini markazda va quti ichida ko'rsatish funksiyasi
def add_labels(x, y):
    for i in range(len(x)):
        # y[i] // 2 - balandlikning yarmi
        # bbox - oq rangli, 0.5 shaffoflikdagi quticha
        plt.text(i, y[i] // 2, y[i], ha='center', 
                 bbox=dict(facecolor='white', alpha=0.5))

x = ["Engineering", "BSc", "MBA", "Bcom", "BBA", "MSc"]
y = [9330, 4050, 3030, 5500, 8040, 4560]

# Grafik o'lchamini belgilash
plt.figure(figsize=(10, 5))

# Ko'k rangli ustunli diagramma yaratish
plt.bar(x, y, color='blue')

# Qiymat belgilarini qo'shish
add_labels(x, y)

# Sarlavha va o'qlarga nom berish
plt.title("College Admission")
plt.xlabel("Courses")
plt.ylabel("Number of Admissions")

plt.show()
```

**Izoh:**
*   `add_labels(x, y)` funksiyasi raqamli belgilarni ustun balandligining yarmiga (`y[i] // 2`) yarim shaffof oq quticha ichida joylashtiradi.
*   `plt.bar(x, y, color='blue')` ko‘k rangli ustunlarni hosil qiladi.
*   `plt.figure(figsize=(10, 5))` grafik o‘lchamini moslashtirib beradi.
*   `ha='center'` parametri matnning gorizontal ravishda ustun o‘rtasiga tushishini kafolatlaydi.

### 3.3. To‘rlarni sozlash (Configuring Grid)

Matplotlib'dagi **to‘r chiziqlari (grids)** ma’lumot nuqtalarini aniqroq ko‘rish uchun asos vazifasini o‘taydigan kesishuvchi chiziqlardir. Ular grafikning o‘qilishini va ma’lumotlarni tahlil qilishni sezilarli darajada yaxshilaydi. Pyplot modulidagi `grid()` funksiyasi grafikdagi ushbu to‘r chiziqlarini sozlash uchun ishlatiladi.

### Sintaksis
`matplotlib.pyplot.grid(b=True, color='grey', linewidth=1.4, axis='', linestyle='-.')`

**Parametrlar:**
*   **`b` (True/False):** To‘r chiziqlari ko‘rsatilishini belgilaydi.
*   **`color`:** To‘r chiziqlarining rangi.
*   **`linewidth`:** Chiziqlarning qalinligi.
*   **`axis`:** To‘r qaysi o‘qqa qo‘llanilishi (`'x'`, `'y'` yoki `'both'`).
*   **`linestyle`:** Chiziq uslubi (masalan, uzuq-uzuq yoki nuqtali).

### Grafikga to‘r chiziqlarini qo‘shish
`grid()` funksiyasi yordamida to‘rni yoqish yoki o‘chirish, asosiy yoki yordamchi to‘r chiziqlarini ko‘rsatish hamda ularning ko‘rinishini (rangi, qalinligi, uslubi) moslashtirish mumkin.

**1-misol: To‘rli va to‘rsiz grafiklar farqi**

Ushbu misol to‘r chiziqlari grafikni qanchalik tushunarli qilishini ko‘rsatadi:

```python
# Ma’lumotlarni tayyorlash
x1 = np.linspace(0, 5, 100)
y1 = np.cos(2 * np.pi * x1) * np.exp(-x1)

# Ikkita yonma-yon grafik yaratish
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Birinchi grafik: to‘rsiz
ax1.plot(x1, y1)
ax1.set_title('To‘rsiz grafik')

# Ikkinchi grafik: to‘r bilan
ax2.plot(x1, y1)
ax2.set_title("To‘rli grafik")
ax2.grid(True) # To‘r chiziqlarini yoqish

plt.show()
```

**Izoh:**
*   **ax1:** Grafikni to‘r chiziqlarsiz ko‘rsatadi.
*   **ax2.grid(True):** Ikkinchi grafik uchun to‘r chiziqlarini faollashtiradi, natijada ma’lumotlar nuqtalarini koordinata o‘qlari bilan solishtirish osonlashadi.

### To‘r chiziqlarini moslashtirish

Vizual tushunarlilikni oshirish maqsadida to‘r chiziqlarining rangi (`color`), qalinligi (`linewidth`) va uslubini (`linestyle`) o‘zgartirishingiz mumkin.

**2-misol: Moslashtirilgan to‘r chiziqlari**


Ushbu misol Matplotlib-da to‘r chiziqlarining ko‘rinishini qanday qilib batafsil sozlash mumkinligini ko‘rsatib beradi:

```python
# 0 dan 2π gacha bo‘lgan 400 ta nuqta yaratish
x = np.linspace(0, 2 * np.pi, 400)
# x qiymatlari kvadratining sinusini hisoblash
y = np.sin(x ** 2)

plt.plot(x, y, 'green')
plt.title("Moslashtirilgan to‘r chiziqlariga ega grafik")

# To‘r chiziqlarini kulrang, qalinligi 1.4 va chiziq-nuqta uslubida ko‘rsatish
plt.grid(True, color='grey', linewidth=1.4, linestyle='-.')

plt.show()
```

**Izoh:**

*   **`plt.grid(True, color='grey', linewidth=1.4, linestyle='-.')`:** To‘r chiziqlarini yoqadi va ularni kulrang, 1.4 qalinlikda hamda chiziq-nuqta (`-.`) uslubida aks ettiradi.

Grafikdagi to‘r chiziqlarining bu kabi aniq sozlanishi, ayniqsa murakkab matematik funksiyalarni yoki zich ma’lumotlar to‘plamini tahlil qilishda yordam beradi.

### 3.4. Quyi grafiklarni yaratish (Creating Subplots)

**Matplotlib** — bu Python dasturlash tilida ma’lumotlarni vizuallashtirish uchun keng qo‘llaniladigan kutubxona bo‘lib, u turli xil grafiklar yaratish uchun kuchli vositalarni taqdim etadi. Matplotlib-ning eng foydali xususiyatlaridan biri — bu `plt.subplots()` metodi yordamida bitta oyna (figure) ichida bir nechta grafiklar (subplots) yaratish imkoniyatidir. Bu foydalanuvchilarga bir-biriga bog‘liq bo‘lgan bir nechta vizuallashtirishni yonma-yon ko‘rsatish imkonini beradi, bu esa ma’lumotlar tahlilini yanada chuqurroq va samaraliroq qiladi.

### Matplotlib-da subplots nima?

Matplotlib-dagi **subplots** deganda bitta rasm (figure) ichida joylashtirilgan bir nechta grafiklar tushuniladi. `matplotlib.pyplot.subplots()` metodi bir nechta grafikli oynani yaratishning oson usulini taqdim etadi. Qatorlar va ustunlar soni berilganda, u `(fig, ax)` ko‘rinishidagi kortejni (tuple) qaytaradi:
*   **`fig`**: Butun bir rasm obyekti.
*   **`ax`**: Alohida kichik grafiklarni (subplots) ifodalovchi o‘qlar obyektlarining massivi.

### `plt.subplots()` yordamida kichik grafiklar (subplots) yaratish

Matplotlib-da kichik grafiklarni yaratishning turli usullari mavjud. Quyida ularning qo‘llanilishini ko‘rsatuvchi misollar keltirilgan.

#### 1. Grafiklar to‘rini (grid) yaratish

Ushbu misol 3x3 o‘lchamdagi kichik grafiklar to‘rini yaratadi va har birining ustida takrorlanish (iteratsiya) orqali tasodifiy chiziqlarni chizadi.



```python
# Kichik grafiklarni (subplots) yaratish
# 3 qator va 3 ustundan iborat to'r
fig, ax = plt.subplots(3, 3)

# Har bir subplot ichida tasodifiy ma’lumotlarni chizish
for row in ax:
    for col in row:
        # NumPy yordamida 0 dan 5 gacha bo‘lgan tasodifiy sonlar olish
        col.plot(np.random.randint(0, 5, 5), np.random.randint(0, 5, 5))

plt.show()
```

**Izoh:**
*   Ushbu kod Matplotlib yordamida 3x3 o‘lchamdagi kichik grafiklar to‘rini yaratadi.
*   So‘ngra u har bir subplot bo‘ylab takrorlanadi va NumPy yordamida tasodifiy butun sonli ma’lumotlarni chizadi.
*   Har bir subplot o‘ziga xos tasodifiy qiymatlar to‘plamini qabul qiladi, bu esa to‘r ichida bir-biridan farq qiluvchi grafiklarni hosil qiladi.

Ushbu uslub yordamida siz juda ko‘p sonli grafiklar bilan ishlayotganda har biriga alohida kod yozib o‘tirmasdan, sikllar yordamida ularni avtomatik tarzda shakllantirishingiz mumkin.

### 2. Turli matematik funksiyalar uchun kichik grafiklar yaratish

Matplotlib yordamida bitta oyna ichida bir nechta matematik funksiyalarni yonma-yon ko‘rsatish ma’lumotlarni qiyosiy tahlil qilish uchun juda qulaydir. Quyidagi misolda 2x2 o‘lchamdagi to‘r yaratilib, unda sinus, kosinus, tangens va sink funksiyalari turli chiziq uslublari hamda ranglarda tasvirlanadi.


```python
# Kichik grafiklarni (subplots) yaratish: 2 qator va 2 ustun
fig, ax = plt.subplots(2, 2)

# Ma’lumotlarni shakllantirish
x = np.linspace(0, 10, 1000)

# Har bir subplot uchun funksiyalarni turli uslublarda chizish
ax[0, 0].plot(x, np.sin(x), 'r-.', label='sin(x)') # Qizil chiziq-nuqta
ax[0, 1].plot(x, np.cos(x), 'g--', label='cos(x)') # Yashil uzuq-uzuq chiziq
ax[1, 0].plot(x, np.tan(x), 'y-', label='tan(x)')  # Sariq tekis chiziq
ax[1, 1].plot(x, np.sinc(x), 'c.-', label='sinc(x)') # Havorang nuqtali chiziq

# Har bir grafik uchun izoh (legend) qo‘shish va oynani ko‘rsatish
for axes in ax.flat:
    axes.legend()

# Grafiklar orasidagi masofani avtomatik sozlash
plt.tight_layout()
plt.show()
```

**Kodni o‘rganish:**
*   **Grid yaratish:** Matplotlib yordamida 2×2 o‘lchamdagi kichik grafiklar to‘ri yaratiladi.
*   **Funksiyalar:** Sinus, kosinus, tangens va sink kabi turli matematik funksiyalar NumPy orqali yaratilgan `x` qiymatlari asosida hisoblanadi.
*   **Uslublar:** Har bir funksiya o‘ziga xos chiziq uslubi (linestyle) va rangga ega bo‘lgan maxsus subplotga biriktiriladi.
*   **Joylashuv:** `plt.tight_layout()` funksiyasi grafiklar va yozuvlar bir-birining ustiga chiqib ketmasligini ta’minlaydi.

### 3. Kichik grafiklarda chiziqli grafiklar (Line plots)

Ushbu misol sinus, kosinus va tangens funksiyalarini hosil qiladi va ularni alohida kichik grafiklarda (subplots) chizib ko‘rsatadi.

```python
# Kichik grafiklar uchun ma’lumotlar hosil qilish
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

# Chiziqli grafiklar bilan subplotlar yaratish
# nrows=1, ncols=3 - bitta qatorda uchta ustun
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(12, 4))

axes[0].plot(x, y1, color='blue', label='sin(x)')
axes[1].plot(x, y2, color='green', label='cos(x)')
axes[2].plot(x, y3, color='orange', label='tan(x)')

# Sarlavhalar va izohlarni (legends) qo‘shish
axes[0].set_title('Sine Function')
axes[1].set_title('Cosine Function')
axes[2].set_title('Tangent Function')

for ax in axes:
    ax.legend()

# Yaxshiroq masofa uchun joylashuvni sozlash
plt.tight_layout()

# Shaklni ko‘rsatish
plt.show()
```

**Izoh:** Ushbu kod Matplotlib yordamida bir qatorli va uch ustunli subplot joylashuvini yaratadi hamda `sin(x)`, `cos(x)` va `tan(x)` funksiyalarini turli ranglarda chizadi. Aniqlik uchun sarlavhalar va izohlar qo‘shiladi, `plt.tight_layout()` esa grafik ko‘rsatilishidan oldin tegishli masofalarni ta’minlaydi.

### 4. Kichik grafiklarda ustunli diagrammalar (Bar plots)

Ushbu misolda `DataFrame` yaratiladi va kategoriyali ma’lumotlarni vizuallashtirish uchun uchta ustunli diagramma hosil qilinadi.


```python
import pandas as pd

# Tasodifiy kategoriyali ma’lumotlar bilan DataFrame yaratish
data = {
    'Category': ['A', 'B', 'C', 'D'],
    'Value1': np.random.randint(1, 10, 4),
    'Value2': np.random.randint(1, 10, 4),
    'Value3': np.random.randint(1, 10, 4)
}
df = pd.DataFrame(data)

# Ustunli diagrammalar bilan subplotlar yaratish
fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(12, 4))

# Har bir subplotga alohida ustunli grafik chizish
df.plot(kind='bar', x='Category', y='Value1', color='skyblue', ax=axes[0])
df.plot(kind='bar', x='Category', y='Value2', color='lightgreen', ax=axes[1])
df.plot(kind='bar', x='Category', y='Value3', color='coral', ax=axes[2])

# Sarlavhalar qo‘shish
axes[0].set_title('Value1 Distribution')
axes[1].set_title('Value2 Distribution')
axes[2].set_title('Value3 Distribution')

# Masofalarni sozlash
plt.tight_layout()

# Shaklni ko‘rsatish
plt.show()
```

### 6. GridSpec yordamida subplotlarni moslashtirish

Ushbu misol Matplotlib-dagi `GridSpec` vositasi orqali nostandart subplot joylashuvini yaratishni ko‘rsatadi. Unda to‘rtta subplot (chiziqli, tarqoq, ustunli va doiraviy diagrammalar) turli o‘lchamdagi kataklarda joylashtiriladi.


```python
import matplotlib.gridspec as gridspec

# Turli subplot o‘lchamlari bilan maxsus joylashuv yaratish
fig = plt.figure(figsize=(12, 6))

# Joylashuvni aniqlash uchun gridspec dan foydalanish
gs = gridspec.GridSpec(2, 3, width_ratios=[1, 2, 1], height_ratios=[2, 1])

# Joylashuv asosida subplotlarni yaratish
ax1 = plt.subplot(gs[0, 0])
ax2 = plt.subplot(gs[0, 1])
ax3 = plt.subplot(gs[0, 2])
ax4 = plt.subplot(gs[1, :]) # Oxirgi subplot butun pastki qatorni egallaydi

# 1-subplot: Chiziqli grafik
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
ax1.plot(x, y1, color='blue')
ax1.set_title('Line Plot - Sine Function')

# 2-subplot: Tarqoq grafik (Scatter Plot)
x_rand = np.random.rand(30)
y2 = 3 * x_rand + np.random.randn(30)
ax2.scatter(x_rand, y2, color='green')
ax2.set_title('Scatter Plot')

# 3-subplot: Ustunli diagramma
categories = ['A', 'B', 'C', 'D']
values = np.random.randint(1, 10, 4)
ax3.bar(categories, values, color='orange')
ax3.set_title('Bar Plot')

# 4-subplot: Doiraviy diagramma
labels = ['Category 1', 'Category 2', 'Category 3']
sizes = np.random.randint(1, 10, 3)
ax4.pie(sizes, labels=labels, autopct='%1.1f%%', colors=['lightcoral', 'lightblue', 'lightgreen'])
ax4.set_title('Pie Chart')

# Joylashuvni sozlash
plt.tight_layout()
plt.show()
```

**Izoh:** Ushbu kod `GridSpec` yordamida o‘zgaruvchan o‘lchamdagi kataklardan iborat 2×3 formatidagi subplot joylashuvini yaratadi. Unda alohida subplotlarda sinus to‘lqini, tarqoq grafik, ustunli diagramma va doiraviy diagramma chiziladi. Tushunarli bo‘lishi uchun sarlavhalar qo‘shiladi va `plt.tight_layout()` orqali masofalar muvofiqlashtiriladi.

### 3.7. Grafik shaffofligini sozlash (Adjust Plot Transparency)

Matplotlib grafiklarida shaffoflikni (transparency) o‘zgartirish, ayniqsa ma’lumotlar bir-birining ustiga tushib qolganda (overlap), vizual aniqlikni oshiradi. Shaffoflik **0 (to‘liq shaffof)** va **1 (to‘liq ochiq/xira)** oralig‘idagi qiymat orqali boshqariladi. Ushbu sozlamani chiziqlar, ustunlar, tarqoq nuqtalar va bo‘yalgan sohalarga grafik yaratish jarayonida yoki undan keyin grafik obyektini tahrirlash orqali qo‘llash mumkin. Quyidagi misollar Python-da Matplotlib yordamida shaffoflikni sozlashning turli usullarini ko‘rsatib beradi.

### `alpha` parametridan foydalanish

`alpha` parametri — Matplotlib-da grafiklar yaratishda shaffoflikni boshqarishning eng to‘g‘ridan-to‘g‘ri usulidir. U deyarli barcha grafik funksiyalari, jumladan `plot()`, `scatter()`, `bar()`, `hist()` va boshqalar tomonidan qo‘llab-quvvatlanadi. Grafik funksiyasida `alpha` qiymatini ko‘rsatish orqali har bir elementning qanchalik shaffof ko‘rinishini nazorat qilishingiz mumkin.

**1-misol: Past `alpha` qiymatiga ega chiziqli grafik**

```python
import matplotlib.pyplot as plt

x = [0, 5, 10, 15]
y = [0, 1, 2, 3]

# Chiziqni ko'k rangda va 0.3 shaffoflik darajasida chizish
plt.plot(x, y, color='blue', alpha=0.3)

plt.title("0.3 shaffoflikka ega chiziqli grafik")
plt.show()
```

**Izoh:**
Ushbu misolda `alpha=0.3` qiymati chiziqni sezilarli darajada shaffof qiladi, bu esa grafikda bir nechta chiziqlar ustma-ust tushganda ularni bir-biridan ajratib olishga yordam beradi.

**2-misol: Shaffof ustunli gistogramma**

```python
import matplotlib.pyplot as plt
import numpy as np

# 1000 ta tasodifiy ma'lumot yaratish
d = np.random.randn(1000)

# Gistogrammani to'q sariq rangda va 0.4 shaffoflikda chizish
plt.hist(d, bins=30, color='orange', alpha=0.4)
plt.title("Shaffof ustunli gistogramma")
plt.show()
```

**Izoh:** To‘q sariq rangli ustunlar bir-birining ustiga tushgan qismlar (binlar) ko‘rinib turishi uchun biroz shaffof qilingan.

**3-misol: Har xil shaffoflikka ega ustunli diagramma**

```python
import matplotlib.pyplot as plt

x = ['A', 'B', 'C']
y = [5, 8, 6]

for i in range(len(x)):
    # Har bir ustun uchun uning indeksiga asoslanib alohida shaffoflik o'rnatish
    plt.bar(x[i], y[i], color='green', alpha=(i + 1)/4)

plt.title("Har xil shaffoflikka ega ustunli diagramma")
plt.show()
```

**Izoh:** Har bir ustun o'z indeksiga ko'ra turlicha shaffoflikka ega (0.25, 0.5, 0.75).

**`set_alpha()` dan foydalanish**

`set_alpha()` metodi grafik elementlari yaratib bo'lingandan so'ng, ularning shaffofligini dinamik ravishda o'zgartirish kerak bo'lganda foydalidir. Dastlabki chizish jarayonida shaffoflikni o'rnatish o'rniga, siz qaytarilgan artist obyektiga (masalan, chiziq, tarqoq nuqta yoki bo'yalgan soha) murojaat qilishingiz va unga `.set_alpha(qiymat)` metodini qo'llashingiz mumkin.

**1-misol: Chiziq chizilgandan so'ng alpha'ni o'rnatish**

```python
import matplotlib.pyplot as plt

# Chiziq obyektini 'line' o'zgaruvchisiga saqlab olish
line, = plt.plot([1, 2, 3], [4, 5, 6])

# set_alpha() yordamida shaffoflikni o'zgartirish
line.set_alpha(0.5)

plt.title("set_alpha() yordamida o'rnatilgan shaffoflik")
plt.show()
```

**Izoh:** Chizilgan chiziq `set_alpha()` yordamida yarim shaffof holatga keltirilgan.

**2-misol: Tarqoq grafikda (Scatter Plot) alpha’ni o‘rnatish**


```python
import matplotlib.pyplot as plt

# Tarqoq grafik obyektini 'sc' o'zgaruvchisiga saqlash
sc = plt.scatter([1, 2, 3], [4, 5, 6], c='blue')

# set_alpha() yordamida shaffoflikni 0.3 ga sozlash
sc.set_alpha(0.3)

plt.title("Scatter grafikda set_alpha qo‘llanilishi")
plt.show()
```

**Izoh:** Barcha tarqoq nuqtalar 30% xiralik darajasiga ega.

**3-misol: Bo‘yalgan soha (Filled Area) uchun alpha’ni o‘rnatish**


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

# Sohani bo'yash obyektini 'fill' o'zgaruvchisiga saqlash
fill = plt.fill_between(x, y, color='red')

# set_alpha() orqali shaffoflikni 0.4 ga o'rnatish
fill.set_alpha(0.4)

plt.title("set_alpha yordamida shaffof bo‘yalgan soha")
plt.show()
```

**Izoh:** Chiziq va x o‘qi orasidagi qizil soha biroz shaffof ko‘rinishga ega.

### 3.9. Shkala yozuvlari shrift o‘lchamini sozlash (Setting Tick Label Font Size)

**Matplotlib** — bu Python dasturlash tilidagi eng keng tarqalgan ma'lumotlar vizualizatsiyasi kutubxonalaridan biri bo'lib, u yuqori sifatli 2D grafiklar yaratish uchun turli xil usullarni taqdim etadi. Grafikni tushunarli va estetik jihatdan chiroyli qilishning muhim jihatlaridan biri bu koordinata belgilari (tick labels) formatini, jumladan ularning shrift o'lchamini to'g'ri sozlashdir.

### Asosiy tushunchalar

Koordinata belgilarining shrift o'lchamini o'zgartirish usullarini ko'rib chiqishdan oldin, quyidagi tushunchalarni anglab olish muhim:

*   **Koordinata belgilari (Tick Labels):** Koordinata o'qlaridagi ma'lum nuqtalarni ko'rsatuvchi markerlar "ticks" deb ataladi, ularga biriktirilgan matnli yozuvlar esa "labels" (belgilar) hisoblanib, ular grafikdagi qiymatlarni tushunishga yordam beradi.
*   **Shrift o'lchami (Font Size):** Ushbu parametr koordinata belgilaridagi matn hajmini nazorat qiladi hamda grafikning aniqligi va o'qishli bo'lishini ta'minlaydi.




### `xticks()` va `yticks()` funksiyalarida `fontsize` parametridan foydalanish

Koordinata belgilarining shrift o‘lchamini o‘zgartirishning eng oddiy va to‘g‘ridan-to‘g‘ri usuli — bu `plt.xticks()` va `plt.yticks()` funksiyalari ichida `fontsize` parametridan foydalanishdir. Ushbu usul boshqa koordinata xususiyatlariga tegmagan holda, faqat shrift o‘lchamini belgilash kerak bo‘lganda juda qulaydir.


```python
import matplotlib.pyplot as plt

# Namunaviy ma'lumotlar
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 50]

plt.plot(x, y)

# X va Y o'qlari koordinata belgilarining shrift o'lchamini o'rnatish
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.show()
```

**Natija**

**Izoh:** Ma’lumotlar `plt.plot(x, y)` yordamida chizilgandan so‘ng, `plt.xticks(fontsize=12)` va `plt.yticks(fontsize=12)` funksiyalaridagi `fontsize` parametri orqali har ikkala o‘qdagi koordinata belgilarining shrift o‘lchami moslashtiriladi.

### `tick_params()` yordamida yanada aniqroq boshqarish

`tick_params()` metodi koordinata belgilarining ko'rinishini (tick label appearance) yanada mukammalroq nazorat qilish imkonini beradi. Uning yordamida asosiy (major) va qo'shimcha (minor) belgilar uchun shrift o'lchamini alohida belgilash mumkin. Bu usul asosiy va qo'shimcha belgilar bilan ishlashda va ularning har biri uchun turlicha o'lcham kerak bo'lganda juda foydalidir.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(x, y)

# Asosiy va qo'shimcha belgilar uchun shrift o'lchamini o'rnatish
ax.tick_params(axis='both', which='major', labelsize=14)
ax.tick_params(axis='both', which='minor', labelsize=10)

plt.show()
```

**Izoh:** Ma'lumotlar `ax.plot(x, y)` yordamida chizilgandan so'ng, `ax.tick_params()` orqali koordinata belgilarining shrift o'lchamlari moslashtiriladi. Bunda `axis='both'` argumenti orqali har ikkala o'qdagi asosiy belgilar uchun `labelsize=14`, qo'shimcha belgilar uchun esa `labelsize=10` o'lchami qo'llaniladi.

### `get_xticklabels()` va `get_yticklabels()` bilan `set_fontsize()` dan foydalanish

Ushbu yondashuv har bir koordinata belgisini alohida o‘zgartirish orqali yanada katta moslashuvchanlikni ta’minlaydi. `get_xticklabels()` va `get_yticklabels()` bo‘ylab takrorlash (iteratsiya) orqali siz har bir belgiga o‘ziga xos uslubni qo‘llashingiz mumkin, bu esa koordinata belgilarining maxsus formatlanishi bilan ishlashda juda foydalidir.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(x, y)

# Koordinata belgilari shrift o‘lchamlarini alohida-alohida o‘zgartirish
for label in ax.get_xticklabels():
    label.set_fontsize(12)

for label in ax.get_yticklabels():
    label.set_fontsize(12)

plt.show()
```

**Izoh:** Ma’lumotlar chizilgandan so‘ng, har ikkala o‘qda izchil o‘qishni ta’minlash uchun `ax.get_xticklabels()` va `ax.get_yticklabels()` bo‘ylab takrorlanib, har bir belgi uchun `fontsize=12` o‘rnatish orqali shrift o‘lchamlari alohida-alohida moslashtiriladi.

### Global shrift o‘lchamini o‘zgartirish uchun `rcParams` dan foydalanish

Agar siz shrift o‘lchami sozlamasini har safar qayta ko‘rsatmasdan bir nechta grafiklarda qo‘llamoqchi bo‘lsangiz, `rcParams` dan foydalanish eng yaxshi usul hisoblanadi. U Matplotlib-ning standart sozlamalarini o‘zgartiradi, bu esa bir nechta chizmalarda izchillikni saqlash uchun juda foydalidir.

```python
import matplotlib.pyplot as plt
import matplotlib as mpl

# X va Y o‘qlari koordinata belgilari o‘lchamini global tarzda o‘rnatish
mpl.rcParams['xtick.labelsize'] = 12
mpl.rcParams['ytick.labelsize'] = 12

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

**Izoh:** Ma’lumotlar chizilgandan so‘ng, `mpl.rcParams` yordamida koordinata belgilarining shrift o‘lchami global miqyosda o‘rnatiladi, bunda `'xtick.labelsize'` va `'ytick.labelsize'` qiymatlari 12 ga teng qilib belgilanadi.

Grafikimiz chiroyli ko‘rinishi uchun fon rangini o‘rnatmoqchi bo‘lsak, grafik chizilgandan so‘ng `axes()` atributi yordamida o‘qlar (axes) obyektini yaratishimiz kerak.

    <img src="images/fig.png" alt="Matplotlib grafik elementlari tuzilishi">

**Yondashuv:**
* Modulni yuklash.
* Ma’lumotlarni yuklash yoki yaratish.
* Oddiy grafikni chizish.
* O‘qlar (axes) obyektini yaratish.
* `set_facecolor()` atributiga kerakli rangni berish. Ushbu atribut rangning nomini ham, kodini ham qabul qiladi.

Mavzuni yaxshiroq tushunish uchun quyidagi misolga e’tibor bering.

**Misol: Standart rangdagi grafik**

```python
import matplotlib.pyplot as plt

# Talabalar baholari va natijalari
student_marks = [50, 60, 70, 80, 90]
student_grade = ['B', 'B', 'B+', 'B+', 'A']

plt.plot(student_marks, student_grade)

# O‘q yozuvlarini qalin shriftda o‘rnatish
plt.xlabel("student_marks", fontweight='bold')
plt.ylabel("student_grade", fontweight='bold')

plt.title("Student Marks v/s Student Grade")
plt.show()
```

**2-misol: Fon rangini sariqqa o‘zgartirish**

Ushbu misolda talabalar baholari grafigi chiziladi va uning fon rangi sariq (`yellow`) qilib belgilanadi. Bunda `plt.gca()` yordamida joriy o‘qlar obyekti olinadi va `set_facecolor()` metodi qo‘llaniladi.


```python
import matplotlib.pyplot as plt

# Baholarni raqamlarga moslashtirish
grade_mapping = {'B': 2, 'B+': 3, 'A': 4}
student_marks = [50, 60, 70, 80, 90]
student_grade = ['B', 'B', 'B+', 'B+', 'A']
numeric_grades = [grade_mapping[grade] for grade in student_grade]

# Ma’lumotlarni chizish
plt.plot(student_marks, numeric_grades, color='black')
plt.xlabel("Student Marks", fontweight='bold')
plt.ylabel("Student Grade", fontweight='bold')

# Fon rangini sariqqa o‘rnatish
ax = plt.gca() 
ax.set_facecolor("yellow")

plt.title("Student Marks v/s Student Grade")

# Y o‘qi belgilarini (ticks) matnli ko‘rinishga qaytarish
plt.yticks(ticks=list(grade_mapping.values()), labels=list(grade_mapping.keys()))
plt.show()
```

**3-misol: Fon rangini binafsha (violet) rangga o‘zgartirish**

Ushbu misolda sinus to‘lqini grafigi chiziladi va uning foni binafsha (`violet`) rangga bo‘yaladi.


```python
import matplotlib.pyplot as plt
import numpy as np

# X va Y o‘qlari uchun qiymatlar yaratish
x = np.arange(0, 10, 0.1)
y = np.sin(x)

# Grafikni chizish
plt.plot(x, y)

# O‘q nomlarini o‘rnatish
plt.xlabel("X", fontweight='bold')
plt.ylabel("sin(x)", fontweight='bold')

# Grafikning fon rangini o‘rnatish
ax = plt.gca()  # Joriy o‘qlarni (axes) olish
ax.set_facecolor("violet")

# Grafikni ko‘rsatish
plt.show()
```

### Grafikning tashqi va ichki ranglarini sozlash

Grafikning nafaqat ichki fonini, balki uning tashqi qismini ham rangli qilish imkoniyati mavjud. Grafikning ham foni, ham tashqi sohasi uchun ranglarni belgilashda, grafikni chizishni boshlashdan oldin `plt.figure(facecolor='rang')` funksiyasini qo‘shish kifoya qiladi.

**1-misol:**

```python
import matplotlib.pyplot as plt
import numpy as np

# X va Y o‘qlari uchun qiymatlar yaratish
x = np.arange(0, 10, 0.1)
y = np.sin(x)

# Shaklning (figure) tashqi soha rangini sariqqa o‘rnatish
plt.figure(facecolor='yellow')

plt.plot(x, y)
plt.xlabel("X", fontweight='bold')

# Joriy o‘qlarga (axes) murojaat qilish va uning fon rangini sozlash
ax = plt.gca()  
ax.set_facecolor("violet")

plt.ylabel('sin(x)', fontweight='bold')
plt.show()
```

**Natija tahlili:**

*   **Tashqi soha (Outer area):** `plt.figure(facecolor='yellow')` buyrug‘i yordamida butun grafik atrofidagi maydon sariq rangga bo‘yaladi.
*   **Ichki soha (Inner area):** `ax.set_facecolor("violet")` metodi orqali chiziq chizilgan ichki fon binafsha rangga o‘zgartiriladi.
*   **Koordinata o‘qlari:** `plt.gca()` (get current axes) orqali biz ayni vaqtdagi o‘qlar obyektiga kirish huquqini olamiz va uning xususiyatlarini o‘zgartiramiz.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 10, 0.1)
y = np.sin(x)

# Shaklning (figure) tashqi soha rangini HTML kodi yordamida o‘rnatish
plt.figure(facecolor='#94F008')

plt.plot(x, y)
plt.xlabel("X", fontweight='bold')

# Joriy o‘qlarni (axes) olish va uning fon rangini kod orqali sozlash
ax = plt.gca() 
ax.set_facecolor("#1CC4AF")

plt.ylabel('sin(x)', fontweight='bold')
plt.show()
```

### 3.11. O‘q yorliqlari joyini o‘zgartirish (Move Axis Labels)

Matplotlib — grafiklar va diagrammalar yaratish uchun mo‘ljallangan kuchli Python kutubxonasi hisoblanadi. Standart holatda u o‘q yozuvlarini odatiy joylarga joylashtiradi, biroq yaxshiroq o‘qilishi yoki chiroyli dizayn uchun ularni ko‘chirish talab qilinishi mumkin. Quyida grafiklar ko‘rinishini yaxshilash uchun o‘q yozuvlari joylashuvini sozlashning oson usullari keltirilgan.

### `labelpad` yordamida o‘q yozuvlari joylashuvini sozlash

`set_xlabel()` va `set_ylabel()` metodlaridagi **labelpad** parametri o‘q va unga mos keladigan yozuv orasidagi masofani boshqarish imkonini beradi. Ushbu metod yozuvlar o‘qqa juda yaqin bo‘lib qolganda va ular orasida ko‘proq joy kerak bo‘lganda juda foydalidir.

```python
import matplotlib.pyplot as plt

# Namunaviy grafik yaratish
fig, ax = plt.subplots()
ax.plot([0, 1, 2], [0, 1, 4])

# labelpad yordamida yozuv holatini sozlash
ax.set_xlabel("X-o‘qi yozuvi", labelpad=20)  # O‘qdan masofani oshirish
ax.set_ylabel("Y-o‘qi yozuvi", labelpad=30)

plt.show()
```

**Natija tahlili:**

*   **labelpad vazifasi:** `ax.set_xlabel()` va `ax.set_ylabel()` funksiyalaridagi `labelpad` yozuvning o‘qdan qanchalik uzoqda turishini belgilaydi. 
*   **O‘qish qulayligi:** Yuqori qiymatlar masofani oshirib, matnlarning o‘q chiziqlari yoki raqamlar bilan ustma-ust tushib qolishini oldini oladi va o‘qishni osonlashtiradi. 
*   **Misolda:** Ushbu kodda x-o‘qi uchun `labelpad=20` va y-o‘qi uchun `labelpad=30` qiymatlari grafikning aniqligini oshirish uchun qo‘llangan.

Grafikdagi yozuvlarni nafaqat o‘qdan uzoqlashtirish, balki ularni o‘q bo‘ylab chapga yoki o‘ngga surish (alignment) usullari haqida ham ma’lumot beraymi?

### `set_position()` yordamida yozuv holatini qo‘lda sozlash

Agar o‘q yozuvlarining joylashuvi ustidan to‘liq va aniq nazorat kerak bo‘lsa, `set_position()` metodidan foydalanish mumkin. Ushbu metod o‘q yozuvlari uchun aniq koordinatalarni belgilash imkonini beradi.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot([0, 1, 2], [0, 1, 4])

# set_position() yordamida yozuv holatini sozlash
ax.xaxis.label.set_position((0.5, -0.1))  # (x, y) koordinatalari
ax.yaxis.label.set_position((-0.2, 0.5))

plt.xlabel("X-o‘qi yozuvi")
plt.ylabel("Y-o‘qi yozuvi")
plt.show()
```

**Natija tahlili:**

*   `ax.xaxis.label` va `ax.yaxis.label` tarkibidagi `set_position()` metodi yozuv koordinatalarini o‘qlarga nisbatan o‘rnatadi.
*   `(0.5, -0.1)` koordinatasi x-o‘qi yozuvini gorizontal markazda saqlagan holda pastroqqa joylashtiradi.
*   `(-0.2, 0.5)` koordinatasi esa y-o‘qi yozuvini chapga suradi va uni vertikal ravishda markazda saqlaydi.

### Yaxshiroq ko‘rinish uchun koordinata o‘qi yozuvlarini burish

Uzun yozuvlar yoki zich joylashgan grafik (layout) bilan ishlashda, o‘q yozuvlarini burish ularning o‘qishli bo‘lishini sezilarli darajada yaxshilaydi. Siz `set_rotation()` metodidan foydalanib, yozuvlarni o‘zingiz belgilagan burchak ostida burishingiz mumkin.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot([0, 1, 2], [0, 1, 4])

# Yozuvlarni o'rnatish
ax.set_xlabel("X-o‘qi yozuvi", labelpad=10)
ax.set_ylabel("Y-o‘qi yozuvi", labelpad=10)

# Yozuvlarni burish
ax.xaxis.label.set_rotation(30)  # X o‘qi yozuvini burish
ax.yaxis.label.set_rotation(90)  # Y o‘qi yozuvini burish

plt.show()
```

**Natija tahlili:**

*   `ax.xaxis.label` va `ax.yaxis.label` tarkibidagi **`set_rotation()`** metodi o‘q yozuvlarini yaxshiroq o‘qish uchun buradi.
*   Ushbu misolda **30°** qiymati x-o‘qi yozuvini diagonal tarzda qiyshaytiradi.
*   **90°** qiymati esa y-o‘qi yozuvini vertikal holatga keltiradi.
*   Bu usul yozuvlarning bir-biriga xalaqit berishini (ustma-ust tushishini) oldini oladi va zich grafiklarda aniqlikni oshiradi.

### Subplotlar (ichki grafiklar) bilan o‘q yozuvlarini sozlash

Bir nechta subplotlar bilan ishlaganda, o‘q yozuvlari orasida to‘g‘ri masofani saqlash o‘ta muhimdir. Matplotlib-dagi `tight_layout()` funksiyasi yozuvlarning bir-biri bilan yoki qo‘shni grafiklar bilan ustma-ust tushib qolishini oldini olishga yordam beradi.



```python
import matplotlib.pyplot as plt

# 8x6 o‘lchamdagi 2x2 subplotlar tarmog‘ini yaratish
fig, axs = plt.subplots(2, 2, figsize=(8, 6))

# Har bir subplot bo‘ylab takrorlash (loop)
for ax in axs.flat:
    ax.plot([0, 1, 2], [0, 1, 4]) 
    ax.set_xlabel("X-Axis", labelpad=15)  # Masofa (padding) bilan x-o‘qi yozuvini o‘rnatish
    ax.set_ylabel("Y-Axis", labelpad=15)  # Masofa (padding) bilan y-o‘qi yozuvini o‘rnatish

# Grafiklar orasidagi masofani avtomatik sozlash
plt.tight_layout()
plt.show() 
```

**Natija tahlili:**

*   **`plt.subplots(2, 2, figsize=(8, 6))`**: Ushbu buyruq 2 qator va 2 ustundan iborat jami 4 ta subplotni yaratadi.
*   **`axs.flat`**: Bu xususiyat tarmoqdagi barcha subplotlar bo‘ylab takrorlanib (iteratsiya), har biriga bir xil sozlamalarni qo‘llash imkonini beradi.
*   **`labelpad=15`**: O‘q yozuvlarining o‘qdan masofasini belgilaydi, bu esa matnlarning grafik elementlari bilan aralashib ketmasligini ta’minlaydi.
*   **`plt.tight_layout()`**: Subplotlar orasidagi bo‘shliqlarni avtomatik ravishda moslashtiradi, natijada yozuvlar ustma-ust tushmaydi va tartibli ko‘rinishga keladi.

### Chegaralar va oq bo‘shliqlarni olib tashlash

Biz `plt.axis('off')` buyrug‘ini qo‘llaganimizda o‘qlar yashiriladi, biroq grafik atrofida oq bo‘shliqlar va shakl chegaralari baribir saqlanib qolishi mumkin. Ularni butunlay olib tashlash va grafikni samarali qirqib olish (crop) uchun `savefig()` metodida `bbox_inches='tight'` va `pad_inches=0` parametrlaridan foydalanamiz.



**Misol:**

```python
import numpy as np
import matplotlib.pyplot as plt

x = ['Science', 'Maths', 'English', 'History', 'Geography']
y = [75, 85, 88, 78, 74]

plt.bar(x, y)
plt.xlabel("Subject")
plt.ylabel("Ram's marks out of 100")

# O‘qlarni yashirish
plt.axis('off')

# Chegarasiz va oq bo‘shliqlarsiz rasmni saqlash
plt.savefig('image.png', bbox_inches='tight', pad_inches=0)

plt.show()
```

### O‘q ko‘rinishini moslashtirishning qo‘shimcha usullari

Grafik elementlarini yanada aniqroq nazorat qilish uchun faqatgina o‘qlarni to‘liq o‘chirish bilan cheklanib qolmasdan, ularning ayrim qismlarini ham boshqarish mumkin.

### 1. O‘q chegaralarini (spines) yashirish

**Spines** — bu grafik atrofida chegara hosil qiluvchi chiziqlardir. Matplotlib-da ushbu chiziqlarni tanlab olib tashlash imkoniyati mavjud.


```python
import numpy as np
import matplotlib.pyplot as plt

# Ma'lumotlar: Fan nomlari va Alining to'plagan ballari
# Inglizcha nomlar o'rniga o'zbekcha fan nomlarini kiritdik
fanlar = ['Tabiiy fan', 'Matematika', 'Ingliz tili', 'Tarix', 'Geografiya']
ballar = [75, 85, 88, 78, 74]

fig, ax = plt.subplots()
ax.bar(fanlar, ballar)

# O'qlar nomini o'zbek tilida o'rnatish
ax.set_xlabel("Fanlar")
ax.set_ylabel("Alining 100 ballik tizimdagi baholari")

# Grafik atrofidagi ma'lum bir chiziqlarni (spines) yashirish
# Bu grafikni zamonaviy va minimalistik ko'rinishga keltiradi
ax.spines['top'].set_visible(False)    # Yuqori chegarani yashirish
ax.spines['right'].set_visible(False)  # O'ng chegarani yashirish

# Grafikni ekranga chiqarish
plt.show()
```

**Natija tahlili:**

*   Ushbu misolda `ax.spines['top'].set_visible(False)` va `ax.spines['right'].set_visible(False)` buyruqlari grafikning yuqori va o‘ng chegaralarini olib tashlaydi.
*   Chap (left) va pastki (bottom) chegaralar esa ma’lumotlar uchun asosiy tayanch nuqtasi bo‘lib xizmat qilishi uchun ko‘rinadigan holatda qoldiriladi.
*   Bu usul grafikni yanada zamonaviy va "ochiq" ko‘rinishga keltirish uchun keng qo‘llaniladi.

### 2. Koordinata belgilari (ticks) va yozuvlarini yashirish uchun `plt.gca()` dan foydalanish

`plt.gca()` (get current axes) buyrug‘i joriy o‘qlar obyektini oladi va ularning ko‘rinishini (visibility) yanada aniqroq boshqarish imkonini beradi.


```python
import numpy as np
import matplotlib.pyplot as plt

# O'zbek tilidagi ma'lumotlar
fanlar = ['Tabiiy fan', 'Matematika', 'Ingliz tili', 'Tarix', 'Geografiya']
ballar = [75, 85, 88, 78, 74]

plt.bar(fanlar, ballar)
plt.xlabel("Fanlar")
plt.ylabel("Ramning 100 ballik tizimdagi baholari")

# Joriy o'qlarni (axes) olish
ax = plt.gca()

# O'qlarni yashirish
ax.xaxis.set_visible(False)  # X-o'qini yashirish
ax.yaxis.set_visible(False)  # Y-o'qini yashirish

plt.show()
```

**Izoh:** 
Bu yerda `ax.xaxis.set_visible(False)` va `ax.yaxis.set_visible(False)` buyruqlari diagrammaning o‘zini ko‘rinadigan holatda qoldirib, har ikkala o‘qni ham yashiradi. Ushbu usul shaklning (figure) o‘ziga yoki tashqi sohalariga ta’sir qilmasdan, faqatgina o‘q elementlarini to‘liq olib tashlash uchun juda qulaydir.
