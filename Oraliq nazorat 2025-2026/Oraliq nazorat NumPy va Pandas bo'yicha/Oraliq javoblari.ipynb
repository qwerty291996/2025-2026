### 1. Hodisalarga asoslangan (Event-driven) yondashuv nima?

An'anaviy (protsedurali) dasturlashda dastur qadam-baqadam o‘z ishini bajaradi va tugaydi. Uning qachon nima qilishini dasturchi yozgan qat'iy ketma-ketlik hal qiladi.

**Hodisalarga asoslangan dasturlashda** esa dasturning ishlash oqimini foydalanuvchi yoki tizim yaratadigan "hodisalar" (events) boshqaradi. Dastur ishga tushgach, orqa fonda **"Hodisalar sikli" (Event Loop)** deb ataladigan cheksiz sikl aylanib, kutish rejimiga o‘tadi. 



Hodisalarga quyidagilar kirishi mumkin:
*   Foydalanuvchi sichqoncha bilan tugmani bosishi.

*   Oynaning o‘lchami o‘zgartirilishi.
*   Tizimdan kelgan xabar (masalan, taymer tugashi).

Hodisa yuz berganda, Event Loop uni darhol tutib oladi va shu hodisaga biriktirilgan maxsus funksiyani (Event Handler) ishga tushiradi. Funksiya o‘z ishini tugatgach, dastur yana kutish rejimiga qaytadi.

### 2. PyQt5 dagi Signal va Slot mexanizmi qanday ishlaydi?

PyQt5 (va umuman Qt freymvorki) yuqoridagi "hodisalarni tutib olish va unga javob qaytarish" jarayonini o‘zining mashhur va juda qulay bo‘lgan **Signal va Slot** arxitekturasi orqali amalga oshiradi.



Bu mexanizm uchta asosiy qismdan iborat:

*   **Signal (Signal):** Vidjet bilan biror voqea sodir bo‘lganda u tarqatadigan xabar (emit). Masalan: `QPushButton` (tugma) bosilganda `clicked` signalini tarqatadi, matn maydoni o‘zgarganda `textChanged` signalini beradi.
*   **Slot (Slot):** Bu ma'lum bir signal tarqatilganda ishga tushishi kerak bo‘lgan oddiy Python funksiyasi yoki obyekti (metodi). U reaksiyani, ya'ni hodisaga javobni o‘zida saqlaydi.
*   **Bog‘lash (Connection):** Signal tarqatilganda aynan qaysi slot ishga tushishini ta'minlovchi ko‘prik. Bu `connect()` metodi orqali amalga oshiriladi.

#### Amaliy misolda ko‘ramiz:

Tasavvur qiling, bizda bitta tugma bor va uni bosganimizda ekranga xabar chiqishi kerak.

```python
!pip install pyqt5
```

```python
import sys
from PyQt5.QtWidgets import QApplication, QPushButton, QWidget, QVBoxLayout

# 1. SLOT yaratamiz (bu shunchaki voqeaga javob beruvchi funksiya)
def xabar_chiqar():
    print("Tugma muvaffaqiyatli bosildi!")

app = QApplication(sys.argv)
oyna = QWidget()
layout = QVBoxLayout()

# Tugma yaratamiz
tugma = QPushButton("Meni bos!")

# 2. SIGNAL va SLOTni bog‘laymiz
# tugma.clicked - bu SIGNAL (tugma bosilganda kelib chiqadi)
# .connect(xabar_chiqar) - bu SIGNALni bizning SLOT funksiyamizga ulash
tugma.clicked.connect(xabar_chiqar)

layout.addWidget(tugma)
oyna.setLayout(layout)
oyna.show()

# 3. Hodisalar siklini (Event Loop) ishga tushiramiz
sys.exit(app.exec_())
```

#### Bu yondashuvning afzalliklari nimada?

*   **Yuqori darajadagi mustaqillik (Loose Coupling):** Signal tarqatayotgan vidjet (masalan, tugma) o‘zining signalini kim va qanday qabul qilayotgani bilan ishi yo‘q. Slot (funksiya) ham o‘zini qaysi vidjet chaqirganini bilishi shart emas. Bu kodni o‘zgartirish va kengaytirishni juda osonlashtiradi.
*   **Ko‘pga-Ko‘p (Many-to-Many) munosabatlar:** Bitta signalga bir nechta turli xil slotlarni ulashingiz mumkin (bitta tugmani bossangiz ham oyna yopiladi, ham ma'lumot saqlanadi). Shuningdek, turli xil signallarni bitta yagona slotga ulash ham mumkin.

**Javob:**
PyQt5 (va umuman Qt freymvorki) arxitekturasining eng asosiy tamoyillaridan biri bu **"Obyektlar daraxti" (Object Tree)** mexanizmidir. Bunda har bir vidjet (element) o‘zining "otasi"ga (parent) ega bo‘lishi va o‘z navbatida boshqa vidjetlarga "ota" bo‘lishi mumkin. 

Bu "Ota-bola" yondashuvi asosan ikkita juda muhim vazifani hal qilib beradi:

### 1. Xotirani avtomatik boshqarish (Memory Management)
PyQt5 orqasida C++ tilida yozilgan Qt kutubxonasi yotadi. Odatda, pastki darajadagi tillarda xotirani qo‘lda boshqarish, kerak bo‘lmagan obyektlarni xotiradan tozalash juda mashaqqatli jarayon. PyQt5 da esa bu "Ota-bola" tizimi orqali juda nafis hal qilingan:

*   **Avtomatik tozalash (Garbage Collection muqobili):** Agar siz asosiy oynani (ota vidjetni) yopsangiz yoki xotiradan o‘chirsangiz, u o‘ziga tegishli bo‘lgan barcha "bola" vidjetlarni (ichidagi barcha tugmalar, matnlar, rasmlar va h.k.) xotiradan **avtomatik ravishda o‘chiradi**.
*   **Xavfsizlik:** Dasturchi har bir tugma yoki yozuv uchun xotirani bo‘shatish kodini alohida yozib o‘tirishi shart emas. Bu xotira sizib chiqishi (memory leak) xavfini keskin kamaytiradi.

### 2. Dizayn ierarxiyasi va vizual joylashuv
"Ota-bola" munosabati nafaqat orqa fondagi xotirani, balki vidjetlarning ekranda qanday ko‘rinishi va o‘zini qanday tutishini ham to‘g‘ridan-to‘g‘ri belgilaydi:

*   **Chegaralanish (Clipping):** Bola vidjet har doim ota vidjetning ichida (uning chegaralarida) chiziladi. Agar bolaning o‘lchami kattalashib ketsa ham, u ota oynasidan tashqariga chiqib keta olmaydi (kesib tashlanadi).
*   **Nisbiy koordinatalar:** Bola vidjetning `(x, y)` koordinatalari kompyuterning butun ekraniga nisbatan emas, balki aynan **ota vidjetning yuqori chap burchagiga nisbatan** hisoblanadi. Bu oynani siljitganda ichidagi barcha elementlar ham birga siljishini ta'minlaydi.
*   **Holatni meros qilib olish:** Agar ota vidjet yashirilsa (`hide()`), uning barcha bolalari ham ko‘zdan yo‘qoladi. Agar ota oyna nofaol holatga o‘tkazilsa (`setEnabled(False)`), uning ichidagi barcha tugmalar va kiritish maydonlari ham bosilmaydigan holatga keladi.

#### Amaliy misol:

Quyidagi kodda "Ota-bola" bog‘liqligi qanday o‘rnatilishini ko‘rishingiz mumkin:

```python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QPushButton

app = QApplication(sys.argv)

# 1. OTA vidjet (Asosiy oyna) yaratamiz
asosiy_oyna = QWidget()
asosiy_oyna.setWindowTitle("Ota Oyna")
asosiy_oyna.resize(300, 200)

# 2. BOLA vidjet (Tugma) yaratamiz
# E'tibor bering: ikkinchi argument sifatida 'asosiy_oyna' ni beryapmiz.
# Shu orqali tugma o‘zining otasi kimligini taniydi.
tugma = QPushButton("Men bola vidjetman", asosiy_oyna)

# Bu koordinatalar (50, 50) kompyuter ekraniga emas, 
# balki 'asosiy_oyna'ning ichki qismiga nisbatan olinadi.
tugma.move(50, 50) 

asosiy_oyna.show()

# Asosiy oyna yopilganda, 'tugma' ham xotiradan avtomatik o‘chadi.
sys.exit(app.exec_())
```

### 2. Muammoni QThread bilan qanday hal qilamiz?

Yechim juda oddiy: Asosiy oqimni (Event Loop'ni) faqat va faqat GUI bilan ishlashga (rasm chizish, foydalanuvchini eshitish) qoldirish kerak. Og‘ir vazifani esa **Parallel Oqimga (Background Thread)** o‘tkazish kerak. PyQt5 da bu maqsad uchun **`QThread`** klassi ishlatiladi.

Bu jarayonni to‘g‘ri amalga oshirishning qat'iy qoidalari bor:

1.  **Vazifani ajratish:** Og‘ir hisoblash jarayonini yoki fayl yuklash kodini `QThread` dan meros olgan yangi klassning `run()` metodiga yozasiz.
2.  **To‘g‘ridan-to‘g‘ri yozish taqiqlanadi (Oltin qoida):** Orqa fon oqimidan (QThread) turib hech qachon ekrandagi vidjetlarga to‘g‘ridan-to‘g‘ri o‘zgartirish kiritmang (masalan, `self.label.setText()`). GUI faqat Asosiy oqimda yangilanishi shart, aks holda dastur qulaydi (crash).
3.  **Aloqa bog‘lash (Signal va Slot):** QThread orqa fonda ishlarni bajarayotganda, o‘zining qancha qismini tugatgani yoki qanday natija olgani haqida faqat **Signal** tarqatishi (`emit()`) kerak. Asosiy oqim bu signalni o‘zining **Slot'iga** ulangan holda kutib oladi va Asosiy oqimning o‘zi ekrandagi yozuvni yoki ProgressBar'ni yangilaydi.

#### Qisqacha misol:

```python
from PyQt5.QtCore import QThread, pyqtSignal
import time

# 1. Orqa fonda ishlovchi oqim yaratamiz
class OgirIshOqimi(QThread):
    # Bu signal orqali Asosiy oqimga foizlarni yuboramiz
    progress_signal = pyqtSignal(int) 

    def run(self):
        for i in range(1, 101):
            time.sleep(0.1) # Qandaydir og‘ir hisoblash yoki yuklash jarayoni
            self.progress_signal.emit(i) # Asosiy oqimga xabar beramiz
```

Kompilyator `.exe` yig‘ayotganda bu yashirin kutubxonalarni ko‘rmay qoladi va tayyor dasturga qo‘shmaydi. Natijada dastur ishga tushganda "ModuleNotFoundError" xatosi chiqadi.

**11-savol:** Tarmoq dasturlashda ma’lumotlarni uzatishning TCP va UDP protokollari qanday fundamental prinsiplarga asoslanadi va har bir protokol qanday real loyihalarda tanlanishi kerak?

---

**Javob:**
Internet tarmog‘ida ma'lumotlar (paketlar) asosan ikkita fundamental transport protokoli orqali uzatiladi: **TCP** (Transmission Control Protocol) va **UDP** (User Datagram Protocol). Garchi ikkalasi ham ma'lumotni A nuqtadan B nuqtaga yetkazishga xizmat qilsa-da, ularning ishlash falsafasi bir-biriga butunlay qarama-qarshidir.

Buni tushunish uchun ularni oddiy hayotdagi aloqa vositalariga qiyoslaymiz.

### 1. TCP (Transmission Control Protocol) – "Ishonchli Telefon Qo‘ng‘irog‘i"

**Fundamental prinsiplari:**
TCP ning asosiy maqsadi — **Ishonchlilik (Reliability)**. U ma'lumotning biror bayti ham yo‘qolmasligini va aynan jo‘natilgan ketma-ketlikda yetib borishini 100% kafolatlaydi.
*   **Aloqa o‘rnatish (Connection-oriented):** Ma'lumot yuborishdan oldin TCP "Uch tomonlama kelishuv" (Three-way handshake) o‘tkazadi. Ya'ni: *"Salom, eshityapsanmi?" -> "Ha, eshityapman, o‘zing qaleysan?" -> "Zo‘r, hozir ma'lumot yuborishni boshlayman!"*.
*   **Tasdiqlash (Acknowledgment - ACK):** TCP da jo‘natilgan har bir paket uchun qabul qiluvchidan tasdiq javobi kutiladi. *"1-paketni oldingmi? Ha. Endi 2-paketni yuboraman."*
*   **Qayta jo‘natish (Retransmission):** Agar tarmoqda uzilish bo‘lib, qaysidir paket yo‘qolib qolsa (tasdiq kelmasa), TCP uzoqroq kutadi va o‘sha yo‘qolgan paketni topib, yana qaytadan yuboradi.

**Qanday loyihalarda tanlanadi?**
Aniqlik tezlikdan muhimroq bo‘lgan har qanday loyihada:
*   **Veb-saytlar (HTTP/HTTPS):** Saytdagi bitta harf yoki kod tushib qolsa, sayt ishlamay qoladi.
*   **Fayl yuklash (FTP):** Rasm yoki videoning o‘rtasidan 1 Kilobayt yo‘qolsa ham fayl buziladi (corrupted).
*   **Elektron pochta (SMTP/IMAP) va Chatlar:** Xabar to‘liq va o‘zgarmas yetib borishi shart.
*   **Ma'lumotlar bazasi bilan ishlash:** Tranzaksiyalar va SQL so‘rovlarda.

---

### 2. UDP (User Datagram Protocol) – "Tezkor Pochta Kaptari" (yoki Karnay)

**Fundamental prinsiplari:**
UDP ning asosiy maqsadi — **Tezlik (Speed)**. U ma'lumotni yetib borgan-bormaganligiga umuman qiziqmaydi, tasdiq so‘ramaydi va yo‘qolgan paketlarni qayta jo‘natmaydi.
*   **Aloqasiz (Connectionless):** Oldindan hech qanday kelishuv o‘tkazilmaydi. Shunchaki manzilni yozadi-yu, ma'lumotni to‘xtovsiz otaveradi (Fire-and-forget).
*   **Kam xarajat (Low overhead):** Tasdiqlash, tartibga solish va xatolarni tekshirish kabi mexanizmlar yo‘qligi sababli, UDP paketlari juda yengil bo‘ladi va protsessorni charchatmaydi.

**Qanday loyihalarda tanlanadi?**
Ma'lumotlarning aniqligidan ko‘ra ularning real vaqtda (Real-time), tez yetib kelishi muhim bo‘lgan loyihalarda (yo‘qolgan paketlarga ahamiyat berilmaydi):
*   **Onlayn o‘yinlar (CS:GO, PUBG):** o‘yinchining qayerda turgani har millisoniyada yangilanadi. Agar bitta paket yo‘qolsa, uni qayta so‘rashning keragi yo‘q, chunki o‘yinchi allaqachon boshqa joyga o‘tib ketgan. Eskirgan ma'lumot kerak emas.
*   **Jonli video va audio (Zoom, YouTube Live, Telegram Calls):** Internet qotib qolganda video biroz xiralashadi yoki ovoz uziladi (bu paket yo‘qolganining belgisi). Lekin qo‘ng‘iroq to‘xtab qolmaydi, jonli efir davom etaveradi. Agar bu TCP da bo‘lganda, yo‘qolgan paket yetib kelgunicha ekraningiz 5-10 soniya "Loading" bo‘lib qotib turar edi.
*   **IoT (Aqlli qurilmalar):** Har soniyada temperaturani serverga jo‘natib turuvchi sensorlar (bitta qiymat yetib bormasa, keyingi soniyada yangisi boraveradi).

---

### Qisqacha Taqqoslash Jadvali

| Xususiyat | TCP (Transmission Control Protocol) | UDP (User Datagram Protocol) |
| :--- | :--- | :--- |
| **Maqsad** | 100% Ishonchlilik va Kafolat | Maksimal Tezlik va Jonli efir |
| **Tasdiqlash (ACK)** | Har bir paket tasdiqlanadi | Tasdiqlanmaydi (Umuman) |

**12-savol:** Mijoz-server (Client-Server) arxitekturasida serverning asosiy mas’uliyati nimalardan iborat bo‘ladi va bitta to‘liq tarmoq aloqasi sikli qanday bosqichlardan o‘tadi?

---

**Javob:**
Mijoz-server (Client-Server) arxitekturasi butun zamonaviy internetning (veb-saytlar, mobil ilovalar, o‘yinlar) asosiy poydevori hisoblanadi. Unda rollar qat'iy taqsimlangan: **Mijoz** (xizmat so‘rovchi) va **Server** (xizmat ko‘rsatuvchi).

### 1. Serverning asosiy mas'uliyatlari nimalardan iborat?

Server bu — oddiy qilib aytganda, 24/7 ishlab turuvchi, yuqori quvvatli kompyuter va unda yozilgan dastur. Uning bo‘ynida quyidagi o‘ta muhim mas'uliyatlar bor:

*   **Tinglash va Kutish (Listening):** Server doimiy ravishda ma'lum bir portda (masalan, veb-saytlar uchun 80 yoki 443-port) mijozlardan keladigan so‘rovlarni quloq tutib kutib turadi.
*   **So‘rovlarni tahlil qilish (Parsing/Routing):** Mijozdan xat kelganda, server u nima so‘rayotganini tahlil qiladi. "Bu foydalanuvchi rasmni ochmoqchimi yoki akkauntiga kirmoqchimi?" degan savolga manzilni (URL) o‘qib javob topadi.
*   **Mantiqiy amallar va Baza bilan ishlash (Business Logic & Database):** Agar foydalanuvchi "Parolim to‘g‘rimi?" deb so‘rasa, server o‘zining Ma'lumotlar bazasiga (Database) kirib, parolni tekshiradi, hisob-kitoblarni amalga oshiradi.
*   **Xavfsizlik va Ruxsat (Authentication/Authorization):** So‘rov yuborayotgan mijoz haqiqatan ham tizimga kirganmi? Uning bu ma'lumotni o‘qishga huquqi bormi? Server har bir qadamda shu xavfsizlikni tekshiradi.
*   **Javob shakllantirish (Response generation):** Barcha jarayon tugagach, server mijoz tushunadigan tilda (HTML, JSON, XML yoki fayl ko‘rinishida) tayyor javobni paketlab, ortga yuboradi.

---

### 2. Bitta to‘liq tarmoq aloqasi sikli qanday bosqichlardan o‘tadi?

Tasavvur qiling, brauzeringizda `google.com` deb yozib, Enter tugmasini bosdingiz. Shu soniyaning o‘zida quyidagi **to‘liq sikl (Request-Response Cycle)** sodir bo‘ladi:

1.  **Mijoz so‘rov yaratishi:** Brauzer nima kerakligini belgilab, xat (HTTP Request) yozishni boshlaydi.
2.  **Manzilni aniqlash (DNS Lookup):** Kompyuter `google.com` ning IP manzilini (masalan, `142.250.190.46`) topish uchun DNS serverga murojaat qiladi (xuddi telefon kitobchasidan raqam qidirgandek).
3.  **Aloqa o‘rnatish (TCP Handshake):** Mijoz va Server o‘rtasida ishonchli ko‘prik quriladi. (Oldingi savoldagi kabi: "Eshityapsanmi? -> Ha -> Gapirdim!").
4.  **So‘rovni jo‘natish (Request Transmission):** Yozilgan xat Internet tarmoqlari, kabellar va ruterlar orqali servergacha yetib boradi.
5.  **Serverda qayta ishlash (Server Processing):** Server xatni ochadi, bazaga qaraydi, kerakli HTML sahifani va rasmlarni tayyorlaydi.
6.  **Javobni qaytarish (Response Transmission):** Server tayyor ma'lumotni Internet orqali mijozga qaytarib yuboradi. Odatda bu yerda "200 OK" (Hammasi zo‘r) yoki "404 Not Found" (Topilmadi) kabi kodlar qo‘shib yuboriladi.
7.  **Mijozda chizish (Rendering):** Brauzer kelgan kodni o‘qiydi va ekraningizda chiroyli tugmalar, matnlar va rasmlarni chizib beradi.
8.  **Aloqani uzish (Connection Close):** Barcha ma'lumot olinib bo‘lgach, TCP ko‘prigi yopiladi (yoki keyingi so‘rovlar uchun biroz kutib turadi).

**13-savol:** Operatsion tizim darajasida “soket” (socket) tushunchasi o‘zi nima va Python dagi socket kutubxonasi yozgan dasturimiz bilan kompyuter tarmog‘i o‘rtasida qanday qilib muloqot ko‘prigini o‘rnatib beradi?

---

**Javob:**
"Soket" (inglizchada *socket* — rozetka, uya) kompyuter tarmoqlari va internetning eng quyi, poydevor g‘ishtidir. Biz har kuni ishlatadigan brauzerlar, Telegram, onlayn o‘yinlar — bularning barchasi orqa fonda aynan soketlar orqali ishlaydi.

Keling, bu tushunchani operatsion tizim (OS) darajasida va Python misolida ajratib tushunamiz.

### 1. Operatsion tizim darajasida "Soket" o‘zi nima?

Operatsion tizimlar (Windows, Linux, macOS) kompyuterning jismoniy tarmoq kartasini (Wi-Fi, LAN kabel) to‘g‘ridan-to‘g‘ri boshqaradi. Oddiy dasturlar u yerga bevosita aralasha olmaydi. 

Soket — bu Operatsion Tizim tomonidan dasturlarga taqdim etiladigan **dasturiy eshik** yoki **aloqa nuqtasi (endpoint)**.

*   **Uy va Eshik o‘xshatishi:** Agar kompyuterning **IP manzili** ko‘p qavatli binoning umumiy manzili bo‘lsa, **Port** (masalan, 80 yoki 443) o‘sha binodagi aniq bir xonaning (dasturning) eshigidir. **Soket** esa o‘sha eshikning qulfi va u orqali o‘tkazilgan aloqa trubasidir.
*   **Fayl falsafasi:** Ayniqsa Linux tizimlarida "hamma narsa fayl" tamoyili ishlaydi. Tizim uchun internetdan kelayotgan ma'lumotni qabul qilish yoki jo‘natish — bu shunchaki bitta maxsus faylga (soketga) matn yozish (`write`) va undan o‘qish (`read`) bilan barobar. Ichki murakkab jarayonlarni (paketlarni TCP ga o‘rash, xatolarni tekshirish) Operatsion Tizimning o‘zi (Network Stack) bo‘yniga oladi.

### 2. Python dagi `socket` kutubxonasi qanday ko‘prik o‘rnatadi?

Python dagi `socket` moduli — bu OS ichida C tilida yozilgan, tushunish juda qiyin bo‘lgan pastki darajadagi API (Berkeley Sockets) buyruqlarini Python'ning qulay, insonbop obyektlariga aylantirib beruvchi **tarjimon**.

Siz Python'da bir necha qator kod yozganingizda, u orqa fonda OS bilan qanday gaplashishini bosqichma-bosqich ko‘ramiz:

1.  **`s = socket.socket()` (Yaratish):** 
    Python OS ga: *"Menga internetga chiqish uchun bitta bo‘sh truba (soket) yasab ber"* deydi. OS unga maxsus ID (fayl deskriptor) ajratib, trubaning bir uchini Pythonga beradi.
2.  **`s.bind(('IP', 8080))` (Bog‘lash):** 
    Python OS ga: *"Men 8080-raqamli portni (eshikni) band qildim. Shu eshikka kim xat tashlasa, boshqa dasturlarga bermay, to‘g‘ridan-to‘g‘ri mening trubamga (soketimga) tashlaysan"* deydi.
3.  **`s.listen()` (Quloq tutish):** 
    Python eshikni ochib qo‘yadi va OS ga: *"Men mijozlarni kutishni boshladim, kimdir ulanmoqchi bo‘lsa menga xabar ber"* deb navbatchilikka o‘tadi.
4.  **`s.accept()` (Qabul qilish):** 
    Tashqaridan kimdir ulanganda, OS darchani ochadi. Bu funksiya orqali Python o‘sha aniq mijoz bilan gaplashish uchun maxsus, yakkama-yakka yangi soket (liniya) yaratib oladi.
5.  **`s.send()` va `s.recv()` (Ma'lumot almashish):** 
    Endi eng qiziq joyi: Siz Python'da shunchaki `s.send(b"Salom")` deb yozasiz. Python bu 5 ta baytni OS ga beradi. OS esa uni o‘zining tarmog‘i orqali qadoqlab (TCP paketlar), internet kiygizib, dunyoning narigi chekkasidagi kompyuterga uchirib yuboradi.

**Xulosa:**
Soket — bu kompyuteringiz tashqi dunyo bilan gaplashadigan "og‘iz va quloq" bo‘lsa, Python dagi `socket` kutubxonasi OS ning miyasiga to‘g‘ridan-to‘g‘ri buyruq beruvchi "asab tolasi" hisoblanadi. U tarmoqdagi eng og‘ir pastki daraja (low-level) amallarini yashirib, dasturchiga faqat ma'lumotni o‘qish va yozishdek oson vazifani qoldiradi.

**Javob:**
Kompyuterlararo tarmoq muloqotida (soketlar orqali) ma'lumotlarni bevosita Python matni yoki obyekti (masalan, `list` yoki `dict`) ko‘rinishida yuborishning iloji yo‘q. Buning o‘ziga xos juda muhim fundamental va fizik sabablari bor. 

Bu holatni tushunish uchun **Lego qasri o‘xshatishidan** foydalanamiz. Siz do‘stingizga o‘zingiz qurgan ulkan Lego qasrini (Python obyekti) kichkina quvur (Tarmoq kabeli) orqali yubormoqchisiz. ### 1. Tarmoq faqatgina "1" va "0" larni taniydi (Fizik sabab)

Ular faqatgina yorug‘lik chaqnashlari yoki elektr signallari (elektr bor = 1, elektr yo‘q = 0) orqaligina ma'lumot tashiydi. Shuning uchun kompyuterdan chiqib ketayotgan har qanday narsa oxir-oqibat universal tilga — **baytlar (bytes) oqimiga** aylanishi majburiydir.

### 2. Dasturlash tillari orasidagi farq (Mantiqiy sabab)
Tasavvur qiling, siz Python dasturidan turib Java yoki C++ da yozilgan serverga xabar jo‘natyapsiz. 
*   Pythondagi `Dictionary` (lug‘at) obyekti xotirada qanday saqlanishini faqatgina Python'ning o‘zi tushunadi. 
*   Agar siz bu obyektni xomligicha jo‘natsangiz, C++ serveri bu kelgan narsani o‘qiy olmaydi (chunki unda Dictionary o‘rniga Map degan boshqacha tuzilma bor).
**Baytlarga o‘girish (Encoding / Serialization)** ma'lumotni hech qaysi tilga tegishli bo‘lmagan, barcha uchun tushunarli bo‘lgan neytral (xalqaro) formatga tushirib beradi.

### 3. Kodlash standartlari (Encoding)
Oddiy matnni (`"Salom"`) olaylik. Kompyuterlar harflarni bilmaydi, ular raqamlarni biladi xolos. "S" harfi qaysi raqamga teng ekanligini ikkala kompyuter ham bir xil tushunishi kerak. Shuning uchun matnlar **UTF-8** kabi xalqaro standartlar yordamida baytlarga o‘giriladi (`.encode('utf-8')`). 
*   Siz "S" ni yuborasiz -> Encode uni `0x53` (bayt) ga aylantiradi.
*   Tarmoq `01010011` ni tashiydi.
*   Qabul qiluvchi `01010011` ni olib, uni `0x53` deb tushunadi va uni Decode (`.decode('utf-8')`) qilib ekranga yana "S" harfini chiqaradi.

### Xulosa: Amaliyotda bu qanday qilinadi?

1.  **Oddiy matnlar uchun:** `xabar.encode('utf-8')` qilib baytga (`b"xabar"`) aylantiramiz. Olganda esa `xabar.decode('utf-8')` qilamiz.
2.  **Murakkab Python obyektlari uchun (Masalan ro‘yxatlar, lug‘atlar):** Avval ularni oddiy matnga (JSON) o‘giramiz (`json.dumps(obj)`), keyin esa u matnni baytga o‘giramiz. Olganda esa jarayonni teskarisiga bajaramiz.

**15-savol:** Server dasturini ishga tushirishda IP manzil hamda Port tushunchalarining aniq vazifasi nima va nega serverni lokal tarmoqda barchaga ko‘rinadigan qilish uchun IP manzilni “0.0.0.0” ga sozlash qo‘llaniladi?

*(Sizda savollar raqami biroz adashib ketgan ko‘rinadi, xronologiya bo‘yicha buni 15-savol deb belgilab oldim).*

---

**Javob:**
Tarmoq dasturlashda IP manzil va Port tushunchalari server kompyuterining internetdagi (yoki lokal tarmoqdagi) aniq **geografik va jismoniy koordinatalari** hisoblanadi. Ularning vazifasini hayotiy analogiya bilan juda oson tushunish mumkin:

### 1. IP manzil va Portning aniq vazifasi nima?

Tasavvur qiling, sizga kuryer orqali muhim posilka (tarmoq xabari) kelishi kerak.
*   **IP Manzil (Masalan: `192.168.1.15` yoki `142.250.190.46`):** Bu siz yashaydigan **ko‘p qavatli binoning aniq manzili** (Toshkent shahar, Navoiy ko‘chasi, 10-uy). IP manzil kompyuterni (yoki serverni) butun boshli tarmoq ichidan topib borish uchun xizmat qiladi. Kuryer aynan shu IP manzil orqali qaysi kompyuterga borishni biladi.
*   **Port (Masalan: `80`, `443`, `8080`):** Bu o‘sha ko‘p qavatli binodagi **sizning kvartirangiz (eshigingiz) raqami**. Bitta kompyuter ichida bir vaqtning o‘zida ham veb-sayt (80-port), ham ma'lumotlar bazasi (5432-port), ham o‘yin serveri (27015-port) ishlashi mumkin. Kuryer binoga kelgach, xatni aynan qaysi dasturga (qaysi xonaga) berish kerakligini Port orqali aniqlaydi.

### 2. Nega IP manzil "0.0.0.0" ga sozlanadi?

Dasturchilar ko‘pincha server kodini yozib ishga tushirayotganda `app.run(host='0.0.0.0', port=8080)` degan qatorni yozishadi. Bu nima degani?

Kompyuterda odatda bitta emas, bir nechta "tarmoq eshiklari" (tarmoq interfeyslari) bo‘ladi. Serverni ishga tushirishda IP manzilni qanday tanlash, uning kimlarga xizmat ko‘rsatishini belgilab beradi:

*   **`127.0.0.1` (Localhost - Faqat o‘zim uchun):** Agar serverni shu manzil bilan ishga tushirsangiz, kompyuter faqat **o‘zining ichidan** kelgan so‘rovlarni qabul qiladi. Agar siz telefoningizdan (shu Wi-Fi da bo‘lsangiz ham) kompyuterga ulanmoqchi bo‘lsangiz, server sizni kiritmaydi. Bu kod yozish va xavfsiz sinovdan o‘tkazish uchun kerak.
*   **`192.168.1.15` (Lokal IP - Faqat bitta eshik):** Agar serverni faqatgina Wi-Fi ga ulangan IP orqali ishga tushirsangiz, u faqatgina Wi-Fi orqali kelganlarni qabul qiladi. Agar kimdir LAN kabeli orqali (boshqa IP bilan) ulansa, ishlamaydi.
*   **`0.0.0.0` (Barcha eshiklar lang ochiq):** Bu **"Wildcard"** (maxsus belgi) hisoblanadi. U kompyuterga shunday buyruq beradi: *"Menda qanday tarmoq interfeysi bo‘lsa (Wi-Fi, kabel, Localhost, VPN) — hammasidan kelayotgan mijozlarni qabul qilaver!"*. 

**Hayotiy ssenariy:** Siz noutbukda yangi veb-sayt (Python+PyQt5 yoki Flask) yozdingiz va uni telefoningizda qanday ko‘rinishini sinab ko‘rmoqchisiz. Agar serverni `127.0.0.1` da yoqsangiz, telefon uni ko‘rmaydi. Agar `0.0.0.0` da yoqsangiz, endi telefon brauzeriga kompyuterning IP manzilini yozish orqali bemalol ulanishingiz va saytni telefoningizda ham ko‘rishingiz mumkin bo‘ladi.

**16-savol:** Bitta server dasturiga bir vaqtning o‘zida o‘nlab mijozlar (client) ulana olishi va ma’lumot almashishi uchun arxitekturani qanday tashkil etish kerak va bunda qanday muammolar kelib chiqadi?

---

**Javob:**
Oddiy tarzda yozilgan tayanch server dasturi (sinxron server) bir vaqtning o‘zida faqat **bitta** mijozga xizmat ko‘rsata oladi. Chunki uning ishlash mantig‘i ketma-ketlikka asoslangan: u bitta mijozni qabul qiladi (`accept()`), uning xabarini kutadi (`recv()`) va javob qaytarmaguncha boshqa mijozlarni eshik tagida (navbatda) kutib turishga majbur qiladi.

Bir vaqtning o‘zida o‘nlab, yuzlab mijozlarga parallel xizmat ko‘rsatish uchun server arxitekturasini tubdan o‘zgartirish talab etiladi. Buning asosan ikkita mashhur yondashuvi mavjud:

### 1. Ko‘p oqimli arxitektura (Multi-threading)
Bu an'anaviy va tushunish uchun eng oson yondashuv.
*   **Qanday ishlaydi?** Serverning bitta "Asosiy oqimi" (Main Thread) doimiy ravishda eshik og‘zida poyloqchilik qiladi (`accept()`). Yangi mijoz kelishi bilan u darhol yangi, mustaqil "Ishchi oqim" (Worker Thread) yaratadi va mijozni o‘sha ishchiga topshirib, o‘zi yana eshikka qaytadi. Shunday qilib, 50 ta mijoz kelsa, 50 ta mustaqil parallel oqim paydo bo‘ladi.
*   
*   **Qachon ishlatiladi?** Har bir mijoz bilan aloqa uzoq vaqt davom etadigan va uzilmaydigan jarayonlarda (masalan, fayl yuklab olish yoki jonli chatlar). Python'da bu `threading` moduli orqali amalga oshiriladi.

### 2. Asinxron voqealar sikli (Asynchronous I/O)
Bu zamonaviy, o‘ta yuqori unumdorlikka ega yondashuv (Node.js, Nginx va Python'ning `asyncio` moduli shu asosda ishlaydi).
*   **Qanday ishlaydi?** Bu yerda o‘nlab oqimlar yaratilmaydi. Faqat bitta oqim va bitta "Voqealar sikli" (Event Loop) bo‘ladi. Server mijozdan xabar kutayotganda qotib turmaydi, balki darhol boshqa mijozning ishiga o‘tib ketadi. Ya'ni server o‘z vaqtini mijozlar o‘rtasida millisoniyalarda taqsimlab, xuddi hammaga bir vaqtda xizmat ko‘rsatayotgandek tasavvur uyg‘otadi.
*   
*   **Qachon ishlatiladi?** Bir vaqtning o‘zida 10,000 lab mijozlar ulanishi kerak bo‘lgan, lekin xabarlar juda tez va qisqa uzatiladigan tizimlarda (masalan, Telegram xabarlari, onlayn o‘yin signallari).

### Kelib chiqadigan xavf va muammolar

Parallel arxitekturani qurish o‘ziga yarasha juda murakkab "bosh og‘riqlarni" keltirib chiqaradi:

#### A. Ma'lumotlar to‘qnashuvi (Race Conditions)
Aytaylik, serverdagi bitta bank hisobida 100 ming so‘m pul bor. Ikkita turli joydan (ikkita parallel thread'dan) bir vaqtning o‘zida mijozlar 100 ming so‘mni yechib olish tugmasini bosdi. Ikkala thread ham bir xil millisoniyada hisobni tekshiradi (pul yetarli!) va yechib oladi. Natijada mijoz 200 ming so‘m yechib oladi, bazada esa xatolik yuz beradi. 
*   **Yechimi:** Bitta obyektni bir nechta oqim o‘zgartirmoqchi bo‘lsa, "Qulflar" (Locks / Mutex) ishlatish kerak. Ya'ni birinchi mijoz hisobga kirdimi, baza qulflanadi, ikkinchi mijoz u chiqib ketmaguncha kutib turadi.

#### B. Resurslarning tugab qolishi (Thread Exhaustion)
Har bir yangi Thread yaratish kompyuterning operativ xotirasidan (RAM) ma'lum joy va protsessor vaqtini oladi. Agar serveringizga bir vaqtda 5000 ta mijoz ulansa va siz 5000 ta thread yaratsangiz, operatsion tizim ularni boshqara olmay (Context Switching qilaverib) qulab tushadi (Crash).
*   **Yechimi:** Har bir mijoz uchun yangi oqim yaratmasdan, "Thread Pool" (Oqimlar hovuzi) ishlatish. Masalan, oldindan qat'iy qilib 100 ta ishchi oqim yaratib qo‘yiladi. 5000 ta mijoz kelsa ham, shu 100 ta ishchi ularga navbat bilan xizmat ko‘rsatadi.

#### C. Qotib qolish (Deadlocks)
Bu holat ikkita parallel jarayon bir-birini cheksiz kutib qolganda yuz beradi. Masalan: A mijoz B mijozdan pul o‘tkazmasini kutayotgani uchun tranzaksiyani ochiq ushlab turibdi, B mijoz esa A mijoz o‘zining tizimidan chiqishini kutib turibdi. Ikkisi ham bir-biriga yo‘l bermaydi va butun server qotib qoladi.


**17-savol:** Soketlar bilan ishlashda bloklanuvchi (blocking) rejim va kutib turmaydigan (non-blocking) rejim dasturning ishlash mantig‘iga va tezligiga qanday ta’sir ko‘rsatadi?

---

**Javob:**
Soketlardagi "Bloklanuvchi" (Blocking) va "Kutib turmaydigan" (Non-blocking) rejimlar dasturning tarmoqqa nisbatan naqadar **"sabrli"** yoki **"shoshqaloq"** ekanligini belgilab beruvchi eng asosiy tushunchalardir.

Ular dastur logikasi va tezligini mutlaqo ikki xil o‘zanga burib yuboradi. Keling, ularni hayotiy misol — **"Pochtalyonni kutish"** orqali tushunamiz:

### 1. Bloklanuvchi rejim (Blocking Mode) – "Sabrli kutish"

Python'da soket yaratganingizda u sukut bo‘yicha (default) aynan shu rejimda ishlaydi. 
*   **Qanday ishlaydi?** Siz dasturda `mijoz, manzil = server.accept()` (yangi mijozni kutish) yoki `ma_lumot = mijoz.recv(1024)` (xabar kelishini kutish) deb yozsangiz, dastur aynan shu qatorga kelganda **to‘xtaydi (muzlaydi)**. Toki tarmoqdan haqiqatdan ham kimdir ulanmaguncha yoki xabar yubormaguncha, dastur keyingi qatorga o‘tmaydi. Xuddi pochtalyon kelmaguncha eshik tagida qimirlamay kutib o‘tirgandek.
*   **Mantiqqa ta'siri:** Kodni yozish va o‘qish juda oson (ketma-ketlik chiroyli). Lekin, bitta mijozdan xabar kutayotganda dastur qotib qolgani uchun, u boshqa mijozlarga xizmat ko‘rsata olmaydi. Agar GUI (oyna) dastur bo‘lsa, oyna "Not Responding" bo‘lib oqarib qoladi.
*   **Tezlikka ta'siri:** Bitta mijoz bilan ishlashda yaxshi, lekin ko‘p mijozlarga xizmat ko‘rsatish kerak bo‘lganda (agar Multi-threading ishlatilmasa) tezlik falaj bo‘ladi.

### 2. Kutib turmaydigan rejim (Non-blocking Mode) – "Shoshqaloq tekshirish"

Bu rejimni yoqish uchun dasturda maxsus `socket.setblocking(False)` buyrug‘i yoziladi.
*   **Qanday ishlaydi?** Dastur `recv()` ga kelganda tarmoqqa qaraydi: ma'lumot bormi? Bor bo‘lsa darhol oladi. **Yo‘q bo‘lsa, umuman kutib o‘tirmaydi!** U darhol operatsion tizimdan "Hozircha hech narsa yo‘q" degan xatoni (`BlockingIOError`) oladi va darhol keyingi qatorga o‘tib ketib, boshqa ishlarni qilaveradi. Xuddi pochtani pochta qutisidan tekshirib, bo‘sh bo‘lsa darhol uy ishlarini qilishda davom etgan kabi.
*   **Mantiqqa ta'siri:** Kod ancha murakkablashadi. Siz tinimsiz ravishda xatoliklarni ushlab turuvchi `try/except` yoki maxsus `select`/`epoll` modullarini ishlatib, cheksiz sikl ichida barcha mijozlarni bitta-bitta aylanib tekshirib (Polling) chiqishingiz kerak bo‘ladi. 
*   **Tezlikka ta'siri:** Bu rejim orqali yozilgan serverlar **juda tez va yengil** bo‘ladi. Bitta oqimning o‘zi minglab mijozlardan bir soniya ichida xabar kelgan-kelmaganini so‘rab chiqishga ulguradi va xotira (RAM) tejab qolinadi. Zamonaviy `asyncio` kutubxonalari aynan shu rejim ustiga qurilgan.

**18-savol:** Internet yoki tarmoq uzilib qolishi, yoxud mijozning birdan javob bermay qo‘yishi (timeout) kabi kutilmagan favqulodda holatlarda dastur “qulab tushmasligi” uchun qanday ehtiyot choralari ko‘riladi?

---

**Javob:**
Tarmoq dasturlashdagi eng katta va eng xavfli xato — bu **"Internet doim barqaror ishlaydi"** deb o‘ylashdir. Aslida tarmoq har doim ishonchsiz: kabel uzilib ketishi mumkin, Wi-Fi o‘chib qolishi mumkin yoki mijozning kompyuteri to‘satdan qotib qolishi (yoki tokdan uzilishi) mumkin.

Bunday holatlarda dastur qulab tushmasligi (Crash bo‘lmasligi) va o‘zini tiklay olishi (Resilience) uchun quyidagi 4 ta asosiy "himoya yostiqchalari" qo‘llaniladi:

### 1. Xatoliklarni ushlab qolish (Try/Except bloklari)
Bu eng birinchi va eng muhim qadam. Tarmoqqa tegishli har qanday operatsiya (`connect`, `send`, `recv`) albatta `try...except` bloki ichiga olinishi shart.
*   Agar dastur tarmoq uzilgan vaqtda nimanidir jo‘natmoqchi bo‘lsa, Operatsion Tizim darhol xatolik otadi (`ConnectionResetError`, `BrokenPipeError`).
*   Agar `try/except` bo‘lmasa, butun dasturingiz (va uning oynasi) yopilib ketadi.
*   Bu blok orqali biz xatoni jimgina "yutib yuboramiz" va dasturga: *"Hechqisi yo‘q, tarmoq uzilibdi, ishlashda davom etaver"* deb buyruq beramiz.

### 2. Kutish vaqtini chegaralash (Timeouts)
Tasavvur qiling, server mijozga fayl yubordi va undan "Oldim" degan javobni kutmoqda (`recv()`). Aynan shu soniyada mijozning chirog‘i o‘chib qoldi. Mijoz kompyuteri aloqani rasman uzishga ulgurmadi. Buni **"Zombi ulanish" (Half-open connection)** deyiladi.
*   Agar dastur bloklanuvchi (blocking) rejimda bo‘lsa, u cheksiz muddatgacha — soatlab, kunlab o‘sha mijozdan javob kutib qotib qolaveradi.
*   **Yechim:** `socket.settimeout(5.0)` buyrug‘i beriladi. Bu dasturga: *"Faqat 5 soniya kut, agar mijozdan sado chiqmasa, uni 'o‘lik' deb hisobla va ulanishni majburan uzib tashla"* degan buyruqdir. Bu tizimni qotib qolishdan asraydi.

### 3. "Tomir urishi" (Heartbeat / Keep-Alive) mexanizmi
Ba'zida tarmoq ulangan, lekin internet ishlamayotgan bo‘lishi mumkin. Dastur buni darhol seza olmaydi.
*   **Yechim:** Ikkala tomon (Server va Mijoz) o‘zaro kelishib oladi va fonda har 10 soniyada bir-biriga bo‘m-bo‘sh, juda kichik paket: *"Men tirikman (Ping)"* deb yuborib turadi. Ikkinchi tomon bunga *"Men ham (Pong)"* deb javob qaytaradi.
*   Agar ketma-ket 3 ta "Ping" ga "Pong" javobi kelmasa, dastur aloqa uzilganini o‘zi tushunib yetadi va foydalanuvchiga xabar chiqaradi.

### 4. Avtomatik qayta ulanish va Backoff (Auto-Reconnect)
Dastur aloqa uzilganini tushunib yetgach, vahimaga tushib yopilib ketmaydi. U o‘zini **"Kutish rejimiga"** o‘tkazadi va orqa fonda serverga qayta ulanishga harakat qila boshlaydi.
*   **Exponential Backoff (o‘sib boruvchi kutish):** Serverni tinimsiz so‘rovlar bilan qiynamaslik uchun aqlli dasturlar qayta ulanish oralig‘ini oshirib boradi: avval 1 soniya kutadi, o‘xshamasa 2 soniya, keyin 4 soniya, 8 soniya va hokazo. Internet qaytib kelishi bilan dastur hech narsa ko‘rmagandek o‘z ishini davom ettirib ketaveradi.

---

**Javob:**
Oddiy TCP yoki UDP soketlar orqali yuborilgan ma'lumotlar tarmoqda **ochiq matn (plain text)** ko‘rinishida sayohat qiladi. Bu degani, siz "Parolim: 12345" deb yuborsangiz, u kabeldan aynan shu harflar ko‘rinishida o‘tadi. Bu esa xavfsizlik nuqtai nazaridan halokatli hisoblanadi.

### 1. Ochiq ma'lumot almashish qanday xavflarni tug‘diradi?

Tarmoqdagi bu zaiflik asosan **"o‘rtadagi Odam" (Man-in-the-Middle - MitM)** hujumlariga yo‘l ochib beradi.

*   **Tutib olish (Packet Sniffing):** Siz kafelardagi bepul Wi-Fi dan foydalanayotganingizda, o‘sha tarmoqqa ulangan har qanday havaskor xaker "Wireshark" kabi maxsus dasturlar yordamida sizning kompyuteringizdan ruterga qarab ketayotgan barcha xatlarni havoda tutib olishi va o‘qishi mumkin. Sizning yozishmalaringiz, bank kartalaringiz, parollaringiz ochiq-oydin ko‘rinib turadi.
*   **Ma'lumotni o‘zgartirish (Tampering):** Xaker nafaqat xatni o‘qishi, balki uni o‘zgartirib yuborishi ham mumkin. Masalan, siz serverga "Akamning hisobiga 100$ o‘tkaz" degan xat yubordingiz. Xaker havoda uni ushlab, "Xakerning hisobiga 1000$ o‘tkaz" deb tahrirlaydi va serverga yuboradi. Server esa buni siz yubordingiz deb o‘ylaydi.
*   **Soxta Server (Spoofing):** Xaker o‘zini siz ulanmoqchi bo‘lgan haqiqiy server (masalan, Google) qilib ko‘rsatishi va barcha ma'lumotlaringizni o‘ziga yig‘ib olishi mumkin.

### 2. SSL/TLS protokollarini ulash nima maqsadda qilinadi?

Yuqoridagi dahshatli muammolarni hal qilish uchun **SSL (Secure Sockets Layer)** va uning zamonaviy avlodi **TLS (Transport Layer Security)** o‘ylab topilgan. Biz har kuni ko‘radigan brauzerdagi qulf belgisi (`https://`) aynan shu texnologiya ishlashini bildiradi.

Python'da yozgan oddiy soketingizni `ssl.wrap_socket()` funksiyasi orqali SSL qobig‘iga o‘rasangiz, u quyidagi uchta mo‘jizani yaratadi:

1.  **Harbiy darajadagi Shifrlash (Encryption):**
    Siz yuborgan "Parolim: 12345" xati kompyuteringizdan chiqishidanoq o‘qib bo‘lmaydigan abrakadabraga (masalan, `x8f@!z9Q...`) aylanadi. Xaker uni ushlab olsa ham, kalitsiz bu matnni yechish uchun dunyodagi eng kuchli superkompyuterga minglab yillar kerak bo‘ladi.
2.  **Yaxlitlik (Data Integrity):**
    SSL har bir paketga raqamli muhr (MAC) bosadi. Agar xaker yo‘lda xatning bitta harfini o‘zgartirsa ham, muhr buziladi va Server bu xatni qabul qilmay, darhol aloqani uzadi.
3.  **Haqiqiylikni tasdiqlash (Authentication):**
    Server mijozga o‘zining "Raqamli Sertifikati"ni (Pasportini) ko‘rsatadi. Bu sertifikat ishonchli xalqaro tashkilotlar (CA) tomonidan tasdiqlangan bo‘ladi. Mijoz bu orqali rostdan ham xakerning soxta serveriga emas, balki aynan haqiqiy manzilga ulanganiga 100% ishonch hosil qiladi.

**20-savol:** Eng katta muammolardan biri sifatida, tarmoqdan ma’lumot kutib turgan funksiyalarni PyQt5 ning vizual interfeysi bilan qanday qilib shunday birlashtirish kerakki, interfeys hech qachon qotib qolmasin?

---

**Javob:**
Bu muammo (GUI qotib qolishi) har bir boshlovchi dasturchi to‘qnash keladigan eng katta to‘siqdir. Muammoning ildizi shundaki: **PyQt5 ning oynasi (Event Loop) ham, tarmoqdan xabar kutuvchi `recv()` funksiyasi ham "Bloklanuvchi" (kutib turuvchi) tabiatga ega.** Agar ularni bitta joyga (Asosiy oqimga) qamab qo‘ysangiz, ular bir-birini falaj qilib qo‘yadi.

Interfeys **hech qachon qotib qolmasligi** uchun, arxitekturani **"QThread + pyqtSignal"** qolipida, qat'iy uchta qoidaga amal qilgan holda qurish kerak:

### 1-qoida: Qora ishlarni orqa fonga suring (QThread)
Hech qachon `socket.recv()`, `requests.get()` yoki uzoq davom etuvchi bazaga ulanish jarayonlarini asosiy oynangizdagi tugma bosilganda ishlaydigan funksiya ichiga yozmang.
Buning o‘rniga, tarmoq bilan gaplashadigan barcha kodlarni `QThread` dan meros olgan mutlaqo alohida sinf (class) ning `run()` metodi ichiga yashiring. Bu "Ishchi oqim" (Worker Thread) deb ataladi.

### 2-qoida: Qat'iy chegara (Asosiy oqimga teginmang)
Orqa fondagi Ishchi oqim (QThread) tarmoqdan ma'lumotni qabul qilib olgach, u to‘g‘ridan-to‘g‘ri ekrandagi oynaga yozishga haqqi yo‘q. Ya'ni, QThread ichidan turib `self.label.setText("Xabar keldi")` deyish qat'iyan taqiqlanadi. Bu xotira to‘qnashuviga (Crash) olib keladi. Interfeys elementlariga faqatgina Asosiy oqim (Main Thread) tegishi mumkin.

### 3-qoida: Xavfsiz ko‘prik (pyqtSignal)
Ishchi oqim ma'lumotni oldi, lekin oynaga yozolmaydi. Unda nima qiladi? U PyQt5 ning maxsus xabarchisi — **`pyqtSignal`** dan foydalanadi. 
Ishchi oqim ma'lumotni signalga o‘raydi va havoga otadi (`emit()`). Asosiy oqimdagi interfeys esa o‘sha signalni o‘zining tayyor xaltachasi (`Slot`) orqali tutib oladi va oynani **o‘zi** yangilaydi.

### Arxitekturani kodda ko‘ramiz (Oltin Qolip)

Quyidagi kod har qanday og‘ir tarmoq dasturida GUI qotmasligi uchun andoza (shablon) hisoblanadi:

```python
from PyQt5.QtCore import QThread, pyqtSignal
from PyQt5.QtWidgets import QApplication, QWidget, QVBoxLayout, QPushButton, QLabel
import time, sys

# 1. TARMOQ ISHCHISI (Orqa fon)
class TarmoqOqimi(QThread):
    # Asosiy oqimga matn yuborish uchun signal yaratamiz
    xabar_signali = pyqtSignal(str) 

    def run(self):
        # Bu yerda tarmoqdan javob kutish jarayoni ketadi
        time.sleep(3) # Masalan, 3 soniya kutilmoqda (Blocking!)
        yangi_xabar = "Serverdan javob keldi: 200 OK!"
        
        # 2. To‘g‘ridan-to‘g‘ri UI ga yozmasdan, signal orqali yuboramiz
        self.xabar_signali.emit(yangi_xabar)

# 3. ASOSIY OYNA (Interfeys)
class AsosiyOyna(QWidget):
    def __init__(self):
        super().__init__()
        self.layout = QVBoxLayout(self)
        
        self.label = QLabel("Kutmoqda...")
        self.tugma = QPushButton("Tarmoqqa ulanish")
        self.tugma.clicked.connect(self.tarmoqni_boshlash)
        
        self.layout.addWidget(self.label)
        self.layout.addWidget(self.tugma)
        
        # Oqimni tayyorlab qo‘yamiz
        self.ishchi = TarmoqOqimi()
        # Ishchidan kelgan signalni ekranni yangilovchi slotga ulaymiz
        self.ishchi.xabar_signali.connect(self.ekranni_yangila)

    def tarmoqni_boshlash(self):
        self.label.setText("Yuklanmoqda (oyna qotgani yo‘q!)...")
        self.ishchi.start() # Ishchini ishga tushiramiz

    def ekranni_yangila(self, xabar):
        self.label.setText(xabar) # Kelgan xabarni ekranga yozamiz

app = QApplication(sys.argv)
oyna = AsosiyOyna()
oyna.show()
sys.exit(app.exec_())
```
