**3-bob**

**Saralash (*Sorting*)**

Saralash algoritm tuzishdagi fundamental masalalardan biridir. Ko‘plab samarali algoritmlar saralashdan quyi dastur (*subroutine*) sifatida foydalanadi, chunki elementlar saralangan tartibda bo‘lsa, ma'lumotlarni qayta ishlash ko‘pincha osonroq kechadi.

Masalan, "Massivda ikkita bir xil element bormi?" degan masalani saralash yordamida hal qilish juda oson. Agar massivda ikkita bir xil element bo‘lsa, saralashdan keyin ular yonma-yon kelib qoladi va ularni topish qiyin bo‘lmaydi. Shuningdek, "Massivdagi eng ko‘p uchraydigan element qaysi?" masalasini ham xuddi shunday hal qilish mumkin.

Saralash uchun juda ko‘p algoritmlar mavjud va ular turli xil algoritmik yondashuvlarni qo‘llash bo‘yicha yaxshi misol bo‘la oladi. Samarali umumiy saralash algoritmlari $O(n \log n)$ vaqtda ishlaydi va saralashni quyi dastur sifatida ishlatadigan ko‘plab algoritmlar ham aynan shunday vaqt murakkabligiga ega.

**Saralash nazariyasi (*Sorting theory*)**

Saralashdagi asosiy masala quyidagicha:
$n$ ta elementdan iborat massiv berilgan, sizning vazifangiz elementlarni o‘sish tartibida saralashdir.
Masalan, `1 3 8 2 9 2 5 6` massivi saralangandan keyin quyidagicha ko‘rinishga keladi: `1 2 2 3 5 6 8 9`

**$O(n^2)$ algoritmlar**

Massivni saralash uchun oddiy algoritmlar $O(n^2)$ vaqtda ishlaydi. Bunday algoritmlar qisqa bo‘ladi va odatda ikkita ichma-ich tsikldan iborat bo‘ladi. Mashhur $O(n^2)$ saralash algoritmlaridan biri bu "Pufakchali saralash" (*bubble sort*) bo‘lib, unda elementlar o‘z qiymatlariga ko‘ra massiv bo‘ylab "pufakcha" kabi tepaga qalqib chiqadi.

Pufakchali saralash $n$ ta bosqichdan iborat. Har bir bosqichda algoritm massiv elementlari bo‘ylab aylanib chiqadi. Qachonki noto‘g‘ri tartibda turgan ikkita ketma-ket element topilsa, algoritm ularning o‘rnini almashtiradi (*swap*).

1-bosqichdan so‘ng eng katta element o‘zining to‘g‘ri joyiga borib qoladi va umuman olganda, $k$ ta bosqichdan so‘ng eng katta $k$ ta element to‘g‘ri pozitsiyalarda bo‘ladi. Shunday qilib, $n$ ta bosqichdan keyin butun massiv saralanadi.

**Inversiyalar (*Inversions*)**

Pufakchali saralash massivdagi doimo ketma-ket turgan elementlar o‘rnini almashtiradigan algoritmga misoldir. Bunday algoritmning vaqt murakkabligi har doim kamida $O(n^2)$ bo‘lishi ma'lum bo‘ldi, chunki eng yomon holatda massivni saralash uchun $O(n^2)$ ta almashtirish (*swaps*) talab qilinadi.

Saralash algoritmlarini tahlil qilishda foydali tushuncha bu inversiyadir: $a < b$ va $array[a] > array[b]$ shartni qanoatlantiruvchi $(array[a], array[b])$ massiv elementlari juftligi, ya'ni noto‘g‘ri tartibda turgan elementlar. Masalan, `1 2 2 6 3 5 9 8` massivida uchta inversiya mavjud: $(6,3), (6,5)$ va $(9,8)$. Inversiyalar soni massivni saralash uchun qancha ish bajarish kerakligini ko‘rsatadi. Inversiyalar qolmaganda massiv to‘liq saralangan bo‘ladi. Boshqa tomondan, agar massiv elementlari teskari tartibda bo‘lsa, inversiyalar soni eng katta bo‘ladi: $O(n^2)$.
Ketma-ket turgan noto‘g‘ri tartibdagi elementlar o‘rnini almashtirish massivdan aynan bitta inversiyani olib tashlaydi.

**$O(n \log n)$ algoritmlar**

Massivni faqat ketma-ket elementlarni almashtirish bilan cheklanmagan algoritmlar yordamida $O(n \log n)$ vaqtda samarali saralash mumkin. Shunday algoritmlardan biri rekursiyaga asoslangan qo‘shilishli saralashdir (*merge sort*).

Merge sort $array[a...b]$ qism-massivini quyidagicha saralaydi:

1. Agar $a = b$ bo‘lsa, hech narsa qilmang, chunki qism-massiv allaqachon saralangan.
2. O‘rta element o‘rnini hisoblang: $k = \lfloor(a+b)/2\rfloor$.
3. $array[a...k]$ qism-massivini rekursiv ravishda saralang.
4. $array[k+1...b]$ qism-massivini rekursiv ravishda saralang.
5. Saralangan $array[a...k]$ va $array[k+1...b]$ qism-massivlarini bitta saralangan $array[a...b]$ qism-massiviga birlashtiring (*merge*).

Merge sort juda samarali algoritmdir, chunki u har bir qadamda qism-massiv hajmini yarmiga qisqartiradi. Rekursiya $O(\log n)$ ta darajadan iborat va har bir darajani qayta ishlash $O(n)$ vaqt oladi. Qism-massivlarni birlashtirish chiziqli vaqtda amalga oshadi, chunki ular allaqachon saralangan.

**Saralashning quyi chegarasi (*Sorting lower bound*)**

Massivni $O(n \log n)$ dan tezroq saralash mumkinmi? Agar biz faqat massiv elementlarini bir-biri bilan taqqoslashga asoslangan algoritmlar bilan cheklansak, bu imkonsiz ekanligi ma'lum bo‘ladi. Taqqoslashlar daraxti modeliga ko‘ra eng yomon holatdagi qadamlar soni kamida $n \log n$ bo‘lishi isbotlangan.

**Sanash orqali saralash (*Counting sort*)**

$n \log n$ quyi chegarasi elementlarni bir-biri bilan taqqoslamaydigan algoritmlarga taalluqli emas. Bunga misol sifatida, agar massivdagi har bir element $0...c$ oralig‘idagi butun son bo‘lsa va $c = O(n)$ bo‘lsa, massivni $O(n)$ vaqtda saralaydigan *Counting sort* (Sanash orqali saralash) algoritmini keltirish mumkin. Algoritm indekslari asl massiv elementlariga teng bo‘lgan sanoq massivini yaratadi va har bir element necha marta uchrashini sanab chiqadi.

**C++ da saralash (*Sorting in C++*)**

Musobaqalarda o‘z qo‘lingiz bilan yozilgan saralash algoritmidan foydalanish deyarli har doim yomon g‘oyadir, chunki dasturlash tillarida tayyor va yaxshi optimallashtirilgan funksiyalar mavjud. Masalan, C++ kutubxonasida `sort` funksiyasi bor:

```cpp
vector<int> v = {4,2,5,3,5,8,3};
sort(v.begin(), v.end()); // O'sish tartibida
sort(v.rbegin(), v.rend()); // Kamayish (teskari) tartibida

```

C++ da Juftliklar (*pair*) va Kortejlar (*tuple*) avtomatik ravishda birinchi elementi bo‘yicha, agar ular teng bo‘lsa, ikkinchi elementi bo‘yicha o‘sish tartibida saralanadi.
Foydalanuvchi tomonidan yaratilgan tuzilmalar (*User-defined structs*) uchun taqqoslash operatori `operator<` ni yozish kerak. Yoki maxsus taqqoslash funksiyasini (*comparison function*) `sort` ga uchinchi parametr sifatida berish mumkin.

**Binar qidiruv (*Binary search*)**

Elementni qidirishning odatiy usuli bu massiv bo‘ylab aylanib chiquvchi `for` tsiklidir ($O(n)$). Lekin agar massiv saralangan bo‘lsa, qidiruvni ancha tezroq amalga oshirish mumkin. Binar qidiruv saralangan massivdan elementni $O(\log n)$ vaqtda izlab topadi.

* **1-usul:** Faol qismni yarmiga bo‘lib borish (`while (a <= b) { int k = (a+b)/2; ... }`).
* **2-usul:** Sakrashlar orqali (*Jump method*). Qidiruv chapdan o‘ngga qarab sakraydi va maqsadli elementga yaqinlashganda tezlikni (sakrash masofasini) kamaytiradi. Boshlang‘ich sakrash $n/2$, keyin $n/4, n/8 \dots 1$ gacha.

**C++ funksiyalari:**

* `lower_bound`: qiymati kamida $x$ ga teng bo‘lgan birinchi elementga ko‘rsatkich qaytaradi.
* `upper_bound`: qiymati qat'iy $x$ dan katta bo‘lgan birinchi elementga ko‘rsatkich qaytaradi.

**Eng kichik yechimni topish (*Finding the smallest solution*)**
Binar qidiruvning eng muhim qo‘llanilishi — bu funksiya qiymati o‘zgaradigan pozitsiyani topishdir. Faraz qilaylik, biz masala uchun to‘g‘ri yechim bo‘la oladigan eng kichik $k$ ni topmoqchimiz. Bizda $x$ yaroqli bo‘lsa `true` qaytaradigan `ok(x)` predikati bor:
`false, false, ..., false, true, true, ...`
Binar qidiruv yordamida biz oxirgi `false` ni topib, unga $+1$ qo‘shish orqali birinchi `true` ($k$) ni $O(\log z)$ vaqtda topa olamiz. (Buni "Javob bo'yicha binar qidiruv" deyishadi).

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

**1. Kutubxonadagi `std::sort` nega bunchalik zo'r?**

Musobaqada hech qachon o'zingiz Merge Sort yoki Quick Sort yozmaysiz. Chunki C++ dagi `std::sort` bu oddiy saralash emas, u **IntroSort** deb ataladigan gibrid algoritmdir.
U ishlashni tezkor Quick Sort dan boshlaydi. Agar rekursiya juda chuqurlashib ketsa (eng yomon holat yuzaga kelsa), u darhol Heap Sort ga o'tib ketadi. Kichik qismlar qolganda esa Insertion Sort ishlatadi. Shuning uchun u har qanday holatda ham barqaror va $O(n \log n)$ tezlikni kafolatlaydi.

**2. Custom Comparator yozishdagi o'lim xatosi (Strict Weak Ordering)**

C++ da struct yoki maxsus saralash funksiyasini yozayotganda `return a <= b;` deb yozish **qatiyan taqiqlanadi!**
C++ standarti `Strict Weak Ordering` ni talab qiladi. Ya'ni ikkita bir xil element kelib qolsa, sizning funksiyangiz qat'iy ravishda `false` qaytarishi shart (faqatgina `<` yoki `>` ishlating). Agar `<=` ishlatsangiz, Codeforces'da kodingiz `Runtime Error` (Segfault) berib qulaydi, chunki saralash algoritmi cheksiz xotiraga o'tib ketadi.

**3. Inversiyalarni hisoblash — Klassik CP masalasi**

Matnda aytilgan "Inversiya" tushunchasi musobaqalarda tez-tez keladi: "Massivni saralash uchun eng kamida nechta qo'shni element almashtirilishi kerak?". O'lcham $N = 10^5$ bo'lsa, buni $O(n^2)$ da sanash imkonsiz.
*CP Yechimi:* Buni hisoblash uchun odatda **Merge Sort** algoritmi biroz o'zgartiriladi. Birlashtirish (merge) jarayonida o'ng massivdagi element chapdagi elementdan kichik bo'lsa, chap massivda qolgan barcha elementlar soni qadar inversiya qo'shiladi. Yoki **Fenwick Tree (BIT)** yordamida $O(n \log n)$ da juda oson sanaladi.

**4. Binary Search 2-usuli (Binary Lifting style)**

Kitobda berilgan *Method 2* (b ni 2 ga bo'lib borish) g'alati tuyulishi mumkin, lekin u CP da juda mashhur. Bu kod asosan grafiklarda va Fenwick Tree'da $O(\log n)$ da kerakli elementni qidirib topish (Binary Lifting) uchun standart andoza hisoblanadi. Unda biz `x` ni 0 dan boshlaymiz va unga turli kuchlarni (powers of 2) qo'shib ko'ramiz. Agar u shartni qanoatlantirsa, uni olamiz, yo'qsa tashlab ketamiz.

**5. `lower_bound` o'rniga nima ishlatish kerak? (Tuzoqlar)**

Agar siz elementlarni oddiy `std::vector` yoki array'da qidirsangiz `std::lower_bound` eng yaxshisidir.
LEKIN, agar siz `std::set` yoki `std::map` ishlatsangiz, ularga umumiy funksiyani BERMANG (`std::lower_bound(s.begin(), s.end(), x)` yozsangiz $O(N)$ ga aylanib TLE beradi!). Buning o'rniga, daraxtning o'ziga tegishli metodni chaqirishingiz shart: `s.lower_bound(x);`. Bu har doim kafolatlangan $O(\log n)$ da ishlaydi.


Asosiy savol shunday: **"Nega biz massivni $O(n \log n)$ dan tezroq saralay olmaymiz?"**

Ushbu isbot faqat **taqqoslashga asoslangan** (*comparison-based*) saralash algoritmlariga (masalan, Merge Sort, Quick Sort, Bubble Sort) tegishli. Ya'ni, algoritm elementlarni faqatgina `x < y` mi yoki yo‘qmi deb so‘rash orqali saralashi kerak. (Counting Sort kabi elementning qiymatini to‘g‘ridan-to‘g‘ri indeks sifatida ishlatadigan algoritmlar bu qoidaga kirmaydi).

### 1. Qarorlar daraxti (*Decision Tree*)

Tasavvur qiling, sizda 3 ta elementdan iborat massiv bor: `[A, B, C]`. Siz ularni o‘sish tartibida joylashtirishingiz kerak. Siz ko‘zingiz bog‘langan holda faqatgina "A, B dan kichikmi?" degan savolni bera olasiz.

Sizning har bir savolingiz va unga olingan "Ha" yoki "Yo‘q" javobingiz bitta **Qarorlar daraxtini** hosil qiladi:

* Asosiy tugunda turibsiz: `A < B` mi?
* Agar **Ha** bo‘lsa, chapga yurasiz. Keyingi savol: `B < C` mi?
* Agar **Yo‘q** bo‘lsa, o‘ngga yurasiz. Keyingi savol: `A < C` mi?

### 2. Barglar (Mumkin bo‘lgan natijalar)

3 ta elementni necha xil usulda ketma-ket joylashtirish mumkin? Matematikada bu o‘rin almashtirishlar (*permutations*) deb ataladi va $n!$ (Faktorial) orqali topiladi.
$3! = 1 \times 2 \times 3 = 6$.
Demak, 3 ta element uchun jami 6 xil har xil boshlang‘ich holat bo‘lishi mumkin: `[A,B,C]`, `[A,C,B]`, `[B,A,C]`, `[B,C,A]`, `[C,A,B]`, `[C,B,A]`.

Saralash algoritmi shu 6 ta holatning qaysi biri berilganidan qat'i nazar, to‘g‘ri javobni topib bera olishi kerak. Demak, biz qurgan Qarorlar daraxtining **eng pastki qismida (barglarida) kamida $n!$ ta nuqta** bo‘lishi shart.

### 3. Daraxt balandligi va Vaqt murakkabligi

Daraxtning balandligi (eng yuqoridan eng pastki barggacha bo‘lgan qadamlar soni) algoritm qanchalik uzoq ishlashini, ya'ni uning **eng yomon holatdagi vaqt murakkabligini (Worst-case time complexity)** bildiradi.

Ikkilik daraxtda (*Binary tree*) har bir qadamda shoxlar 2 taga ko‘payadi: $1 \rightarrow 2 \rightarrow 4 \rightarrow 8 \dots$
Agar daraxtning balandligi $h$ bo‘lsa, u ko‘pi bilan $2^h$ ta bargga ega bo‘lishi mumkin.
Bizning daraxtda kamida $n!$ ta barg bo‘lishi majburiy ekan, demak quyidagi tengsizlik o‘rinli:


$$2^h \ge n!$$

Endi har ikkala tomondan 2 asosga ko‘ra logarifm olamiz:


$$h \ge \log_2(n!)$$


Demak, bizning eng zo‘r algoritmimiz ham eng yomon holatda kamida $\log_2(n!)$ ta taqqoslash amaliga majbur.

### 4. Matematik Fokus: $\log_2(n!)$ o‘zi qanchaga teng?

Kitobdagi matnning eng murakkab joyi shu yerda. Keling, logarifmning ko‘paytma xossasidan foydalanib uni yoyib yozamiz:


$$\log_2(n!) = \log_2(1 \times 2 \times 3 \times \dots \times n)$$

$$\log_2(n!) = \log_2(1) + \log_2(2) + \dots + \log_2(n)$$

Bu yig‘indi aynan qanchaga tengligini bilish qiyin, lekin bizga uning **quyi chegarasi** (kamida qancha bo‘lishi) kerak. Quyi chegarani topish uchun kitob ajoyib mantiq ishlatadi:

1. Yig‘indining boshidagi kichik sonlarni ($1$ dan $n/2$ gacha) butunlay tashlab yuboramiz. (Biz qiymatni ataylab kichraytiryapmiz).
2. Faqat yig‘indining ikkinchi yarmini, ya'ni eng katta $n/2$ ta hadni qoldiramiz:

$$\log_2(n/2) + \log_2(n/2 + 1) + \dots + \log_2(n)$$


3. Qolgan har bir had qat'iyan $\log_2(n/2)$ dan katta yoki unga teng. Shuning uchun barcha hadlarni eng kichigi — $\log_2(n/2)$ bilan almashtiramiz:

$$\underbrace{\log_2(n/2) + \log_2(n/2) + \dots + \log_2(n/2)}_{n/2 \text{ marta}}$$



Bizda $n/2$ ta shunday had bo‘lgani uchun ularning yig‘indisi quyidagiga teng bo‘ladi:


$$\frac{n}{2} \cdot \log_2\left(\frac{n}{2}\right)$$

### Xulosa

Biz isbotladikki:


$$h \ge \log_2(n!) \ge \frac{n}{2} \cdot \log_2\left(\frac{n}{2}\right)$$


Big O notation qoidalariga ko‘ra doimiylar (konstantalar, masalan $1/2$) tashlab yuboriladi. Natijada **$O(n \log n)$** qoladi.

Bu matematik tilda shuni anglatadiki: **Siz qanday aqlli algoritm o‘ylab topmang, agar u faqat elementlarni taqqoslashga asoslangan bo‘lsa, u matematik jihatdan eng yomon holatda $N \log N$ qadamdan tezroq ishlay olmaydi.**

Qarorlar daraxti qanday ishlashini amalda tasavvur qilish uchun quyidagi interaktiv vidjetdan foydalaning. Unda 3 ta element qanday qilib savol-javoblar orqali $3! = 6$ xil holatga ajralishi ko'rsatilgan:
