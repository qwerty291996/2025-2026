**Ochko‘z algoritmlar (*Greedy algorithms*)**
Ochko‘z algoritm har doim o‘sha daqiqada eng yaxshi ko‘ringan tanlovni amalga oshirish orqali masalaning yechimini quradi. Ochko‘z algoritm hech qachon o‘z tanlovlaridan ortga qaytmaydi, balki to‘g‘ridan-to‘g‘ri yakuniy yechimni qurib boradi. Aynan shu sababli, ochko‘z algoritmlar odatda juda samarali (tezkor) ishlaydi.

Ochko‘z algoritmlarni loyihalashdagi qiyinchilik — bu har doim masalaning optimal yechimini beradigan ochko‘z strategiyani topishdadir. Ochko‘z algoritmdagi lokal optimal tanlovlar global darajada ham optimal bo‘lishi kerak. Ochko‘z algoritmning to‘g‘ri ishlashini isbotlash ko‘pincha qiyin vazifa hisoblanadi.

**Tanga masalasi (*Coin problem*)**
Birinchi misol sifatida, bizga bir nechta tangalar to‘plami berilgan va biz tangalardan foydalanib $n$ miqdordagi pul yig‘indisini hosil qilishimiz kerak bo‘lgan masalani ko‘rib chiqamiz. Tangalarning qiymatlari $coins = \{c_1, c_2, \dots, c_k\}$ va har bir tangadan xohlagancha foydalanish mumkin. Bunga eng kamida nechta tanga kerak bo‘ladi?

Masalan, agar tangalar yevro tangalari (sentlarda) bo‘lsa:
$\{1, 2, 5, 10, 20, 50, 100, 200\}$
va $n = 520$ bo‘lsa, bizga kamida to‘rtta tanga kerak bo‘ladi. Optimal yechim — yig‘indisi 520 ga teng bo‘lgan $200 + 200 + 100 + 20$ tangalarini tanlashdir.

**Ochko‘z algoritm**
Masalaning oddiy ochko‘z algoritmi har doim talab qilingan pul yig‘indisi hosil bo‘lguniga qadar imkon qadar eng katta tangani tanlaydi. Bu algoritm misoldagi holatda ishlaydi, chunki biz avval ikkita 200 sentlik tangani, so‘ngra bitta 100 sentlik va oxirida bitta 20 sentlik tangani tanlaymiz. Lekin bu algoritm doim ishlaydimi?

Ma’lum bo‘lishicha, agar tangalar yevro tangalari bo‘lsa, ochko‘z algoritm har doim ishlaydi, ya’ni u doimo eng kam tangalar sonidan iborat yechimni beradi. Algoritmning to‘g‘riligini quyidagicha ko‘rsatish mumkin:
Birinchidan, har bir 1, 5, 10, 50 va 100 tangalari optimal yechimda ko‘pi bilan bir marta qatnashadi, chunki agar yechim ikkita shunday tangani o‘z ichiga olsa, ularni bitta tangaga almashtirib, yaxshiroq yechim olishimiz mumkin edi. Masalan, agar yechim $5+5$ tangalarini o‘z ichiga olsa, biz ularni 10 tangasiga almashtira olardik.
Xuddi shu tarzda, 2 va 20 tangalari optimal yechimda ko‘pi bilan ikki marta qatnashadi, chunki biz $2+2+2$ tangalarini $5+1$ ga va $20+20+20$ tangalarini $50+10$ ga almashtirishimiz mumkin edi. Bundan tashqari, optimal yechim $2+2+1$ yoki $20+20+10$ tangalarini o‘z ichiga olmaydi, chunki biz ularni mos ravishda 5 va 50 ga almashtira olamiz.
Ushbu kuzatishlardan foydalanib, har bir $x$ tangasi uchun $x$ yig‘indisini yoki undan kattaroq yig‘indini faqatgina $x$ dan kichik bo‘lgan tangalar yordamida optimal shakllantirish imkonsiz ekanligini ko‘rsatishimiz mumkin. Shunday qilib, har doim eng katta tangani tanlaydigan ochko‘z algoritm optimal yechimni beradi. Bu misol shuni ko‘rsatadiki, garchi algoritmning o‘zi oddiy bo‘lsa ham, uning to‘g‘ri ishlashini isbotlash qiyin bo‘lishi mumkin.

**Umumiy holat (*General case*)**
Umumiy holatda, tangalar to‘plami har qanday tangalarni o‘z ichiga olishi mumkin va ochko‘z algoritm har doim ham optimal yechim bermasligi mumkin.
Biz ochko‘z algoritm ishlamasligini algoritm xato javob beradigan qarshi misol (counterexample) ko‘rsatish orqali isbotlashimiz mumkin. Agar tangalar $\{1, 3, 4\}$ bo‘lsa va maqsadli yig‘indi 6 ga teng bo‘lsa, ochko‘z algoritm $4+1+1$ yechimini beradi, vaholanki optimal yechim $3+3$ dir.
Umumiy tanga masalasi har qanday ochko‘z algoritm yordamida hal qilinishi mumkinmi yoki yo‘qmi, noma’lum. Biroq, 7-bobda ko‘rib chiqadiganimizdek, ayrim hollarda umumiy masalani har doim to‘g‘ri javob beradigan Dinamik Dasturlash (DP) algoritmi orqali samarali hal qilish mumkin.

**Jadvallashtirish (*Scheduling*)**
Ko‘plab jadvallashtirish masalalarini ochko‘z algoritmlar yordamida hal qilish mumkin. Klassik masala quyidagicha: boshlanish va tugash vaqtlari berilgan $n$ ta tadbir orasidan imkon qadar ko‘p tadbirlarni o‘z ichiga oladigan jadvalni toping (tadbirlarni qisman tanlash mumkin emas).

Buning uchun turli xil ochko‘z algoritmlarni o‘ylab topish mumkin, lekin ulardan qaysi biri har doim ishlaydi?

* **1-algoritm:** Imkon qadar eng qisqa tadbirlarni tanlash. (Xato: Bitta qisqa tadbir ikkita uzunroq tadbirning o‘rtasiga tushib qolib, ularni bloklab qo‘yishi mumkin).
* **2-algoritm:** Imkon qadar erta boshlanadigan tadbirni tanlash. (Xato: Birinchi boshlangan tadbir o‘ta uzoq davom etsa, boshqa barcha tadbirlarni bloklab qo‘yadi).
* **3-algoritm:** Imkon qadar **erta tugaydigan** tadbirni tanlash. Ma’lum bo‘lishicha, aynan shu algoritm **har doim optimal yechimni beradi**. Bunga sabab, birinchi navbatda eng erta tugaydigan tadbirni tanlash har doim optimal tanlovdir. Shundan so‘ng, qolgan tadbirlar uchun ham xuddi shu strategiyadan foydalanib keyingi tadbirni tanlash eng maqbul yo‘ldir.

**Vazifalar va muddatlar (*Tasks and deadlines*)**
Bizga davomiyligi (duration) va muddati (deadline) berilgan $n$ ta vazifa berilgan bo‘lsin. Har bir vazifa uchun biz $d - x$ ochko ishlaymiz, bu yerda $d$ — vazifaning muddati va $x$ — biz vazifani tugatgan vaqt. Qanday qilib eng katta umumiy ochkoni yig‘ishimiz mumkin?
Hayratlanarlisi shundaki, masalaning optimal yechimi muddatlarga umuman bog‘liq emas! To‘g‘ri ochko‘z strategiya — bu shunchaki **vazifalarni ularning davomiyligi bo‘yicha o‘sish tartibida saralab bajarishdir**. Eng qisqa vazifalar birinchi bajarilishi kerak.

**Yig‘indilarni minimallashtirish (*Minimizing sums*)**
Bizga $a_1, a_2, \dots, a_n$ sonlari berilgan va vazifamiz quyidagi yig‘indilarni minimallashtiradigan $x$ ni topish:

* **$c=1$ holati (Modullar yig‘indisi):** $|a_1 - x| + |a_2 - x| + \dots + |a_n - x|$. Bu holda $x$ uchun eng yaxshi tanlov massivning **medianasi** (saralangandan keyingi o‘rtadagi element) hisoblanadi.
* **$c=2$ holati (Kvadratlar yig‘indisi):** $(a_1 - x)^2 + (a_2 - x)^2 + \dots + (a_n - x)^2$. Bu holda $x$ uchun eng yaxshi tanlov sonlarning **o‘rta arifmetigi** (average) hisoblanadi.

**Ma’lumotlarni siqish va Xaffman kodlashi (*Data compression & Huffman coding*)**
Xaffman kodlashi (Huffman coding) — bu berilgan satrni siqish uchun optimal kod (ikkilik daraxt) quradigan ochko‘z algoritmdir. Algoritm satrdagi belgilari chastotasiga (necha marta uchrashganiga) asoslanib binar daraxt quradi va eng ko‘p uchraydigan belgilarga eng qisqa bitlarni, kam uchraydiganlariga esa uzunroq bitlarni o‘zlashtiradi.

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

Ochko‘z algoritm (Greedy) musobaqalardagi eng makkor va qiyin mavzulardan biridir. Chunki uning to‘g‘ri ishlashini isbotlash juda mushkul. Kodni yozish 2 daqiqa oladi, lekin uning nega ishlashini (yoki nega xato berishini) tushunish soatlab vaqt olishi mumkin.

**1. Dinamik Dasturlash (DP) vs Greedy**
"Tanga masalasi" CP dagi eng klassik misoldir. Codeforces musobaqalarida masala berilganda, u Greedy yechiladimi yoki DP bilanmi, shuni ajrata olish muhim.
Qoida: Agar optimal tanlov **har doim yagona bo‘lmasa** yoki qilingan bitta tanlov kelajakdagi boshqa qadamlarni **butunlay o‘zgartirib yuborsa**, Greedy qulab tushadi (Counterexample). Tanga masalasida $\{1, 3, 4\}$ kabi "g‘alati" qiymatlar berilganda Greedy xato beradi. Shuning uchun bunday masalalarni har doim DP bilan yechish tavsiya etiladi.

**2. Jadvallashtirish siri: Har doim o‘ng tamonni (End Time) saralang!**
*Scheduling* masalasi musobaqalarda yuzlab marta uchraydi (masalan, "xonalarga imkon qadar ko‘p odamni joylash", "imkon qadar ko‘p kinolarni ko‘rish"). Boshlang‘ich dasturchilar har doim intervalning boshlanish vaqti (`start time`) bo‘yicha massivni saralashadi va xatoga (WA) tushishadi.
**Oltin qoida:** Har qanday interval / jadvallashtirish masalasida **har doim tugash vaqti (`end time`) bo‘yicha o‘sish tartibida saralang!** **3. Mediana vs O‘rta arifmetik (Median vs Average)**
Bu juda ko‘p CP marafondalarini hayratda qoldiradigan sof matematik faktdir:

* Agar masalada barcha nuqtalardan biror markaziy nuqtagacha bo‘lgan **masofalar yig‘indisini** (Absolute difference sum) eng qisqa qilish so‘ralsa, javob har doim massivning **MEDIANASI** bo‘ladi.
* Agar masalada masofalarning **kvadratlari yig‘indisini** (Sum of squared differences) minimallashtirish so‘ralsa (masalan, Fizika yoki ML dagi *Least Squares*), u holda optimal javob **O‘RTA ARIFMETIK** (Average) qiymat bo‘ladi.

**4. Xaffman kodlashining CP dagi qiyofasi (Priority Queue)**
Xaffman algoritmi o‘z-o‘zidan matnni kompress qilish uchungina emas. CP da u ko‘pincha: *"N ta tayoqcha berilgan. Ikkita tayoqchani birlashtirish ularning uzunliklari yig‘indisiga teng energiya talab qiladi. Barcha tayoqchalarni bitta qilish uchun eng kam energiya sarflang"* degan shaklda beriladi. Bu sof Xaffman daraxtidir. Buni yechish uchun massivni saralab o‘tirmaysiz, balki barcha elementlarni **Min-Priority Queue** (Min-heap) ga tashlaysiz va tsikl ichida har doim eng kichik 2 ta elementni sug‘urib olib, yig‘indisini yana Heap'ga qaytarib solaverasiz. Vaqt murakkabligi: $O(N \log N)$.

## 1. Yandex taxi

Shohruh asosan avtobusdan foydalanadi. Lekin ba'zi payt Yandex taxiga murojaat qiladi, Bilamizki Yandex taxi da haydovchiga 1 dan 5 tagacha yulduzcha bilan baho qo'yish mumkin va Shohruh bu ishni doim amalga oshiradi. Bir kuni u ilova orqali necha marta safar qilgani va umumiy nechta yulduzcha qo'yganini ko'rib qoldi. Shohruh 5 ta yulduzcha qo'ygan taxi larning soni minimum va maksimum nechta bo'lishi mumkinligiga qiziqib qoldi.

Shohruhga u 5 ta yulduz qo'ygan taksilarning minimum va maksimum sonini hisoblashda yordam bering.

### Kiruvchi ma'lumotlar:

Kirish faylida ikkita butun son N va M $(1 \le N, M \le 10^{18})$

$(1 \le N, M \le 10^{18})$ - yulduzchalar va safarlar soni kiritiladi.

### Chiquvchi ma'lumotlar:

Chiqish faylida ikkita sonni chop eting: minimum va maksimum son. Agar buning iloji bo'lmasa yoki hisoblashda xatolik mavjud bo'lsa “−1  −1”

“−1  −1” (qo'shtirnoqlarsiz) ni chop eting.

### Misollar

| # | input.txt | output.txt |
| --- | --- | --- |
| **1** | 14 4 | 0 2 |
| **2** | 100 1 | -1 -1 |

Masalani tahlil qilib chiqaylik:

**Masala mohiyati:**
- N ta yulduzcha, M ta safar
- Har bir safarga 1-5 yulduzcha berilgan
- 5 yulduzcha berilgan safarlar min va max nechta bo'lishi mumkin?

**Validatsiya:**
- Minimum yulduzcha: M×1 = M (hammaga 1 bersak)
- Maksimum yulduzcha: M×5 = 5M (hammaga 5 bersak)
- Agar N < M yoki N > 5×M bo'lsa → **-1 -1**

**Maximum topish:**
- 5 yulduzcha berilganlarni maksimal qilish uchun: qolganlariga **1** yulduzcha beramiz
- Agar `k` ta safarga 5 yulduzcha bersak: `5k + 1*(M-k) = N` → `4k = N - M` → `k = (N-M)/4`
- Agar `(N-M) % 4 != 0` bo'lsa, butun son bo'lmaydi → max = `(N-M)/4` + agar qoldiq bo'lsa 0

Aniqroq: `max = (N - M) / 4` (pastga yaxlitlash), lekin tekshirish kerak

**Minimum topish:**
- 5 yulduzcha berilganlarni minimal qilish uchun: qolganlariga **4** yulduzcha beramiz (maksimal, lekin 5 emas)
- Agar `k` ta safarga 5 yulduzcha, qolganiga 4: `5k + 4*(M-k) = N` → `k = N - 4M`
- Agar `N - 4M < 0` bo'lsa, k=0 va tekshiramiz

**Umumiy formula:**

| | Boshqa safarlarga |
|---|---|
| **Maximum k** | Hammaga 1 yulduz → `k = (N - M) / 4` (butun bo'lishi shart emas, pastga) |
| **Minimum k** | Hammaga 4 yulduz → `k = max(0, N - 4*M)` |

```cpp
#include <bits/stdc++.h>
using namespace std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);
    
    long long N, M;
    cin >> N >> M;
    
    // Validatsiya: N yulduzcha M safar uchun mumkinmi?
    if (N < M || N > 5 * M) {
        cout << "-1 -1" << endl;
        return 0;
    }
    
    // MINIMUM: qolganlariga imkon qadar ko'p (4) yulduz beramiz
    // 5*k + 4*(M-k) >= N bo'lishi kerak
    // k >= N - 4*M
    long long minVal = max(0LL, N - 4 * M);
    
    // MAXIMUM: qolganlariga minimum (1) yulduz beramiz
    // 5*k + 1*(M-k) <= N
    // 4*k <= N - M
    // k <= (N - M) / 4
    long long maxVal = (N - M) / 4;
    
    cout << minVal << " " << maxVal << endl;
    
    return 0;
}
```

**Misollar tekshiruvi:**

**Misol 1:** N=14, M=4
- Validatsiya: 4 ≤ 14 ≤ 20 ✓
- min = max(0, 14 - 16) = max(0, -2) = **0**
- max = (14 - 4) / 4 = 10/4 = **2**
- Javob: `0 2` ✅

**Misol 2:** N=100, M=1
- Validatsiya: 100 > 5×1 = 5 → **-1 -1** ✅

**Qo'shimcha testlar:**
- N=5, M=1 → min=max(0,1)=1, max=(4)/4=1 → `1 1` (bitta safar, 5 yulduz)
- N=3, M=1 → min=max(0,-1)=0, max=2/4=0 → `0 0` (bitta safar, 3 yulduz)

## 2. Bomba masalasi



Shohruh yaqinda R radiusli bomba ixtiro qildi. Katakcha bomba radiusida joylashgan deb ataladi, qachonki gorizontal va vertikal masofalar farqining minimali R dan oshmasa. Boshqacha aytganda, agar bomba (a, b) koordinatada, katakcha (c, d) koordinatada joylashgan va min(|a−c|, |b−d|) ≤ R bo'lsa, shu katakcha bomba radiusida joylashgan bo'ladi. N∗M maydon berilgan, har bir katakcha kamida bomba radiusida joylashishi uchun minimal nechta bomba kerak bo'ladi?

**Kiruvchi ma'lumotlar:** Kirish faylining yagona qatorida 3 ta butun son — N, M (1 ≤ N, M ≤ 1000) va R (0 ≤ R ≤ 1000) kiritiladi.

**Chiquvchi ma'lumotlar:** Chiqish faylida kerak bo'ladigan minimal bombalar sonini chop eting.

|---|---|---|
| 1 | 1 7 3 | 1 |
| 2 | 5 5 0 | 5 |

## Matematik yechim

### 1-qadam: Shartni tahlil qilish

```
min(|a−c|, |b−d|) ≤ R
```

Bu shart quyidagiga ekvivalent:

```
|a−c| ≤ R   YOKI   |b−d| ≤ R
```

### 2-qadam: Bomba qamrov sohasini tushunish

Bomba `(a, b)` koordinataga qo'yilganda:

- `|b−d| ≤ R` bo'lgan barcha `d` ustunlar uchun — **qaysi satrdan qat'iy nazar** butun ustun yopiladi
- Ya'ni `b−R ≤ d ≤ b+R` oraliqidagi **barcha ustunlar to'liq** qamrab olinadi

```
     b-R        b        b+R
      |←————————|————————→|
satr1 ✅ ✅ ✅ ✅ ✅ ✅ ✅
satr2 ✅ ✅ ✅ ✅ ✅ ✅ ✅
satr3 ✅ ✅ ✅ ✅ ✅ ✅ ✅
...   ✅ ✅ ✅ ✅ ✅ ✅ ✅
```

Bitta bomba `2R+1` ta ketma-ket ustunni **to'liq** yopadi.

### 3-qadam: 2D → 1D ga kamaytirish

Satrlar soni `N` ahamiyatsiz — faqat ustunlarni yopish kifoya. Xuddi shunday mantiq bilan satrlarni yopsak ham bo'ladi. Shuning uchun **kichik o'lchamni** olamiz:

```
k = min(N, M)
```

Endi masala: **k ta pozitsiyani, har biri 2R+1 kenglikdagi oyna bilan minimal nechta oyna bilan yopish mumkin?**

### 4-qadam: Greedy interval yopish formulasi

```
Birinchi yopilmagan pozitsiya: p = 1
Bomba qo'yish joyi: p + R  (maksimal foyda uchun)
Bu bomba yopadi: p dan p + 2R gacha
Keyingi yopilmagan: p + 2R + 1
```

Natijada:

$$\text{Javob} = \left\lceil \dfrac{\min(N,\, M)}{2R + 1} \right\rceil$$

### 5-qadam: Misollarni tekshirish

**Misol 1:** N=1, M=7, R=3
```
min(1, 7) = 1
2R+1 = 7
⌈ 1/7 ⌉ = 1 ✅
```

**Misol 2:** N=5, M=5, R=0
```
min(5, 5) = 5
2R+1 = 1
⌈ 5/1 ⌉ = 5 ✅
```

## C++ yechimi

```cpp
#include <bits/stdc++.h>
using using std;

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    long long N, M, R;
    cin >> N >> M >> R;

    // Kichik o'lchamni olamiz
    long long k = min(N, M);

    // Har bir bomba (2R+1) ta pozitsiyani yopadi
    long long cover = 2 * R + 1;

    // Yuqoriga yaxlitlangan bo'linma: ⌈k / cover⌉
    long long ans = (k + cover - 1) / cover;

    cout << ans << endl;

    return 0;
}
```

### Kodning ishlash tartibi

```
N=5, M=5, R=0 uchun:

k = min(5,5) = 5
cover = 2*0+1 = 1

ans = (5 + 1 - 1) / 1 = 5/1 = 5 ✅


N=1, M=7, R=3 uchun:

k = min(1,7) = 1
cover = 2*3+1 = 7

ans = (1 + 7 - 1) / 7 = 7/7 = 1 ✅
```
