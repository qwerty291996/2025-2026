**Dinamik massivlar (*Dynamic arrays*)**
Dinamik massiv — o‘lchami dastur ishlashi davomida o‘zgartirilishi mumkin bo‘lgan massivdir. C++ da eng mashhur dinamik massiv `vector` tuzilmasi bo‘lib, undan deyarli oddiy massiv kabi foydalanish mumkin.

Quyidagi kod bo‘sh vektor yaratadi va unga uchta element qo‘shadi:

```cpp
vector<int> v;
v.push_back(3); // [3]
v.push_back(2); // [3, 2]
v.push_back(5); // [3, 2, 5]

```

Shundan so‘ng, elementlarga xuddi oddiy massivdagidek murojaat qilish mumkin:

```cpp
cout << v[0] << "\n"; // 3
cout << v[1] << "\n"; // 2
cout << v[2] << "\n"; // 5

```

`size` funksiyasi vektordagi elementlar sonini qaytaradi. Quyidagi kod vektor bo‘ylab aylanib chiqadi va undagi barcha elementlarni ekranga chop etadi:

```cpp
for (int i = 0; i < v.size(); i++) {
    cout << v[i] << "\n";
}

```

Vektor bo‘ylab aylanib chiqishning qisqaroq usuli quyidagicha:

```cpp
for (auto x : v) {
    cout << x << "\n";
}

```

`back` funksiyasi vektordagi eng oxirgi elementni qaytaradi, `pop_back` funksiyasi esa oxirgi elementni o‘chirib tashlaydi:

```cpp
vector<int> v;
v.push_back(5);
v.push_back(2);
cout << v.back() << "\n"; // 2
v.pop_back();
cout << v.back() << "\n"; // 5

```

Quyidagi kod beshta elementdan iborat vektor yaratadi:

```cpp
vector<int> v = {2, 4, 2, 5, 1};

```

Vektorni yaratishning yana bir usuli — elementlar sonini va har bir element uchun boshlang‘ich qiymatni berishdir:

```cpp
// o'lchami 10, boshlang'ich qiymati 0
vector<int> v(10);
// o'lchami 10, boshlang'ich qiymati 5
vector<int> v(10, 5);

```

Vektorning ichki ishlash mexanizmi oddiy massivdan foydalanadi. Agar vektorning o‘lchami oshib, massiv juda kichiklik qilib qolsa, yangi massiv uchun xotira ajratiladi (*allocated*) va barcha elementlar yangi massivga ko‘chiriladi. Biroq, bu holat tez-tez yuz bermaydi va `push_back` amaliyotining o‘rtacha (*amortized*) vaqt murakkabligi $O(1)$ ni tashkil qiladi.

`string` (satr) tuzilmasi ham xuddi vektor kabi ishlatilishi mumkin bo‘lgan dinamik massivdir. Bunga qo‘shimcha ravishda, satrlar uchun boshqa ma’lumotlar tuzilmalarida mavjud bo‘lmagan maxsus sintaksis mavjud. Satrlarni `+` belgisi yordamida birlashtirish mumkin. `substr(k, x)` funksiyasi $k$ pozitsiyadan boshlanuvchi va $x$ uzunlikdagi qism-satrni (*substring*) qaytaradi, `find(t)` funksiyasi esa $t$ qism-satrining birinchi marta uchrashgan pozitsiyasini topib beradi.

Quyidagi kodda ba’zi satr operatsiyalari ko‘rsatilgan:

```cpp
string a = "hatti";
string b = a + a;
cout << b << "\n"; // hattihatti
b[5] = 'v';
cout << b << "\n"; // hattivatti
string c = b.substr(3, 4);
cout << c << "\n"; // tiva

```

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

**1. `v.size() - 1` qopqog‘i (Eng mashhur xatolik)**
C++ da `v.size()` funksiyasi oddiy manfiy bo‘lishi mumkin bo‘lgan `int` tipida emas, balki `unsigned int` (`size_t`) tipida qiymat qaytaradi.
Agar vektor bo‘sh bo‘lsa (`v.size() == 0`), va siz tsiklda `v.size() - 1` deb yozsangiz, u `-1` ga emas, balki **$18446744073709551615$** kabi ulkan musbat songa aylanib ketadi! Buning oqibatida tsiklingiz xotiraning mavjud bo‘lmagan joyiga murojaat qilib, darhol **Runtime Error (Segfault)** yoki **Memory Limit Exceeded** beradi.
*Yechim:* Tsikllarda har doim kasrni manfiyga o‘tkazib qo‘ying: `(int)v.size() - 1` deb ishlating.

**2. Satrlarni birlashtirishdagi "TLE" tuzog‘i (`+` vs `+=`)**
Kitobda aytilganidek satrlarni `+` yordamida birlashtirish mumkin, lekin buni tsikl ichida qilish o‘ta xavfli:

```cpp
string s = "";
for (int i = 0; i < n; i++) {
    s = s + "a"; // XATO! Bu O(N^2) ga aylanadi
}

```

Nega? Chunki `s = s + "a"` ifodasi har safar yangi xotira yaratadi va eski satrni to‘liq nusxalab chiqadi. $N = 10^5$ bo‘lganda kodingiz 1 soniyadan oshib (TLE) ketadi.
*To‘g‘ri usul:* Faqatgina **`s += "a";`** yoki **`s.push_back('a');`** ishlating! Bu xuddi vektordagidek *amortized* $O(1)$ vaqt oladi.

**3. Vektorni oldindan rejalashtirish (`v.reserve()`)**
Matnda "vektor hajmi to‘lganda, u o‘zini ikki barobar kattalashtirib, eskisini yangisiga ko‘chiradi" deyilgan. Garchi u o‘rtacha $O(1)$ bo‘lsa ham, qattiq optimizatsiya qilinayotgan masalalarda xotira ko‘chirishi jarayoni *Constant Factor* ni oshirib yuboradi. Agar siz massivga aynan $N$ ta element qo‘shishingizni oldindan bilsangiz, tsikldan oldin `v.reserve(n);` funksiyasini chaqirib qo‘ying. Bu ortiqcha xotira ko‘chishlarining oldini oladi va kodingiz ancha tezroq ishlaydi.

**To‘plam tuzilmalari (*Set structures*)**

To‘plam (*set*) — bu elementlar majmuasini o‘zida saqlab turuvchi ma’lumotlar tuzilmasidir. To‘plamlarning asosiy operatsiyalari elementni qo‘shish, qidirish va o‘chirishdan iborat.

C++ standart kutubxonasida to‘plamning ikkita realizatsiyasi mavjud: `set` tuzilmasi muvozanatlangan binar daraxtga asoslangan va uning operatsiyalari $O(\log n)$ vaqtda ishlaydi.

`unordered_set` tuzilmasi esa heshlashdan (*hashing*) foydalanadi va uning operatsiyalari o‘rtacha $O(1)$ vaqtda ishlaydi.

Qaysi to‘plam realizatsiyasini tanlash ko‘pincha didga bog‘liq. `set` tuzilmasining afzalligi shundaki, u elementlarning tartibini saqlaydi va `unordered_set` da mavjud bo‘lmagan funksiyalarni taqdim etadi. Boshqa tomondan, `unordered_set` samaraliroq bo‘lishi mumkin.

Quyidagi kod butun sonlarni saqlovchi to‘plam yaratadi va ba’zi operatsiyalarni ko‘rsatadi. `insert` funksiyasi to‘plamga element qo‘shadi, `count` funksiyasi to‘plamda elementning uchrashishlar sonini qaytaradi, `erase` funksiyasi esa elementni to‘plamdan o‘chiradi.

```cpp
set<int> s;
s.insert(3);
s.insert(2);
s.insert(5);
cout << s.count(3) << "\n"; // 1
cout << s.count(4) << "\n"; // 0
s.erase(3);
s.insert(4);
cout << s.count(3) << "\n"; // 0
cout << s.count(4) << "\n"; // 1

```

To‘plamdan ko‘pincha vektor kabi foydalanish mumkin, biroq elementlarga `[]` belgisi orqali murojaat qilib bo‘lmaydi. Quyidagi kod to‘plam yaratadi, undagi elementlar sonini chop etadi va keyin barcha elementlar bo‘ylab aylanib chiqadi:

```cpp
set<int> s = {2, 5, 6, 8};
cout << s.size() << "\n"; // 4
for (auto x : s) {
    cout << x << "\n";
}

```

To‘plamlarning muhim xususiyati shundaki, ularning barcha elementlari takrorlanmas (*distinct*) bo‘ladi. Shunday qilib, `count` funksiyasi har doim yoki 0 (element to‘plamda yo‘q) yoki 1 (element to‘plamda bor) qaytaradi va `insert` funksiyasi element allaqachon to‘plamda mavjud bo‘lsa, uni hech qachon qayta qo‘shmaydi. Quyidagi kod buni tasvirlaydi:

```cpp
set<int> s;
s.insert(5);
s.insert(5);
s.insert(5);
cout << s.count(5) << "\n"; // 1

```

C++ shuningdek, boshqa jihatdan `set` va `unordered_set` kabi ishlaydigan, lekin o‘zida bitta elementning bir nechta nusxalarini saqlay oladigan `multiset` va `unordered_multiset` tuzilmalarini ham o‘z ichiga oladi. Masalan, quyidagi kodda 5 sonining barcha uchta nusxasi `multiset` ga qo‘shiladi:

```cpp
multiset<int> s;
s.insert(5);
s.insert(5);
s.insert(5);
cout << s.count(5) << "\n"; // 3

```

`erase` funksiyasi `multiset` dan elementning barcha nusxalarini o‘chirib tashlaydi:

```cpp
s.erase(5);
cout << s.count(5) << "\n"; // 0

```

Ko‘pincha, faqat bitta nusxani o‘chirish kerak bo‘ladi va buni quyidagicha amalga oshirish mumkin:

```cpp
s.erase(s.find(5));
cout << s.count(5) << "\n"; // 2

```

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar: Set va Multiset tuzoqlari

Ushbu bob C++ ning eng kuchli, ammo eng ko‘p mantiqiy xatolar (Bug) keltirib chiqaradigan ma'lumotlar tuzilmalaridan birini yoritgan.

**1. `unordered_set` ning qora yuzi (Hash Collision TLE)**
Matnda `unordered_set` o‘rtacha $O(1)$ vaqtda ishlashi aytilgan. Ammo Codeforces musobaqalarida **hech qachon sof `unordered_set` (yoki `unordered_map`) ishlatmang!**
Sababi shundaki, Codeforces testlarini tuzuvchilar C++ ning standart xesh-funksiyasini buzadigan "Anti-hash" testlarni qasddan kiritishadi. Natijada sizning $O(1)$ deb o‘ylagan algoritmingiz eng yomon holatga tushib, har bir qidiruv uchun $O(n)$ vaqt oladi va tayyor **TLE (Time Limit Exceeded)** qaytaradi.
*Yechim:* Yoki doim $O(\log n)$ da ishonchli ishlaydigan oddiy `set` ishlating, yoki Custom Hash funksiyasini yozishni o‘rganing.

**2. `multiset` da bitta elementni o‘chirish fokusi qanday ishlaydi?**
Kitobning oxiridagi kod eng ko‘p qilinadigan xatoning yechimidir: `s.erase(5)` deb yozsangiz, massivdagi yuzta 5 bo‘lsa, yuzalasini ham bittada qirib tashlaydi (va bu kutilmaganda juda sekin ishlaydi).
Nega `s.erase(s.find(5))` to‘g‘ri ishlaydi?
Chunki `s.find(5)` qiymat qaytarmaydi, u 5 joylashgan **xotira manzilini (iteratorni)** qaytaradi. `erase()` funksiyasiga aniq manzil berilganda, u faqatgina aynan o‘sha bitta tugunni (node) kesib tashlaydi.

**3. `set` aslida avtomatik saralovchi mashinadir**
To‘plam nafaqat takroriylikni yo‘qotadi, balki barcha ma'lumotlarni doimiy ravishda o‘sish tartibida saqlaydi!
Agar sizga eng kichik element kerak bo‘lsa, u doim tayyor: `*s.begin()`
Eng katta element kerak bo‘lsa: `*s.rbegin()`
Bu xususiyat `set` ni `priority_queue` ga ajoyib muqobil qiladi. Farqi shundaki, `set` ning o‘rtasidan xohlagan elementni osonlikcha o‘chirib tashlash mumkin, `priority_queue` da esa bunday imkoniyat yo‘q.

**4. `std::lower_bound` tuzog‘i**
3-bobda binar qidiruv bo‘yicha o‘rgangan `std::lower_bound(s.begin(), s.end(), x)` kodingizni `set` ga ishlatsangiz, u kompilyatsiya bo‘ladi, lekin **$O(n)$** ga aylanib ketib qotib qoladi (chunki Set xotirada ketma-ket joylashmagan).
Buning o‘rniga har doim Set'ning o‘ziga tegishli metodni ishlating: **`s.lower_bound(x)`**. Bu har qanday holatda kafolatlangan $O(\log n)$ tezlikda ishlaydi!

**Map (Lug‘at) tuzilmalari (*Map structures*)**

Map — bu kalit-qiymat (*key-value*) juftliklaridan iborat bo‘lgan umumiylashtirilgan massivdir. Oddiy massivdagi kalitlar (indekslar) har doim ketma-ket keluvchi $0, 1, \dots, n-1$ butun sonlar (bu yerda $n$ massiv o‘lchami) bo‘lsa, map ichidagi kalitlar har qanday ma’lumotlar tipida bo‘lishi mumkin va ular ketma-ket qiymatlar bo‘lishi shart emas.

C++ standart kutubxonasida, xuddi to‘plamlar (*set*) kabi, map'ning ham ikkita realizatsiyasi mavjud: `map` tuzilmasi muvozanatlangan binar daraxtga asoslangan va elementlarga murojaat qilish $O(\log n)$ vaqtni oladi, `unordered_map` tuzilmasi esa heshlashdan foydalanadi va elementlarga murojaat qilish o‘rtacha $O(1)$ vaqt oladi.

Quyidagi kod kalitlari satr (*string*), qiymatlari esa butun son (*int*) bo‘lgan map yaratadi:

```cpp
map<string, int> m;
m["monkey"] = 4;
m["banana"] = 3;
m["harpsichord"] = 9;
cout << m["banana"] << "\n"; // 3

```

Agar biron bir kalitning qiymati so‘ralsa-yu, lekin u map ichida mavjud bo‘lmasa, bu kalit avtomatik ravishda standart qiymat bilan (masalan, sonlar uchun 0) map'ga qo‘shib qo‘yiladi. Masalan, quyidagi kodda 0 qiymatiga ega bo‘lgan "aybabtu" kaliti map'ga qo‘shiladi:

```cpp
map<string, int> m;
cout << m["aybabtu"] << "\n"; // 0

```

`count` funksiyasi kalitning map'da mavjud yoki yo‘qligini tekshiradi:

```cpp
if (m.count("aybabtu")) {
    // kalit mavjud
}

```

Quyidagi kod map'dagi barcha kalitlar va qiymatlarni ekranga chop etadi:

```cpp
for (auto x : m) {
    cout << x.first << " " << x.second << "\n";
}

```

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

Map tuzilmasi musobaqalarda elementlar chastotasini (necha marta qatnashganini) sanash uchun eng zo‘r qurol hisoblanadi. Ayniqsa kalitlar $10^9$ dan katta sonlar yoki satrlar bo‘lganda, oddiy massiv (Frequency Array) ochib bo‘lmaydi. Ammo `map` ishlatishning o‘ziga xos juda xavfli tuzoqlari bor.

**1. Avtomatik qo‘shilish tuzog‘i (Memory & Time Limit)**
Kitobdagi *“agar kalit mavjud bo‘lmasa, avtomatik qo‘shiladi”* degan qoida CP dagi eng ko‘p yashirin xatolarni (bug) keltirib chiqaradi.
Ko‘pchilik dasturchilar element bor-yo‘qligini tekshirish uchun shunday yozishadi:

```cpp
if (m["olma"] == 1) { ... } // YOMON USUL!

```

Agar `"olma"` map ichida bo‘lmasa ham, bu ifoda uni map'ga `0` qiymati bilan **yangi tugun (node) qilib yaratib qo‘yadi**. Agar siz buni million marta aylanadigan tsikl ichida qilsangiz, map keraksiz axlatga to‘lib, xotira tugaydi (MLE) va vaqt chegarasidan o‘tib ketadi (TLE).
**To‘g‘ri usul:** Hech qachon bor-yo‘qligini bilish uchun `[]` ishlatingmang! Uning o‘rniga har doim `if (m.count("olma"))` yoki `if (m.find("olma") != m.end())` ishlating.

**2. `unordered_map` dagi "Anti-hash" hujumi**
Oldingi bobda `unordered_set` uchun berilgan ogohlantirish `unordered_map` uchun ham 100% amal qiladi. Codeforces'da xesh-jadvallarni ataylab "sindiradigan" testlar mavjud bo‘lib, `unordered_map` o‘rtacha $O(1)$ dan eng yomon $O(N)$ tezlikka tushib ketadi va kodingiz TLE oladi.
Musobaqalarda asosan oddiy `std::map` ishlating. Agar vaqt juda qattiq chegaralangan bo‘lsa va $O(1)$ muhim bo‘lsa, maxsus xesh-funksiya (custom hash) yozishingiz shart.

**3. Kalitlar har doim saralangan bo‘ladi**
Xuddi `set` kabi, `std::map` ham o‘z ichidagi barcha ma’lumotlarni **kalitlari bo‘yicha o‘sish tartibida** saqlaydi.
Agar siz quyidagicha tsikl yozsangiz:

```cpp
m[100] = 5;
m[10] = 2;
m[50] = 8;
for (auto x : m) cout << x.first << " ";

```

Ekranga `10 50 100` chiqadi. Ya'ni elementlar qo‘shilgan ketma-ketlikda emas, balki saralangan tartibda chiqadi. Bu masalalarni yechishda, masalan, kichik qiymatlardan kattasiga qarab yurish kerak bo‘lganda juda qo‘l keladi. Xuddi shunday eng oxirgi (eng katta) kalitni `m.rbegin()->first` orqali $O(1)$ da olishingiz mumkin.

**Iteratorlar va oraliqlar (*Iterators and ranges*)**

C++ standart kutubxonasidagi ko‘plab funksiyalar iteratorlar bilan ishlaydi. Iterator — bu ma’lumotlar tuzilmasidagi elementga ko‘rsatib turuvchi (*points*) o‘zgaruvchidir.

Ko‘p ishlatiladigan `begin` va `end` iteratorlari ma’lumotlar tuzilmasidagi barcha elementlarni o‘z ichiga oluvchi oraliqni (*range*) belgilab beradi. `begin` iteratori ma’lumotlar tuzilmasidagi birinchi elementga ko‘rsatadi, `end` iteratori esa eng oxirgi elementdan keyingi pozitsiyaga ko‘rsatadi.
Holat quyidagicha ko‘rinishga ega:

```text
{ 3, 4, 6, 8, 12, 13, 14, 17 }
  ↑                            ↑
s.begin()                    s.end()

```

Iteratorlardagi bu asimmetriyaga e'tibor bering: `s.begin()` ma’lumotlar tuzilmasi ichidagi elementga ko‘rsatadi, `s.end()` esa ma’lumotlar tuzilmasidan tashqariga ko‘rsatadi. Shunday qilib, iteratorlar tomonidan belgilangan oraliq yarim-ochiq (*half-open*) hisoblanadi.

**Oraliqlar bilan ishlash (*Working with ranges*)**
Iteratorlar C++ standart kutubxonasida ma’lumotlar tuzilmasidagi elementlar oralig‘i beriladigan funksiyalarda qo‘llaniladi. Odatda biz ma’lumotlar tuzilmasidagi barcha elementlarni qayta ishlashni xohlaymiz, shuning uchun funksiyaga `begin` va `end` iteratorlari beriladi.

Masalan, quyidagi kod vektorlarni `sort` funksiyasi yordamida saralaydi, so‘ngra `reverse` funksiyasi orqali elementlar tartibini teskarisiga o‘giradi va nihoyat, `random_shuffle` funksiyasi orqali elementlar tartibini aralashtirib yuboradi.

```cpp
sort(v.begin(), v.end());
reverse(v.begin(), v.end());
random_shuffle(v.begin(), v.end());

```

Ushbu funksiyalar oddiy massiv bilan ham ishlatilishi mumkin. Bu holda funksiyalarga iteratorlar o‘rniga massiv ko‘rsatkichlari (*pointers*) beriladi:

```cpp
sort(a, a+n);
reverse(a, a+n);
random_shuffle(a, a+n);

```

**To‘plam iteratorlari (*Set iterators*)**
Iteratorlar ko‘pincha to‘plam (*set*) elementlariga murojaat qilish uchun ishlatiladi. Quyidagi kod to‘plamdagi eng kichik elementga ko‘rsatuvchi `it` iteratorini yaratadi:

```cpp
set<int>::iterator it = s.begin();

```

Kodni yozishning qisqaroq usuli quyidagicha:

```cpp
auto it = s.begin();

```

Iterator ko‘rsatayotgan elementga `*` (yulduzcha) belgisi yordamida murojaat qilish mumkin. Masalan, quyidagi kod to‘plamdagi birinchi elementni ekranga chop etadi:

```cpp
auto it = s.begin();
cout << *it << "\n";

```

Iteratorlarni `++` (oldinga) va `--` (orqaga) operatorlari yordamida siljitish mumkin, bu iterator to‘plamdagi keyingi yoki oldingi elementga o‘tishini anglatadi.
Quyidagi kod barcha elementlarni o‘sish tartibida chop etadi:

```cpp
for (auto it = s.begin(); it != s.end(); it++) {
    cout << *it << "\n";
}

```

Quyidagi kod to‘plamdagi eng katta elementni chop etadi:

```cpp
auto it = s.end(); it--;
cout << *it << "\n";

```

`find(x)` funksiyasi qiymati $x$ ga teng bo‘lgan elementga ko‘rsatuvchi iterator qaytaradi. Biroq, agar to‘plamda $x$ mavjud bo‘lmasa, iterator `end` bo‘ladi.

```cpp
auto it = s.find(x);
if (it == s.end()) {
    // x topilmadi
}

```

`lower_bound(x)` funksiyasi to‘plamdagi qiymati kamida $x$ ga teng bo‘lgan eng kichik elementga iterator qaytaradi, va `upper_bound(x)` funksiyasi to‘plamdagi qiymati qat’iy $x$ dan katta bo‘lgan eng kichik elementga iterator qaytaradi. Ikkala funksiyada ham, agar bunday element mavjud bo‘lmasa, qaytariladigan qiymat `end` bo‘ladi. Elementlar tartibini saqlamaydigan `unordered_set` tuzilmasi bu funksiyalarni qo‘llab-quvvatlamaydi.

Masalan, quyidagi kod $x$ ga eng yaqin bo‘lgan elementni topadi:

```cpp
auto it = s.lower_bound(x);
if (it == s.begin()) {
    cout << *it << "\n";
} else if (it == s.end()) {
    it--;
    cout << *it << "\n";
} else {
    int a = *it; it--;
    int b = *it;
    if (x - b < a - x) cout << b << "\n";
    else cout << a << "\n";
}

```

Kod to‘plam bo‘sh emas deb faraz qiladi va barcha ehtimoliy holatlarni `it` iteratori orqali ko‘rib chiqadi. Birinchidan, iterator qiymati kamida $x$ ga teng bo‘lgan eng kichik elementga ko‘rsatadi. Agar `it` `begin` ga teng bo‘lsa, u holda mos element $x$ ga eng yaqin hisoblanadi. Agar u `end` ga teng bo‘lsa, u holda to‘plamdagi eng katta element $x$ ga eng yaqindir. Agar oldingi holatlarning hech biri o‘rinli bo‘lmasa, $x$ ga eng yaqin element yo `it` ko‘rsatayotgan element, yo undan bitta oldingi element bo‘ladi.

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

**1. Nima uchun `s.end()` oxirgi elementga teng emas?**
Bu C++ ning eng genial arxitekturasidir. Uni $[begin, end)$ ko'rinishidagi ochiq yarim-oraliq deyiladi.
Agar `s.end()` haqiqatan ham oxirgi element bo'lganida nima yuz berardi? Biz bo'sh (empty) massivni ifodalay olmas edik! Bo'sh massivda `begin == end` bo'lishi aynan shuning uchun kerak. Qolaversa, tsikllarda `for(auto it = s.begin(); it != s.end(); it++)` deb yozish, indekslardagi `< n` qoidasi kabi xatoliklarni kamaytiradi.

**2. Pointers vs Iterators**
Massiv uchun ko'rsatkichlar (`a`, `a+n`) nima bo'lsa, STL tuzilmalar uchun iteratorlar (`begin()`, `end()`) ham aynan shu narsadir. Farqi shundaki, massiv xotirada 100% ketma-ket (contiguous) joylashadi, shuning uchun pointerga $+1$ qo'shib sakrash mumkin. Lekin `set` kabi daraxtsimon tuzilmalarda xotira tarqoq bo'ladi. Iteratorlar — shu daraxtni "simlab chiqadigan" aqlli ko'rsatkichlardir.

**3. Iteratorni uzib qo'yish xavfi (Invalidation)**
`vector` da tsikl aylanayotganda `v.push_back()` qilsangiz, ba'zan barcha iteratorlaringiz (va pointerlaringiz) uzilib, xato ishlab ketadi (Invalidation). Chunki vektor xotira ko'chirishini amalga oshirgan bo'lishi mumkin.
Lekin Node-based tuzilmalarda (masalan, `set`, `map`) iteratorlar o'chirilmaguncha o'z joyidan jilmaydi.

**4. `std::next` va `std::prev` dan foydalanish**
Siz C++11 (va undan yuqori) da `it--` yoki `it++` deb yozish o'rniga, asl holatni o'zgartirmasdan bitta oldingi yoki keyingi elementni olishni istasangiz, `<iterator>` kutubxonasidagi maxsus funksiyalardan foydalanishingiz mumkin:

```cpp
auto next_it = std::next(it);
auto prev_it = std::prev(it);

```

Bu ayniqsa `set` dagi elementning "qo'shnilarini" tekshirish uchun o'ta muhim usuldir.

C++ da `std::set` elementlarni har doim o'sish tartibida (saralangan holda) saqlaydi. Bu xususiyat binar qidiruvga asoslangan `lower_bound` va `upper_bound` funksiyalarini juda samarali ($O(\log N)$ tezlikda) ishlatish imkonini beradi.

Bu ikki funksiyaning asosiy farqi shundaki, biri **qatiy bo'lmagan (>=)** shartni, ikkinchisi esa **qat'iy (>)** shartni tekshiradi.

### 📚 Asosiy qoidalar:

1. **`s.lower_bound(x)`**: Qiymati **kamida $x$ ga teng** (ya'ni $\ge x$) bo'lgan eng birinchi elementga iterator (ko'rsatkich) qaytaradi.
2. **`s.upper_bound(x)`**: Qiymati **qat'iy $x$ dan katta** (ya'ni $> x$) bo'lgan eng birinchi elementga iterator qaytaradi.
3. Agar shartni qanoatlantiruvchi element topilmasa, ikkala funksiya ham **`s.end()`** ni qaytaradi.

Quyidagi kodda ushbu ikki funksiyaning turli holatlarda qanday ishlashini batafsil ko'rishingiz mumkin:

```cpp
#include <iostream>
#include <set>

using namespace std;

int main() {
    // Set yaratamiz (avtomatik saralanadi: 10, 20, 30, 40, 50)
    set<int> s = {10, 20, 30, 40, 50};

    // ----------------------------------------------------
    // 1-HOLAT: Qidirilayotgan son set'da mavjud bo'lsa
    // ----------------------------------------------------
    auto it1 = s.lower_bound(30); 
    // 30 ga teng yoki undan katta birinchi element -> 30
    cout << "lower_bound(30) = " << *it1 << "\n"; // Natija: 30

    auto it2 = s.upper_bound(30); 
    // Qat'iy 30 dan katta birinchi element -> 40
    cout << "upper_bound(30) = " << *it2 << "\n"; // Natija: 40


    // ----------------------------------------------------
    // 2-HOLAT: Qidirilayotgan son set'da mavjud bo'lmasa
    // ----------------------------------------------------
    auto it3 = s.lower_bound(25); 
    // 25 ga teng yoki undan katta birinchi element -> 30
    cout << "lower_bound(25) = " << *it3 << "\n"; // Natija: 30

    auto it4 = s.upper_bound(25); 
    // Qat'iy 25 dan katta birinchi element -> 30
    cout << "upper_bound(25) = " << *it4 << "\n"; // Natija: 30


    // ----------------------------------------------------
    // 3-HOLAT: Son set'dagi eng katta elementdan ham katta bo'lsa
    // ----------------------------------------------------
    auto it5 = s.lower_bound(60); 
    if (it5 == s.end()) {
        cout << "lower_bound(60) topilmadi, s.end() qaytdi.\n";
    }

    return 0;
}

```

**Boshqa tuzilmalar (*Other structures*)**

**Bitset**
Bitset — har bir qiymati yo 0, yo 1 bo‘lgan massivdir. Masalan, quyidagi kod 10 ta elementni o‘zida saqlovchi bitset yaratadi:

```cpp
bitset<10> s;
s[1] = 1;
s[3] = 1;
s[4] = 1;
s[7] = 1;
cout << s[4] << "\n"; // 1
cout << s[5] << "\n"; // 0

```

Bitsetlardan foydalanishning afzalligi shundaki, ular oddiy massivlarga qaraganda kamroq xotira talab qiladi, chunki bitsetdagi har bir element xotiraning atigi bir bitini egallaydi. Masalan, agar $n$ ta bit `int` massivida saqlansa, xotiraning $32n$ ta biti band qilinadi, lekin xuddi shunday bitset atigi $n$ bit xotira talab qiladi. Bunga qo‘shimcha ravishda, bitset qiymatlarini bitli operatorlar (*bit operators*) yordamida samarali boshqarish mumkin, bu esa bitsetlar yordamida algoritmlarni optimallashtirish imkonini beradi.

Quyidagi kod yuqoridagi bitsetni yaratishning boshqa usulini ko‘rsatadi:

```cpp
bitset<10> s(string("0010011010")); // o'ngdan chapga qarab
cout << s[4] << "\n"; // 1
cout << s[5] << "\n"; // 0

```

`count` funksiyasi bitsetdagi birlar sonini qaytaradi:

```cpp
bitset<10> s(string("0010011010"));
cout << s.count() << "\n"; // 4

```

Quyidagi kodda bitli amallardan (*bit operations*) foydalanishga misollar keltirilgan:

```cpp
bitset<10> a(string("0010110110"));
bitset<10> b(string("1011011000"));
cout << (a&b) << "\n"; // 0010010000 (AND)
cout << (a|b) << "\n"; // 1011111110 (OR)
cout << (a^b) << "\n"; // 1001101110 (XOR)

```

**Dek (*Deque - Double Ended Queue*)**
Dek — o‘lchami massivning har ikki chetidan ham samarali o‘zgartirilishi mumkin bo‘lgan dinamik massivdir. Vektor singari, dek ham `push_back` va `pop_back` funksiyalarini taqdim etadi, lekin uning tarkibida vektorda mavjud bo‘lmagan `push_front` va `pop_front` funksiyalari ham bor.

```cpp
deque<int> d;
d.push_back(5); // [5]
d.push_back(2); // [5,2]
d.push_front(3); // [3,5,2]
d.pop_back(); // [3,5]
d.pop_front(); // [5]

```

Dekning ichki realizatsiyasi vektorga qaraganda murakkabroq va shu sababli dek vektorga nisbatan sekinroq ishlaydi. Shunga qaramay, har ikki chetdan elementlarni qo‘shish va o‘chirish o‘rtacha $O(1)$ vaqt oladi.

**Stek (*Stack*)**
Stek ikkita $O(1)$ vaqtli operatsiyani taqdim etuvchi ma’lumotlar tuzilmasidir: elementni eng yuqoriga (*top*) qo‘shish va eng yuqoridagi elementni o‘chirish. Stekda faqatgina eng yuqoridagi elementga murojaat qilish mumkin (LIFO qoidasi - *Last In, First Out*).

```cpp
stack<int> s;
s.push(3);
s.push(2);
s.push(5);
cout << s.top(); // 5
s.pop();
cout << s.top(); // 2

```

**Navbat (*Queue*)**
Navbat ham ikkita $O(1)$ vaqtli operatsiyani taqdim etadi: elementni navbatning oxiriga qo‘shish va navbatdagi eng birinchi elementni o‘chirish. Navbatda faqatgina birinchi va oxirgi elementlarga murojaat qilish mumkin (FIFO qoidasi - *First In, First Out*).

```cpp
queue<int> q;
q.push(3);
q.push(2);
q.push(5);
cout << q.front(); // 3
q.pop();
cout << q.front(); // 2

```

**Ustuvor navbat (*Priority queue*)**
Ustuvor navbat elementlar majmuasini o‘zida saqlaydi. Qo‘llab-quvvatlanadigan operatsiyalar — qo‘shish va, navbatning turiga qarab, eng kichik yoki eng katta elementni ajratib olish va o‘chirishdir. Qo‘shish va o‘chirish $O(\log n)$ vaqt, ajratib olish esa $O(1)$ vaqt oladi.
Tartiblangan to‘plam (*ordered set*) ustuvor navbatning barcha operatsiyalarini samarali bajara olsa-da, ustuvor navbatdan foydalanishning afzalligi shundaki, u kichikroq doimiy koeffitsiyentlarga (*constant factors*) ega. Ustuvor navbat odatda tartiblangan to‘plamda qo‘llaniladigan muvozanatlangan binar daraxtga qaraganda ancha soddaroq bo‘lgan **to‘da (*heap*)** tuzilmasi yordamida yaratiladi.

Odatiy holatda, C++ dagi ustuvor navbatda elementlar kamayish tartibida saqlanadi, va navbatdagi eng katta elementni topib, o‘chirish imkoni mavjud bo‘ladi.

```cpp
priority_queue<int> q;
q.push(3);
q.push(5);
q.push(7);
q.push(2);
cout << q.top() << "\n"; // 7
q.pop();
cout << q.top() << "\n"; // 5
q.pop();
q.push(6);
cout << q.top() << "\n"; // 6
q.pop();

```

Agar biz eng kichik elementni qidirish va o‘chirishni qo‘llab-quvvatlaydigan ustuvor navbat yaratmoqchi bo‘lsak, buni quyidagicha amalga oshirishimiz mumkin:

```cpp
priority_queue<int, vector<int>, greater<int>> q;

```

`g++` kompilyatori C++ standart kutubxonasiga kirmaydigan ba’zi ma’lumotlar tuzilmalarini ham qo‘llab-quvvatlaydi. Bunday tuzilmalar siyosatga asoslangan ma’lumotlar tuzilmalari (PBDS) deb ataladi. Ulardan foydalanish uchun kodga quyidagi qatorlarni qo‘shish kerak:

Shundan so‘ng biz `set` kabi ishlaydigan, biroq xuddi massiv kabi indekslash mumkin bo‘lgan `indexed_set` ma’lumotlar tuzilmasini aniqlashimiz mumkin.

```cpp
typedef tree<int, null_type, less<int>, rb_tree_tag, tree_order_statistics_node_update> indexed_set;

```

### ❌ 1-Yechim: Oddiy Vektor bilan (TLE - Time Limit Exceeded)

Eng sodda fikr: elementlarni `std::vector` ga qo‘shamiz va har safar so‘rov kelganda uni saralaymiz.

```cpp
vector<int> v;
// 1-so'rov: Qo'shish -> O(1)
v.push_back(X); 

// 2-so'rov: K-chi elementni topish -> O(N log N)
sort(v.begin(), v.end());
cout << v[K - 1] << "\n";

// 3-so'rov: X dan kichiklarni sanash -> O(N log N)
sort(v.begin(), v.end());
cout << lower_bound(v.begin(), v.end(), X) - v.begin() << "\n";

```

* **Natija:** Vektorda indeks bo‘yicha murojaat $O(1)$ bo‘lsa-da, har so‘rovda massivni saralash kerak. Dastur 1 soniya emas, daqiqalab qotib qoladi.

### ⚠️ 2-Yechim: `std::set` bilan (TLE - Time Limit Exceeded)

Agar elementlar doim saralangan turishi kerak bo‘lsa, `set` ishlatamiz. U binar daraxt bo‘lgani uchun qo‘shish doim $O(\log N)$ da ishlaydi!

* **Natija:** `set` ning eng katta kamchiligi — uning indekslari yo‘q! K-chi elementga borish uchun daraxtning boshidan boshlab bittalab $O(K)$ tezlikda yurib chiqish kerak. Vaqt murakkabligi baribir $O(Q^2)$ bo‘lib qoladi.

### 🛠 3-Yechim: Fenwick Tree / Segment Tree (AC - Accepted, lekin azobli)

Oltin qoidaga ko‘ra, har qadamda $O(\log N)$ ishlashimiz shart. Buning uchun Fenwick Tree quramiz. Ammo $X$ qiymati $10^9$ gacha bo‘lishi mumkin! Fenwick Tree uchun $10^9$ o‘lchamli massiv ocha olmaymiz (Memory Limit).

1. Oldin barcha so‘rovlarni yig‘ib olib, **Coordinate Compression** (Koordinatalarni siqish) algoritmini ishlatamiz.
2. Keyin Fenwick Tree ga qo‘shamiz (`update(idx, 1)`).
3. K-chi elementni topish uchun daraxt bo‘ylab **Binary Lifting** qilamiz.

* **Natija:** Dastur $O(Q \log Q)$ da juda tez ishlaydi. Lekin bu kodni yozish uchun sizga kamida 50-60 qator, 3 ta alohida murakkab algoritm (Siqish + Fenwick + Binary Search) va 20 daqiqa vaqt kerak bo‘ladi. Kichkina xato qilsangiz, musobaqada vaqtingiz kuyadi.

### 💎 4-Yechim: Siyosatga asoslangan tuzilma (PBDS - Eng Optimal)

Yuqoridagi barcha azoblarni C++ ning maxfiy quroli — **Indexed Set** o‘z bo‘yniga oladi. U `set` ning barcha avtomatik saralash xususiyatlarini va massivning $O(1)$ ga o‘xshash indekslash logikasini bitta $O(\log N)$ funksiyaga birlashtirgan.

```cpp
#include <iostream>
#include <ext/pb_ds/assoc_container.hpp>

using namespace std;
using namespace __gnu_pbds;

// PBDS shabloni (Kitobdagidek)
typedef tree<int, null_type, less<int>, rb_tree_tag, tree_order_statistics_node_update> indexed_set;

int main() {
    indexed_set s;
    int Q;
    cin >> Q;
    
    while(Q--) {
        int type, val;
        cin >> type >> val;
        
        if (type == 1) {
            // Qo'shish: O(log N)
            s.insert(val);
        } 
        else if (type == 2) {
            // K-chi elementni topish: O(log N)
            // (0-indeksdan boshlangani uchun K-1)
            cout << *s.find_by_order(val - 1) << "\n";
        } 
        else if (type == 3) {
            // X dan qat'iy kichik elementlar soni: O(log N)
            cout << s.order_of_key(val) << "\n";
        }
    }
    return 0;
}

```

* **Natija:** Barcha operatsiyalar qat'iy $O(\log N)$ da ishlaydi. Kod hajmi bor-yo‘g‘i bir necha qator. Hech qanday koordinatalarni siqish yoki murakkab daraxtlarni noldan qurish shart emas. Vaqt murakkabligi jami $O(Q \log Q)$ va musobaqada yechim 100% muvaffaqiyatli (AC) o‘tadi.

---

PBDS haqiqatan ham murakkab Data Structure masalalarini o'yinchoqqa aylantirib qo'yadi. Agar ushbu masalada xodimlarning reytinglari bir xil bo'lib qolishi mumkin bo'lsa (ya'ni duplicate qiymatlar ruxsat etilsa), `less<int>` o'rniga qanday parametr ishlatish qulayroq bo'lar edi?

Albatta! Mana `std::less<int>` o'rniga qanday parametr ishlatish kerakligi haqidagi qisqacha javob:

Agar siz **Indexed Set (PBDS)** ichida bir xil reytingga ega bo'lgan xodimlarni (ya'ni duplicate qiymatlarni) ham saqlamoqchi bo'lsangiz, `less<int>` o'rniga **`less_equal<int>`** ishlatishingiz kerak bo'ladi.

Ya'ni shablon quyidagicha o'zgaradi:

```cpp
typedef tree<int, null_type, less_equal<int>, rb_tree_tag, tree_order_statistics_node_update> indexed_multiset;

```

Bu orqali sizning PBDS tuzilmangiz oddiy `set` dan `multiset` ko'rinishiga o'tadi va barcha operatsiyalar $O(\log N)$ tezlikda takrorlanuvchi elementlar bilan ham to'g'ri ishlayveradi. Mantiq shundaki, u bir xil qiymatlarni xotirada "chap yoki o'ngga" joylashtirishni `>=` (yoki `<=`) deb qabul qilib, alohida tugun sifatida saqlab qoladi.

**Saralash bilan taqqoslash (*Comparison to sorting*)**
Masalani ko‘pincha ma’lumotlar tuzilmalari yoki saralash yordamida hal qilish mumkin. Ba’zida bu yondashuvlarning amaldagi samaradorligida sezilarli farqlar bo‘ladi va bu farqlar ularning vaqt murakkabliklari (*time complexities*) orasida yashiringan bo‘lishi mumkin.

Keling, ikkalasi ham $n$ ta elementdan iborat bo‘lgan A va B ro‘yxatlari berilgan masalani ko‘rib chiqamiz. Bizning vazifamiz ikkala ro‘yxatga ham tegishli bo‘lgan elementlar sonini hisoblashdir. Masalan:
A = [5, 2, 8, 9, 4] va B = [3, 2, 9, 5]
ro‘yxatlari uchun javob 3 ga teng, chunki 2, 5 va 9 sonlari ikkala ro‘yxatda ham mavjud.

Masalaning eng to‘g‘ridan-to‘g‘ri yechimi — bu barcha elementlar juftliklarini $O(n^2)$ vaqtda tekshirib chiqishdir, lekin biz ancha samaraliroq algoritmlarga e’tibor qaratamiz.

**1-algoritm**
A ro‘yxatdagi elementlardan to‘plam (*set*) quramiz va shundan so‘ng B ro‘yxati elementlari bo‘ylab aylanib chiqib, har bir element A da ham bor-yo‘qligini tekshiramiz. Bu samarali usul, chunki A dagi elementlar to‘plamda saqlanmoqda. `set` tuzilmasidan foydalangan holda, algoritmning vaqt murakkabligi $O(n \log n)$ ni tashkil qiladi.

**2-algoritm**
Tartiblangan to‘plamni saqlash shart emas, shuning uchun `set` o‘rniga `unordered_set` tuzilmasidan ham foydalanishimiz mumkin. Bu algoritmni yanada samarali qilishning oson yo‘li, chunki biz faqat asosiy ma’lumotlar tuzilmasini o‘zgartiramiz xolos. Yangi algoritmning vaqt murakkabligi $O(n)$ ga teng.

**3-algoritm**
Ma’lumotlar tuzilmalari o‘rniga biz saralashdan foydalanishimiz mumkin. Birinchi navbatda, A va B ro‘yxatlarini saralaymiz. Shundan so‘ng, ikkala ro‘yxat bo‘ylab bir vaqtning o‘zida aylanib chiqamiz va umumiy elementlarni topamiz. Saralashning vaqt murakkabligi $O(n \log n)$ ga teng, algoritmning qolgan qismi esa $O(n)$ vaqtda ishlaydi, shuning uchun umumiy vaqt murakkabligi $O(n \log n)$ ni tashkil qiladi.

**Samaradorlikni taqqoslash (*Efficiency comparison*)**
Quyidagi jadvalda ro‘yxat elementlari 1...10^9 oralig‘idagi tasodifiy butun sonlar bo‘lganda va n qiymati o‘zgarganda yuqoridagi algoritmlar qanchalik samarali ishlashi ko‘rsatilgan:

| n | 1-algoritm | 2-algoritm | 3-algoritm |
| --- | --- | --- | --- |
| 10^6 | 1.5 s | 0.3 s | 0.2 s |
| 2·10^6 | 3.7 s | 0.8 s | 0.3 s |
| 3·10^6 | 5.7 s | 1.3 s | 0.5 s |
| 4·10^6 | 7.7 s | 1.7 s | 0.7 s |
| 5·10^6 | 10.0 s | 2.3 s | 0.9 s |

1 va 2-algoritmlar faqatgina har xil to‘plam tuzilmalaridan foydalanishi bilan farq qiladi. Ushbu masalada bu tanlov ishlash vaqtiga juda muhim ta’sir ko‘rsatadi, chunki 2-algoritm 1-algoritmga qaraganda 4–5 marta tezroq.

Biroq, eng samarali algoritm bu saralashdan foydalanuvchi 3-algoritmdir. U 2-algoritmga qaraganda atigi yarim barobar kam vaqt sarflaydi. Qizig‘i shundaki, 1-algoritm va 3-algoritmning vaqt murakkabligi bir xil ($O(n \log n)$), lekin shunga qaramay, 3-algoritm o‘n barobar tezroq ishlaydi. Buni shunday tushuntirish mumkinki, saralash ancha sodda jarayon va u 3-algoritmning faqatgina boshida bir marta bajariladi, algoritmning qolgan qismi esa chiziqli vaqtda ishlaydi. Boshqa tomondan, 1-algoritm butun ish jarayoni davomida murakkab muvozanatlangan binar daraxtni saqlab borishga majbur.

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

Ushbu qism kompyuter fanlaridagi Big-O notatsiyasining "yolg'onlari" va dasturlash musobaqalaridagi eng katta sirlardan birini ochib beradi.

**1. Kesh-xotira lokalligi (*Cache Locality*) — Tezlikning asl kaliti**
Nima uchun $O(N)$ da ishlaydigan `unordered_set` $O(N \log N)$ da ishlaydigan oddiy saralash (`sort`) va massivdan 2 marta sekin?
Buning sababi kompyuterning protsessor (CPU) keshida yashiringan. Massivdagi elementlar kompyuter xotirasida **qat'iy ketma-ket** joylashadi. Protsessor massivning birinchi elementini o'qiyotganda, avtomatik ravishda keyingi o'nlab elementlarni ham o'ta tezkor L1 keshiga yuklab oladi.
`set` yoki `unordered_set` kabi tugunlarga (node) asoslangan tuzilmalarda esa elementlar operativ xotirada sochilib yotadi. Protsessor har bir pointer (ko'rsatkich) orqali xotiraning u yer-bu yeriga sakrashga majbur bo'ladi. Bu **Cache Miss** (Keshga tusha olmaslik) deb ataladi va protsessorni juda qattiq sekinlashtiradi.

**2. Constant Factor (Doimiy koeffitsiyent)**
Big-O tahlilida $O(2N)$ ham, $O(50N)$ ham shunchaki $O(N)$ deb yoziladi. Lekin amaliyotda o'sha 50 va 2 ning farqi osmon bilan yercha!
1-algoritm ($O(N \log N)$) har bir elementni daraxtga qo'shish uchun xotiradan yangi joy ajratadi, pointerlarni ulaydi, daraxtning chap va o'ng qismlarini muvozanatlash uchun aylantirishlar (rotations) bajaradi.
3-algoritm ($O(N \log N)$) esa bitta xotira blokida faqatgina oddiy matematik taqqoslash amallarini bajaradi. Yozuvda ikkisi bir xil bo'lsa-da, Constant Factor jihatidan 1-algoritm juda og'ir hisoblanadi.

**3. "Ikki Ko'rsatkich" usuli (*Two Pointers technique*)**
3-algoritmning ikkinchi qismida aytib o'tilgan *"ikkala ro'yxat bo'ylab bir vaqtning o'zida aylanib chiqamiz"* iborasi CP dagi mashhur **Two Pointers** texnikasidir. Ikkita massiv ham o'sish tartibida saralangan bo'lsa, bitta $i$ ko'rsatkichini A massivga, $j$ ko'rsatkichini B massivga qo'yib, faqat bitta tsikl yordamida $O(N)$ tezlikda ularni solishtirib chiqish — Xesh-jadvallarsiz eng kuchli va eng tez xotira-do'st yechim hisoblanadi. Qachonki Codeforces'da masalaning Time Limiti juda qattiq (masalan, 0.5 soniya) bo'lsa, `map` yoki `set` o'rniga doim `std::sort` + *Two Pointers* dan foydalaning.
