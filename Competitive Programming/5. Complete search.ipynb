**To‘liq qidiruv (*Complete search*)**
To‘liq qidiruv (yoki *Brute force*) — bu deyarli har qanday algoritmik masalani hal qilish uchun ishlatilishi mumkin bo‘lgan umumiy usuldir. Uning asosiy g‘oyasi "qo‘pol kuch" (*brute force*) yordamida masalaning barcha ehtimoliy yechimlarini hosil qilish va keyin, masalaning turiga qarab, eng yaxshi yechimni tanlash yoki yechimlar sonini sanab chiqishdan iborat.

Agar barcha yechimlarni ko‘rib chiqish uchun yetarli vaqt bo‘lsa, to‘liq qidiruv juda yaxshi texnika hisoblanadi, chunki qidiruvni dasturlash odatda oson kechadi va u har doim to‘g‘ri javobni kafolatlaydi. Agar to‘liq qidiruv o‘ta sekin ishlasa, u holda ochko‘z algoritmlar (*greedy algorithms*) yoki dinamik dasturlash (*dynamic programming*) kabi boshqa texnikalarni qo‘llashga to‘g‘ri kelishi mumkin.

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

Garchi bu bob eng sodda algoritm haqida bo‘lsa-da, tajribali CP ustalari (Grandmasterlar) *Complete Search* ni o‘zlarining eng kuchli qurollaridan biri deb bilishadi.

**1. Chegaralar barchasini hal qiladi (*Constraints rule*)**
Musobaqada masala shartining eng oxirida berilgan $N$ ning o‘lchami sizga To‘liq qidiruv ishlatish mumkin yoki yo‘qligini "qichqirib" aytib turadi:

* Agar $N \le 10$ bo‘lsa: Hech o‘ylanmang, sizdan barcha o‘rin almashtirishlarni (*permutations*) $O(N!)$ da tekshirish kutilmoqda.
* Agar $N \le 20$ bo‘lsa: Barcha qism-to‘plamlarni (*subsets*) tekshirish $O(2^N)$ kutilyapti (bitmask yoki rekursiya orqali).
* Agar $N \le 100$ bo‘lsa: Uchta ichma-ich tsikl $O(N^3)$ to‘liq qidiruvi o‘tib ketadi.
Agar chegaralar shunday kichik bo‘lsa, aqlli va murakkab algoritm o‘ylashga vaqt sarflamang, shunchaki *Brute force* yozib tashlang!

**2. Stress Testing (Avtomatik testlash)**
Tasavvur qiling, siz murakkab $O(N \log N)$ tezlikda ishlaydigan algoritm yozdingiz, lekin u qayerdadir xato (WA) beryapti. Qayerdaligini topolmaysiz.
Shu paytda CP ustalari darhol o‘sha masala uchun qisqa va aniq ishlaydigan To‘liq qidiruv (*Brute force*) kodini yozishadi (masalan $O(N^3)$ da). Keyin avtomatik test generator yozib, ikkala kodni minglab kichik tasodifiy testlarda yurishtirishadi. Qaysi testda murakkab kodingiz To‘liq qidiruvning javobidan farq qilsa, demak o‘sha test sizning xatoyingizni ushlagan test bo‘ladi. Bunga musobaqalarda **Stress Testing** deyiladi va u *Brute force* siz umuman amalga oshmaydi.

**3. Qirqish san'ati (*Pruning / Backtracking*)**
To‘liq qidiruv har doim ham "ko‘r-ko‘rona" barcha variantlarni tekshirish degani emas. Agar rekursiv qidiruv paytida siz joriy yechim hech qachon to‘g‘ri javobga olib bormasligini tushunib yetsangiz, tsiklni darhol to‘xtatib, orqaga qaytishingiz kerak. Bu usul **Backtracking** va shoxlarni qirqish (*Pruning*) deb ataladi. Yaxshi yozilgan Pruning yordamida $O(2^N)$ deb o‘ylangan algoritmni aslida ancha kattaroq sonlar uchun ham ishlata oladigan darajada optimallashtirish mumkin.

**Qism-to‘plamlarni yaratish (*Generating subsets*)**
Biz dastlab $n$ ta elementdan iborat to‘plamning barcha qism-to‘plamlarni yaratish masalasini ko‘rib chiqamiz. Masalan, $\{0, 1, 2\}$ to‘plamining qism-to‘plamlari quyidagilar: $\emptyset$, $\{0\}$, $\{1\}$, $\{2\}$, $\{0, 1\}$, $\{0, 2\}$, $\{1, 2\}$ va $\{0, 1, 2\}$.
Qism-to‘plamlarni yaratishning ikkita keng tarqalgan usuli mavjud: biz yo rekursiv qidiruvni amalga oshirishimiz mumkin, yoki butun sonlarning bitli ifodalanishidan (*bit representation*) foydalanishimiz mumkin.

**1-usul**
To‘plamning barcha qism-to‘plamlari bo‘ylab aylanib chiqishning nafis usullaridan biri rekursiyadan foydalanishdir. Quyidagi `search` funksiyasi $\{0, 1, \dots, n-1\}$ to‘plamining qism-to‘plamlarini yaratadi. Funksiya har bir qism-to‘plamning elementlarini o‘z ichiga oladigan `subset` (qism-to‘plam) vektorini saqlab boradi. Qidiruv jarayoni funksiya 0 parametri bilan chaqirilganda boshlanadi.

```cpp
void search(int k) {
    if (k == n) {
        // qism-to'plamni qayta ishlash
    } else {
        search(k+1);
        subset.push_back(k);
        search(k+1);
        subset.pop_back();
    }
}

```

`search` funksiyasi $k$ parametri bilan chaqirilganda, u $k$ elementini qism-to‘plamga qo‘shish yoki qo‘shmaslikni hal qiladi va har ikkala holatda ham o‘zini $k+1$ parametri bilan qayta chaqiradi. Biroq, agar $k = n$ bo‘lsa, funksiya barcha elementlar ko‘rib chiqilganini va qism-to‘plam to‘liq shakllanganini anglab yetadi.

Quyidagi daraxt $n=3$ bo‘lgandagi funksiya chaqiruvlarini tasvirlaydi. Biz doimo yo chap shoxni (bu yerda $k$ qism-to‘plamga kiritilmaydi) yoki o‘ng shoxni ($k$ qism-to‘plamga kiritiladi) tanlashimiz mumkin.

```text
                      search(0)
             /                        \
        search(1)                  search(1)
       /         \                /         \
  search(2)   search(2)      search(2)   search(2)
   /    \       /    \        /    \       /    \
 s(3)  s(3)   s(3)  s(3)    s(3)  s(3)   s(3)  s(3)
  ∅     {2}    {1}  {1,2}    {0}  {0,2}  {0,1} {0,1,2}

```

**2-usul**
Qism-to‘plamlarni yaratishning yana bir usuli butun sonlarning bitli ifodalanishiga asoslangan. $n$ ta elementdan iborat to‘plamning har bir qism-to‘plami $n$ ta bitdan iborat ketma-ketlik sifatida ifodalanishi mumkin, va bu $0...2^n-1$ oralig‘idagi butun songa mos keladi. Bitlar ketma-ketligidagi 1 lar qaysi elementlar qism-to‘plamga kiritilganligini ko‘rsatadi.

Odatdagi qoidaga ko‘ra, eng oxirgi bit (o‘ngdan birinchi) 0-elementga, oxiridan bitta oldingi bit 1-elementga mos keladi va hokazo. Masalan, 25 ning bitli ifodasi 11001 bo‘lib, bu $\{0, 3, 4\}$ qism-to‘plamiga mos keladi.

Quyidagi kod $n$ ta elementdan iborat to‘plamning barcha qism-to‘plamlari bo‘ylab aylanib chiqadi:

```cpp
for (int b = 0; b < (1<<n); b++) {
    // qism-to'plamni qayta ishlash
}

```

Quyidagi kod bitlar ketma-ketligiga mos keladigan qism-to‘plam elementlarini qanday ajratib olishimiz mumkinligini ko‘rsatadi. Har bir qism-to‘plamni qayta ishlash jarayonida, kod qism-to‘plamdagi elementlarni o‘zida saqlovchi vektor quradi.

```cpp
for (int b = 0; b < (1<<n); b++) {
    vector<int> subset;
    for (int i = 0; i < n; i++) {
        if (b&(1<<i)) subset.push_back(i);
    }
}

```

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

Ushbu qism barcha musobaqalardagi masalalarning kamida 10% ini tashkil qiluvchi **Bitmask** (Bitli niqob) mavzusiga kirishdir.

**1. Qaysi usul kuchliroq? (Rekursiya vs Bitmask)**
Ikkala usul ham $O(2^N)$ tezlikda ishlaydi, biroq Bitmask usuli (2-usul) doimo ustunroq va CP ustalari tomonidan afzal ko'riladi. Nima uchun?

1. U tezroq ishlaydi (Constant Factor), chunki kompyuter bitli amallarni (`<<`, `&`) oddiy matematik hisob-kitoblarga yoki rekursiv chaqiruvlarga qaraganda ancha tezroq bajaradi.
2. Kod yozish hajmi 2 barobar qisqa. Hech qanday qo‘shimcha funksiya yoki global o‘zgaruvchilar ochish shart emas.

**2. Sehrli operatorlar sirqanadi: `(1<<n)` nima degani?**
Kitobda `(1<<n)` ishlatilgan. Bu C++ tilida $2^n$ ni hisoblashning eng tezkor va eng xavfsiz yo‘lidir (hech qachon `pow(2, n)` ishlatmang, u double qiymat qaytaradi va aniqlikni yo‘qotadi!). Agar sizda 3 ta element bo‘lsa, `1<<3` = 8. Tsikl 0 dan 7 gacha aylanadi va to‘liq 8 ta qism-to‘plamni hosil qilib beradi.

**3. Bitlarni tekshirish logikasi `if (b&(1<<i))**`
Bu CP dagi eng ko‘p yoziladigan qatorlardan biri. U "b sonining o'ngdan i-chi biti 1 ga tengmi?" degan savolga javob beradi.
Agar sizda $b = 5$ (iklikda: `101`) bo‘lsa va 0-chi elementni tekshirayotgan bo‘lsangiz:
`1<<0` bu `001`. `101 & 001 = 001` (Rost, demak 0-chi element qism-to‘plamda bor).
2-chi elementni tekshirganda:
`1<<2` bu `100`. `101 & 100 = 100` (Rost, demak 2-chi element ham bor).
Lekin 1-chi elementni tekshirganda `000` (Yolg‘on) chiqadi.
Shunday qilib bu oddiy matematik sikl to‘plamlarni mukammal avtomatlashtiradi.

**O‘rin almashtirishlarni yaratish (*Generating permutations*)**
Endi biz $n$ ta elementdan iborat to‘plamning barcha o‘rin almashtirishlarini (permutatsiyalarini) yaratish masalasini ko‘rib chiqamiz. Masalan, $\{0, 1, 2\}$ ning o‘rin almashtirishlari quyidagilar: $(0, 1, 2)$, $(0, 2, 1)$, $(1, 0, 2)$, $(1, 2, 0)$, $(2, 0, 1)$ va $(2, 1, 0)$. Yana ikkita yondashuv mavjud: biz rekursiyadan foydalanishimiz yoki o‘rin almashtirishlar bo‘ylab iterativ tarzda (tsikl orqali) aylanib chiqishimiz mumkin.

**1-usul**
Qism-to‘plamlar singari, o‘rin almashtirishlarni ham rekursiya yordamida yaratish mumkin. Quyidagi `search` funksiyasi $\{0, 1, \dots, n-1\}$ to‘plamining o‘rin almashtirishlari bo‘ylab aylanib chiqadi. Funksiya o‘rin almashtirishni o‘zida saqlovchi `permutation` vektorini quradi va qidiruv jarayoni funksiya parametrlarsiz chaqirilganda boshlanadi.

```cpp
void search() {
    if (permutation.size() == n) {
        // o'rin almashtirishni qayta ishlash
    } else {
        for (int i = 0; i < n; i++) {
            if (chosen[i]) continue;
            chosen[i] = true;
            permutation.push_back(i);
            search();
            chosen[i] = false;
            permutation.pop_back();
        }
    }
}

```

Har bir funksiya chaqiruvi `permutation` vektoriga yangi element qo‘shadi. `chosen` (tanlangan) massivi qaysi elementlar o‘rin almashtirishga allaqachon kiritilganligini ko‘rsatadi. Agar `permutation` vektorining o‘lchami to‘plam o‘lchamiga teng bo‘lsa, bitta o‘rin almashtirish to‘liq yaratilgan bo‘ladi.

**2-usul**
O‘rin almashtirishlarni yaratishning yana bir usuli — $\{0, 1, \dots, n-1\}$ o‘rin almashtirishidan boshlash va doimiy ravishda o‘sish (leksikografik) tartibidagi keyingi o‘rin almashtirishni tuzadigan funksiyadan foydalanishdir. C++ standart kutubxonasida buning uchun ishlatilishi mumkin bo‘lgan `next_permutation` funksiyasi mavjud:

```cpp
vector<int> permutation;
for (int i = 0; i < n; i++) {
    permutation.push_back(i);
}
do {
    // o'rin almashtirishni qayta ishlash
} while (next_permutation(permutation.begin(), permutation.end()));

```

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

Bu qismda C++ ning eng katta "cheat-code" (yashirin ustunlik) laridan biri yashiringan!

**1. Nima uchun musobaqalarda 1-usulni (Rekursiya) deyarli hech kim ishlatmaydi?**
O‘rin almashtirishlar hosil qilishda hech qachon noldan rekursiya yozib yurmang (agar masala N-farzin kabi qattiq shoxlarni qirqishni (pruning) talab qilmasa). `std::next_permutation` — bu CP olamidagi mo‘jizaviy funksiyadir. U nafaqat tezroq ishlaydi, balki eng muhim bir muammoni avtomatik hal qiladi: **Takrorlanuvchi elementlar**.
Agar sizda `{1, 1, 2}` massivi bo‘lsa va 1-usulni (rekursiya) ishlatsangiz, u sizga bir xil kombinatsiyalarni qayta-qayta chiqarib, jami 6 ta javob beradi. Ammo `next_permutation` matematik jihatdan mukammal tuzilgan — u faqat takrorlanmas noyob (unique) kombinatsiyalarnigina (`1 1 2`, `1 2 1`, `2 1 1`) $O(1)$ amortized tezlikda ajratib beradi!

**2. Oltin qoidalar: Nega `do-while` va nega oldin `sort`?**
Kitobdagi 2-usulda aynan `do-while` tsikli ishlatilganiga e’tibor bering.
Agar siz odatiy `while(next_permutation(...))` qilsangiz, tsikl eng birinchi (boshlang‘ich) holatni tekshirmasdan sakrab o‘tib ketadi va siz 1 ta javobni yo‘qotasiz. Shuningdek, `next_permutation` doim **leksikografik (alifbo) o‘sish** tartibida keyingi qadamni topadi. Shuning uchun barcha kombinatsiyalarni ko‘rish uchun massiv **eng boshida o‘sish tartibida saralangan (`sort`) bo‘lishi shart!** Aks holda to‘liq ro‘yxat chiymaydi.

**3. $O(N!)$ qachon TLE (Time Limit Exceeded) beradi?**
Vaqt chegarasini (1 soniya) his qilishingiz uchun faktorial tezliklarni yodlab oling:

* $N = 10 \implies 10! = 3,628,800$ (Bemalol 0.1 soniyada o‘tadi)
* $N = 11 \implies 11! \approx 3.9 \times 10^7$ (1 soniya atrofida arang o‘tadi)
* $N \ge 12 \implies 12! \approx 4.7 \times 10^8$ (**Kafolatlangan TLE!**)
Demak, musobaqada $N = 11$ yoki $12$ berilsa, hech qachon to‘liq permutatsiya qilmang, bu masalada DP (Dynamic Programming with Bitmask) yoki Meet-in-the-Middle yashiringan bo‘ladi.

**Orqaga qaytish (*Backtracking*)**
Backtracking algoritmi bo‘sh yechimdan boshlanadi va yechimni qadamma-qadam kengaytirib boradi. Qidiruv jarayoni yechimni qurish mumkin bo‘lgan barcha turli xil usullar bo‘ylab rekursiv tarzda aylanib chiqadi.

Misol tariqasida, $n \times n$ o‘lchamdagi shaxmat doskasiga $n$ ta farzinni (*queen*) hech qaysi ikkitasi bir-biriga hujum qilmaydigan tarzda joylashtirish usullari sonini hisoblash masalasini ko‘rib chiqaylik. Masalan, $n = 4$ bo‘lganda, ikkita ehtimoliy yechim mavjud:

```text
. Q . .        . . Q .
. . . Q        Q . . .
Q . . .        . . . Q
. . Q .        . Q . .

```

Bu masalani backtracking yordamida farzinlarni doskaga qatorma-qator joylashtirish orqali hal qilish mumkin. Aniqroq aytganda, har bir qatorga aynan bitta farzin joylashtiriladi, bunda u o‘zidan oldin joylashtirilgan farzinlarning hech biriga hujum qilmasligi kerak. Barcha $n$ ta farzin doskaga joylashtirib bo‘linganda bitta to‘liq yechim topilgan bo‘ladi.

Masalan, $n = 4$ bo‘lganda, backtracking algoritmi tomonidan hosil qilingan ba’zi chala (qisman) yechimlar quyidagicha ko‘rinishga ega:

```text
Q . . .    Q . . .    Q . . .    . Q . .
Q . . .    . Q . .    . . Q .    . . . Q
. . . .    . . . .    . . . .    . . . .
. . . .    . . . .    . . . .    . . . .
 xato       xato       xato      to'g'ri

```

Eng quyi darajada, dastlabki uchta holat yaroqsiz (*illegal*), chunki farzinlar bir-biriga hujum qilyapti. Biroq, to‘rtinchi holat yaroqli (*valid*) va unga yana ikkita farzin qo‘shish orqali uni to‘liq yechimgacha kengaytirish mumkin. Qolgan ikkita farzinni joylashtirishning faqat bitta usuli mavjud.

Algoritmni quyidagicha dasturlash mumkin:

```cpp
void search(int y) {
    if (y == n) {
        count++;
        return;
    }
    for (int x = 0; x < n; x++) {
        if (column[x] || diag1[x+y] || diag2[x-y+n-1]) continue;
        
        column[x] = diag1[x+y] = diag2[x-y+n-1] = 1;
        search(y+1);
        column[x] = diag1[x+y] = diag2[x-y+n-1] = 0;
    }
}

```

Qidiruv `search(0)` ni chaqirish bilan boshlanadi. Doskaning o‘lchami $n \times n$ bo‘lib, kod yechimlar sonini `count` o‘zgaruvchisiga hisoblab boradi.

Kod doskaning qator va ustunlari $0$ dan $n-1$ gacha raqamlangan deb faraz qiladi. `search` funksiyasi $y$ parametri bilan chaqirilganda, u $y$-qatorga farzin joylashtiradi va keyin o‘zini $y+1$ parametri bilan qayta chaqiradi. Agar $y = n$ ga teng bo‘lsa, yechim topilgan hisoblanadi va `count` o‘zgaruvchisi bittaga oshiriladi.

`column` massivi farzin joylashgan ustunlarni hisobga olib boradi, `diag1` va `diag2` massivlari esa diagonallarni kuzatib boradi. Allaqachon farzin joylashtirilgan ustun yoki diagonalga yana boshqa farzin qo‘shishga ruxsat berilmaydi. Masalan, $4 \times 4$ doskaning ustunlari va diagonallari quyidagicha raqamlanadi:

```text
0 1 2 3      0 1 2 3      3 4 5 6
0 1 2 3      1 2 3 4      2 3 4 5
0 1 2 3      2 3 4 5      1 2 3 4
0 1 2 3      3 4 5 6      0 1 2 3
column        diag1        diag2

```

Faraz qilaylik, $q(n)$ ifodasi $n \times n$ shaxmat doskasiga $n$ ta farzinni joylashtirish usullari sonini bildirsin. Yuqoridagi backtracking algoritmi bizga, masalan, $q(8) = 92$ ekanligini ko‘rsatadi. $n$ ortgan sari qidiruv tezda sekinlashadi, chunki yechimlar soni eksponensial ravishda o‘sib boradi. Masalan, yuqoridagi algoritm yordamida $q(16) = 14,772,512$ ekanligini hisoblash zamonaviy kompyuterda taxminan bir daqiqa vaqt oladi.

Bu bob algoritmik fikrlashning eng cho‘qqisi hisoblangan Backtracking (Orqaga qaytish) paradigmasini tushuntiradi.

**1. Qulf va Kalit (State Reversion)**
Backtracking'ning eng muhim qatori bu rekursiyadan qaytishdagi holatni tozalashdir:
`column[x] = diag1[...] = diag2[...] = 0;`
Agar siz qidiruvning chuquriga kirayotganda holatni o‘zgartirsangiz (1 qilib qulflasangiz), u yerdan javob topib yoki xatoga uchrab orqaga qaytganingizda, xuddi shu qulfni **albatta ochib qo‘yishingiz (0 qilishingiz)** shart. **2. Diagonallar arifmetikasi (Matematik Fokus)**
2D matritsada diagonallarni tekshirish ko‘pchilik uchun bosh og‘rig‘i. Lekin kitobdagi formula buni $O(1)$ da hal qiladi:

* **Chap-pastdan O‘ng-tepaga ketuvchi diagonallar (`diag1`):** Har doim qator va ustun indekslari yig‘indisi ($x + y$) bir xil bo‘ladi.
* **Chap-tepadan O‘ng-pastga ketuvchi diagonallar (`diag2`):** Har doim ularning ayirmasi ($x - y$) bir xil bo‘ladi. Manfiy indeks bo‘lib qolmasligi uchun unga oddiygina $+ (n-1)$ qo‘shib qo‘yiladi.
Bu matematik faktni yodda saqlang, u 2D Grid (To‘r) masalalarida yuzlab marta asqotadi.

Bitli operatsiyalar orqali $q(16)$ ni bir daqiqada emas, balki soniyaning bir qismida hisoblab chiqish mumkin. Backtracking va Bitmask musobaqalarda har doim birga yuruvchi mukammal juftlikdir.

**Qidiruvni qirqish (*Pruning the search*)**
Ko‘pincha qidiruv daraxtini qirqish orqali orqaga qaytishni (backtracking) optimallashtirishimiz mumkin. G‘oya algoritmga "aql" qo‘shishdan iboratki, u qisman yechimni to‘liq yechimgacha kengaytirib bo‘lmasligini imkon qadar tezroq sezsin. Bunday optimizatsiyalar qidiruv samaradorligiga ulkan ta’sir ko‘rsatishi mumkin.
*(Izoh: $q(n)$ ning kattaroq qiymatlarini samarali hisoblashning ma’lum usuli yo‘q. Hozirgi rekord $q(27) = 234907967154122528$ bo‘lib, u 2016-yilda hisoblangan).*

Keling, $n \times n$ to‘rdagi (grid) yuqori-chap burchakdan pastki-o‘ng burchakkacha bo‘lgan va har bir katakka aynan bir marta tashrif buyuradigan yo‘llar (Gamilton yo‘llari) sonini hisoblash masalasini ko‘rib chiqamiz. Masalan, $7 \times 7$ to‘rda 111,712 ta shunday yo‘l mavjud.

Biz $7 \times 7$ holatiga e’tibor qaratamiz, chunki uning qiyinchilik darajasi bizning ehtiyojlarimizga mos keladi. Biz eng to‘g‘ridan-to‘g‘ri backtracking algoritmidan boshlaymiz va qidiruvni qanday qilib qirqish (*prune*) mumkinligi haqidagi kuzatishlar yordamida uni qadamma-qadam optimallashtiramiz. Har bir optimizatsiyadan so‘ng, qidiruv samaradorligiga ta’sirni aniq ko‘rish uchun algoritmning ishlash vaqtini va rekursiv chaqiruvlar sonini o‘lchaymiz.

**Asosiy algoritm (*Basic algorithm*)**
Algoritmning birinchi versiyasi hech qanday optimizatsiyani o‘z ichiga olmaydi. Biz shunchaki yuqori-chap burchakdan pastki-o‘ng burchakkacha bo‘lgan barcha mumkin bo‘lgan yo‘llarni yaratish va ularning sonini sanash uchun backtracking’dan foydalanamiz.

* ishlash vaqti: 483 soniya
* rekursiv chaqiruvlar soni: 76 milliard

**1-optimizatsiya**
Har qanday yechimda biz birinchi qadamni pastga yoki o‘ngga tashlaymiz. Birinchi qadamdan so‘ng to‘rning diagonali bo‘ylab doimo ikkita simmetrik yo‘l bo‘ladi.
Shu sababli, biz har doim birinchi bo‘lib bir qadam pastga (yoki o‘ngga) yurishni belgilab olishimiz va yakunda topilgan yechimlar sonini ikkiga ko‘paytirishimiz mumkin.

* ishlash vaqti: 244 soniya
* rekursiv chaqiruvlar soni: 38 milliard

**2-optimizatsiya**
Agar yo‘l to‘rdagi barcha boshqa kataklarga tashrif buyurmasdan turib pastki-o‘ng katakka yetib borsa, yechimni oxiriga yetkazib bo‘lmasligi aniq. Ushbu kuzatishdan foydalanib, agar pastki-o‘ng katakka juda erta yetib borsak, qidiruvni darhol to‘xtatishimiz mumkin.

* ishlash vaqti: 119 soniya
* rekursiv chaqiruvlar soni: 20 milliard

**3-optimizatsiya**
Agar yo‘l devorga borib taqalib, faqat chapga yoki o‘ngga burilishi mumkin bo‘lsa, u holda to‘r hali tashrif buyurilmagan kataklarni o‘z ichiga olgan ikkita alohida qismga bo‘linib qoladi. Bunday holda, biz endi barcha kataklarga tashrif buyura olmaymiz, shuning uchun qidiruvni to‘xtatishimiz mumkin. Bu juda foydali optimizatsiya:

* ishlash vaqti: 1.8 soniya
* rekursiv chaqiruvlar soni: 221 million

**4-optimizatsiya**
3-optimizatsiya g‘oyasini umumlashtirish mumkin: agar yo‘l oldinga davom eta olmasa, lekin chapga yoki o‘ngga burilishi mumkin bo‘lsa, u holda to‘r tashrif buyurilmagan kataklarga ega ikkita qismga ajraladi. Bunday holda qidiruvni darhol to‘xtatamiz. Ushbu optimizatsiyadan so‘ng qidiruv o‘ta samarali ishlaydi:

* ishlash vaqti: 0.6 soniya
* rekursiv chaqiruvlar soni: 69 million

Endi algoritmni optimallashtirishni to‘xtatib, nimaga erishganimizni ko‘rishning ayni vaqti. Dastlabki algoritmning ishlash vaqti 483 soniya edi, optimizatsiyalardan keyin esa ishlash vaqti atigi 0.6 soniyani tashkil qilyapti. Shunday qilib, optimizatsiyalar natijasida algoritm deyarli **1000 marta tezlashdi**.

Bu backtracking algoritmlarida tez-tez uchraydigan holat, chunki qidiruv daraxti odatda juda katta bo‘ladi va hatto oddiy kuzatishlar ham qidiruvni samarali qirqishga (qisqartirishga) imkon beradi. Ayniqsa algoritmning dastlabki qadamlarida, ya’ni qidiruv daraxtining yuqori qismida yuz beradigan optimizatsiyalar eng katta foyda keltiradi.

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

Ushbu qism masalalarni "Time Limit" (TLE) dan qutqarib qolishning eng amaliy sirlarini ochib beradi.

**1. Simmetriyani buzish (*Symmetry Breaking*)**
1-optimizatsiya CP dagi o‘ta kuchli usuldir. Agar masala to‘liq simmetrik bo‘lsa (masalan, matritsada xoh chapdan, xoh o‘ngdan boshlash bir xil natija bersa), qidiruvni ikkala tomonga ham yurgizish shart emas. Bir tomonni qat’iy qulflab (*hardcode* qilib), chiqqan natijani 2 ga ko‘paytirish orqali ishlash vaqtini va xotirani bir soniya ichida **50% ga** qisqartirasiz.

**2. Ulanishni yo‘qotish (*Connectivity / Articulation Points*)**
3 va 4-optimizatsiyalar aslida Graflar nazariyasidagi o‘ziga xos qoidaga asoslanadi. Agar siz yo‘lda ketyotib matritsani "ikkiga kesib" qo‘ysangiz (shunday joyga bossangizki, natijada yurmagingiz kerak bo‘lgan bo‘sh hujralar devor bilan ikkiga ajralib qolsa), bu shoxda qidiruvni davom ettirish mantiqsizdir. Chunki o‘yin qoidasiga ko‘ra har bir katakka faqat 1 marta kiriladi, siz esa bitta yo‘l bilan bir vaqtning o‘zida ajralib qolgan ikkala qismga kira olmaysiz. Buni tekshirish kodingizga atigi 2-3 qator `if` qo‘shishni talab qiladi, lekin natijada funksiya chaqiruvlari 20 milliarddan 69 millionga (300 barobar) qisqaradi!

**3. Tepadan qirqish qoidasi (*Top-level Pruning*)**
Daraxtning eng pastki barglariga borganda xatoni tushunish sizga yuzlab ortiqcha iteratsiyalarga tushadi. Ammo shartni daraxtning eng yuqori qismidayoq (birinchi 2-3 qadamda) tekshirib qirqsangiz (Prune), shu bitta tekshiruv ostidagi millionlab ehtimoliy shoxchalar xotiradan butunlay o‘chib ketadi. Shuning uchun optimizatsiya shartlarini (*if-statements*) rekursiya funksiyasining doimo **eng boshiga** (qadamlar boshlanishidan ham oldinga) yozish shart.


---


**O‘rtada uchrashish (*Meet in the middle*)**
O‘rtada uchrashish — bu qidiruv maydoni taxminan teng ikkita qismga bo‘linadigan texnikadir. Ikkala qism uchun ham alohida qidiruv amalga oshiriladi va yakunda qidiruv natijalari birlashtiriladi.

Agar qidiruv natijalarini birlashtirishning samarali usuli mavjud bo‘lsa, ushbu texnikadan foydalanish mumkin. Bunday vaziyatda, bitta katta qidiruvga qaraganda ikkita qidiruv kamroq vaqt talab qilishi mumkin. Odatda, o‘rtada uchrashish texnikasidan foydalanib biz $2^n$ koeffitsiyentini $2^{n/2}$ ga aylantirishimiz mumkin.

Misol tariqasida, bizga $n$ ta sondan iborat ro‘yxat va $x$ soni berilgan va biz ro‘yxatdan yig‘indisi $x$ ga teng bo‘ladigan qilib bir nechta sonlarni tanlab olish mumkin yoki yo‘qligini aniqlash masalasini ko‘rib chiqaylik. Masalan, `[2, 4, 5, 9]` ro‘yxati va $x = 15$ berilgan bo‘lsa, biz $2 + 4 + 9 = 15$ hosil qilish uchun `[2, 4, 9]` sonlarini tanlashimiz mumkin. Biroq, xuddi shu ro‘yxat uchun $x = 10$ bo‘lsa, bunday yig‘indini hosil qilib bo‘lmaydi.

Masalaning oddiy algoritmi elementlarning barcha qism-to‘plamlari bo‘ylab aylanib chiqish va ulardan birortasining yig‘indisi $x$ ga teng yoki yo‘qligini tekshirishdir. Bunday algoritmning ishlash vaqti $O(2^n)$ ga teng, chunki jami $2^n$ ta qism-to‘plam mavjud. Biroq, o‘rtada uchrashish texnikasi yordamida biz ancha samarali bo‘lgan $O(2^{n/2})$ vaqtli algoritmga erishishimiz mumkin. E’tibor bering, $O(2^n)$ va $O(2^{n/2})$ har xil murakkabliklardir, chunki $2^{n/2}$ ifodasi $\sqrt{2^n}$ ga tengdir. *(Izoh: Bu g‘oya 1974 yilda E. Horowitz va S. Sahni tomonidan kiritilgan).*

Asosiy g‘oya ro‘yxatni ikkala qismi ham sonlarning taxminan yarmini o‘z ichiga oladigan `A` va `B` ro‘yxatlarga bo‘lishdir. Birinchi qidiruv `A` ning barcha qism-to‘plamlarini hosil qiladi va ularning yig‘indilarini `SA` ro‘yxatida saqlaydi. Mos ravishda, ikkinchi qidiruv `B` dan `SB` ro‘yxatini yaratadi. Shundan so‘ng, ularning yig‘indisi $x$ ga teng bo‘ladigan qilib `SA` dan bitta elementni va `SB` dan boshqa elementni tanlash mumkin yoki yo‘qligini tekshirish kifoya. Bu faqatgina dastlabki ro‘yxat sonlari yordamida $x$ yig‘indisini hosil qilish usuli mavjud bo‘lgandagina mumkin bo‘ladi.

Masalan, ro‘yxat `[2, 4, 5, 9]` va $x = 15$ deb faraz qilaylik. Dastlab biz ro‘yxatni `A = [2, 4]` va `B = [5, 9]` ga bo‘lamiz. Shundan so‘ng biz `SA = [0, 2, 4, 6]` va `SB = [0, 5, 9, 14]` ro‘yxatlarini yaratamiz. Ushbu holatda $x = 15$ yig‘indisini hosil qilish mumkin, chunki `SA` da 6 yig‘indisi bor, `SB` da 9 yig‘indisi bor va $6 + 9 = 15$. Bu `[2, 4, 9]` yechimiga mos keladi.

Biz algoritmni uning vaqt murakkabligi $O(2^{n/2})$ bo‘ladigan qilib amalga oshirishimiz mumkin. Birinchi navbatda, saralangan `SA` va `SB` ro‘yxatlarini yaratamiz, buni qo‘shilishga (merge) o‘xshash texnika yordamida $O(2^{n/2})$ vaqtda bajarish mumkin. Shundan so‘ng, ro‘yxatlar saralangan bo‘lgani uchun, `SA` va `SB` orqali $x$ yig‘indisini hosil qilish mumkin yoki yo‘qligini $O(2^{n/2})$ vaqtda tekshirib chiqishimiz mumkin.

### 💡 Sport Dasturlashdagi (CP) o‘ta muhim xulosalar

**1. Qachon "Meet in the Middle" ishlatiladi? (Sehrli $N=40$ chegarasi)**
Musobaqalarda masalaning chegaralari (Constraints) sizga yechimni topishga katta yordam beradi.
Agar masala "Subset" (qism-to‘plam) ga oid bo‘lsa va **$N \le 20$** bo‘lsa — siz oddiy Bitmask yoki Recursion ishlata olasiz ($O(2^N) \approx 10^6$ operatsiya).
Lekin to‘satdan **$N \le 40$** berilib qolsa, $2^{40} \approx 10^{12}$ bo‘lib qoladi va TLE (Time Limit Exceeded) olasiz. Aynan $N=40$ yoki $N=36$ kabi chegaralarni ko‘rishingiz bilan miyangizda darhol **Meet in the Middle** yonishi kerak. Chunki $2^{40/2} = 2^{20} \approx 1,048,576$ bo‘lib, u yana qaytib $0.1$ soniyada ishlaydigan mukammal algoritmga aylanadi.

**2. Ikkita massivni ulash algoritmlari**
Siz $SA$ va $SB$ massivlarini yaratganingizdan so‘ng, ular ichidan yig‘indisi $X$ ga teng bo‘lgan juftlikni topishingiz kerak. Buning eng mashhur 2 ta CP usuli bor:

* **Binar qidiruv (Binary Search):** $SB$ massivni saralaysiz (`sort`). Keyin $SA$ bo‘ylab bittadan aylanib chiqib, $SB$ ichidan `std::binary_search` yoki `std::lower_bound` yordamida $(X - SA[i])$ ni qidirasiz.
* **Ikki ko‘rsatkich (Two Pointers):** Ikkala massivni ham saralaysiz. Bitta ko‘rsatkichni $SA$ ning boshiga, ikkinchisini $SB$ ning oxiriga qo‘yasiz. Agar yig‘indi $X$ dan katta bo‘lsa, $SB$ ko‘rsatkichini bitta chapga surasiz, kichik bo‘lsa $SA$ ni bitta o‘ngga surasiz. Bu usul Binar qidiruvdan ham tezroq ishlaydi!

**3. "Subset Sum" (Knapsack-dagi ekvivalenti)**
Matnda keltirilgan masala aslida mashhur *Subset Sum* muammosidir. Odatda bu Dinamik Dasturlash (DP) orqali hal qilinadi. DP yechimining vaqti $O(N \cdot X)$ ga teng. Agar $X$ (izlanayotgan yig‘indi) kichik bo‘lsa ($10^5$ atrofida), DP ishlatgan ma'qul. Lekin CP masalalarida g‘irromlik qilishib, $X = 10^{15}$ gacha berib yuborishadi. Bunda xotiraga $10^{15}$ sig‘maydi (Memory Limit) va DP qulab tushadi. Ana shunday vaziyatda **Meet in the Middle** yagona qutqaruvchi algoritm hisoblanadi.
